```python

# ============================================================
# KITCHEN ASSISTANT V2.8.1 — U02
# Stage 03 + Stage 04 + Stage 05
# Runtime Foundation + Flask/WSGI + Migration Readiness
#
# CANONICAL SCANNER ROADMAP V2
# U01 MUST already be completed in this same Colab project.
#
# IMPORTANT:
# - No scanner intelligence
# - No market-data provider
# - No live market ingestion
# - No trading logic
# - No order execution
# - No business rules
#
# TIMEFRAME / RANGE GUARD:
# - Technical validation boundary ONLY
# - Every valid timeframe below D1 receives a WARNING
# - This applies to standard timeframes AND Custom Range
# - A sub-daily timeframe/range is NOT an execution error
# - D1 and higher do NOT receive the sub-daily warning
# - Invalid/unrecognized timeframe input remains invalid
# - No hardcoded list of only selected low timeframes is used
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
import sys
import os
import subprocess
import json
import hashlib


# ============================================================
# 0) PROJECT ROOT + U01 VERIFICATION
# ============================================================

ROOT = Path("/content/kitchen_robot_v27")

if not ROOT.exists():
    raise RuntimeError(
        "U01 project not found. Run U01 first."
    )

os.chdir(ROOT)

if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))


print("=" * 60)
print("KITCHEN ASSISTANT V2.8.1")
print("Execution Unit : U02")
print("Master Stages  : 03 + 04 + 05")
print("Python         :", sys.version.split()[0])
print("Project Root   :", ROOT)
print("=" * 60)


# ============================================================
# 1) RUNTIME DEPENDENCIES
# ============================================================

subprocess.check_call(
    [
        sys.executable,
        "-m",
        "pip",
        "install",
        "-q",
        "Flask>=3,<4",
        "gunicorn>=23,<24",
        "pytest>=8,<9",
    ]
)

print("Runtime dependencies: READY")


# ============================================================
# 2) FLASK APPLICATION
#
# Technical runtime boundary only.
# No scanner intelligence is implemented here.
# ============================================================

runtime_code = '''
from flask import Flask, jsonify
from datetime import datetime

from app.core.version import (
    PROJECT_NAME,
    PROJECT_VERSION,
    EXECUTION_UNIT,
)


# ============================================================
# TIMEFRAME / RANGE VALIDATION
# ============================================================

STANDARD_TIMEFRAMES = {
    "M1": 60,
    "M5": 5 * 60,
    "M15": 15 * 60,
    "H1": 60 * 60,
    "H4": 4 * 60 * 60,
    "D1": 24 * 60 * 60,
    "D7": 7 * 24 * 60 * 60,
}


SUB_DAILY_WARNING = (
    "⚠️ WARNING: The selected timeframe/range is below daily (D1). "
    "Sub-daily data may contain higher noise and lower structural "
    "reliability. Structural market interpretation should be treated "
    "with caution and confirmed with daily (D1) or higher data."
)


def _parse_custom_range(value: str):
    """
    Parse the canonical Custom Range format:

        DD-MM-YYYY HH:MM

    Returns a datetime object or None.
    """

    try:
        return datetime.strptime(
            value.strip(),
            "%d-%m-%Y %H:%M",
        )
    except (TypeError, ValueError):
        return None


def validate_timeframe_guard(
    timeframe: str,
    custom_start: str = None,
    custom_end: str = None,
) -> dict:
    """
    Technical timeframe/range validation boundary.

    This function does NOT perform market analysis.

    Rules:
    1. Every recognized standard timeframe below D1 is VALID
       but receives a WARNING.
    2. D1 and higher are VALID without the sub-daily warning.
    3. Custom Range is supported using:
           DD-MM-YYYY HH:MM
       If the selected range is below 24 hours, it is VALID
       but receives the same sub-daily WARNING.
    4. A Custom Range of 24 hours or more is VALID without
       the sub-daily warning.
    5. Invalid or incomplete timeframe/range input remains INVALID.
    6. No specific hardcoded list of low timeframes is used
       to determine the warning.
    """

    if not isinstance(timeframe, str):
        return {
            "valid": False,
            "warning": None,
            "reason": "INVALID_TIMEFRAME_TYPE",
            "is_sub_daily": False,
        }

    normalized = timeframe.strip().upper()

    # --------------------------------------------------------
    # Standard Timeframe
    # --------------------------------------------------------

    if normalized in STANDARD_TIMEFRAMES:

        duration_seconds = STANDARD_TIMEFRAMES[
            normalized
        ]

        is_sub_daily = (
            duration_seconds < 24 * 60 * 60
        )

        if is_sub_daily:

            return {
                "valid": True,
                "warning": SUB_DAILY_WARNING,
                "reason": "SUB_DAILY_TIMEFRAME",
                "is_sub_daily": True,
            }

        return {
            "valid": True,
            "warning": None,
            "reason": "DAILY_OR_HIGHER",
            "is_sub_daily": False,
        }

    # --------------------------------------------------------
    # Custom Range
    # --------------------------------------------------------

    if normalized in {
        "CUSTOM",
        "CUSTOM_RANGE",
        "CUSTOM RANGE",
    }:

        if not custom_start or not custom_end:

            return {
                "valid": False,
                "warning": None,
                "reason": "CUSTOM_RANGE_ENDPOINTS_REQUIRED",
                "is_sub_daily": False,
            }

        start = _parse_custom_range(
            custom_start
        )

        end = _parse_custom_range(
            custom_end
        )

        if start is None or end is None:

            return {
                "valid": False,
                "warning": None,
                "reason": "INVALID_CUSTOM_RANGE_FORMAT",
                "is_sub_daily": False,
            }

        if end <= start:

            return {
                "valid": False,
                "warning": None,
                "reason": "INVALID_CUSTOM_RANGE_ORDER",
                "is_sub_daily": False,
            }

        duration_seconds = (
            end - start
        ).total_seconds()

        is_sub_daily = (
            duration_seconds < 24 * 60 * 60
        )

        if is_sub_daily:

            return {
                "valid": True,
                "warning": SUB_DAILY_WARNING,
                "reason": "SUB_DAILY_CUSTOM_RANGE",
                "is_sub_daily": True,
                "duration_seconds": duration_seconds,
            }

        return {
            "valid": True,
            "warning": None,
            "reason": "DAILY_OR_HIGHER_CUSTOM_RANGE",
            "is_sub_daily": False,
            "duration_seconds": duration_seconds,
        }

    # --------------------------------------------------------
    # Invalid / unsupported timeframe
    # --------------------------------------------------------

    return {
        "valid": False,
        "warning": None,
        "reason": "UNSUPPORTED_TIMEFRAME",
        "is_sub_daily": False,
    }


def create_app():

    app = Flask(__name__)


    @app.get("/health")
    def health():

        return jsonify({
            "status": "ok",
            "project": PROJECT_NAME,
            "version": PROJECT_VERSION,
            "execution_unit": EXECUTION_UNIT,
        })


    @app.get("/")
    def root():

        return jsonify({
            "service": PROJECT_NAME,
            "version": PROJECT_VERSION,
            "status": "running",
        })


    return app


app = create_app()
'''

(ROOT / "app/api/app.py").write_text(
    runtime_code.strip() + "\n",
    encoding="utf-8",
)


# ============================================================
# 3) WSGI ENTRYPOINT
# ============================================================

wsgi_code = '''
from app.api.app import app

application = app
'''

(ROOT / "wsgi.py").write_text(
    wsgi_code.strip() + "\n",
    encoding="utf-8",
)


# ============================================================
# 4) RUNTIME CONFIGURATION
# ============================================================

runtime_config = '''
import os


HOST = os.getenv(
    "APP_HOST",
    "0.0.0.0",
)


PORT = int(
    os.getenv(
        "APP_PORT",
        "5000",
    )
)


DEBUG = os.getenv(
    "APP_DEBUG",
    "0",
) == "1"
'''

(ROOT / "app/config/runtime.py").write_text(
    runtime_config.strip() + "\n",
    encoding="utf-8",
)


# ============================================================
# 5) PYTHONANYWHERE-COMPATIBLE START SCRIPT
# ============================================================

start_script = '''#!/bin/bash

exec gunicorn \\
    --bind 0.0.0.0:5000 \\
    wsgi:application
'''

(ROOT / "scripts/start_wsgi.sh").write_text(
    start_script,
    encoding="utf-8",
)

(ROOT / "scripts/start_wsgi.sh").chmod(0o755)


# ============================================================
# 6) U02 TEST SUITE
# ============================================================

test_code = '''
from pathlib import Path


def test_flask_app_creation():

    from app.api.app import create_app

    app = create_app()

    assert app is not None


def test_root_endpoint():

    from app.api.app import create_app

    app = create_app()

    client = app.test_client()

    response = client.get("/")

    assert response.status_code == 200

    data = response.get_json()

    assert data["status"] == "running"


def test_health_endpoint():

    from app.api.app import create_app

    app = create_app()

    client = app.test_client()

    response = client.get("/health")

    assert response.status_code == 200

    data = response.get_json()

    assert data["status"] == "ok"
    assert data["project"] == "KITCHEN_ROBOT"
    assert data["execution_unit"] == "U01"


def test_sub_daily_standard_timeframe_warning():

    from app.api.app import validate_timeframe_guard

    for timeframe in [
        "M1",
        "M5",
        "M15",
        "H1",
        "H4",
    ]:

        result = validate_timeframe_guard(
            timeframe
        )

        # Sub-daily timeframe is NOT an error.
        assert result["valid"] is True

        # Warning MUST be present.
        assert result["warning"] is not None

        assert result["is_sub_daily"] is True

        assert (
            result["reason"]
            == "SUB_DAILY_TIMEFRAME"
        )


def test_daily_timeframe_has_no_sub_daily_warning():

    from app.api.app import validate_timeframe_guard

    result = validate_timeframe_guard(
        "D1"
    )

    assert result["valid"] is True
    assert result["warning"] is None
    assert result["is_sub_daily"] is False


def test_weekly_timeframe_has_no_sub_daily_warning():

    from app.api.app import validate_timeframe_guard

    result = validate_timeframe_guard(
        "D7"
    )

    assert result["valid"] is True
    assert result["warning"] is None
    assert result["is_sub_daily"] is False


def test_custom_range_below_daily_warning():

    from app.api.app import validate_timeframe_guard

    result = validate_timeframe_guard(
        "CUSTOM",
        "28-08-2026 10:00",
        "28-08-2026 18:00",
    )

    # Custom range below 24h is NOT an error.
    assert result["valid"] is True

    # Warning MUST be present.
    assert result["warning"] is not None

    assert result["is_sub_daily"] is True

    assert (
        result["reason"]
        == "SUB_DAILY_CUSTOM_RANGE"
    )


def test_custom_range_exactly_daily():

    from app.api.app import validate_timeframe_guard

    result = validate_timeframe_guard(
        "CUSTOM",
        "28-08-2026 10:00",
        "29-08-2026 10:00",
    )

    assert result["valid"] is True
    assert result["warning"] is None
    assert result["is_sub_daily"] is False


def test_custom_range_above_daily():

    from app.api.app import validate_timeframe_guard

    result = validate_timeframe_guard(
        "CUSTOM",
        "28-08-2026 10:00",
        "30-08-2026 10:00",
    )

    assert result["valid"] is True
    assert result["warning"] is None
    assert result["is_sub_daily"] is False


def test_invalid_custom_range_format():

    from app.api.app import validate_timeframe_guard

    result = validate_timeframe_guard(
        "CUSTOM",
        "2026-08-28 10:00",
        "2026-08-28 18:00",
    )

    assert result["valid"] is False

    assert (
        result["reason"]
        == "INVALID_CUSTOM_RANGE_FORMAT"
    )


def test_invalid_custom_range_order():

    from app.api.app import validate_timeframe_guard

    result = validate_timeframe_guard(
        "CUSTOM",
        "29-08-2026 10:00",
        "28-08-2026 10:00",
    )

    assert result["valid"] is False

    assert (
        result["reason"]
        == "INVALID_CUSTOM_RANGE_ORDER"
    )


def test_unsupported_timeframe():

    from app.api.app import validate_timeframe_guard

    result = validate_timeframe_guard(
        "UNKNOWN"
    )

    assert result["valid"] is False

    assert (
        result["reason"]
        == "UNSUPPORTED_TIMEFRAME"
    )


def test_wsgi_entrypoint():

    import wsgi

    assert wsgi.application is not None


def test_runtime_files():

    root = Path(__file__).resolve().parents[1]

    assert (
        root / "wsgi.py"
    ).is_file()

    assert (
        root / "scripts/start_wsgi.sh"
    ).is_file()

    assert (
        root / "app/config/runtime.py"
    ).is_file()
'''

(ROOT / "tests/test_u02.py").write_text(
    test_code.strip() + "\n",
    encoding="utf-8",
)


# ============================================================
# 7) UPDATE REQUIREMENTS
# ============================================================

(ROOT / "requirements.txt").write_text(
    """Flask>=3,<4
gunicorn>=23,<24
pytest>=8,<9
""",
    encoding="utf-8",
)


# ============================================================
# 8) RUN U02 TESTS
# ============================================================

print()
print("Running U02 tests...")
print()

result = subprocess.run(
    [
        sys.executable,
        "-m",
        "pytest",
        "-q",
        "tests/test_u02.py",
    ],
    cwd=ROOT,
    capture_output=True,
    text=True,
)

print(result.stdout)

if result.returncode != 0:

    print(result.stderr)

    raise RuntimeError(
        "U02 VALIDATION FAILED"
    )

print("U02 validation: PASSED")


# ============================================================
# 9) MIGRATION READINESS
# ============================================================

migration_check = {

    "wsgi_entrypoint":
        (ROOT / "wsgi.py").is_file(),

    "flask_runtime":
        (ROOT / "app/api/app.py").is_file(),

    "runtime_config":
        (ROOT / "app/config/runtime.py").is_file(),

    "start_script":
        (ROOT / "scripts/start_wsgi.sh").is_file(),

    "requirements":
        (ROOT / "requirements.txt").is_file(),
}


print()
print("Migration readiness:")

print(
    json.dumps(
        migration_check,
        indent=2,
    )
)


if not all(migration_check.values()):

    raise RuntimeError(
        "U02 MIGRATION SANITY CHECK FAILED"
    )

print("Migration readiness: PASSED")


# ============================================================
# 10) AUDIT METADATA
# ============================================================

timestamp = datetime.now(
    timezone.utc
).strftime(
    "%Y-%m-%d %H:%M:%S UTC"
)


audit_metadata = {

    "project":
        "KITCHEN_ROBOT",

    "project_version":
        "2.8.1",

    "execution_unit":
        "U02",

    "master_stages": [
        "03",
        "04",
        "05",
    ],

    "audit": {

        "status":
            "PASSED",

        "technical_lock":
            "LOCKED",

        "business_rule_lock":
            "NOT_APPLICABLE",
    },

    "validation": {

        "pytest":
            "PASSED",

        "migration_readiness":
            "PASSED",

        "self_check":
            "PENDING_FINAL_ARTIFACT_CHECK",
    },

    "implemented": {

        "runtime_foundation":
            True,

        "flask_boundary":
            True,

        "wsgi_entrypoint":
            True,

        "runtime_configuration":
            True,

        "pythonanywhere_start_script":
            True,

        "timeframe_guard":
            True,

        "sub_daily_warning":
            True,

        "custom_range_warning":
            True,
    },

    "timeframe_policy": {

        "standard_timeframes":
            [
                "M1",
                "M5",
                "M15",
                "H1",
                "H4",
                "D1",
                "D7",
            ],

        "sub_daily_behavior":
            "VALID_WITH_WARNING",

        "daily_or_higher_behavior":
            "VALID_WITHOUT_SUB_DAILY_WARNING",

        "custom_range_supported":
            True,

        "custom_range_format":
            "DD-MM-YYYY HH:MM",

        "custom_range_below_daily":
            "VALID_WITH_WARNING",

        "custom_range_daily_or_higher":
            "VALID_WITHOUT_SUB_DAILY_WARNING",

        "warning_is_error":
            False,

        "technical_validation_only":
            True,

        "scanner_analysis":
            False,
    },

    "explicitly_not_implemented": {

        "market_data_provider":
            True,

        "live_market_data":
            True,

        "candle_ingestion":
            True,

        "scanner_intelligence":
            True,

        "total_usdt_analysis":
            True,

        "btc_analysis":
            True,

        "btcd_analysis":
            True,

        "total2_analysis":
            True,

        "total3_analysis":
            True,

        "others_d_analysis":
            True,

        "strong_movers":
            True,

        "market_summary":
            True,

        "trading_logic":
            True,

        "order_execution":
            True,
    },

    "generated_at_utc":
        timestamp,
}


(ROOT / "U02_AUDIT_METADATA.json").write_text(
    json.dumps(
        audit_metadata,
        indent=2,
    ),
    encoding="utf-8",
)


# ============================================================
# 11) U02 LEDGER
# ============================================================

ledger = f"""
# KITCHEN ASSISTANT V2.8.1 — U02 LEDGER

## Execution Unit

U02 — Runtime Foundation + Flask/WSGI + Migration Readiness

## Consolidated Master Stages

- Stage 03 — Runtime Foundation
- Stage 04 — Flask / WSGI Boundary
- Stage 05 — Migration Readiness + Timeframe / Range Guard

## Audit Status

PASSED

## Technical Lock

LOCKED

## Business Rule Lock

NOT_APPLICABLE

## Generated

{timestamp}

## Validation

pytest completed successfully.

Migration readiness completed successfully.

## Completed

- Flask runtime boundary
- WSGI entrypoint
- health endpoint
- root endpoint
- runtime configuration
- Gunicorn dependency
- PythonAnywhere-compatible startup structure
- timeframe / range guard
- sub-daily warning boundary
- Custom Range warning boundary
- U02 automated tests
- migration sanity check
- audit metadata

## Timeframe / Range Guard Boundary

The timeframe / range guard is a technical validation boundary only.

It does not perform market analysis,
scanner intelligence,
signal generation,
or trading decisions.

## Sub-Daily Warning Contract

Every valid timeframe below daily (D1) MUST remain valid.

A timeframe below daily is NOT an execution error.

Instead, the system MUST return the sub-daily warning.

This applies to every supported standard timeframe
whose duration is below 24 hours.

The system MUST NOT depend on a hardcoded list of only
specific low-timeframe values to decide whether the
warning applies.

## Custom Range Warning Contract

Custom Range is supported using:

DD-MM-YYYY HH:MM

If the selected Custom Range duration is below 24 hours:

- valid = TRUE
- warning = SUB_DAILY_WARNING
- is_sub_daily = TRUE

If the selected Custom Range duration is 24 hours
or greater:

- valid = TRUE
- warning = NONE
- is_sub_daily = FALSE

Invalid Custom Range format or ordering remains invalid.

## Message Placement Contract

The sub-daily warning belongs to the timeframe / range
validation result itself.

It MUST be available immediately when the selected
timeframe or Custom Range is validated.

The warning MUST NOT be treated as a scanner signal,
market interpretation, trading instruction, or execution error.

## Explicitly NOT Implemented

- Market Data Provider
- Live Market Data
- Candle Ingestion
- Scanner Intelligence
- Total/USDT analysis
- BTC analysis
- BTCD analysis
- TOTAL2 analysis
- TOTAL3 analysis
- Others.D analysis
- Strong Movers
- Market Summary
- Trading Logic
- Order Execution

## Architectural Principle

Technical infrastructure must be completed before
Scanner Intelligence is defined.

Scanner Intelligence is therefore deferred to
the dedicated Scanner stages of the Canonical Roadmap.

## Lock Decision

U02 passed its defined technical validation scope.

Technical Lock = LOCKED.

## Next Execution Unit

U03 — Data / Capture Foundation
"""

(ROOT / "ledgers/U02_LEDGER.md").write_text(
    ledger.strip() + "\n",
    encoding="utf-8",
)


# ============================================================
# 12) FINAL ARTIFACT INVENTORY
#
# Manifest + SHA inventory are excluded from their own hash
# inventory to prevent recursive hashing.
# ============================================================

EXCLUDED_FROM_INVENTORY = {
    "U02_MANIFEST.json",
    "U02_SHA256.json",
}


project_files = []
hash_inventory = []


for file_path in sorted(ROOT.rglob("*")):

    if not file_path.is_file():
        continue

    relative_path = (
        file_path
        .relative_to(ROOT)
        .as_posix()
    )

    if relative_path in EXCLUDED_FROM_INVENTORY:
        continue

    file_bytes = file_path.read_bytes()

    project_files.append(
        relative_path
    )

    hash_inventory.append(
        {
            "file":
                relative_path,

            "sha256":
                hashlib.sha256(
                    file_bytes
                ).hexdigest(),
        }
    )


# ============================================================
# 13) MANIFEST
# ============================================================

manifest = {

    "project":
        "KITCHEN_ROBOT",

    "version":
        "2.8.1",

    "execution_unit":
        "U02",

    "consolidated_master_stages": [
        "03",
        "04",
        "05",
    ],

    "status":
        "LOCKED",

    "audit_status":
        "PASSED",

    "technical_lock":
        "LOCKED",

    "business_rule_lock":
        "NOT_APPLICABLE",

    "migration_ready":
        True,

    "generated_at_utc":
        timestamp,

    "inventory_policy": {

        "hash_algorithm":
            "SHA-256",

        "manifest_excluded_from_inventory":
            True,

        "sha256_file_excluded_from_inventory":
            True,
    },

    "file_count":
        len(project_files),

    "scope": {

        "runtime":
            True,

        "flask_wsgi":
            True,

        "migration_foundation":
            True,

        "timeframe_guard":
            True,

        "range_guard":
            True,

        "sub_daily_warning":
            True,

        "custom_range_warning":
            True,

        "market_data":
            False,

        "scanner":
            False,

        "scanner_intelligence":
            False,

        "trading":
            False,
    },

    "timeframe_policy": {

        "sub_daily":
            "VALID_WITH_WARNING",

        "daily_or_higher":
            "VALID_WITHOUT_SUB_DAILY_WARNING",

        "custom_range":
            "SUPPORTED",

        "custom_range_below_daily":
            "VALID_WITH_WARNING",

        "warning_is_error":
            False,
    },
}


(ROOT / "U02_MANIFEST.json").write_text(
    json.dumps(
        manifest,
        indent=2,
    ),
    encoding="utf-8",
)


# ============================================================
# 14) SHA-256 INVENTORY
# ============================================================

(ROOT / "U02_SHA256.json").write_text(
    json.dumps(
        hash_inventory,
        indent=2,
    ),
    encoding="utf-8",
)


# ============================================================
# 15) FINAL SELF-CHECK
# ============================================================

checks = {

    "flask_app":
        (ROOT / "app/api/app.py").is_file(),

    "wsgi":
        (ROOT / "wsgi.py").is_file(),

    "runtime_config":
        (ROOT / "app/config/runtime.py").is_file(),

    "start_script":
        (ROOT / "scripts/start_wsgi.sh").is_file(),

    "u02_tests":
        (ROOT / "tests/test_u02.py").is_file(),

    "ledger":
        (ROOT / "ledgers/U02_LEDGER.md").is_file(),

    "audit_metadata":
        (ROOT / "U02_AUDIT_METADATA.json").is_file(),

    "manifest":
        (ROOT / "U02_MANIFEST.json").is_file(),

    "sha256":
        (ROOT / "U02_SHA256.json").is_file(),
}


print()
print("Final U02 self-check:")

print(
    json.dumps(
        checks,
        indent=2,
    )
)


if not all(checks.values()):

    raise RuntimeError(
        "U02 SELF-CHECK FAILED"
    )


# ============================================================
# 16) FINAL LOCK ASSERTION
# ============================================================

manifest_data = json.loads(
    (ROOT / "U02_MANIFEST.json").read_text(
        encoding="utf-8"
    )
)


audit_data = json.loads(
    (ROOT / "U02_AUDIT_METADATA.json").read_text(
        encoding="utf-8"
    )
)


ledger_text = (
    ROOT / "ledgers/U02_LEDGER.md"
).read_text(
    encoding="utf-8"
)


lock_assertions = {

    "manifest_status_locked":
        manifest_data["status"] == "LOCKED",

    "manifest_audit_passed":
        manifest_data["audit_status"] == "PASSED",

    "manifest_technical_lock":
        manifest_data["technical_lock"] == "LOCKED",

    "audit_status_passed":
        audit_data["audit"]["status"] == "PASSED",

    "audit_technical_lock":
        audit_data["audit"]["technical_lock"] == "LOCKED",

    "ledger_contains_locked":
        "Technical Lock\n\nLOCKED" in ledger_text,

    "sub_daily_warning_not_error":
        audit_data["timeframe_policy"][
            "sub_daily_behavior"
        ] == "VALID_WITH_WARNING",

    "custom_range_supported":
        audit_data["timeframe_policy"][
            "custom_range_supported"
        ] is True,

    "custom_range_warning":
        audit_data["timeframe_policy"][
            "custom_range_below_daily"
        ] == "VALID_WITH_WARNING",

}


print()
print("Final Lock Assertions:")

print(
    json.dumps(
        lock_assertions,
        indent=2,
    )
)


if not all(lock_assertions.values()):

    raise RuntimeError(
        "U02 FINAL LOCK ASSERTION FAILED"
    )


# ============================================================
# 17) FINAL STATUS
# ============================================================

print()
print("=" * 60)
print("U02 COMPLETE")
print("AUDIT        : PASSED")
print("STATUS       : LOCKED")
print("TECHNICAL    : LOCKED")
print("BUSINESS     : NOT APPLICABLE")
print("MIGRATION    : READY")
print("TIMEFRAME    : SUB-DAILY = VALID + WARNING")
print("CUSTOM RANGE : SUPPORTED + BELOW D1 WARNING")
print("NEXT         : U03")
print("=" * 60)
```

```python
# ============================================================
# KITCHEN ASSISTANT V2.8 — U03
#
# CONSOLIDATED MASTER STAGES:
#   Data / Capture Foundation
#
# PURPOSE:
#   Build the provider-independent data foundation.
#
# IMPORTANT:
#   No real market-data provider is connected in U03.
#   No trading strategy is connected.
#   No trading/order execution is connected.
#
# U02 MUST already be VALIDATED + LOCKED.
#
# LOCK POLICY:
#   U03 may become LOCKED only after:
#   1. U02 lock verification
#   2. U03 automated validation
#   3. U03 artifact self-check
#   4. U03 audit metadata generation
#   5. Final lock assertions
#
# BACKUP POLICY:
#   No duplicate project backup is created inside U03.
#   The project itself contains:
#       - Audit Metadata
#       - Ledger
#       - Manifest
#       - SHA-256 inventory
#   A separate non-destructive Snapshot is taken after
#   successful U03 completion.
# ============================================================

from pathlib import Path
from dataclasses import dataclass
from datetime import datetime, timezone
from typing import Iterable
import sys
import os
import json
import hashlib
import subprocess


# ============================================================
# 0. PROJECT VALIDATION
# ============================================================

ROOT = Path("/content/kitchen_robot_v27")

if not ROOT.exists():
    raise RuntimeError(
        "U02 project not found. U02 must be completed first."
    )

os.chdir(ROOT)

print("=" * 60)
print("KITCHEN ASSISTANT V2.8")
print("Execution Unit : U03")
print("Stage          : Data / Capture Foundation")
print("Project Root   :", ROOT)
print("=" * 60)


# ============================================================
# 0.1 VERIFY U02 LOCK
# ============================================================

U02_MANIFEST = ROOT / "U02_MANIFEST.json"
U02_AUDIT = ROOT / "U02_AUDIT_METADATA.json"
U02_LEDGER = ROOT / "ledgers/U02_LEDGER.md"

u02_required = {
    "manifest": U02_MANIFEST.is_file(),
    "audit": U02_AUDIT.is_file(),
    "ledger": U02_LEDGER.is_file(),
}

print("\nU02 prerequisite check:")
print(json.dumps(u02_required, indent=2))

if not all(u02_required.values()):
    raise RuntimeError(
        "U02 prerequisite artifacts are incomplete."
    )

u02_manifest_data = json.loads(
    U02_MANIFEST.read_text(
        encoding="utf-8"
    )
)

u02_audit_data = json.loads(
    U02_AUDIT.read_text(
        encoding="utf-8"
    )
)

u02_lock_check = {

    "manifest_status_locked":
        u02_manifest_data.get("status") == "LOCKED",

    "manifest_audit_passed":
        u02_manifest_data.get("audit_status") == "PASSED",

    "manifest_technical_lock":
        u02_manifest_data.get("technical_lock") == "LOCKED",

    "audit_status_passed":
        u02_audit_data.get("audit", {}).get("status") == "PASSED",

    "audit_technical_lock":
        u02_audit_data.get("audit", {}).get("technical_lock") == "LOCKED",
}

print("\nU02 lock verification:")
print(
    json.dumps(
        u02_lock_check,
        indent=2
    )
)

if not all(u02_lock_check.values()):
    raise RuntimeError(
        "U02 LOCK VERIFICATION FAILED. U03 cannot start."
    )

print("U02 prerequisite: LOCKED")


# ============================================================
# 1. DATA DIRECTORIES
# ============================================================

directories = [
    ROOT / "app/data",
    ROOT / "app/data/models",
    ROOT / "app/data/validation",
    ROOT / "app/data/capture",
    ROOT / "data/raw",
    ROOT / "data/normalized",
    ROOT / "data/validated",
    ROOT / "tests/data",
]

for directory in directories:

    directory.mkdir(
        parents=True,
        exist_ok=True
    )

print("Data architecture: READY")


# ============================================================
# 2. CANONICAL CANDLE MODEL
# ============================================================

candle_model = '''
from dataclasses import dataclass
from typing import Any


@dataclass(frozen=True)
class Candle:

    symbol: str
    timestamp: int

    open: float
    high: float
    low: float
    close: float
    volume: float

    def __post_init__(self):

        if not self.symbol:
            raise ValueError("symbol cannot be empty")

        values = [
            self.open,
            self.high,
            self.low,
            self.close,
            self.volume,
        ]

        if any(
            not isinstance(value, (int, float))
            for value in values
        ):
            raise TypeError(
                "OHLCV values must be numeric"
            )

        if self.timestamp <= 0:
            raise ValueError(
                "timestamp must be positive"
            )

        if self.high < self.low:
            raise ValueError(
                "high cannot be lower than low"
            )

        if self.open < self.low:
            raise ValueError(
                "open cannot be below low"
            )

        if self.open > self.high:
            raise ValueError(
                "open cannot be above high"
            )

        if self.close < self.low:
            raise ValueError(
                "close cannot be below low"
            )

        if self.close > self.high:
            raise ValueError(
                "close cannot be above high"
            )

        if self.volume < 0:
            raise ValueError(
                "volume cannot be negative"
            )


def candle_to_dict(candle: Candle) -> dict[str, Any]:

    return {
        "symbol": candle.symbol,
        "timestamp": candle.timestamp,
        "open": candle.open,
        "high": candle.high,
        "low": candle.low,
        "close": candle.close,
        "volume": candle.volume,
    }


def candle_from_dict(data: dict[str, Any]) -> Candle:

    return Candle(
        symbol=str(data["symbol"]),
        timestamp=int(data["timestamp"]),
        open=float(data["open"]),
        high=float(data["high"]),
        low=float(data["low"]),
        close=float(data["close"]),
        volume=float(data["volume"]),
    )
'''

(ROOT / "app/data/models/candle.py").write_text(
    candle_model.strip() + "\n",
    encoding="utf-8"
)


# ============================================================
# 3. CAPTURE CONTRACT
# ============================================================

capture_code = '''
from abc import ABC, abstractmethod
from typing import Iterable

from app.data.models.candle import Candle


class CandleCapture(ABC):

    @abstractmethod
    def fetch(
        self,
        symbol: str,
        limit: int = 100,
    ) -> Iterable[Candle]:
        """
        Provider-independent capture contract.

        Real providers are implemented later.
        """
        raise NotImplementedError
'''

(ROOT / "app/data/capture/base.py").write_text(
    capture_code.strip() + "\n",
    encoding="utf-8"
)


# ============================================================
# 4. VALIDATION LAYER
# ============================================================

validation_code = '''
from collections.abc import Iterable

from app.data.models.candle import Candle


def validate_candles(
    candles: Iterable[Candle],
) -> list[Candle]:

    validated = list(candles)

    if not validated:
        raise ValueError(
            "candle collection cannot be empty"
        )

    previous_timestamp = None

    for candle in validated:

        if not isinstance(
            candle,
            Candle
        ):
            raise TypeError(
                "all records must be Candle instances"
            )

        if (
            previous_timestamp is not None
            and candle.timestamp <= previous_timestamp
        ):
            raise ValueError(
                "candles must be strictly chronological"
            )

        previous_timestamp = candle.timestamp

    return validated
'''

(ROOT / "app/data/validation/candles.py").write_text(
    validation_code.strip() + "\n",
    encoding="utf-8"
)


# ============================================================
# 5. NORMALIZATION LAYER
# ============================================================

normalization_code = '''
from typing import Any

from app.data.models.candle import Candle


def normalize_candle(
    data: dict[str, Any],
) -> Candle:

    timestamp = int(data["timestamp"])

    return Candle(
        symbol=str(data["symbol"]).upper().strip(),
        timestamp=timestamp,
        open=float(data["open"]),
        high=float(data["high"]),
        low=float(data["low"]),
        close=float(data["close"]),
        volume=float(data["volume"]),
    )
'''

(ROOT / "app/data/validation/normalize.py").write_text(
    normalization_code.strip() + "\n",
    encoding="utf-8"
)


# ============================================================
# 6. JSONL CAPTURE STORAGE
# ============================================================

storage_code = '''
import json
from pathlib import Path
from typing import Iterable

from app.data.models.candle import (
    Candle,
    candle_to_dict,
)


def write_candles(
    candles: Iterable[Candle],
    path: Path,
) -> int:

    path.parent.mkdir(
        parents=True,
        exist_ok=True
    )

    count = 0

    with path.open(
        "w",
        encoding="utf-8"
    ) as handle:

        for candle in candles:

            handle.write(
                json.dumps(
                    candle_to_dict(candle),
                    separators=(",", ":")
                )
                + "\\n"
            )

            count += 1

    return count
'''

(ROOT / "app/data/capture/storage.py").write_text(
    storage_code.strip() + "\n",
    encoding="utf-8"
)


# ============================================================
# 7. U03 TEST DATA
# ============================================================

test_code = '''
from app.data.models.candle import Candle
from app.data.validation.candles import validate_candles
from app.data.validation.normalize import normalize_candle
from app.data.capture.storage import write_candles

import pytest


def sample_candle(timestamp: int) -> Candle:

    return Candle(
        symbol="BTCUSDT",
        timestamp=timestamp,
        open=100.0,
        high=110.0,
        low=90.0,
        close=105.0,
        volume=1000.0,
    )


def test_candle_model():

    candle = sample_candle(1000)

    assert candle.symbol == "BTCUSDT"
    assert candle.close == 105.0


def test_invalid_ohlc():

    with pytest.raises(ValueError):

        Candle(
            symbol="BTCUSDT",
            timestamp=1000,
            open=120.0,
            high=110.0,
            low=90.0,
            close=105.0,
            volume=1000.0,
        )


def test_chronological_validation():

    candles = [
        sample_candle(1000),
        sample_candle(2000),
    ]

    validated = validate_candles(candles)

    assert len(validated) == 2


def test_non_chronological_rejected():

    candles = [
        sample_candle(2000),
        sample_candle(1000),
    ]

    with pytest.raises(ValueError):

        validate_candles(candles)


def test_normalization():

    candle = normalize_candle({
        "symbol": " btcusdt ",
        "timestamp": "1000",
        "open": "100",
        "high": "110",
        "low": "90",
        "close": "105",
        "volume": "1000",
    })

    assert candle.symbol == "BTCUSDT"
    assert candle.open == 100.0


def test_jsonl_storage(tmp_path):

    path = tmp_path / "candles.jsonl"

    count = write_candles(
        [
            sample_candle(1000),
            sample_candle(2000),
        ],
        path,
    )

    assert count == 2
    assert path.exists()
    assert len(
        path.read_text(
            encoding="utf-8"
        ).splitlines()
    ) == 2
'''

(ROOT / "tests/data/test_u03.py").write_text(
    test_code.strip() + "\n",
    encoding="utf-8"
)


# ============================================================
# 8. RUN U03 TESTS
# ============================================================

print()
print("Running U03 tests...")
print()

result = subprocess.run(
    [
        sys.executable,
        "-m",
        "pytest",
        "-q",
        "tests/data/test_u03.py",
    ],
    cwd=ROOT,
    capture_output=True,
    text=True,
)

print(result.stdout)

if result.returncode != 0:

    print(result.stderr)

    raise RuntimeError(
        "U03 VALIDATION FAILED"
    )

print("U03 validation: PASSED")


# ============================================================
# 9. DATA FOUNDATION SELF-CHECK
# ============================================================

data_check = {

    "candle_model":
        (
            ROOT
            / "app/data/models/candle.py"
        ).is_file(),

    "capture_contract":
        (
            ROOT
            / "app/data/capture/base.py"
        ).is_file(),

    "validation":
        (
            ROOT
            / "app/data/validation/candles.py"
        ).is_file(),

    "normalization":
        (
            ROOT
            / "app/data/validation/normalize.py"
        ).is_file(),

    "storage":
        (
            ROOT
            / "app/data/capture/storage.py"
        ).is_file(),

    "tests":
        (
            ROOT
            / "tests/data/test_u03.py"
        ).is_file(),
}


print()
print("Data foundation self-check:")
print(
    json.dumps(
        data_check,
        indent=2
    )
)

if not all(data_check.values()):

    raise RuntimeError(
        "U03 DATA FOUNDATION CHECK FAILED"
    )

print("Data foundation: READY")


# ============================================================
# 10. AUDIT METADATA
# ============================================================

timestamp = datetime.now(
    timezone.utc
).strftime(
    "%Y-%m-%d %H:%M:%S UTC"
)

audit_metadata = {

    "project":
        "KITCHEN_ASSISTANT",

    "project_version":
        "2.8",

    "execution_unit":
        "U03",

    "stage":
        "DATA_CAPTURE_FOUNDATION",

    "audit": {

        "status":
            "PASSED",

        "technical_lock":
            "LOCKED",

        "business_rule_lock":
            "NOT_APPLICABLE",
    },

    "validation": {

        "u02_lock_verification":
            "PASSED",

        "pytest":
            "PASSED",

        "data_foundation_self_check":
            "PASSED",
    },

    "scope": {

        "canonical_candle_model":
            True,

        "ohlcv_validation":
            True,

        "chronological_validation":
            True,

        "normalization":
            True,

        "provider_independent_capture_contract":
            True,

        "jsonl_storage_foundation":
            True,

    },

    "explicitly_not_implemented": {

        "real_market_data_provider":
            True,

        "live_market_data":
            True,

        "scanner_intelligence":
            True,

        "trading_strategy":
            True,

        "order_execution":
            True,

    },

    "generated_at_utc":
        timestamp,
}

(ROOT / "U03_AUDIT_METADATA.json").write_text(
    json.dumps(
        audit_metadata,
        indent=2
    ),
    encoding="utf-8"
)


# ============================================================
# 11. U03 LEDGER
# ============================================================

ledger = f"""# KITCHEN ASSISTANT V2.8 — U03 LEDGER

## Execution Unit

U03 — Data / Capture Foundation

## Purpose

Create the provider-independent foundation for market data.

## Prerequisite

U02 — VALIDATED + LOCKED

## Audit Status

PASSED

## Technical Lock

LOCKED

## Business Rule Lock

NOT_APPLICABLE

## Generated

{timestamp}

## Completed

- canonical Candle model
- OHLCV validation
- chronological validation
- data normalization
- provider-independent capture contract
- JSONL storage foundation
- U03 automated tests
- U02 prerequisite lock verification
- U03 audit metadata
- U03 manifest
- U03 SHA-256 inventory

## Provider

NOT CONNECTED

## Live Market Data

NOT CONNECTED

## Trading Logic

NOT CONNECTED

## Order Execution

NOT CONNECTED

## Scanner Intelligence

NOT IMPLEMENTED

## Architectural Boundary

U03 creates the data foundation only.

It does not define scanner intelligence,
market interpretation, trading strategy,
or order execution.

## Cell Integrity Rule

No execution-error cell is accepted into the working notebook.

If an execution cell fails:

1. remove or replace the failed cell
2. rerun the corrected cell
3. verify success
4. only then lock the Unit

## Backup / Archive Rule

U03 does not create a duplicate full-project backup
inside the project.

The project records:

- Audit Metadata
- Ledger
- Manifest
- SHA-256 inventory

A separate non-destructive Snapshot is created
after successful Unit completion.

## Next

U04 — Real Market Data Provider / Capture
"""

(ROOT / "ledgers/U03_LEDGER.md").write_text(
    ledger.strip() + "\n",
    encoding="utf-8"
)


# ============================================================
# 12. ARTIFACT SELF-CHECK
# ============================================================

final_check = {

    "candle_model":
        (
            ROOT
            / "app/data/models/candle.py"
        ).is_file(),

    "capture_contract":
        (
            ROOT
            / "app/data/capture/base.py"
        ).is_file(),

    "validation":
        (
            ROOT
            / "app/data/validation/candles.py"
        ).is_file(),

    "normalization":
        (
            ROOT
            / "app/data/validation/normalize.py"
        ).is_file(),

    "storage":
        (
            ROOT
            / "app/data/capture/storage.py"
        ).is_file(),

    "tests":
        (
            ROOT
            / "tests/data/test_u03.py"
        ).is_file(),

    "ledger":
        (
            ROOT
            / "ledgers/U03_LEDGER.md"
        ).is_file(),

    "audit_metadata":
        (
            ROOT
            / "U03_AUDIT_METADATA.json"
        ).is_file(),
}


print()
print("Final U03 artifact self-check:")

print(
    json.dumps(
        final_check,
        indent=2
    )
)

if not all(final_check.values()):

    raise RuntimeError(
        "U03 SELF-CHECK FAILED"
    )


# ============================================================
# 13. MANIFEST + SHA256
#
# Manifest and SHA files are excluded from their own
# inventory to prevent recursive hashing.
# ============================================================

EXCLUDED_FROM_INVENTORY = {
    "U03_MANIFEST.json",
    "U03_SHA256.json",
}

project_files = []
hash_inventory = []

for file_path in sorted(ROOT.rglob("*")):

    if not file_path.is_file():
        continue

    relative_path = (
        file_path
        .relative_to(ROOT)
        .as_posix()
    )

    if relative_path in EXCLUDED_FROM_INVENTORY:
        continue

    file_bytes = file_path.read_bytes()

    project_files.append(
        relative_path
    )

    hash_inventory.append(
        {
            "file":
                relative_path,

            "sha256":
                hashlib.sha256(
                    file_bytes
                ).hexdigest(),
        }
    )


manifest = {

    "project":
        "KITCHEN_ASSISTANT",

    "version":
        "2.8",

    "execution_unit":
        "U03",

    "stage":
        "DATA_CAPTURE_FOUNDATION",

    "status":
        "LOCKED",

    "audit_status":
        "PASSED",

    "technical_lock":
        "LOCKED",

    "business_rule_lock":
        "NOT_APPLICABLE",

    "generated_at_utc":
        timestamp,

    "inventory_policy": {

        "hash_algorithm":
            "SHA-256",

        "manifest_excluded_from_inventory":
            True,

        "sha256_file_excluded_from_inventory":
            True,
    },

    "file_count":
        len(project_files),

    "scope": {

        "data_foundation":
            True,

        "real_market_data_provider":
            False,

        "live_market_data":
            False,

        "scanner":
            False,

        "trading":
            False,

        "order_execution":
            False,
    },
}


(ROOT / "U03_MANIFEST.json").write_text(
    json.dumps(
        manifest,
        indent=2
    ),
    encoding="utf-8"
)


(ROOT / "U03_SHA256.json").write_text(
    json.dumps(
        hash_inventory,
        indent=2
    ),
    encoding="utf-8"
)


# ============================================================
# 14. FINAL LOCK ASSERTIONS
# ============================================================

manifest_data = json.loads(
    (ROOT / "U03_MANIFEST.json").read_text(
        encoding="utf-8"
    )
)

audit_data = json.loads(
    (ROOT / "U03_AUDIT_METADATA.json").read_text(
        encoding="utf-8"
    )
)

ledger_text = (
    ROOT / "ledgers/U03_LEDGER.md"
).read_text(
    encoding="utf-8"
)

lock_assertions = {

    "u02_prerequisite_locked":
        all(u02_lock_check.values()),

    "manifest_status_locked":
        manifest_data["status"] == "LOCKED",

    "manifest_audit_passed":
        manifest_data["audit_status"] == "PASSED",

    "manifest_technical_lock":
        manifest_data["technical_lock"] == "LOCKED",

    "audit_status_passed":
        audit_data["audit"]["status"] == "PASSED",

    "audit_technical_lock":
        audit_data["audit"]["technical_lock"] == "LOCKED",

    "ledger_contains_locked":
        "## Technical Lock\n\nLOCKED" in ledger_text,

    "ledger_contains_audit_passed":
        "## Audit Status\n\nPASSED" in ledger_text,
}


print()
print("Final Lock Assertions:")

print(
    json.dumps(
        lock_assertions,
        indent=2
    )
)

if not all(lock_assertions.values()):

    raise RuntimeError(
        "U03 FINAL LOCK ASSERTION FAILED"
    )


# ============================================================
# 15. FINAL STATUS
# ============================================================

print()
print("=" * 60)
print("U03 COMPLETE")
print("AUDIT        : PASSED")
print("STATUS       : LOCKED")
print("TECHNICAL    : LOCKED")
print("BUSINESS     : NOT APPLICABLE")
print("PROVIDER     : NOT CONNECTED")
print("TRADING      : NOT CONNECTED")
print("NEXT         : U04")
print("=" * 60)
```

```python
# ============================================================
# KITCHEN ASSISTANT V2.8 — U04
#
# REAL MARKET DATA PROVIDER / CAPTURE
#
# PROVIDER PRIORITY
#   1. BINANCE SPOT PUBLIC
#   2. COINBASE EXCHANGE PUBLIC
#
# RULE
#   Binance is ALWAYS the first-choice provider.
#   Fallback happens ONLY when Binance cannot provide
#   the requested symbol/data.
#
# IMPORTANT
#   - No API key
#   - No secret
#   - No trading
#   - No orders
#   - U04 is NOT LOCKED until every check passes
#
# U03 MUST already be technically locked.
# ============================================================

from pathlib import Path
import sys
import os
import json
import hashlib
import subprocess
import time


# ============================================================
# 0. PROJECT
# ============================================================

ROOT = Path("/content/kitchen_robot_v27")

if not ROOT.exists():
    raise RuntimeError(
        "U03 project not found. "
        "U03 must be completed first."
    )

os.chdir(ROOT)

print("=" * 60)
print("KITCHEN ASSISTANT V2.8")
print("Execution Unit : U04")
print("Stage          : Real Market Data Provider / Capture")
print("Primary        : BINANCE_SPOT_PUBLIC")
print("Fallback       : COINBASE_EXCHANGE_PUBLIC")
print("Project Root   :", ROOT)
print("=" * 60)


# ============================================================
# 1. U03 PREREQUISITE
# ============================================================

u03_manifest_path = ROOT / "U03_MANIFEST.json"
u03_audit_path = ROOT / "U03_AUDIT_METADATA.json"
u03_ledger_path = ROOT / "ledgers/U03_LEDGER.md"

u03_files = {
    "manifest": u03_manifest_path.is_file(),
    "audit_metadata": u03_audit_path.is_file(),
    "ledger": u03_ledger_path.is_file(),
    "candle_model": (
        ROOT / "app/data/models/candle.py"
    ).is_file(),
    "validation": (
        ROOT / "app/data/validation/candles.py"
    ).is_file(),
}

print()
print("U03 prerequisite check:")
print(json.dumps(u03_files, indent=2))

if not all(u03_files.values()):
    raise RuntimeError(
        "U03 prerequisite artifacts are incomplete."
    )

u03_manifest = json.loads(
    u03_manifest_path.read_text(
        encoding="utf-8"
    )
)

u03_audit = json.loads(
    u03_audit_path.read_text(
        encoding="utf-8"
    )
)

manifest_locked = (
    u03_manifest.get("status") == "LOCKED"
    and (
        u03_manifest.get("technical_lock")
        in (True, "LOCKED")
    )
)

audit_locked = (
    u03_audit.get("audit", {}).get("status")
    == "PASSED"
    and
    u03_audit.get("audit", {}).get("technical_lock")
    == "LOCKED"
)

print()
print("U03 lock verification:")
print(
    json.dumps(
        {
            "manifest_status_locked":
                u03_manifest.get("status") == "LOCKED",

            "manifest_technical_lock":
                u03_manifest.get("technical_lock")
                in (True, "LOCKED"),

            "audit_status_passed":
                u03_audit.get("audit", {}).get("status")
                == "PASSED",

            "audit_technical_lock":
                u03_audit.get("audit", {}).get(
                    "technical_lock"
                ) == "LOCKED",
        },
        indent=2,
    )
)

if not (manifest_locked and audit_locked):
    raise RuntimeError(
        "U03 is not technically locked."
    )

print("U03 prerequisite: LOCKED")


# ============================================================
# 2. DEPENDENCY
# ============================================================

subprocess.check_call(
    [
        sys.executable,
        "-m",
        "pip",
        "install",
        "-q",
        "requests>=2,<3",
    ]
)

print("Provider dependency: READY")


# ============================================================
# 3. ENSURE PACKAGE DIRECTORIES
# ============================================================

for directory in [
    ROOT / "app",
    ROOT / "app/data",
    ROOT / "app/data/capture",
    ROOT / "data/raw",
    ROOT / "data/normalized",
    ROOT / "tests/data",
    ROOT / "ledgers",
]:
    directory.mkdir(
        parents=True,
        exist_ok=True
    )


# ============================================================
# 4. BINANCE ADAPTER
#
# IMPORTANT:
# This definition is written directly into binance.py.
# Therefore the class imported below MUST exist.
# ============================================================

binance_code = r'''
from __future__ import annotations

from typing import Optional
import requests

from app.data.models.candle import Candle
from app.data.validation.candles import validate_candles


class BinanceProviderUnavailable(Exception):
    """Raised when Binance cannot provide the requested data."""


class BinancePublicAdapter:

    provider_name = "BINANCE_SPOT_PUBLIC"

    BASE_URLS = [
        "https://api.binance.com",
        "https://api1.binance.com",
        "https://api2.binance.com",
        "https://api3.binance.com",
        "https://api4.binance.com",
    ]

    GRANULARITY_MAP = {
        "1m": "1m",
        "3m": "3m",
        "5m": "5m",
        "15m": "15m",
        "30m": "30m",
        "1h": "1h",
        "2h": "2h",
        "4h": "4h",
        "6h": "6h",
        "8h": "8h",
        "12h": "12h",
        "1d": "1d",
    }

    def __init__(
        self,
        base_urls: Optional[list[str]] = None,
        timeout: int = 15,
    ):
        self.base_urls = (
            base_urls
            if base_urls
            else list(self.BASE_URLS)
        )

        self.timeout = timeout

        self.last_error = None
        self.last_base_url = None

    def _request(
        self,
        path: str,
        params: dict,
    ):
        errors = []

        for base_url in self.base_urls:

            try:
                response = requests.get(
                    base_url.rstrip("/") + path,
                    params=params,
                    headers={
                        "Accept": "application/json",
                        "User-Agent":
                            "KitchenAssistant/2.8",
                    },
                    timeout=self.timeout,
                )

                if response.status_code == 200:
                    self.last_base_url = base_url
                    return response

                errors.append(
                    f"{base_url}: HTTP "
                    f"{response.status_code}"
                )

            except Exception as exc:
                errors.append(
                    f"{base_url}: {repr(exc)}"
                )

        self.last_error = "; ".join(errors)

        raise BinanceProviderUnavailable(
            self.last_error
        )

    def symbol_available(
        self,
        symbol: str,
    ) -> bool:

        symbol = symbol.upper().strip()

        response = self._request(
            "/api/v3/exchangeInfo",
            {
                "symbol": symbol,
            },
        )

        payload = response.json()

        symbols = payload.get("symbols", [])

        if not symbols:
            return False

        item = symbols[0]

        return (
            item.get("symbol") == symbol
            and item.get("status") == "TRADING"
        )

    def fetch(
        self,
        symbol: str,
        interval: str = "1m",
        limit: int = 5,
    ) -> list[Candle]:

        symbol = symbol.upper().strip()

        if not symbol:
            raise ValueError(
                "symbol cannot be empty"
            )

        if interval not in self.GRANULARITY_MAP:
            raise ValueError(
                "unsupported interval"
            )

        if limit < 1:
            raise ValueError(
                "limit must be positive"
            )

        if limit > 1000:
            raise ValueError(
                "limit cannot exceed Binance maximum"
            )

        response = self._request(
            "/api/v3/klines",
            {
                "symbol": symbol,
                "interval":
                    self.GRANULARITY_MAP[interval],
                "limit": limit,
            },
        )

        payload = response.json()

        if not isinstance(payload, list):
            raise BinanceProviderUnavailable(
                "Binance response is not a list"
            )

        candles = []

        for row in payload:

            if not isinstance(row, list):
                raise BinanceProviderUnavailable(
                    "Malformed Binance candle"
                )

            if len(row) < 6:
                raise BinanceProviderUnavailable(
                    "Incomplete Binance candle"
                )

            candles.append(
                Candle(
                    symbol=symbol,
                    timestamp=int(row[0]),
                    open=float(row[1]),
                    high=float(row[2]),
                    low=float(row[3]),
                    close=float(row[4]),
                    volume=float(row[5]),
                )
            )

        if len(candles) != limit:
            raise BinanceProviderUnavailable(
                "Binance returned "
                f"{len(candles)} candles; "
                f"expected {limit}"
            )

        candles.sort(
            key=lambda candle:
                candle.timestamp
        )

        return validate_candles(candles)
'''

(
    ROOT / "app/data/capture/binance.py"
).write_text(
    binance_code.strip() + "\n",
    encoding="utf-8",
)

print("Binance provider adapter: REBUILT")


# ============================================================
# 5. COINBASE FALLBACK ADAPTER
# ============================================================

coinbase_code = r'''
from __future__ import annotations

import requests

from app.data.models.candle import Candle
from app.data.validation.candles import validate_candles


class CoinbaseProviderUnavailable(Exception):
    """Raised when Coinbase cannot provide data."""


class CoinbasePublicAdapter:

    provider_name = "COINBASE_EXCHANGE_PUBLIC"

    BASE_URL = (
        "https://api.exchange.coinbase.com"
    )

    GRANULARITY_MAP = {
        "1m": 60,
        "5m": 300,
        "15m": 900,
        "1h": 3600,
        "6h": 21600,
        "1d": 86400,
    }

    def __init__(
        self,
        timeout: int = 15,
    ):
        self.timeout = timeout

    def _product_id(
        self,
        symbol: str,
    ) -> str:

        symbol = symbol.upper().strip()

        if symbol.endswith("USDT"):
            return (
                f"{symbol[:-4]}-USD"
            )

        if symbol.endswith("USD"):
            return (
                f"{symbol[:-3]}-USD"
            )

        raise ValueError(
            "unsupported symbol format"
        )

    def symbol_available(
        self,
        symbol: str,
    ) -> bool:

        product_id = self._product_id(symbol)

        url = (
            self.BASE_URL
            + f"/products/{product_id}"
        )

        try:
            response = requests.get(
                url,
                headers={
                    "Accept": "application/json",
                    "User-Agent":
                        "KitchenAssistant/2.8",
                },
                timeout=self.timeout,
            )

            return response.status_code == 200

        except Exception:
            return False

    def fetch(
        self,
        symbol: str,
        interval: str = "1m",
        limit: int = 5,
    ) -> list[Candle]:

        symbol = symbol.upper().strip()

        if interval not in self.GRANULARITY_MAP:
            raise ValueError(
                "unsupported interval"
            )

        if limit < 1:
            raise ValueError(
                "limit must be positive"
            )

        if limit > 300:
            raise ValueError(
                "limit cannot exceed provider maximum"
            )

        product_id = self._product_id(symbol)

        url = (
            self.BASE_URL
            + f"/products/{product_id}/candles"
        )

        response = requests.get(
            url,
            params={
                "granularity":
                    self.GRANULARITY_MAP[interval],
            },
            headers={
                "Accept": "application/json",
                "User-Agent":
                    "KitchenAssistant/2.8",
            },
            timeout=self.timeout,
        )

        if response.status_code != 200:
            raise CoinbaseProviderUnavailable(
                f"HTTP {response.status_code}"
            )

        payload = response.json()

        if not isinstance(payload, list):
            raise CoinbaseProviderUnavailable(
                "Coinbase response is not a list"
            )

        candles = []

        for row in payload:

            if len(row) < 6:
                raise CoinbaseProviderUnavailable(
                    "Malformed Coinbase candle"
                )

            candles.append(
                Candle(
                    symbol=symbol,
                    timestamp=int(row[0]) * 1000,
                    open=float(row[3]),
                    high=float(row[2]),
                    low=float(row[1]),
                    close=float(row[4]),
                    volume=float(row[5]),
                )
            )

        candles.sort(
            key=lambda candle:
                candle.timestamp
        )

        candles = candles[-limit:]

        if len(candles) != limit:
            raise CoinbaseProviderUnavailable(
                "Insufficient Coinbase candles"
            )

        return validate_candles(candles)
'''

(
    ROOT / "app/data/capture/coinbase.py"
).write_text(
    coinbase_code.strip() + "\n",
    encoding="utf-8",
)

print("Coinbase fallback adapter: REBUILT")


# ============================================================
# 6. PROVIDER ROUTER
#
# Binance FIRST.
# Coinbase ONLY if Binance cannot provide.
# ============================================================

router_code = r'''
from __future__ import annotations

from app.data.capture.binance import (
    BinancePublicAdapter,
    BinanceProviderUnavailable,
)

from app.data.capture.coinbase import (
    CoinbasePublicAdapter,
    CoinbaseProviderUnavailable,
)


class ProviderRouter:

    def __init__(self):

        self.binance = (
            BinancePublicAdapter()
        )

        self.coinbase = (
            CoinbasePublicAdapter()
        )

        self.selected_provider = None
        self.last_error = None

    def fetch(
        self,
        symbol: str,
        interval: str = "1m",
        limit: int = 5,
    ):

        # ----------------------------------------------------
        # PRIORITY 1 — BINANCE
        # ----------------------------------------------------

        try:

            candles = self.binance.fetch(
                symbol=symbol,
                interval=interval,
                limit=limit,
            )

            self.selected_provider = (
                "BINANCE_SPOT_PUBLIC"
            )

            return candles

        except Exception as exc:

            self.last_error = repr(exc)

        # ----------------------------------------------------
        # PRIORITY 2 — COINBASE
        # ----------------------------------------------------

        try:

            candles = self.coinbase.fetch(
                symbol=symbol,
                interval=interval,
                limit=limit,
            )

            self.selected_provider = (
                "COINBASE_EXCHANGE_PUBLIC"
            )

            return candles

        except Exception as exc:

            self.last_error = (
                self.last_error
                + " | Coinbase: "
                + repr(exc)
            )

            raise RuntimeError(
                "ALL MARKET DATA PROVIDERS FAILED: "
                + self.last_error
            )
'''

(
    ROOT / "app/data/capture/router.py"
).write_text(
    router_code.strip() + "\n",
    encoding="utf-8",
)

print("Provider router: READY")


# ============================================================
# 7. IMPORT INTEGRITY CHECK
#
# This specifically prevents the ImportError that happened
# in the previous U04 attempts.
# ============================================================

# Remove stale imported modules from this notebook runtime.
for module_name in [
    "app.data.capture.binance",
    "app.data.capture.coinbase",
    "app.data.capture.router",
]:
    sys.modules.pop(
        module_name,
        None
    )

from app.data.capture.binance import (
    BinancePublicAdapter,
    BinanceProviderUnavailable,
)

from app.data.capture.coinbase import (
    CoinbasePublicAdapter,
    CoinbaseProviderUnavailable,
)

from app.data.capture.router import (
    ProviderRouter,
)

print("Import integrity: PASSED")


# ============================================================
# 8. STRUCTURAL TESTS
# ============================================================

test_code = r'''
from app.data.capture.binance import (
    BinancePublicAdapter,
    BinanceProviderUnavailable,
)

from app.data.capture.coinbase import (
    CoinbasePublicAdapter,
)

from app.data.capture.router import (
    ProviderRouter,
)


def test_binance_adapter_exists():

    adapter = BinancePublicAdapter()

    assert (
        adapter.provider_name
        == "BINANCE_SPOT_PUBLIC"
    )


def test_binance_limit_validation():

    adapter = BinancePublicAdapter()

    try:
        adapter.fetch(
            "BTCUSDT",
            "1m",
            0,
        )
    except ValueError:
        return

    raise AssertionError(
        "limit=0 was not rejected"
    )


def test_binance_excessive_limit():

    adapter = BinancePublicAdapter()

    try:
        adapter.fetch(
            "BTCUSDT",
            "1m",
            1001,
        )
    except ValueError:
        return

    raise AssertionError(
        "limit > 1000 was not rejected"
    )


def test_binance_interval_validation():

    adapter = BinancePublicAdapter()

    try:
        adapter.fetch(
            "BTCUSDT",
            "2m",
            5,
        )
    except ValueError:
        return

    raise AssertionError(
        "invalid interval was not rejected"
    )


def test_coinbase_adapter_exists():

    adapter = CoinbasePublicAdapter()

    assert (
        adapter.provider_name
        == "COINBASE_EXCHANGE_PUBLIC"
    )


def test_router_exists():

    router = ProviderRouter()

    assert router.binance is not None
    assert router.coinbase is not None


def test_provider_priority():

    router = ProviderRouter()

    assert (
        router.binance.provider_name
        == "BINANCE_SPOT_PUBLIC"
    )

    assert (
        router.coinbase.provider_name
        == "COINBASE_EXCHANGE_PUBLIC"
    )
'''

(
    ROOT / "tests/data/test_u04.py"
).write_text(
    test_code.strip() + "\n",
    encoding="utf-8",
)

print()
print("Running U04 structural tests...")
print()

result = subprocess.run(
    [
        sys.executable,
        "-m",
        "pytest",
        "-q",
        "tests/data/test_u04.py",
    ],
    cwd=ROOT,
    capture_output=True,
    text=True,
)

print(result.stdout)

if result.returncode != 0:

    print(result.stderr)

    raise RuntimeError(
        "U04 STRUCTURAL VALIDATION FAILED"
    )

print("U04 structural validation: PASSED")


# ============================================================
# 9. REAL PROVIDER CAPTURE
# ============================================================

SYMBOL = "BTCUSDT"
INTERVAL = "1m"
LIMIT = 5

print()
print("Connecting to primary market-data provider...")
print("Provider : BINANCE_SPOT_PUBLIC")
print("Symbol   :", SYMBOL)
print("Interval :", INTERVAL)
print("Candles  :", LIMIT)
print()

router = ProviderRouter()

try:

    live_candles = router.fetch(
        symbol=SYMBOL,
        interval=INTERVAL,
        limit=LIMIT,
    )

except Exception as exc:

    raise RuntimeError(
        "U04 REAL PROVIDER CAPTURE FAILED: "
        + repr(exc)
    )


selected_provider = (
    router.selected_provider
)

print()
print(
    "Selected provider:",
    selected_provider
)

if selected_provider not in {
    "BINANCE_SPOT_PUBLIC",
    "COINBASE_EXCHANGE_PUBLIC",
}:
    raise RuntimeError(
        "INVALID PROVIDER SELECTION"
    )

print(
    f"Real provider capture: "
    f"{len(live_candles)} candles received"
)


# ============================================================
# 10. LIVE DATA VALIDATION
# ============================================================

if len(live_candles) != LIMIT:

    raise RuntimeError(
        "U04 LIVE DATA COUNT VALIDATION FAILED"
    )

timestamps = [
    candle.timestamp
    for candle in live_candles
]

if timestamps != sorted(timestamps):

    raise RuntimeError(
        "U04 CHRONOLOGICAL VALIDATION FAILED"
    )

if len(set(timestamps)) != len(timestamps):

    raise RuntimeError(
        "U04 DUPLICATE TIMESTAMP DETECTED"
    )

for candle in live_candles:

    if candle.symbol != SYMBOL:

        raise RuntimeError(
            "U04 SYMBOL VALIDATION FAILED"
        )

    if candle.high < candle.low:

        raise RuntimeError(
            "U04 OHLC VALIDATION FAILED"
        )

    if candle.open < candle.low:
        raise RuntimeError(
            "U04 OPEN/LOW VALIDATION FAILED"
        )

    if candle.open > candle.high:
        raise RuntimeError(
            "U04 OPEN/HIGH VALIDATION FAILED"
        )

    if candle.close < candle.low:
        raise RuntimeError(
            "U04 CLOSE/LOW VALIDATION FAILED"
        )

    if candle.close > candle.high:
        raise RuntimeError(
            "U04 CLOSE/HIGH VALIDATION FAILED"
        )

    if candle.volume < 0:

        raise RuntimeError(
            "U04 VOLUME VALIDATION FAILED"
        )

print("Live data validation: PASSED")


# ============================================================
# 11. STORAGE
# ============================================================

raw_dir = ROOT / "data/raw"
normalized_dir = ROOT / "data/normalized"

raw_dir.mkdir(
    parents=True,
    exist_ok=True
)

normalized_dir.mkdir(
    parents=True,
    exist_ok=True
)

records = []

for candle in live_candles:

    records.append(
        {
            "symbol":
                candle.symbol,

            "timestamp":
                candle.timestamp,

            "open":
                candle.open,

            "high":
                candle.high,

            "low":
                candle.low,

            "close":
                candle.close,

            "volume":
                candle.volume,
        }
    )

raw_file = (
    raw_dir
    / "BTCUSDT_1m_u04.json"
)

normalized_file = (
    normalized_dir
    / "BTCUSDT_1m_u04.jsonl"
)

raw_file.write_text(
    json.dumps(
        records,
        indent=2,
    ),
    encoding="utf-8",
)

with normalized_file.open(
    "w",
    encoding="utf-8",
) as handle:

    for record in records:

        handle.write(
            json.dumps(
                record,
                separators=(",", ":"),
            )
            + "\n"
        )

print("Real capture storage: READY")


# ============================================================
# 12. PROVIDER MANIFEST
# ============================================================

provider_manifest = {

    "project":
        "KITCHEN_ASSISTANT",

    "version":
        "2.8",

    "execution_unit":
        "U04",

    "stage":
        "REAL_MARKET_DATA_PROVIDER_CAPTURE",

    "provider_priority": [
        "BINANCE_SPOT_PUBLIC",
        "COINBASE_EXCHANGE_PUBLIC",
    ],

    "selected_provider":
        selected_provider,

    "symbol":
        SYMBOL,

    "interval":
        INTERVAL,

    "captured_candles":
        len(live_candles),

    "api_key_required":
        False,

    "secret_required":
        False,

    "trading_enabled":
        False,

    "orders_enabled":
        False,

    "technical_lock":
        "LOCKED",

    "audit_status":
        "PASSED",
}

provider_manifest_path = (
    ROOT / "U04_PROVIDER_MANIFEST.json"
)

provider_manifest_path.write_text(
    json.dumps(
        provider_manifest,
        indent=2,
    ),
    encoding="utf-8",
)


# ============================================================
# 13. CAPTURE HASHES
# ============================================================

capture_hashes = {

    "raw_sha256":
        hashlib.sha256(
            raw_file.read_bytes()
        ).hexdigest(),

    "normalized_sha256":
        hashlib.sha256(
            normalized_file.read_bytes()
        ).hexdigest(),
}

capture_hash_path = (
    ROOT / "U04_CAPTURE_SHA256.json"
)

capture_hash_path.write_text(
    json.dumps(
        capture_hashes,
        indent=2,
    ),
    encoding="utf-8",
)


# ============================================================
# 14. U04 AUDIT METADATA
# ============================================================

audit_metadata = {

    "project":
        "KITCHEN_ASSISTANT",

    "version":
        "2.8",

    "execution_unit":
        "U04",

    "stage":
        "REAL_MARKET_DATA_PROVIDER_CAPTURE",

    "audit": {

        "status":
            "PASSED",

        "technical_lock":
            "LOCKED",

        "business_rule_lock":
            "NOT_APPLICABLE",
    },

    "provider_policy": {

        "primary":
            "BINANCE_SPOT_PUBLIC",

        "fallback":
            "COINBASE_EXCHANGE_PUBLIC",

        "priority_rule":
            "BINANCE_FIRST",
    },

    "capture": {

        "symbol":
            SYMBOL,

        "interval":
            INTERVAL,

        "requested_candles":
            LIMIT,

        "received_candles":
            len(live_candles),

        "selected_provider":
            selected_provider,
    },

    "security": {

        "api_key_required":
            False,

        "secret_required":
            False,

        "trading_enabled":
            False,

        "orders_enabled":
            False,
    },

    "validation": {

        "structural_tests":
            "PASSED",

        "import_integrity":
            "PASSED",

        "live_capture":
            "PASSED",

        "chronological":
            "PASSED",

        "duplicate_timestamp":
            "PASSED",

        "ohlc":
            "PASSED",

        "volume":
            "PASSED",
    },
}

audit_path = (
    ROOT / "U04_AUDIT_METADATA.json"
)

audit_path.write_text(
    json.dumps(
        audit_metadata,
        indent=2,
    ),
    encoding="utf-8",
)


# ============================================================
# 15. U04 LEDGER
# ============================================================

ledger = f"""# KITCHEN ASSISTANT V2.8 — U04 LEDGER

## Execution Unit

U04 — Real Market Data Provider / Capture

## Provider Priority

1. BINANCE_SPOT_PUBLIC
2. COINBASE_EXCHANGE_PUBLIC

## Provider Selection Rule

Binance is always the first-choice provider.

Coinbase is used only when Binance cannot provide
the requested market data.

## Selected Provider

{selected_provider}

## Capture

Symbol: {SYMBOL}
Interval: {INTERVAL}
Requested candles: {LIMIT}
Received candles: {len(live_candles)}

## Authentication

No API key required.
No secret stored.

## Validation

- Import integrity
- Structural tests
- Provider response validation
- Candle model validation
- OHLC validation
- Volume validation
- Chronological validation
- Duplicate timestamp validation

## Storage

- Raw capture
- Normalized JSONL capture

## Trading

Trading logic: NOT CONNECTED
Order execution: NOT CONNECTED

## Technical Lock

LOCKED

## Audit

PASSED

## Cell Integrity Rule

No execution-error cell is accepted into the
working notebook.

A failed cell must be removed/replaced,
rerun successfully, and verified before
the Unit is locked.

## Next

U05 — Historical Dataset / Multi-Symbol Capture
"""

ledger_path = (
    ROOT / "ledgers/U04_LEDGER.md"
)

ledger_path.write_text(
    ledger.strip() + "\n",
    encoding="utf-8",
)


# ============================================================
# 16. FINAL ARTIFACT CHECK
# ============================================================

final_check = {

    "binance_adapter":
        (
            ROOT
            / "app/data/capture/binance.py"
        ).is_file(),

    "coinbase_adapter":
        (
            ROOT
            / "app/data/capture/coinbase.py"
        ).is_file(),

    "provider_router":
        (
            ROOT
            / "app/data/capture/router.py"
        ).is_file(),

    "structural_tests":
        (
            ROOT
            / "tests/data/test_u04.py"
        ).is_file(),

    "provider_manifest":
        provider_manifest_path.is_file(),

    "audit_metadata":
        audit_path.is_file(),

    "raw_capture":
        raw_file.is_file(),

    "normalized_capture":
        normalized_file.is_file(),

    "capture_hash":
        capture_hash_path.is_file(),

    "ledger":
        ledger_path.is_file(),
}

print()
print("Final U04 artifact self-check:")
print(
    json.dumps(
        final_check,
        indent=2,
    )
)

if not all(final_check.values()):
    raise RuntimeError(
        "U04 FINAL ARTIFACT CHECK FAILED"
    )


# ============================================================
# 17. FINAL LOCK
# ============================================================

print()
print("=" * 60)
print("U04 COMPLETE")
print("AUDIT        : PASSED")
print("STATUS       : LOCKED")
print("TECHNICAL    : LOCKED")
print("PRIMARY      : BINANCE_SPOT_PUBLIC")
print("FALLBACK     : COINBASE_EXCHANGE_PUBLIC")
print("SELECTED     :", selected_provider)
print("LIVE DATA    : CAPTURED")
print("TRADING      : DISABLED")
print("ORDERS       : DISABLED")
print("NEXT         : U05")
print("=" * 60)
```

```python
# ============================================================
# KITCHEN ASSISTANT V3.1 — U05
#
# REAL MARKET DATA / MULTI-SYMBOL CAPTURE
# FINAL CORRECTED U05
#
# ============================================================
# FINAL U05 CONTRACT
# ============================================================
#
# ASSETS:
#   BTCUSDT
#   ETHUSDT
#   SOLUSDT
#   XRPUSDT
#   ADAUSDT
#
# PROVIDER PRIORITY:
#   1) BINANCE_SPOT_PUBLIC
#   2) COINBASE_EXCHANGE_PUBLIC
#
# VOLUME:
#   - REAL PROVIDER-SUPPLIED TRADED VOLUME
#   - EACH TRADE IS COUNTED ONCE
#   - BUYER + SELLER ARE NOT ADDED TOGETHER
#   - NO CLOSE * VOLUME
#   - NO MARKET-CAP DERIVATION
#   - NO FABRICATION
#   - NO INTERPOLATION
#
# IMPORTANT:
#   CANDLE COUNT != REQUESTED ROLLING TIME RANGE
#
# DATASET:
#   100 REAL OBSERVED CANDLES PER SYMBOL
#
# ROLLING VOLUME:
#   NOW - 60 MINUTES -> NOW
#
#   This is calculated by timestamp range.
#   It is NOT:
#       - the latest completed 1h candle
#       - the last 60 returned rows
#
# PROVIDER PAGINATION:
#   If Coinbase returns fewer candles in one response,
#   U05 continues backward in time until enough REAL
#   observed candles are collected.
#
# MISSING MARKET MINUTES:
#   Coinbase may omit candles for intervals with no trades.
#
#   U05 NEVER creates zero-volume synthetic candles.
#   Such missing intervals remain missing.
#
# TOTAL / TOTAL2 / TOTAL3:
#   NOT INCLUDED IN U05.
#
#   No:
#       inflow_usd
#       outflow_usd
#       netflow_usd
#       TOTAL volume
#       TOTAL2 volume
#       TOTAL3 volume
#
# TRADING:
#   DISABLED
#
# ORDERS:
#   DISABLED
#
# STRATEGY:
#   DISABLED
#
# ============================================================


from pathlib import Path
from datetime import datetime, timezone, timedelta
from decimal import Decimal, InvalidOperation
import sys
import os
import json
import hashlib
import subprocess
import time
import requests


# ============================================================
# 0. PROJECT / NOTEBOOK
# ============================================================

ROOT = Path(
    "/content/drive/MyDrive/Colab Notebooks"
)

if not ROOT.exists():
    raise RuntimeError(
        f"Project root not found: {ROOT}"
    )

os.chdir(ROOT)

NOTEBOOK_PATH = (
    ROOT / "Kitchen Assistant v3.1.ipynb"
)

if not NOTEBOOK_PATH.is_file():
    raise RuntimeError(
        f"Target notebook not found: {NOTEBOOK_PATH}"
    )


print("=" * 72)
print("KITCHEN ASSISTANT V3.1")
print("Execution Unit : U05")
print("Stage          : Real Market Data / Multi-Symbol Capture")
print("Project Root   :", ROOT)
print("Notebook       :", NOTEBOOK_PATH)
print("=" * 72)


# ============================================================
# 1. DEPENDENCY
# ============================================================

subprocess.check_call(
    [
        sys.executable,
        "-m",
        "pip",
        "install",
        "-q",
        "requests>=2,<3",
    ]
)

print("Runtime dependency: READY")


# ============================================================
# 2. CONFIGURATION
# ============================================================

SYMBOLS = [
    "BTCUSDT",
    "ETHUSDT",
    "SOLUSDT",
    "XRPUSDT",
    "ADAUSDT",
]

INTERVAL = "1m"

CANDLES_PER_SYMBOL = 100

REQUESTED_DURATION_MINUTES = 60

REQUESTED_RANGE_MODE = (
    "ROLLING_END_AT_NOW"
)

PROVIDER_PRIORITY = [
    "BINANCE_SPOT_PUBLIC",
    "COINBASE_EXCHANGE_PUBLIC",
]

TRADING_ENABLED = False
ORDERS_ENABLED = False
STRATEGY_ENABLED = False


if CANDLES_PER_SYMBOL < 1:
    raise RuntimeError(
        "CANDLES_PER_SYMBOL must be positive"
    )

if REQUESTED_DURATION_MINUTES < 1:
    raise RuntimeError(
        "REQUESTED_DURATION_MINUTES must be positive"
    )


# ============================================================
# 3. DATASET DIRECTORIES
# ============================================================

DATASET_ROOT = (
    ROOT
    / "data"
    / "historical_u05"
)

RAW_DIR = (
    DATASET_ROOT
    / "raw"
)

NORMALIZED_DIR = (
    DATASET_ROOT
    / "normalized"
)

MANIFEST_DIR = (
    DATASET_ROOT
    / "manifests"
)

QUALITY_DIR = (
    DATASET_ROOT
    / "quality"
)

LEDGER_DIR = (
    ROOT
    / "ledgers"
)

for directory in [
    DATASET_ROOT,
    RAW_DIR,
    NORMALIZED_DIR,
    MANIFEST_DIR,
    QUALITY_DIR,
    LEDGER_DIR,
]:
    directory.mkdir(
        parents=True,
        exist_ok=True
    )


# ============================================================
# 4. TIME HELPERS
# ============================================================

INTERVAL_MS = 60_000

COINBASE_BATCH_LIMIT = 300


def utc_now():
    return datetime.now(
        timezone.utc
    )


def now_ms():
    return int(
        utc_now().timestamp() * 1000
    )


def rolling_range():
    end_dt = utc_now()

    start_dt = (
        end_dt
        - timedelta(
            minutes=
                REQUESTED_DURATION_MINUTES
        )
    )

    return {
        "mode":
            REQUESTED_RANGE_MODE,

        "duration_minutes":
            REQUESTED_DURATION_MINUTES,

        "start_utc":
            start_dt.strftime(
                "%Y-%m-%d %H:%M:%S UTC"
            ),

        "end_utc":
            end_dt.strftime(
                "%Y-%m-%d %H:%M:%S UTC"
            ),

        "start_ms":
            int(
                start_dt.timestamp()
                * 1000
            ),

        "end_ms":
            int(
                end_dt.timestamp()
                * 1000
            ),
    }


# ============================================================
# 5. SYMBOL NORMALIZATION
# ============================================================

def provider_symbol(
    provider_name,
    symbol,
):

    symbol = (
        str(symbol)
        .upper()
        .strip()
    )

    if provider_name == (
        "BINANCE_SPOT_PUBLIC"
    ):
        return symbol

    if provider_name == (
        "COINBASE_EXCHANGE_PUBLIC"
    ):

        if symbol.endswith("USDT"):
            return (
                symbol[:-4]
                + "-USD"
            )

        if symbol.endswith("USD"):
            return (
                symbol[:-3]
                + "-USD"
            )

        raise ValueError(
            f"Unsupported Coinbase symbol: {symbol}"
        )

    raise ValueError(
        f"Unknown provider: {provider_name}"
    )


# ============================================================
# 6. EXACT NUMERIC CONVERSION
# ============================================================

def positive_decimal(
    value,
    field_name,
):

    try:
        value_decimal = Decimal(
            str(value)
        )
    except (
        InvalidOperation,
        ValueError,
        TypeError,
    ) as exc:

        raise RuntimeError(
            f"Invalid numeric {field_name}: "
            f"{value!r}"
        ) from exc

    if value_decimal <= 0:
        raise RuntimeError(
            f"Non-positive {field_name}: "
            f"{value!r}"
        )

    return float(
        value_decimal
    )


def non_negative_decimal(
    value,
    field_name,
):

    try:
        value_decimal = Decimal(
            str(value)
        )
    except (
        InvalidOperation,
        ValueError,
        TypeError,
    ) as exc:

        raise RuntimeError(
            f"Invalid numeric {field_name}: "
            f"{value!r}"
        ) from exc

    if value_decimal < 0:
        raise RuntimeError(
            f"Negative {field_name}: "
            f"{value!r}"
        )

    return float(
        value_decimal
    )


# ============================================================
# 7. BINANCE PUBLIC PROVIDER
# ============================================================

class BinanceProviderUnavailable(
    Exception
):
    pass


class BinancePublicAdapter:

    provider_name = (
        "BINANCE_SPOT_PUBLIC"
    )

    BASE_URLS = [
        "https://api.binance.com",
        "https://api1.binance.com",
        "https://api2.binance.com",
        "https://api3.binance.com",
        "https://api4.binance.com",
    ]

    def __init__(
        self,
        timeout=15,
    ):

        self.timeout = timeout

    def _get(
        self,
        path,
        params,
    ):

        errors = []

        for base_url in self.BASE_URLS:

            url = (
                base_url.rstrip("/")
                + path
            )

            try:

                response = requests.get(
                    url,
                    params=params,
                    headers={
                        "Accept":
                            "application/json",
                        "User-Agent":
                            "KitchenAssistant-V3.1-U05",
                    },
                    timeout=self.timeout,
                )

                if response.status_code == 200:
                    return response

                errors.append(
                    f"{base_url}: "
                    f"HTTP {response.status_code}"
                )

            except Exception as exc:

                errors.append(
                    f"{base_url}: "
                    f"{repr(exc)}"
                )

        raise BinanceProviderUnavailable(
            "Binance unavailable: "
            + " | ".join(errors)
        )

    def fetch_recent(
        self,
        symbol,
        interval,
        limit,
    ):

        response = self._get(
            "/api/v3/klines",
            {
                "symbol":
                    symbol,

                "interval":
                    interval,

                "limit":
                    min(
                        int(limit),
                        1000
                    ),
            },
        )

        payload = response.json()

        if not isinstance(
            payload,
            list
        ):
            raise BinanceProviderUnavailable(
                "Invalid Binance response"
            )

        candles = []

        for row in payload:

            if not isinstance(
                row,
                list
            ):
                raise BinanceProviderUnavailable(
                    "Malformed Binance kline"
                )

            if len(row) < 8:
                raise BinanceProviderUnavailable(
                    "Incomplete Binance kline"
                )

            timestamp = int(
                row[0]
            )

            open_price = (
                positive_decimal(
                    row[1],
                    "open"
                )
            )

            high_price = (
                positive_decimal(
                    row[2],
                    "high"
                )
            )

            low_price = (
                positive_decimal(
                    row[3],
                    "low"
                )
            )

            close_price = (
                positive_decimal(
                    row[4],
                    "close"
                )
            )

            volume_base = (
                non_negative_decimal(
                    row[5],
                    "volume"
                )
            )

            quote_volume = (
                non_negative_decimal(
                    row[7],
                    "quote_volume"
                )
            )

            candles.append(
                {
                    "timestamp":
                        timestamp,

                    "open":
                        open_price,

                    "high":
                        high_price,

                    "low":
                        low_price,

                    "close":
                        close_price,

                    "volume_base":
                        volume_base,

                    "volume_usd":
                        quote_volume,

                    "volume_usd_exact":
                        True,

                    "volume_usd_source":
                        "BINANCE_KLINE_QUOTE_VOLUME",
                }
            )

        candles.sort(
            key=lambda x:
                x["timestamp"]
        )

        return candles


# ============================================================
# 8. COINBASE PUBLIC PROVIDER
# ============================================================

class CoinbaseProviderUnavailable(
    Exception
):
    pass


class CoinbasePublicAdapter:

    provider_name = (
        "COINBASE_EXCHANGE_PUBLIC"
    )

    BASE_URL = (
        "https://api.exchange.coinbase.com"
    )

    GRANULARITY_SECONDS = {
        "1m": 60,
        "5m": 300,
        "15m": 900,
        "1h": 3600,
        "6h": 21600,
        "1d": 86400,
    }

    def __init__(
        self,
        timeout=20,
    ):

        self.timeout = timeout

    def fetch_batch(
        self,
        product_id,
        start_dt,
        end_dt,
        interval,
    ):

        if interval not in (
            self.GRANULARITY_SECONDS
        ):
            raise ValueError(
                f"Unsupported Coinbase interval: "
                f"{interval}"
            )

        url = (
            self.BASE_URL
            + f"/products/{product_id}/candles"
        )

        params = {
            "start":
                start_dt.isoformat(),

            "end":
                end_dt.isoformat(),

            "granularity":
                self.GRANULARITY_SECONDS[
                    interval
                ],
        }

        response = requests.get(
            url,
            params=params,
            headers={
                "Accept":
                    "application/json",

                "User-Agent":
                    "KitchenAssistant-V3.1-U05",
            },
            timeout=self.timeout,
        )

        if response.status_code != 200:

            raise CoinbaseProviderUnavailable(
                "Coinbase HTTP "
                f"{response.status_code}: "
                f"{response.text[:300]}"
            )

        payload = response.json()

        if not isinstance(
            payload,
            list
        ):
            raise CoinbaseProviderUnavailable(
                "Invalid Coinbase response"
            )

        candles = []

        for row in payload:

            if not isinstance(
                row,
                list
            ):
                raise CoinbaseProviderUnavailable(
                    "Malformed Coinbase candle"
                )

            if len(row) < 6:
                raise CoinbaseProviderUnavailable(
                    "Incomplete Coinbase candle"
                )

            # Coinbase format:
            #
            # [time, low, high, open, close, volume]
            #
            # time = Unix seconds

            timestamp = (
                int(row[0])
                * 1000
            )

            low_price = (
                positive_decimal(
                    row[1],
                    "low"
                )
            )

            high_price = (
                positive_decimal(
                    row[2],
                    "high"
                )
            )

            open_price = (
                positive_decimal(
                    row[3],
                    "open"
                )
            )

            close_price = (
                positive_decimal(
                    row[4],
                    "close"
                )
            )

            volume_base = (
                non_negative_decimal(
                    row[5],
                    "volume"
                )
            )

            candles.append(
                {
                    "timestamp":
                        timestamp,

                    "open":
                        open_price,

                    "high":
                        high_price,

                    "low":
                        low_price,

                    "close":
                        close_price,

                    "volume_base":
                        volume_base,

                    # Coinbase candles do NOT provide
                    # an exact USD quote-volume field.
                    #
                    # Therefore:
                    # DO NOT calculate close * volume.
                    "volume_usd":
                        None,

                    "volume_usd_exact":
                        False,

                    "volume_usd_source":
                        None,
                }
            )

        return candles


# ============================================================
# 9. COINBASE PAGINATED REAL-CANDLE CAPTURE
# ============================================================
#
# IMPORTANT:
#
# Coinbase can return fewer rows than requested because:
#
#   - its endpoint has a response limit
#   - some minute intervals may have no trades
#
# We therefore paginate backwards in time.
#
# We NEVER manufacture missing candles.
#
# ============================================================

def fetch_coinbase_real_candles(
    adapter,
    symbol,
    interval,
    required_count,
):

    product_id = provider_symbol(
        "COINBASE_EXCHANGE_PUBLIC",
        symbol,
    )

    collected = {}

    end_dt = utc_now()

    # Each request covers at most 300 minutes.
    # We go farther back as necessary.
    batch_minutes = (
        300
    )

    max_pages = 20

    for page in range(
        max_pages
    ):

        start_dt = (
            end_dt
            - timedelta(
                minutes=batch_minutes
            )
        )

        try:

            batch = (
                adapter.fetch_batch(
                    product_id=
                        product_id,

                    start_dt=
                        start_dt,

                    end_dt=
                        end_dt,

                    interval=
                        interval,
                )
            )

        except Exception as exc:

            if page == 0:
                raise

            raise CoinbaseProviderUnavailable(
                "Coinbase pagination failed "
                f"after {page} page(s): "
                f"{repr(exc)}"
            )

        if not batch:
            break

        for candle in batch:

            timestamp = int(
                candle["timestamp"]
            )

            collected[
                timestamp
            ] = candle

        if len(collected) >= (
            required_count
        ):
            break

        oldest_timestamp = min(
            collected.keys()
        )

        # Move strictly before the oldest
        # returned candle.
        end_dt = (
            datetime.fromtimestamp(
                (
                    oldest_timestamp
                    - INTERVAL_MS
                )
                / 1000,
                tz=timezone.utc,
            )
        )

        # Expand the search window if needed.
        batch_minutes = min(
            batch_minutes * 2,
            300
        )

    candles = sorted(
        collected.values(),
        key=lambda x:
            x["timestamp"]
    )

    if len(candles) < (
        required_count
    ):

        raise CoinbaseProviderUnavailable(
            f"Coinbase could provide only "
            f"{len(candles)} real observed candles "
            f"for {symbol}; "
            f"{required_count} required."
        )

    return candles[
        -required_count:
    ]


# ============================================================
# 10. PROVIDER FETCH
# ============================================================

binance = (
    BinancePublicAdapter()
)

coinbase = (
    CoinbasePublicAdapter()
)


def fetch_from_provider(
    provider_name,
    symbol,
):

    started = time.time()

    if provider_name == (
        "BINANCE_SPOT_PUBLIC"
    ):

        provider_symbol_value = (
            provider_symbol(
                provider_name,
                symbol,
            )
        )

        candles = (
            binance.fetch_recent(
                symbol=symbol,
                interval=INTERVAL,
                limit=CANDLES_PER_SYMBOL,
            )
        )

    elif provider_name == (
        "COINBASE_EXCHANGE_PUBLIC"
    ):

        provider_symbol_value = (
            provider_symbol(
                provider_name,
                symbol,
            )
        )

        candles = (
            fetch_coinbase_real_candles(
                adapter=coinbase,
                symbol=symbol,
                interval=INTERVAL,
                required_count=
                    CANDLES_PER_SYMBOL,
            )
        )

    else:

        raise ValueError(
            f"Unknown provider: {provider_name}"
        )

    elapsed = (
        time.time()
        - started
    )

    if len(candles) != (
        CANDLES_PER_SYMBOL
    ):

        raise RuntimeError(
            f"Provider returned "
            f"{len(candles)} real candles for "
            f"{symbol}; expected "
            f"{CANDLES_PER_SYMBOL}"
        )

    return {
        "provider":
            provider_name,

        "exchange":
            provider_name,

        "provider_symbol":
            provider_symbol_value,

        "records":
            candles,

        "elapsed_seconds":
            round(
                elapsed,
                3
            ),
    }


# ============================================================
# 11. PROVIDER PRIORITY ROUTER
# ============================================================

def capture_with_priority(
    symbol,
):

    attempts = []

    for provider_name in (
        PROVIDER_PRIORITY
    ):

        print()
        print(
            f"Trying provider: "
            f"{provider_name}"
        )

        try:

            result = (
                fetch_from_provider(
                    provider_name=
                        provider_name,

                    symbol=
                        symbol,
                )
            )

            print(
                f"SUCCESS: "
                f"{provider_name}"
            )

            return (
                result,
                attempts,
            )

        except Exception as exc:

            error_record = {

                "provider":
                    provider_name,

                "status":
                    "FAILED",

                "error_type":
                    type(exc).__name__,

                "error":
                    repr(exc),
            }

            attempts.append(
                error_record
            )

            print(
                f"Provider unavailable: "
                f"{provider_name}"
            )

            print(
                "Reason:",
                repr(exc)
            )

    raise RuntimeError(
        "ALL PROVIDERS FAILED for "
        f"{symbol}: "
        + json.dumps(
            attempts,
            indent=2
        )
    )


# ============================================================
# 12. DATASET RECORD
# ============================================================

def build_record(
    candle,
    symbol,
    provider_name,
    provider_symbol_value,
):

    volume_base = (
        non_negative_decimal(
            candle["volume_base"],
            "volume_base"
        )
    )

    record = {

        "project":
            "KITCHEN_ASSISTANT",

        "version":
            "3.1",

        "execution_unit":
            "U05",

        "asset_type":
            "ASSET",

        "symbol":
            symbol,

        "provider":
            provider_name,

        "exchange":
            provider_name,

        "provider_symbol":
            provider_symbol_value,

        "interval":
            INTERVAL,

        "timestamp":
            int(
                candle["timestamp"]
            ),

        "open":
            positive_decimal(
                candle["open"],
                "open"
            ),

        "high":
            positive_decimal(
                candle["high"],
                "high"
            ),

        "low":
            positive_decimal(
                candle["low"],
                "low"
            ),

        "close":
            positive_decimal(
                candle["close"],
                "close"
            ),

        # ====================================================
        # REAL TRADING VOLUME
        # ====================================================

        "volume_base":
            volume_base,

        "volume_definition":
            "TOTAL_TRADED_BASE_ASSET_VOLUME",

        "volume_source":
            "PROVIDER_SUPPLIED",

        "volume_counting_rule":
            "EACH_TRADE_COUNTED_ONCE",

        "volume_double_counting":
            False,

        "buyer_seller_not_double_counted":
            True,

        # ====================================================
        # EXACT USD QUOTE VOLUME
        # ====================================================

        "volume_usd":
            candle.get(
                "volume_usd"
            ),

        "volume_usd_exact":
            bool(
                candle.get(
                    "volume_usd_exact",
                    False
                )
            ),

        "volume_usd_source":
            candle.get(
                "volume_usd_source"
            ),

        # ====================================================
        # TOTAL SERIES
        # ====================================================

        "total_series_included":
            False,

        # ====================================================
        # TRADING
        # ====================================================

        "trading_enabled":
            False,

        "orders_enabled":
            False,

        "strategy_enabled":
            False,
    }

    return record


# ============================================================
# 13. RECORD VALIDATION
# ============================================================

def validate_records(
    records,
    expected_symbol,
):

    if len(records) != (
        CANDLES_PER_SYMBOL
    ):

        raise RuntimeError(
            f"REAL CANDLE COUNT FAILED: "
            f"{expected_symbol}: "
            f"{len(records)} != "
            f"{CANDLES_PER_SYMBOL}"
        )

    timestamps = [
        int(
            record["timestamp"]
        )
        for record in records
    ]

    if timestamps != sorted(
        timestamps
    ):

        raise RuntimeError(
            f"Chronological validation failed: "
            f"{expected_symbol}"
        )

    if len(timestamps) != len(
        set(timestamps)
    ):

        raise RuntimeError(
            f"Duplicate timestamps detected: "
            f"{expected_symbol}"
        )

    for record in records:

        if record["symbol"] != (
            expected_symbol
        ):

            raise RuntimeError(
                f"Symbol mismatch: "
                f"{expected_symbol}"
            )

        open_price = float(
            record["open"]
        )

        high_price = float(
            record["high"]
        )

        low_price = float(
            record["low"]
        )

        close_price = float(
            record["close"]
        )

        volume = float(
            record["volume_base"]
        )

        if open_price <= 0:
            raise RuntimeError(
                f"Invalid open: "
                f"{expected_symbol}"
            )

        if high_price < low_price:
            raise RuntimeError(
                f"Invalid high/low: "
                f"{expected_symbol}"
            )

        if not (
            low_price
            <= open_price
            <= high_price
        ):

            raise RuntimeError(
                f"Open outside candle range: "
                f"{expected_symbol}"
            )

        if not (
            low_price
            <= close_price
            <= high_price
        ):

            raise RuntimeError(
                f"Close outside candle range: "
                f"{expected_symbol}"
            )

        if volume < 0:
            raise RuntimeError(
                f"Negative volume: "
                f"{expected_symbol}"
            )

        if record[
            "volume_source"
        ] != "PROVIDER_SUPPLIED":

            raise RuntimeError(
                "Volume source validation failed"
            )

        if record[
            "volume_counting_rule"
        ] != "EACH_TRADE_COUNTED_ONCE":

            raise RuntimeError(
                "Volume counting rule invalid"
            )

        if record[
            "volume_double_counting"
        ] is not False:

            raise RuntimeError(
                "Volume double counting detected"
            )

        if record[
            "buyer_seller_not_double_counted"
        ] is not True:

            raise RuntimeError(
                "Buyer/seller double-counting rule failed"
            )


# ============================================================
# 14. ROLLING VOLUME
# ============================================================
#
# CRITICAL:
#
# We do NOT use:
#
#   records[-60:]
#
# because "last hour" means:
#
#   NOW - 60m -> NOW
#
# based on actual timestamps.
#
# ============================================================

def calculate_rolling_volume(
    provider_name,
    symbol,
):

    range_info = (
        rolling_range()
    )

    start_ms = int(
        range_info["start_ms"]
    )

    end_ms = int(
        range_info["end_ms"]
    )

    provider_symbol_value = (
        provider_symbol(
            provider_name,
            symbol,
        )
    )

    if provider_name == (
        "BINANCE_SPOT_PUBLIC"
    ):

        # Request enough recent 1m candles
        # to cover the rolling range.
        candles = (
            binance.fetch_recent(
                symbol=symbol,
                interval=INTERVAL,
                limit=100,
            )
        )

    elif provider_name == (
        "COINBASE_EXCHANGE_PUBLIC"
    ):

        product_id = (
            provider_symbol_value
        )

        end_dt = (
            datetime.fromtimestamp(
                end_ms / 1000,
                tz=timezone.utc,
            )
        )

        start_dt = (
            datetime.fromtimestamp(
                start_ms / 1000,
                tz=timezone.utc,
            )
        )

        candles = (
            coinbase.fetch_batch(
                product_id=
                    product_id,

                start_dt=
                    start_dt,

                end_dt=
                    end_dt,

                interval=
                    INTERVAL,
            )
        )

    else:

        raise ValueError(
            f"Unknown provider: "
            f"{provider_name}"
        )

    in_range = [

        candle
        for candle in candles

        if (
            start_ms
            <= int(
                candle["timestamp"]
            )
            <= end_ms
        )
    ]

    rolling_base_volume = sum(
        float(
            candle["volume_base"]
        )
        for candle
        in in_range
    )

    exact_quote_available = all(
        bool(
            candle.get(
                "volume_usd_exact",
                False
            )
        )
        for candle
        in in_range
    ) and bool(in_range)

    if exact_quote_available:

        rolling_usd_volume = sum(
            float(
                candle["volume_usd"]
            )
            for candle
            in in_range
        )

        usd_source = (
            "PROVIDER_QUOTE_VOLUME"
        )

    else:

        rolling_usd_volume = None

        usd_source = None

    return {

        "mode":
            REQUESTED_RANGE_MODE,

        "start_utc":
            range_info["start_utc"],

        "end_utc":
            range_info["end_utc"],

        "start_ms":
            start_ms,

        "end_ms":
            end_ms,

        "duration_minutes":
            REQUESTED_DURATION_MINUTES,

        "interval":
            INTERVAL,

        "provider":
            provider_name,

        "provider_symbol":
            provider_symbol_value,

        "candles_in_time_range":
            len(in_range),

        "rolling_volume_base":
            rolling_base_volume,

        "rolling_volume_definition":
            "TOTAL_TRADED_BASE_ASSET_VOLUME",

        "volume_counting_rule":
            "EACH_TRADE_COUNTED_ONCE",

        "buyer_seller_double_counted":
            False,

        "rolling_volume_source":
            "PROVIDER_SUPPLIED",

        "rolling_volume_usd":
            rolling_usd_volume,

        "rolling_volume_usd_exact":
            exact_quote_available,

        "rolling_volume_usd_source":
            usd_source,

        "not_last_completed_candle":
            True,

        "not_last_n_rows":
            True,

        "no_close_times_volume":
            True,

        "no_market_cap_volume":
            True,
    }


# ============================================================
# 15. DATASET CAPTURE
# ============================================================

dataset_summary = []

all_records = []

provider_statistics = {}

rolling_volume_results = {}

capture_started = time.time()

print()
print("=" * 72)
print("STARTING U05 MULTI-SYMBOL CAPTURE")
print("=" * 72)

print(
    "Candle interval:",
    INTERVAL
)

print(
    "Real candles per symbol:",
    CANDLES_PER_SYMBOL
)

print(
    "Rolling volume range:",
    f"NOW - {REQUESTED_DURATION_MINUTES}m -> NOW"
)

print(
    "Symbols:",
    SYMBOLS
)

print("=" * 72)


for symbol in SYMBOLS:

    print()
    print("-" * 72)

    print(
        f"Asset: {symbol}"
    )

    print(
        f"Interval: {INTERVAL}"
    )

    print(
        "Rolling volume:",
        f"NOW - {REQUESTED_DURATION_MINUTES}m -> NOW"
    )

    print("-" * 72)

    symbol_started = time.time()

    result, attempts = (
        capture_with_priority(
            symbol
        )
    )

    provider_name = (
        result["provider"]
    )

    provider_symbol_value = (
        result["provider_symbol"]
    )

    raw_candles = (
        result["records"]
    )

    records = [

        build_record(
            candle=candle,
            symbol=symbol,
            provider_name=
                provider_name,
            provider_symbol_value=
                provider_symbol_value,
        )

        for candle
        in raw_candles
    ]

    validate_records(
        records,
        symbol,
    )

    # --------------------------------------------------------
    # Rolling volume
    # --------------------------------------------------------

    rolling_volume = (
        calculate_rolling_volume(
            provider_name=
                provider_name,

            symbol=
                symbol,
        )
    )

    rolling_volume_results[
        symbol
    ] = rolling_volume

    provider_statistics.setdefault(
        provider_name,
        0
    )

    provider_statistics[
        provider_name
    ] += 1

    # --------------------------------------------------------
    # Raw JSON
    # --------------------------------------------------------

    raw_file = (
        RAW_DIR
        / f"{symbol}_{INTERVAL}.json"
    )

    raw_payload = {

        "project":
            "KITCHEN_ASSISTANT",

        "version":
            "3.1",

        "execution_unit":
            "U05",

        "symbol":
            symbol,

        "interval":
            INTERVAL,

        "real_candles_requested":
            CANDLES_PER_SYMBOL,

        "real_candles_captured":
            len(records),

        "provider":
            provider_name,

        "exchange":
            result["exchange"],

        "provider_symbol":
            provider_symbol_value,

        "provider_attempts":
            attempts,

        "rolling_volume":
            rolling_volume,

        "volume_contract":
            {
                "definition":
                    "TOTAL_TRADED_BASE_ASSET_VOLUME",

                "each_trade_counted_once":
                    True,

                "buyer_and_seller_not_double_counted":
                    True,

                "provider_supplied":
                    True,

                "fabricated":
                    False,

                "interpolation":
                    False,

                "market_cap_derived":
                    False,

                "close_times_volume":
                    False,

                "last_completed_candle_substitution":
                    False,

                "last_n_rows_substitution":
                    False,
            },

        "total_series":
            {
                "included":
                    False,

                "TOTAL":
                    False,

                "TOTAL2":
                    False,

                "TOTAL3":
                    False,

                "inflow_usd":
                    False,

                "outflow_usd":
                    False,

                "netflow_usd":
                    False,
            },

        "records":
            records,
    }

    raw_file.write_text(
        json.dumps(
            raw_payload,
            indent=2
        ),
        encoding="utf-8"
    )

    # --------------------------------------------------------
    # Normalized JSONL
    # --------------------------------------------------------

    normalized_file = (
        NORMALIZED_DIR
        / f"{symbol}_{INTERVAL}.jsonl"
    )

    with normalized_file.open(
        "w",
        encoding="utf-8"
    ) as handle:

        for record in records:

            handle.write(
                json.dumps(
                    record,
                    separators=(",", ":")
                )
                + "\n"
            )

            all_records.append(
                record
            )

    elapsed = (
        time.time()
        - symbol_started
    )

    summary = {

        "symbol":
            symbol,

        "provider":
            provider_name,

        "exchange":
            result["exchange"],

        "provider_symbol":
            provider_symbol_value,

        "interval":
            INTERVAL,

        "real_candle_count":
            len(records),

        "requested_real_candle_count":
            CANDLES_PER_SYMBOL,

        "first_timestamp":
            records[0]["timestamp"],

        "last_timestamp":
            records[-1]["timestamp"],

        "rolling_volume":
            rolling_volume,

        "volume_definition":
            "TOTAL_TRADED_BASE_ASSET_VOLUME",

        "volume_counting":
            "EACH_TRADE_COUNTED_ONCE",

        "volume_double_counting":
            False,

        "volume_source":
            "PROVIDER_SUPPLIED",

        "volume_usd_policy":
            "EXACT_PROVIDER_QUOTE_VOLUME_ONLY",

        "total_series":
            "NOT_INCLUDED",

        "provider_attempts":
            attempts,

        "elapsed_seconds":
            round(
                elapsed,
                3
            ),

        "status":
            "VALIDATED",
    }

    dataset_summary.append(
        summary
    )

    print()
    print(
        f"{symbol}: "
        f"{len(records)} REAL candles | "
        f"provider={provider_name} | "
        f"VALIDATED"
    )

    print(
        "Rolling base volume:",
        rolling_volume[
            "rolling_volume_base"
        ]
    )

    print(
        "Rolling USD volume:",
        rolling_volume[
            "rolling_volume_usd"
        ]
    )


# ============================================================
# 16. GLOBAL DATASET VALIDATION
# ============================================================

print()
print("=" * 72)
print("GLOBAL DATASET VALIDATION")
print("=" * 72)

expected_count = (
    len(SYMBOLS)
    * CANDLES_PER_SYMBOL
)

actual_count = len(
    all_records
)

# IMPORTANT:
# This check is now based on REAL candles collected
# AFTER provider pagination.
#
# Therefore a partial provider response cannot cause
# an artificial "296 vs 500" failure.
#
# If fewer than 100 REAL candles genuinely exist,
# provider capture itself fails before reaching here.
#

if actual_count != expected_count:

    raise RuntimeError(
        "GLOBAL REAL CANDLE COUNT VALIDATION FAILED: "
        f"expected={expected_count}, "
        f"actual={actual_count}"
    )


captured_symbols = sorted(
    set(
        record["symbol"]
        for record in all_records
    )
)

expected_symbols = sorted(
    SYMBOLS
)

if captured_symbols != (
    expected_symbols
):

    raise RuntimeError(
        "SYMBOL COVERAGE VALIDATION FAILED: "
        f"expected={expected_symbols}, "
        f"actual={captured_symbols}"
    )


for symbol in SYMBOLS:

    symbol_records = [

        record
        for record
        in all_records

        if record["symbol"]
        == symbol
    ]

    validate_records(
        symbol_records,
        symbol,
    )


print(
    "Global real-candle validation: PASSED"
)


# ============================================================
# 17. ROLLING VOLUME VALIDATION
# ============================================================

for symbol in SYMBOLS:

    volume_data = (
        rolling_volume_results[
            symbol
        ]
    )

    if volume_data[
        "mode"
    ] != REQUESTED_RANGE_MODE:

        raise RuntimeError(
            f"Rolling range mode failed: "
            f"{symbol}"
        )

    if volume_data[
        "duration_minutes"
    ] != REQUESTED_DURATION_MINUTES:

        raise RuntimeError(
            f"Rolling duration failed: "
            f"{symbol}"
        )

    if volume_data[
        "not_last_completed_candle"
    ] is not True:

        raise RuntimeError(
            f"Last completed candle substitution "
            f"detected: {symbol}"
        )

    if volume_data[
        "not_last_n_rows"
    ] is not True:

        raise RuntimeError(
            f"Last-N-rows substitution detected: "
            f"{symbol}"
        )

    if volume_data[
        "buyer_seller_double_counted"
    ] is not False:

        raise RuntimeError(
            f"Volume double-counting detected: "
            f"{symbol}"
        )

    if volume_data[
        "rolling_volume_base"
    ] < 0:

        raise RuntimeError(
            f"Negative rolling volume: "
            f"{symbol}"
        )


print(
    "Rolling volume validation: PASSED"
)


# ============================================================
# 18. PROVIDER COVERAGE
# ============================================================

provider_coverage = {

    provider_name: {

        "asset_count":
            provider_statistics.get(
                provider_name,
                0
            ),

        "symbols":
            [
                item["symbol"]
                for item
                in dataset_summary

                if item["provider"]
                == provider_name
            ],
    }

    for provider_name
    in PROVIDER_PRIORITY
}


print()
print("Provider coverage:")

print(
    json.dumps(
        provider_coverage,
        indent=2
    )
)


# ============================================================
# 19. TIMEFRAME CONTRACT
# ============================================================

timeframe_contract = {

    "mode":
        "PRESET",

    "name":
        "U05_VALIDATION",

    "interval":
        INTERVAL,

    "real_candles_per_symbol":
        CANDLES_PER_SYMBOL,

    "supports_custom":
        True,

    "custom_contract":
        {
            "name":
                "CUSTOM",

            "interval":
                "PROVIDER_SUPPORTED_INTERVAL",

            "real_candles":
                "USER_DEFINED_POSITIVE_INTEGER",
        },
}


# ============================================================
# 20. ROLLING VOLUME CONTRACT
# ============================================================

rolling_volume_contract = {

    "enabled":
        True,

    "definition":
        "TOTAL_TRADED_BASE_ASSET_VOLUME",

    "counting_rule":
        "EACH_TRADE_COUNTED_ONCE",

    "buyer_seller_double_counting":
        False,

    "source":
        "PROVIDER_SUPPLIED_TRADING_VOLUME",

    "range":
        "NOW_BACKWARD",

    "range_format":
        "NOW - X MINUTES -> NOW",

    "interval_independent":
        True,

    "not_last_completed_candle":
        True,

    "not_last_n_rows":
        True,

    "fabrication":
        False,

    "interpolation":
        False,

    "market_cap_derivation":
        False,

    "close_times_volume":
        False,
}


# ============================================================
# 21. USD VOLUME CONTRACT
# ============================================================

usd_volume_contract = {

    "metric":
        "EXACT_PROVIDER_QUOTE_VOLUME_ONLY",

    "unit":
        "USD",

    "provider_quote_volume_only":
        True,

    "estimation":
        False,

    "close_times_volume":
        False,

    "market_cap_derivation":
        False,

    "if_not_supplied":
        "UNAVAILABLE",
}


# ============================================================
# 22. TOTAL SERIES CONTRACT
# ============================================================
#
# IMPORTANT:
#
# TOTAL / TOTAL2 / TOTAL3 are deliberately absent
# from the U05 dataset.
#
# We do NOT create placeholder UNAVAILABLE volume fields.
#
# ============================================================

total_contract = {

    "included_in_u05":
        False,

    "TOTAL":
        "NOT_INCLUDED",

    "TOTAL2":
        "NOT_INCLUDED",

    "TOTAL3":
        "NOT_INCLUDED",

    "volume":
        "NOT_INCLUDED",

    "inflow_usd":
        "NOT_INCLUDED",

    "outflow_usd":
        "NOT_INCLUDED",

    "netflow_usd":
        "NOT_INCLUDED",

    "dominance_volume":
        "NOT_INCLUDED",

    "dominance_flow":
        "NOT_INCLUDED",

    "fabrication":
        False,

    "market_cap_change_as_flow":
        False,

    "placeholder_unavailable":
        False,
}


# ============================================================
# 23. DATASET INDEX
# ============================================================

dataset_index = {

    "project":
        "KITCHEN_ASSISTANT",

    "version":
        "3.1",

    "execution_unit":
        "U05",

    "dataset_name":
        "U05_REAL_MARKET_DATA_MULTI_SYMBOL",

    "status":
        "VALIDATED",

    "provider_priority":
        PROVIDER_PRIORITY,

    "provider_rule":
        "BINANCE_FIRST",

    "timeframe":
        timeframe_contract,

    "rolling_volume":
        rolling_volume_contract,

    "usd_volume":
        usd_volume_contract,

    "total_series":
        total_contract,

    "symbols":
        SYMBOLS,

    "symbol_count":
        len(SYMBOLS),

    "real_candles_per_symbol":
        CANDLES_PER_SYMBOL,

    "total_real_candles":
        actual_count,

    "provider_coverage":
        provider_coverage,

    "trading_enabled":
        False,

    "orders_enabled":
        False,

    "strategy_enabled":
        False,
}


index_file = (
    DATASET_ROOT
    / "DATASET_INDEX.json"
)

index_file.write_text(
    json.dumps(
        dataset_index,
        indent=2
    ),
    encoding="utf-8"
)


# ============================================================
# 24. CAPTURE MANIFEST
# ============================================================

capture_manifest = {

    "project":
        "KITCHEN_ASSISTANT",

    "version":
        "3.1",

    "execution_unit":
        "U05",

    "stage":
        "REAL_MARKET_DATA_MULTI_SYMBOL_CAPTURE",

    "status":
        "VALIDATED",

    "provider_policy":
        {
            "primary":
                "BINANCE_SPOT_PUBLIC",

            "fallback":
                "COINBASE_EXCHANGE_PUBLIC",

            "priority":
                PROVIDER_PRIORITY,

            "rule":
                "BINANCE_FIRST",

            "silent_switching":
                False,
        },

    "timeframe":
        timeframe_contract,

    "rolling_volume_contract":
        rolling_volume_contract,

    "rolling_volume_results":
        rolling_volume_results,

    "symbols":
        SYMBOLS,

    "symbol_count":
        len(SYMBOLS),

    "real_candles_per_symbol":
        CANDLES_PER_SYMBOL,

    "total_real_candles":
        actual_count,

    "symbol_results":
        dataset_summary,

    "volume_policy":
        {
            "definition":
                "TOTAL_TRADED_BASE_ASSET_VOLUME",

            "each_trade_counted_once":
                True,

            "buyer_seller_not_double_counted":
                True,

            "fabrication":
                False,

            "interpolation":
                False,

            "close_times_volume":
                False,

            "market_cap_derivation":
                False,
        },

    "total_series_policy":
        total_contract,

    "security":
        {
            "api_key_required":
                False,

            "secret_required":
                False,

            "trading_enabled":
                False,

            "orders_enabled":
                False,

            "strategy_enabled":
                False,
        },

    "generated_at_utc":
        utc_now().strftime(
            "%Y-%m-%d %H:%M:%S UTC"
        ),
}


manifest_file = (
    MANIFEST_DIR
    / "U05_CAPTURE_MANIFEST.json"
)

manifest_file.write_text(
    json.dumps(
        capture_manifest,
        indent=2
    ),
    encoding="utf-8"
)


# ============================================================
# 25. QUALITY REPORT
# ============================================================

quality_report = {

    "execution_unit":
        "U05",

    "status":
        "PASS",

    "total_symbols":
        len(SYMBOLS),

    "expected_real_candles":
        expected_count,

    "actual_real_candles":
        actual_count,

    "symbol_coverage":
        True,

    "chronological_validation":
        True,

    "duplicate_validation":
        True,

    "ohlc_validation":
        True,

    "base_volume_validation":
        True,

    "real_provider_volume":
        True,

    "provider_provenance":
        True,

    "provider_priority":
        "BINANCE_FIRST",

    "silent_provider_switching":
        False,

    "synthetic_data":
        False,

    "interpolation":
        False,

    "live_as_historical":
        False,

    "market_cap_volume":
        False,

    "close_times_volume":
        False,

    "volume_double_counting":
        False,

    "volume_definition":
        "TOTAL_TRADED_BASE_ASSET_VOLUME",

    "volume_counting_rule":
        "EACH_TRADE_COUNTED_ONCE",

    "rolling_range":
        "NOW_BACKWARD",

    "rolling_range_duration_minutes":
        REQUESTED_DURATION_MINUTES,

    "last_completed_candle_substitution":
        False,

    "last_n_rows_substitution":
        False,

    "exact_usd_volume_policy":
        "PROVIDER_QUOTE_VOLUME_ONLY",

    "unsupported_usd_volume":
        "UNAVAILABLE",

    "total_series_in_u05":
        False,

    "total_flow_in_u05":
        False,

    "total_flow_fabrication":
        False,

    "dominance_volume":
        "NOT_INCLUDED",

    "dominance_flow":
        "NOT_INCLUDED",
}


quality_file = (
    QUALITY_DIR
    / "U05_DATA_QUALITY_REPORT.json"
)

quality_file.write_text(
    json.dumps(
        quality_report,
        indent=2
    ),
    encoding="utf-8"
)


# ============================================================
# 26. SHA-256 INVENTORY
# ============================================================

files_to_hash = [

    index_file,

    manifest_file,

    quality_file,
]

for symbol in SYMBOLS:

    files_to_hash.extend([

        RAW_DIR
        / f"{symbol}_{INTERVAL}.json",

        NORMALIZED_DIR
        / f"{symbol}_{INTERVAL}.jsonl",

    ])


hash_inventory = {}

for file_path in files_to_hash:

    if not file_path.is_file():

        raise RuntimeError(
            "Missing U05 artifact: "
            f"{file_path}"
        )

    relative_path = str(
        file_path.relative_to(ROOT)
    )

    hash_inventory[
        relative_path
    ] = hashlib.sha256(
        file_path.read_bytes()
    ).hexdigest()


hash_file = (
    DATASET_ROOT
    / "U05_SHA256.json"
)

hash_file.write_text(
    json.dumps(
        hash_inventory,
        indent=2
    ),
    encoding="utf-8"
)


# ============================================================
# 27. U05 LEDGER
# ============================================================

capture_duration = (
    time.time()
    - capture_started
)

ledger_text = f"""
# KITCHEN ASSISTANT V3.1 — U05 LEDGER

## Execution Unit

U05 — Real Market Data / Multi-Symbol Capture

## Status

VALIDATED

## Provider Policy

Primary: BINANCE_SPOT_PUBLIC

Fallback: COINBASE_EXCHANGE_PUBLIC

Priority: BINANCE_FIRST

Silent provider switching: PROHIBITED

Every selected provider is recorded with:

- provider
- exchange
- provider_symbol
- provider attempts
- capture result

## Candle Dataset

Symbols:

{json.dumps(SYMBOLS, indent=2)}

Interval:

{INTERVAL}

Real candles per symbol:

{CANDLES_PER_SYMBOL}

Total real candles:

{actual_count}

## Candle Integrity

Every stored candle must be provider-supplied.

No synthetic candles.

No interpolation.

No repeated candles.

No timestamp substitution.

No fabricated historical records.

Coinbase pagination is used when a single response does not
contain enough observed real candles.

Missing no-trade intervals are NOT filled with synthetic zero-volume
candles.

## Trading Volume

Definition:

TOTAL_TRADED_BASE_ASSET_VOLUME

The provider's traded volume represents the transaction volume
represented by the candle.

Each trade is counted once.

Buyer and seller sides are NOT added together.

No double counting.

## Rolling Time Range

The requested volume range is:

NOW - {REQUESTED_DURATION_MINUTES} MINUTES -> NOW

This is a rolling timestamp range.

It is NOT:

- the latest completed 1h candle
- the last N returned rows

The candle interval and requested time range are separate concepts.

## USD Volume

Exact provider quote volume is accepted only when the provider
actually supplies an exact quote-volume field.

Binance quote volume may be stored when supplied directly.

Coinbase candle data does not provide an exact quote-volume field
in this U05 contract.

Therefore U05 does NOT calculate:

close * base_volume

and does NOT estimate USD volume.

## TOTAL / TOTAL2 / TOTAL3

NOT PART OF U05 DATASET.

U05 does not create:

- TOTAL volume
- TOTAL2 volume
- TOTAL3 volume
- inflow_usd
- outflow_usd
- netflow_usd
- dominance volume
- dominance flow

No placeholder UNAVAILABLE fields are created for these series.

No market-cap change is treated as capital flow.

## Security

API key: NOT REQUIRED

Secret: NOT REQUIRED

Trading: DISABLED

Orders: DISABLED

Strategy: DISABLED

## Validation

- provider priority validation
- provider provenance validation
- symbol coverage validation
- real candle count validation
- chronological validation
- duplicate timestamp validation
- OHLC validation
- real provider volume validation
- rolling range validation
- buyer/seller double-counting protection
- no synthetic data
- no interpolation
- no close-times-volume substitution
- no market-cap volume
- TOTAL/TOTAL2/TOTAL3 exclusion
- artifact validation
- SHA-256 inventory

## Cell Integrity Rule

No execution-error cell is accepted into the working notebook.

If this Cell fails:

1. the failure must be diagnosed
2. the Cell must be corrected
3. the corrected Cell must be rerun
4. successful validation must be confirmed
5. only then is U05 considered VALIDATED

## Artifacts

- DATASET_INDEX.json
- U05_CAPTURE_MANIFEST.json
- U05_DATA_QUALITY_REPORT.json
- U05_SHA256.json
- Raw JSON per symbol
- Normalized JSONL per symbol
- U05 ledger

## Execution Duration

{round(capture_duration, 3)} seconds

## Technical Lock

YES

## Next

U05 Snapshot then continue to the next execution unit.
""".strip() + "\n"


ledger_file = (
    LEDGER_DIR
    / "U05_LEDGER.md"
)

ledger_file.write_text(
    ledger_text,
    encoding="utf-8"
)


# ============================================================
# 28. FINAL SELF-CHECK
# ============================================================

final_check = {

    "project_root":
        ROOT.is_dir(),

    "target_notebook":
        NOTEBOOK_PATH.is_file(),

    "binance_adapter":
        BinancePublicAdapter is not None,

    "coinbase_adapter":
        CoinbasePublicAdapter is not None,

    "multi_symbol_capture":
        actual_count == expected_count,

    "exact_real_candle_count_per_symbol":
        all(
            item[
                "real_candle_count"
            ]
            == CANDLES_PER_SYMBOL
            for item
            in dataset_summary
        ),

    "symbol_coverage":
        captured_symbols
        == expected_symbols,

    "provider_provenance":
        all(
            bool(
                item["provider"]
            )
            and bool(
                item["exchange"]
            )
            and bool(
                item["provider_symbol"]
            )
            for item
            in dataset_summary
        ),

    "chronological":
        True,

    "unique_timestamps":
        True,

    "ohlc_validation":
        True,

    "base_volume_validation":
        True,

    "real_provider_volume":
        all(
            record[
                "volume_source"
            ]
            == "PROVIDER_SUPPLIED"

            for record
            in all_records
        ),

    "volume_double_counting_disabled":
        all(
            record[
                "volume_double_counting"
            ] is False

            for record
            in all_records
        ),

    "buyer_seller_not_double_counted":
        all(
            record[
                "volume_counting_rule"
            ]
            == "EACH_TRADE_COUNTED_ONCE"

            for record
            in all_records
        ),

    "rolling_range_contract":
        REQUESTED_RANGE_MODE
        == "ROLLING_END_AT_NOW",

    "rolling_range_duration":
        REQUESTED_DURATION_MINUTES
        == 60,

    "not_last_completed_candle":
        all(
            item[
                "rolling_volume"
            ][
                "not_last_completed_candle"
            ] is True

            for item
            in dataset_summary
        ),

    "not_last_n_rows":
        all(
            item[
                "rolling_volume"
            ][
                "not_last_n_rows"
            ] is True

            for item
            in dataset_summary
        ),

    "no_volume_fabrication":
        True,

    "no_interpolation":
        True,

    "no_market_cap_volume":
        True,

    "no_close_times_volume":
        True,

    "total_series_excluded":
        total_contract[
            "included_in_u05"
        ] is False,

    "total_flow_excluded":
        total_contract[
            "inflow_usd"
        ] == "NOT_INCLUDED",

    "dataset_index":
        index_file.is_file(),

    "capture_manifest":
        manifest_file.is_file(),

    "quality_report":
        quality_file.is_file(),

    "raw_dataset":
        all(
            (
                RAW_DIR
                / f"{symbol}_{INTERVAL}.json"
            ).is_file()

            for symbol
            in SYMBOLS
        ),

    "normalized_dataset":
        all(
            (
                NORMALIZED_DIR
                / f"{symbol}_{INTERVAL}.jsonl"
            ).is_file()

            for symbol
            in SYMBOLS
        ),

    "sha256":
        hash_file.is_file(),

    "ledger":
        ledger_file.is_file(),

    "trading_disabled":
        TRADING_ENABLED is False,

    "orders_disabled":
        ORDERS_ENABLED is False,

    "strategy_disabled":
        STRATEGY_ENABLED is False,
}


print()
print("=" * 72)
print("FINAL U05 SELF-CHECK")
print("=" * 72)

print(
    json.dumps(
        final_check,
        indent=2
    )
)


if not all(
    final_check.values()
):

    raise RuntimeError(
        "U05 SELF-CHECK FAILED"
    )


# ============================================================
# 29. FINAL STATUS
# ============================================================

print()
print("=" * 72)
print("U05 COMPLETE")
print("=" * 72)

print(
    "STATUS          : VALIDATED"
)

print(
    "LOCK            : TECHNICAL"
)

print(
    "SYMBOLS         :",
    len(SYMBOLS)
)

print(
    "REAL CANDLES    :",
    actual_count
)

print(
    "CANDLES/SYMBOL  :",
    CANDLES_PER_SYMBOL
)

print(
    "CANDLE INTERVAL :",
    INTERVAL
)

print(
    "VOLUME RANGE    :",
    f"NOW - {REQUESTED_DURATION_MINUTES}m -> NOW"
)

print(
    "VOLUME          :",
    "REAL PROVIDER TRADED VOLUME"
)

print(
    "COUNTING        :",
    "EACH TRADE ONCE"
)

print(
    "DOUBLE COUNTING :",
    "DISABLED"
)

print(
    "PROVIDER POLICY :",
    "BINANCE_FIRST"
)

print(
    "PROVIDER USAGE  :"
)

for provider_name, count in (
    provider_statistics.items()
):

    print(
        f"  {provider_name}: "
        f"{count} symbol(s)"
    )

print(
    "USD VOLUME      :",
    "EXACT PROVIDER QUOTE VOLUME ONLY"
)

print(
    "USD ESTIMATION  :",
    "DISABLED"
)

print(
    "TOTAL SERIES    :",
    "NOT INCLUDED"
)

print(
    "TOTAL FLOW      :",
    "NOT INCLUDED"
)

print(
    "TRADING         :",
    "DISABLED"
)

print(
    "ORDERS          :",
    "DISABLED"
)

print(
    "STRATEGY        :",
    "DISABLED"
)

print(
    "NEXT            :",
    "U05 SNAPSHOT"
)

print("=" * 72)
```

```python

# ============================================================
# KITCHEN ASSISTANT V3.1 — U06
# U05 HISTORICAL DATA QUALITY GATE
#
# FINAL CLEAN / SINGLE-CELL COLAB VERSION
#
# IMPORTANT:
#   - U05 SOURCE DATA IS NOT MODIFIED
#   - U06 validates the ACTUAL U05 schema
#   - U06 uses "volume_base", NOT nonexistent "volume"
#   - No triple-quoted strings
#   - No multiline string literals
#   - No notebook-file dependency
#   - Coinbase timestamp gaps are diagnosed separately
#   - Binance timestamp gaps are treated as integrity failures
#   - U06 locks ONLY after every required test passes
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
from decimal import Decimal, InvalidOperation
import hashlib
import json
import os
import sys
import subprocess
import importlib.util


# ============================================================
# 0. COLAB / DRIVE
# ============================================================

def ensure_drive():
    root = Path("/content/drive/MyDrive/Colab Notebooks")

    if root.is_dir():
        return root

    try:
        from google.colab import drive
        drive.mount("/content/drive")
    except Exception:
        pass

    if root.is_dir():
        return root

    candidates = [
        Path("/content/drive/MyDrive"),
        Path("/content"),
        Path.cwd(),
    ]

    for candidate in candidates:
        if candidate.is_dir():
            if candidate.name == "Colab Notebooks":
                return candidate

            notebooks = candidate / "Colab Notebooks"
            if notebooks.is_dir():
                return notebooks

    raise RuntimeError(
        "U06 project root not found. "
        "Expected /content/drive/MyDrive/Colab Notebooks"
    )


ROOT = ensure_drive()
os.chdir(ROOT)


print("=" * 72)
print("KITCHEN ASSISTANT V3.1 — U06")
print("U05 HISTORICAL DATA QUALITY GATE")
print("=" * 72)
print("PROJECT ROOT:", ROOT)
print()


# ============================================================
# 1. CONFIGURATION
# ============================================================

SYMBOLS = [
    "BTCUSDT",
    "ETHUSDT",
    "SOLUSDT",
    "XRPUSDT",
    "ADAUSDT",
]

INTERVAL = "1m"
EXPECTED_RECORDS_PER_SYMBOL = 100
EXPECTED_TOTAL_RECORDS = (
    len(SYMBOLS) * EXPECTED_RECORDS_PER_SYMBOL
)

DATASET_ROOT = ROOT / "data" / "historical_u05"
RAW_DIR = DATASET_ROOT / "raw"
NORMALIZED_DIR = DATASET_ROOT / "normalized"
MANIFEST_DIR = DATASET_ROOT / "manifests"
QUALITY_DIR = DATASET_ROOT / "quality"
LEDGER_DIR = ROOT / "ledgers"

for directory in [
    DATASET_ROOT,
    RAW_DIR,
    NORMALIZED_DIR,
    MANIFEST_DIR,
    QUALITY_DIR,
    LEDGER_DIR,
]:
    directory.mkdir(parents=True, exist_ok=True)


# ============================================================
# 2. U05 CONTROL ARTIFACTS
# ============================================================

DATASET_INDEX_FILE = (
    DATASET_ROOT / "DATASET_INDEX.json"
)

U05_SHA256_FILE = (
    DATASET_ROOT / "U05_SHA256.json"
)

U05_CAPTURE_MANIFEST_FILE = (
    MANIFEST_DIR / "U05_CAPTURE_MANIFEST.json"
)

U05_QUALITY_FILE = (
    QUALITY_DIR / "U05_DATA_QUALITY_REPORT.json"
)


required_control_artifacts = {
    "DATASET_INDEX.json": DATASET_INDEX_FILE,
    "U05_SHA256.json": U05_SHA256_FILE,
    "U05_CAPTURE_MANIFEST.json": U05_CAPTURE_MANIFEST_FILE,
    "U05_DATA_QUALITY_REPORT.json": U05_QUALITY_FILE,
}


print("=" * 72)
print("U05 CONTROL ARTIFACT VERIFICATION")
print("=" * 72)


control_status = {}

for name, path in required_control_artifacts.items():
    exists = path.is_file()
    control_status[name] = exists
    print(f"{name}: {exists}")


print()

if not all(control_status.values()):
    missing = [
        name
        for name, exists in control_status.items()
        if not exists
    ]

    raise RuntimeError(
        "U05 CONTROL ARTIFACTS MISSING: "
        + ", ".join(missing)
    )


# ============================================================
# 3. LOAD U05 INDEX
# ============================================================

try:
    u05_index = json.loads(
        DATASET_INDEX_FILE.read_text(
            encoding="utf-8"
        )
    )
except Exception as exc:
    raise RuntimeError(
        "Cannot read DATASET_INDEX.json"
    ) from exc


if not isinstance(u05_index, dict):
    raise RuntimeError(
        "DATASET_INDEX.json must contain a JSON object"
    )


if u05_index.get("execution_unit") != "U05":
    raise RuntimeError(
        "DATASET_INDEX.json execution_unit is not U05"
    )


index_symbols = u05_index.get("symbols")

if sorted(index_symbols or []) != sorted(SYMBOLS):
    raise RuntimeError(
        "U05 symbol list does not match controlled symbols"
    )


# ============================================================
# 4. ACTUAL U05 SCHEMA
# ============================================================
#
# IMPORTANT:
#
# U05 stores:
#
#   volume_base
#
# NOT:
#
#   volume
#
# U06 therefore validates volume_base.
#
# This is the key correction that prevents the previous
# false 500-record "missing volume" failure.
# ============================================================

REQUIRED_FIELDS = [
    "project",
    "version",
    "execution_unit",
    "asset_type",
    "symbol",
    "provider",
    "exchange",
    "provider_symbol",
    "interval",
    "timestamp",
    "open",
    "high",
    "low",
    "close",
    "volume_base",
    "volume_definition",
    "volume_source",
    "volume_counting_rule",
    "volume_double_counting",
    "buyer_seller_not_double_counted",
    "volume_usd",
    "volume_usd_exact",
    "volume_usd_source",
    "total_series_included",
    "trading_enabled",
    "orders_enabled",
    "strategy_enabled",
]


# ============================================================
# 5. HELPERS
# ============================================================

def sha256_file(path):
    return hashlib.sha256(
        path.read_bytes()
    ).hexdigest()


def numeric(value, field_name):
    try:
        value_decimal = Decimal(str(value))
    except (
        InvalidOperation,
        ValueError,
        TypeError,
    ) as exc:
        raise ValueError(
            f"Invalid numeric {field_name}: {value!r}"
        ) from exc

    return value_decimal


def validate_ohlc(record):
    open_price = numeric(
        record["open"],
        "open",
    )

    high_price = numeric(
        record["high"],
        "high",
    )

    low_price = numeric(
        record["low"],
        "low",
    )

    close_price = numeric(
        record["close"],
        "close",
    )

    volume = numeric(
        record["volume_base"],
        "volume_base",
    )

    if open_price <= 0:
        return "open_not_positive"

    if high_price <= 0:
        return "high_not_positive"

    if low_price <= 0:
        return "low_not_positive"

    if close_price <= 0:
        return "close_not_positive"

    if high_price < low_price:
        return "high_less_than_low"

    if not (
        low_price
        <= open_price
        <= high_price
    ):
        return "open_outside_range"

    if not (
        low_price
        <= close_price
        <= high_price
    ):
        return "close_outside_range"

    if volume < 0:
        return "negative_volume_base"

    return None


def load_jsonl(path):
    records = []
    errors = []

    with path.open(
        "r",
        encoding="utf-8",
    ) as handle:

        for line_number, line in enumerate(
            handle,
            start=1,
        ):

            stripped = line.strip()

            if not stripped:
                errors.append(
                    {
                        "line": line_number,
                        "error": "empty_line",
                    }
                )
                continue

            try:
                record = json.loads(stripped)
            except Exception as exc:
                errors.append(
                    {
                        "line": line_number,
                        "error": "invalid_json",
                        "detail": repr(exc),
                    }
                )
                continue

            if not isinstance(record, dict):
                errors.append(
                    {
                        "line": line_number,
                        "error": "record_not_object",
                    }
                )
                continue

            records.append(record)

    return records, errors


# ============================================================
# 6. SHA-256 CONTROL
# ============================================================

print("=" * 72)
print("U05 SHA-256 CONTROL VERIFICATION")
print("=" * 72)

try:
    stored_hashes = json.loads(
        U05_SHA256_FILE.read_text(
            encoding="utf-8"
        )
    )
except Exception as exc:
    raise RuntimeError(
        "Cannot read U05_SHA256.json"
    ) from exc


if not isinstance(stored_hashes, dict):
    raise RuntimeError(
        "U05_SHA256.json must contain a JSON object"
    )


sha_mismatches = []
sha_missing = []

hash_targets = [
    DATASET_INDEX_FILE,
    U05_CAPTURE_MANIFEST_FILE,
    U05_QUALITY_FILE,
]

for symbol in SYMBOLS:
    hash_targets.append(
        RAW_DIR / f"{symbol}_{INTERVAL}.json"
    )
    hash_targets.append(
        NORMALIZED_DIR / f"{symbol}_{INTERVAL}.jsonl"
    )


for path in hash_targets:

    if not path.is_file():
        sha_missing.append(
            str(path.relative_to(ROOT))
        )
        continue

    relative = str(
        path.relative_to(ROOT)
    )

    actual_hash = sha256_file(path)
    stored_hash = stored_hashes.get(relative)

    if stored_hash is None:
        sha_missing.append(relative)
        continue

    if actual_hash != stored_hash:
        sha_mismatches.append(
            {
                "file": relative,
                "expected": stored_hash,
                "actual": actual_hash,
            }
        )


sha_valid = (
    len(sha_mismatches) == 0
    and len(sha_missing) == 0
)


print(
    json.dumps(
        {
            "available": True,
            "valid": sha_valid,
            "mismatched": sha_mismatches,
            "missing": sha_missing,
        },
        indent=2,
    )
)


# ============================================================
# 7. DATASET INSPECTION
# ============================================================

print()
print("=" * 72)
print("U06 DATA QUALITY INSPECTION")
print("=" * 72)


symbol_reports = []
all_records = []
total_problems = 0


for symbol in SYMBOLS:

    normalized_file = (
        NORMALIZED_DIR
        / f"{symbol}_{INTERVAL}.jsonl"
    )

    print()
    print(
        f"Inspecting: "
        f"{normalized_file.relative_to(ROOT)}"
    )

    report = {
        "symbol": symbol,
        "file": str(
            normalized_file.relative_to(ROOT)
        ),
        "exists": normalized_file.is_file(),
        "records": 0,
        "timestamp_unit": None,
        "problems": [],
        "problem_count": 0,
        "status": "FAIL",
        "provider": None,
        "schema": "U05_CANONICAL",
    }


    if not normalized_file.is_file():

        report["problems"].append(
            {
                "type": "missing_file",
            }
        )

        report["problem_count"] = 1
        total_problems += 1

        symbol_reports.append(report)

        print("Records: 0")
        print("Status: FAIL")
        print("Problem breakdown:")
        print("  - missing_file")

        continue


    records, parse_errors = load_jsonl(
        normalized_file
    )

    report["records"] = len(records)

    for parse_error in parse_errors:
        report["problems"].append(
            {
                "type": parse_error["error"],
                "detail": parse_error,
            }
        )


    if records:
        all_records.extend(records)


    # --------------------------------------------------------
    # Record count
    # --------------------------------------------------------

    if len(records) != EXPECTED_RECORDS_PER_SYMBOL:
        report["problems"].append(
            {
                "type": "record_count",
                "expected": EXPECTED_RECORDS_PER_SYMBOL,
                "actual": len(records),
            }
        )


    # --------------------------------------------------------
    # Per-record validation
    # --------------------------------------------------------

    timestamps = []
    providers = set()

    for index, record in enumerate(records):

        missing = [
            field
            for field in REQUIRED_FIELDS
            if field not in record
        ]

        if missing:
            report["problems"].append(
                {
                    "type": "missing_fields",
                    "index": index,
                    "line": index + 1,
                    "missing": missing,
                }
            )
            continue


        if record.get("symbol") != symbol:
            report["problems"].append(
                {
                    "type": "symbol_mismatch",
                    "index": index,
                    "actual": record.get("symbol"),
                    "expected": symbol,
                }
            )


        if record.get("execution_unit") != "U05":
            report["problems"].append(
                {
                    "type": "execution_unit_mismatch",
                    "index": index,
                }
            )


        if record.get("interval") != INTERVAL:
            report["problems"].append(
                {
                    "type": "interval_mismatch",
                    "index": index,
                }
            )


        provider = record.get("provider")
        providers.add(provider)


        if not provider:
            report["problems"].append(
                {
                    "type": "missing_provider",
                    "index": index,
                }
            )


        if not record.get("provider_symbol"):
            report["problems"].append(
                {
                    "type": "missing_provider_symbol",
                    "index": index,
                }
            )


        try:
            timestamp = int(
                record["timestamp"]
            )
            timestamps.append(timestamp)
        except Exception:
            report["problems"].append(
                {
                    "type": "invalid_timestamp",
                    "index": index,
                }
            )


        try:
            timestamp = int(
                record["timestamp"]
            )

            if timestamp > 100000000000:
                report["timestamp_unit"] = "milliseconds"
            elif timestamp > 1000000000:
                report["timestamp_unit"] = "seconds"
            else:
                report["timestamp_unit"] = "invalid"

        except Exception:
            pass


        # ----------------------------------------------------
        # Canonical U05 volume validation
        # ----------------------------------------------------

        if (
            record.get("volume_source")
            != "PROVIDER_SUPPLIED"
        ):
            report["problems"].append(
                {
                    "type": "invalid_volume_source",
                    "index": index,
                    "actual": record.get(
                        "volume_source"
                    ),
                }
            )


        if (
            record.get("volume_counting_rule")
            != "EACH_TRADE_COUNTED_ONCE"
        ):
            report["problems"].append(
                {
                    "type": "invalid_volume_counting_rule",
                    "index": index,
                }
            )


        if (
            record.get("volume_double_counting")
            is not False
        ):
            report["problems"].append(
                {
                    "type": "volume_double_counting_flag",
                    "index": index,
                }
            )


        if (
            record.get(
                "buyer_seller_not_double_counted"
            )
            is not True
        ):
            report["problems"].append(
                {
                    "type": "buyer_seller_double_counting_flag",
                    "index": index,
                }
            )


        if (
            record.get("total_series_included")
            is not False
        ):
            report["problems"].append(
                {
                    "type": "total_series_present",
                    "index": index,
                }
            )


        if (
            record.get("trading_enabled")
            is not False
        ):
            report["problems"].append(
                {
                    "type": "trading_enabled",
                    "index": index,
                }
            )


        if (
            record.get("orders_enabled")
            is not False
        ):
            report["problems"].append(
                {
                    "type": "orders_enabled",
                    "index": index,
                }
            )


        if (
            record.get("strategy_enabled")
            is not False
        ):
            report["problems"].append(
                {
                    "type": "strategy_enabled",
                    "index": index,
                }
            )


        # ----------------------------------------------------
        # OHLC / volume
        # ----------------------------------------------------

        try:
            ohlc_error = validate_ohlc(
                record
            )

            if ohlc_error:
                report["problems"].append(
                    {
                        "type": ohlc_error,
                        "index": index,
                    }
                )

        except Exception as exc:
            report["problems"].append(
                {
                    "type": "numeric_validation_error",
                    "index": index,
                    "detail": repr(exc),
                }
            )


    # --------------------------------------------------------
    # Timestamp validation
    # --------------------------------------------------------

    if timestamps:

        if any(
            timestamp < 100000000000
            for timestamp in timestamps
        ):
            report["problems"].append(
                {
                    "type": "timestamp_not_milliseconds",
                }
            )


        if timestamps != sorted(timestamps):
            report["problems"].append(
                {
                    "type": "timestamp_order",
                }
            )


        if len(timestamps) != len(set(timestamps)):
            report["problems"].append(
                {
                    "type": "duplicate_timestamps",
                }
            )


        # ----------------------------------------------------
        # Timestamp gaps
        #
        # IMPORTANT:
        #
        # Coinbase may omit no-trade intervals.
        #
        # Therefore:
        #
        # Coinbase gap -> diagnostic / tolerated
        # Binance gap  -> integrity failure
        # ----------------------------------------------------

        provider_for_gap = None

        if len(providers) == 1:
            provider_for_gap = next(iter(providers))

        gaps = []

        for previous_index in range(
            len(timestamps) - 1
        ):

            previous_timestamp = (
                timestamps[previous_index]
            )

            current_timestamp = (
                timestamps[previous_index + 1]
            )

            delta = (
                current_timestamp
                - previous_timestamp
            )

            if delta != 60000:
                gaps.append(
                    {
                        "previous_timestamp_ms":
                            previous_timestamp,
                        "current_timestamp_ms":
                            current_timestamp,
                        "delta_ms":
                            delta,
                        "expected_delta_ms":
                            60000,
                        "previous_index":
                            previous_index,
                        "current_index":
                            previous_index + 1,
                    }
                )


        if gaps:

            if provider_for_gap == (
                "COINBASE_EXCHANGE_PUBLIC"
            ):

                report["problems"].append(
                    {
                        "type":
                            "timestamp_gaps_tolerated_coinbase",
                        "count":
                            len(gaps),
                        "samples":
                            gaps[:5],
                    }
                )

            else:

                report["problems"].append(
                    {
                        "type":
                            "timestamp_gaps",
                        "count":
                            len(gaps),
                        "samples":
                            gaps[:5],
                    }
                )


    # --------------------------------------------------------
    # Problem count
    #
    # The tolerated Coinbase-gap diagnostic does NOT count
    # as a hard failure.
    # --------------------------------------------------------

    hard_problems = []

    for problem in report["problems"]:

        if (
            problem["type"]
            == "timestamp_gaps_tolerated_coinbase"
        ):
            continue

        hard_problems.append(problem)


    report["hard_problem_count"] = len(
        hard_problems
    )

    report["problem_count"] = len(
        report["problems"]
    )

    if report["hard_problem_count"] == 0:
        report["status"] = "PASS"
    else:
        report["status"] = "FAIL"

    total_problems += (
        report["hard_problem_count"]
    )

    symbol_reports.append(report)


    # --------------------------------------------------------
    # Console
    # --------------------------------------------------------

    print(
        "Records:",
        len(records)
    )

    print(
        "Timestamp unit:",
        report["timestamp_unit"]
    )

    print(
        "Provider:",
        sorted(
            [
                str(provider)
                for provider in providers
            ]
        )
    )

    print(
        "Hard problems:",
        report["hard_problem_count"]
    )

    print(
        "Status:",
        report["status"]
    )

    if report["problems"]:

        breakdown = {}

        for problem in report["problems"]:
            problem_type = problem["type"]

            if (
                problem_type
                == "timestamp_gaps_tolerated_coinbase"
            ):
                continue

            breakdown[problem_type] = (
                breakdown.get(
                    problem_type,
                    0,
                )
                + 1
            )

        if breakdown:
            print("Problem breakdown:")

            for problem_type, count in breakdown.items():
                print(
                    f"  - {problem_type}: {count}"
                )


# ============================================================
# 8. GLOBAL VALIDATION
# ============================================================

print()
print("=" * 72)
print("GLOBAL U06 VALIDATION")
print("=" * 72)


actual_total_records = len(
    all_records
)


symbol_counts = {}

for symbol in SYMBOLS:
    symbol_counts[symbol] = sum(
        1
        for record in all_records
        if record.get("symbol") == symbol
    )


global_record_count_ok = (
    actual_total_records
    == EXPECTED_TOTAL_RECORDS
)


global_symbol_coverage_ok = (
    all(
        symbol_counts[symbol]
        == EXPECTED_RECORDS_PER_SYMBOL
        for symbol in SYMBOLS
    )
)


global_source_unchanged = True


print(
    "Expected total records:",
    EXPECTED_TOTAL_RECORDS
)

print(
    "Actual total records:",
    actual_total_records
)

print(
    "Global record count:",
    global_record_count_ok
)

print(
    "Symbol coverage:",
    global_symbol_coverage_ok
)

print(
    "Source modification by U06:",
    "FALSE"
)


# ============================================================
# 9. TRADING / SAFETY VALIDATION
# ============================================================

print()
print("=" * 72)
print("U06 SAFETY VALIDATION")
print("=" * 72)


trading_disabled = all(
    record.get("trading_enabled") is False
    for record in all_records
)

orders_disabled = all(
    record.get("orders_enabled") is False
    for record in all_records
)

strategy_disabled = all(
    record.get("strategy_enabled") is False
    for record in all_records
)

total_series_excluded = all(
    record.get("total_series_included") is False
    for record in all_records
)

real_provider_volume = all(
    record.get("volume_source")
    == "PROVIDER_SUPPLIED"
    for record in all_records
)

double_counting_disabled = all(
    record.get("volume_double_counting")
    is False
    for record in all_records
)

buyer_seller_not_double_counted = all(
    record.get(
        "buyer_seller_not_double_counted"
    )
    is True
    for record in all_records
)


print(
    "Trading disabled:",
    trading_disabled
)

print(
    "Orders disabled:",
    orders_disabled
)

print(
    "Strategy disabled:",
    strategy_disabled
)

print(
    "TOTAL/TOTAL2/TOTAL3 excluded:",
    total_series_excluded
)

print(
    "Provider volume:",
    real_provider_volume
)

print(
    "Double counting disabled:",
    double_counting_disabled
)

print(
    "Buyer/seller not double-counted:",
    buyer_seller_not_double_counted
)


# ============================================================
# 10. FINAL QUALITY STATUS
# ============================================================

all_symbol_quality_ok = all(
    report["status"] == "PASS"
    for report in symbol_reports
)

validation_status = (
    "PASS"
    if (
        all_symbol_quality_ok
        and global_record_count_ok
        and global_symbol_coverage_ok
        and sha_valid
        and trading_disabled
        and orders_disabled
        and strategy_disabled
        and total_series_excluded
        and real_provider_volume
        and double_counting_disabled
        and buyer_seller_not_double_counted
    )
    else "FAIL"
)


technical_lock = (
    validation_status == "PASS"
)


# ============================================================
# 11. QUALITY REPORT
# ============================================================

quality_report = {
    "project": "KITCHEN_ASSISTANT",
    "version": "3.1",
    "execution_unit": "U06",
    "gate": "U05_HISTORICAL_DATA_QUALITY",
    "validation_status": validation_status,
    "technical_lock": technical_lock,
    "source_data_modified": False,
    "controlled_symbols": SYMBOLS,
    "controlled_symbol_count": len(SYMBOLS),
    "expected_records_per_symbol":
        EXPECTED_RECORDS_PER_SYMBOL,
    "expected_total_records":
        EXPECTED_TOTAL_RECORDS,
    "actual_total_records":
        actual_total_records,
    "hard_total_problems":
        total_problems,
    "u05_schema": {
        "volume_field": "volume_base",
        "volume_source_field": "volume_source",
        "volume_counting_rule_field":
            "volume_counting_rule",
        "double_counting_field":
            "volume_double_counting",
        "buyer_seller_field":
            "buyer_seller_not_double_counted",
    },
    "control_artifacts": control_status,
    "sha256": {
        "available": True,
        "valid": sha_valid,
        "mismatched": sha_mismatches,
        "missing": sha_missing,
    },
    "global_validation": {
        "record_count": global_record_count_ok,
        "symbol_coverage": global_symbol_coverage_ok,
        "source_data_modified":
            global_source_unchanged,
    },
    "safety": {
        "trading_disabled":
            trading_disabled,
        "orders_disabled":
            orders_disabled,
        "strategy_disabled":
            strategy_disabled,
        "total_series_excluded":
            total_series_excluded,
        "real_provider_volume":
            real_provider_volume,
        "double_counting_disabled":
            double_counting_disabled,
        "buyer_seller_not_double_counted":
            buyer_seller_not_double_counted,
    },
    "symbol_reports": symbol_reports,
    "generated_at_utc":
        datetime.now(
            timezone.utc
        ).strftime(
            "%Y-%m-%d %H:%M:%S UTC"
        ),
}


U06_QUALITY_FILE = (
    QUALITY_DIR
    / "U06_DATA_QUALITY_REPORT.json"
)

U06_QUALITY_FILE.write_text(
    json.dumps(
        quality_report,
        indent=2,
    ),
    encoding="utf-8",
)


# ============================================================
# 12. QUALITY SUMMARY
# ============================================================

summary_lines = []

summary_lines.append(
    "KITCHEN ASSISTANT V3.1 — U06 U05 HISTORICAL DATA QUALITY GATE"
)
summary_lines.append("")
summary_lines.append(
    f"GLOBAL STATUS : {validation_status}"
)
summary_lines.append(
    f"TECHNICAL LOCK: {'LOCKED' if technical_lock else 'NOT LOCKED'}"
)
summary_lines.append(
    f"TOTAL RECORDS : {actual_total_records}"
)
summary_lines.append(
    f"TOTAL PROBLEMS: {total_problems}"
)
summary_lines.append("")
summary_lines.append(
    "U05 VOLUME FIELD: volume_base"
)
summary_lines.append(
    "U05 VOLUME SOURCE: PROVIDER_SUPPLIED"
)
summary_lines.append(
    "BUYER/SELLER DOUBLE COUNTING: DISABLED"
)
summary_lines.append("")
summary_lines.append(
    f"SHA-256 VALID: {sha_valid}"
)
summary_lines.append(
    f"TRADING DISABLED: {trading_disabled}"
)
summary_lines.append(
    f"ORDERS DISABLED: {orders_disabled}"
)
summary_lines.append(
    f"STRATEGY DISABLED: {strategy_disabled}"
)
summary_lines.append(
    f"TOTAL SERIES EXCLUDED: {total_series_excluded}"
)
summary_lines.append("")
summary_lines.append("SYMBOL RESULTS:")

for report in symbol_reports:

    summary_lines.append(
        f"{report['symbol']}: "
        f"{report['status']} | "
        f"records={report['records']} | "
        f"hard_problems={report['hard_problem_count']}"
    )


summary_lines.append("")
summary_lines.append(
    "SOURCE DATA MODIFIED: FALSE"
)


summary_file = (
    QUALITY_DIR
    / "U06_DATA_QUALITY_SUMMARY.txt"
)

summary_file.write_text(
    "\n".join(summary_lines)
    + "\n",
    encoding="utf-8",
)


# ============================================================
# 13. U06 LEDGER
# ============================================================

ledger_lines = []

ledger_lines.append(
    "# KITCHEN ASSISTANT V3.1 — U06 LEDGER"
)
ledger_lines.append("")
ledger_lines.append(
    "## Execution Unit"
)
ledger_lines.append("")
ledger_lines.append(
    "U06 — U05 Historical Data Quality Gate"
)
ledger_lines.append("")
ledger_lines.append(
    "## Validation"
)
ledger_lines.append("")
ledger_lines.append(
    f"Global status: {validation_status}"
)
ledger_lines.append(
    f"Technical lock: {'LOCKED' if technical_lock else 'NOT LOCKED'}"
)
ledger_lines.append("")
ledger_lines.append(
    "## Critical Schema Correction"
)
ledger_lines.append("")
ledger_lines.append(
    "U05 canonical traded-volume field is volume_base."
)
ledger_lines.append(
    "U06 does not require a nonexistent volume field."
)
ledger_lines.append("")
ledger_lines.append(
    "## Source Integrity"
)
ledger_lines.append("")
ledger_lines.append(
    "U06 does not modify U05 source datasets."
)
ledger_lines.append(
    "Source data modified: FALSE"
)
ledger_lines.append("")
ledger_lines.append(
    "## Safety"
)
ledger_lines.append("")
ledger_lines.append(
    "Trading: DISABLED"
)
ledger_lines.append(
    "Orders: DISABLED"
)
ledger_lines.append(
    "Strategy: DISABLED"
)
ledger_lines.append(
    "TOTAL/TOTAL2/TOTAL3: NOT INCLUDED"
)
ledger_lines.append("")
ledger_lines.append(
    "## Record Counts"
)
ledger_lines.append("")
ledger_lines.append(
    f"Expected total records: {EXPECTED_TOTAL_RECORDS}"
)
ledger_lines.append(
    f"Actual total records: {actual_total_records}"
)
ledger_lines.append("")
ledger_lines.append(
    "## Artifact"
)
ledger_lines.append("")
ledger_lines.append(
    "U06_DATA_QUALITY_REPORT.json"
)
ledger_lines.append(
    "U06_DATA_QUALITY_SUMMARY.txt"
)
ledger_lines.append("")
ledger_lines.append(
    "## Final Rule"
)
ledger_lines.append("")
ledger_lines.append(
    "U06 is technically locked only when every required gate passes."
)
ledger_lines.append(
    "If validation fails, U06 remains NOT LOCKED."
)
ledger_lines.append("")


ledger_file = (
    LEDGER_DIR
    / "U06_LEDGER.md"
)

ledger_file.write_text(
    "\n".join(ledger_lines),
    encoding="utf-8",
)


# ============================================================
# 14. FINAL SELF-CHECK
# ============================================================

final_check = {
    "project_root":
        ROOT.is_dir(),

    "u05_control_artifacts":
        all(control_status.values()),

    "u05_index_valid":
        u05_index.get("execution_unit") == "U05",

    "symbol_coverage":
        global_symbol_coverage_ok,

    "record_count":
        global_record_count_ok,

    "u05_volume_schema":
        all(
            (
                "volume_base" in record
                and "volume" not in record
            )
            for record in all_records
        ),

    "real_provider_volume":
        real_provider_volume,

    "double_counting_disabled":
        double_counting_disabled,

    "buyer_seller_not_double_counted":
        buyer_seller_not_double_counted,

    "trading_disabled":
        trading_disabled,

    "orders_disabled":
        orders_disabled,

    "strategy_disabled":
        strategy_disabled,

    "total_series_excluded":
        total_series_excluded,

    "sha256_valid":
        sha_valid,

    "source_data_modified":
        False,

    "quality_report_written":
        U06_QUALITY_FILE.is_file(),

    "quality_summary_written":
        summary_file.is_file(),

    "ledger_written":
        ledger_file.is_file(),
}


print()
print("=" * 72)
print("FINAL U06 SELF-CHECK")
print("=" * 72)

print(
    json.dumps(
        final_check,
        indent=2,
    )
)


# ============================================================
# 15. FINAL RESULT
# ============================================================

print()
print("=" * 72)
print("U06 FINAL RESULT")
print("=" * 72)

print(
    "GLOBAL STATUS :",
    validation_status,
)

print(
    "TECHNICAL LOCK:",
    "LOCKED"
    if technical_lock
    else "NOT LOCKED",
)

print(
    "TOTAL RECORDS :",
    actual_total_records,
)

print(
    "TOTAL PROBLEMS:",
    total_problems,
)

print(
    "U06 QUALITY REPORT:",
    U06_QUALITY_FILE.relative_to(ROOT),
)

print(
    "U06 QUALITY SUMMARY:",
    summary_file.relative_to(ROOT),
)

print(
    "U06 LEDGER:",
    ledger_file.relative_to(ROOT),
)

print(
    "SOURCE DATA MODIFIED:",
    "FALSE",
)

print(
    "TRADING:",
    "DISABLED",
)

print(
    "ORDERS:",
    "DISABLED",
)

print(
    "STRATEGY:",
    "DISABLED",
)

print(
    "U06 VALIDATION:",
    validation_status,
)

print("=" * 72)


if validation_status != "PASS":
    raise RuntimeError(
        "U06 VALIDATION FAILED. "
        "U06 remains NOT LOCKED. "
        "Review U06_DATA_QUALITY_REPORT.json "
        "and U06_DATA_QUALITY_SUMMARY.txt."
    )


print("U06 LOCKED — READY FOR U05 SNAPSHOT")
print("=" * 72)
```

```python

# KITCHEN ASSISTANT V3.1 — U06.5
# RULE ZERO: FINAL CODE IS PORTABLE ACROSS COLAB / JUPYTER / PYTHONANYWHERE
# MARKET DATA SOURCE & VALIDATION FOUNDATION
# VERSION 8.4.5 — RELIABILITY-WEIGHTED MULTI-EXCHANGE PRICE FOUNDATION / 8-EXCHANGE CORE / 7-MINIMUM / ORDERBOOK DEPTH / DYNAMIC QUALITY SCORING / CAPABILITY-AWARE LOCK
#
# Contract basis:
#   KITCHEN ASSISTANT V3.1
#   CELL 06.5 — MARKET DATA SOURCE & VALIDATION FOUNDATION
#   FINAL CANONICAL IMPLEMENTATION CONTRACT — VERSION 6.0
#
# Runtime rules:
#   * CMC Keyless is the primary current-data mode.
#   * CMC authenticated is optional when a key is explicitly configured.
#   * CoinGecko is the fallback provider.
#   * TradingView is disabled and is never a runtime provider/reference.
#   * CMC Global Metrics are provider evidence, not Kitchen TOTAL.
#   * Kitchen TOTAL is calculated from the validated dynamic Top-125.
#   * Historical 5m data is never fabricated or silently substituted.
#   * Global market metrics (market cap / dominance / Top-125 / Kitchen TOTAL)
#     remain owned by the GLOBAL layer; exchanges provide trading/price evidence.
#   * Exchange prices are NOT equally weighted.  A dynamic reliability model
#     evaluates data integrity, freshness, traded turnover, order-book depth
#     at multiple price bands, spread, estimated price impact, and cross-market
#     consistency. Missing features are never fabricated.
#   * A canonical live reference price is built from real multi-exchange evidence
#     when the minimum cross-exchange confirmation contract is satisfied.
#   * Historical order-book/liquidity evidence is used only when genuinely
#     available; historical gaps are never backfilled or interpolated.
#   * Multi-exchange Core target is exactly 8 with minimum acceptance 7.
#   * Historical global-index Time Series is required only in HISTORICAL mode.
#   * Cross-source reference evidence is non-blocking unless explicitly promoted
#     to a mandatory contract.
#   * Snapshot persistence is explicit and is NOT automatic.
#   * Downstream cells consume U06_5_DATA_FOUNDATION and do not reacquire
#     market data owned by this cell.
#
# This file is intentionally self-contained and uses only Python standard
# library modules. It is designed to run in Colab, Jupyter, or a normal
# Python environment.

from __future__ import annotations

import ast
import copy
import hashlib
import json
import math
import os
import re
import statistics
import time
import uuid
import traceback
from collections import Counter, defaultdict
from dataclasses import dataclass
from datetime import datetime, timezone, timedelta
from pathlib import Path
from typing import Any, Dict, Iterable, List, Optional, Sequence, Tuple
from urllib.error import HTTPError, URLError
from urllib.parse import urlencode
from urllib.request import Request, urlopen


# =============================================================================
# 0. CONTRACT / CONFIGURATION
# =============================================================================

CELL_ID = "U06.5"
PROJECT = "KITCHEN ASSISTANT V3.1"
VERSION = "8.4.5"
SCHEMA_VERSION = "U06_5_SCHEMA_V6_0"

CMC_BASE_KEYLESS = "https://pro-api.coinmarketcap.com/public-api"
CMC_BASE_AUTH = "https://pro-api.coinmarketcap.com"
COINGECKO_BASE = "https://api.coingecko.com/api/v3"

TOP_N = 125
TIMEFRAME = "5m"
SUPPORTED_TIMEFRAMES = ("1m", "5m", "15m", "1h", "4h", "1d")
DATASET_MODE = os.getenv("KITCHEN_DATASET_MODE", "SNAPSHOT").strip().upper()
if DATASET_MODE not in {"SNAPSHOT", "HISTORICAL"}:
    DATASET_MODE = "SNAPSHOT"

MAX_RETRIES_PER_PROVIDER = 1
HTTP_TIMEOUT_SECONDS = 20
FRESHNESS_THRESHOLD_SECONDS = 15 * 60
MIN_TOP125_COVERAGE = 1.0

TARGET_CANDIDATE_POOL = int(os.getenv("KITCHEN_TARGET_CANDIDATE_POOL", "1250"))
TARGET_EXCHANGE_COUNT = 8
MIN_VALIDATED_EXCHANGE_COUNT = 7
# 8 is the preferred Core; 7 is the minimum acceptance threshold.
# Environment variables cannot silently change the architectural contract.

MULTI_EXCHANGE_PROBE_SYMBOL = "BTCUSDT"
U05_EXCHANGE_SYMBOLS = {
    "BTCUSDT": "BTC-USDT",
    "ETHUSDT": "ETH-USDT",
    "SOLUSDT": "SOL-USDT",
    "XRPUSDT": "XRP-USDT",
    "ADAUSDT": "ADA-USDT",
}
EXCHANGE_KLINE_INTERVAL = "5m"
EXCHANGE_KLINE_LIMIT = int(os.getenv("KITCHEN_EXCHANGE_KLINE_LIMIT", "2"))
HISTORICAL_PAGE_LIMIT = int(os.getenv("KITCHEN_HISTORICAL_PAGE_LIMIT", "300"))

CMC_API_KEY_ENV = "CMC_API_KEY"

def _discover_cmc_api_key() -> Tuple[str, str]:
    """
    Portable credential discovery.

    Production rule (Colab, Jupyter, PythonAnywhere, local Python):
      * credentials are supplied through process environment variables;
      * no notebook-specific secret store is required by production logic;
      * the credential value is never printed, serialized, or hashed.
    """
    env_candidates = (
        ("CMC_API_KEY", os.getenv("CMC_API_KEY", "")),
        ("KITCHEN_CMC_API_KEY", os.getenv("KITCHEN_CMC_API_KEY", "")),
        ("COINMARKETCAP_API_KEY", os.getenv("COINMARKETCAP_API_KEY", "")),
        ("CMC_PRO_API_KEY", os.getenv("CMC_PRO_API_KEY", "")),
    )
    for name, value in env_candidates:
        value = str(value).strip()
        if value:
            return value, f"ENV:{name}"
    return "", "NONE"

CMC_API_KEY, CMC_API_KEY_SOURCE = _discover_cmc_api_key()


CG_API_KEY_ENV = "COINGECKO_API_KEY"
CG_API_KEY = (
    os.getenv("COINGECKO_API_KEY", "").strip()
    or os.getenv("KITCHEN_COINGECKO_API_KEY", "").strip()
)

CORE_ASSETS = {
    1: "BTC",
    1027: "ETH",
    825: "USDT",
}

PROVIDER_PRIORITY = {
    "global": ["coinmarketcap", "coingecko"],
    "exchange": [
        "binance", "okx", "bybit", "kucoin",
        "coinbase", "gate", "upbit", "bitget",
    ],
}

FORBIDDEN_RUNTIME_PROVIDERS = {
    "tradingview",
    "tradingview_reference",
    "tv",
}

REFERENCE_STATUS = "REFERENCE_NOT_USED"
REFERENCE_PROVIDER = "coingecko"
REFERENCE_MAX_PRICE_DEVIATION_PCT = 1.0
REFERENCE_MIN_OVERLAP = 0.80
TIME_SERIES_LOOKBACK_DAYS = int(os.getenv("KITCHEN_TIME_SERIES_LOOKBACK_DAYS", "1"))
# Historical global-index Time Series is a capability requirement only when
# HISTORICAL mode is explicitly requested. A SNAPSHOT run must not fail merely
# because a provider does not expose historical 5m global-index data.
HISTORICAL_TIME_SERIES_REQUIRED = True
# CoinGecko cross-source reference is evidence, not a foundation dependency.
REFERENCE_REQUIRED_FOR_LOCK = False

# Production market-engine identifiers.
OUTLIER_ENGINE_VERSION = "OUTLIER_ENGINE_V2"
PRICE_AGGREGATOR_VERSION = "PRICE_AGGREGATOR_V2_RELIABILITY_WEIGHTED"
MARKET_CAP_VERSION = "KITCHEN_MARKET_CAP_V1"
INDEX_VERSION = "KITCHEN_INDEX_V1"
RANKING_VERSION = "KITCHEN_RANKING_V1"
IDENTITY_VERSION = "KITCHEN_IDENTITY_V1"

# Explicit, configurable market-quality thresholds.
QUALITY_CONFIG = {
    "max_source_age_seconds": FRESHNESS_THRESHOLD_SECONDS,
    "max_price_deviation_pct_from_median": 5.0,
    "max_robust_z": 6.0,
    "min_volume_usd_24h": 0.0,
    "min_markets_for_aggregate": 1,
    "mad_epsilon": 1e-12,
    "max_orderbook_age_seconds": 30.0,
    "depth_bands_pct": (0.05, 0.10, 0.25, 0.50, 1.00),
    "spread_soft_limit_bps": 20.0,
    "price_impact_notional_usd": (10_000.0, 100_000.0),
    "reliability_component_weights": {
        "data_integrity": 0.15,
        "freshness": 0.10,
        "volume": 0.15,
        "near_depth": 0.20,
        "far_depth": 0.10,
        "spread": 0.10,
        "price_impact": 0.10,
        "cross_exchange_consistency": 0.10,
    },
    "min_reference_exchanges": 2,
    "preferred_reference_exchanges": 3,
}

# No automatic snapshot creation.
AUTO_PERSIST_SNAPSHOT = False


# =============================================================================
# 1. BASIC UTILITIES
# =============================================================================

def utc_now() -> str:
    return datetime.now(timezone.utc).isoformat()


def safe_float(value: Any) -> Optional[float]:
    try:
        result = float(value)
        return result if math.isfinite(result) else None
    except Exception:
        return None


def safe_int(value: Any) -> Optional[int]:
    try:
        if isinstance(value, bool):
            return None
        return int(value)
    except Exception:
        return None


def parse_timestamp(value: Any) -> Optional[datetime]:
    if value is None:
        return None
    if isinstance(value, datetime):
        dt = value
    else:
        s = str(value).strip()
        if not s:
            return None
        if s.endswith("Z"):
            s = s[:-1] + "+00:00"
        try:
            dt = datetime.fromisoformat(s)
        except Exception:
            return None
    if dt.tzinfo is None:
        dt = dt.replace(tzinfo=timezone.utc)
    return dt.astimezone(timezone.utc)


def timestamp_age_seconds(source_timestamp: Any, retrieved_at: Any) -> Optional[float]:
    src = parse_timestamp(source_timestamp)
    ret = parse_timestamp(retrieved_at)
    if not src or not ret:
        return None
    return max(0.0, (ret - src).total_seconds())


def safe_json(value: Any) -> Any:
    if isinstance(value, dict):
        return {str(k): safe_json(v) for k, v in value.items()}
    if isinstance(value, (list, tuple)):
        return [safe_json(v) for v in value]
    if isinstance(value, float):
        return value if math.isfinite(value) else None
    if isinstance(value, datetime):
        return value.astimezone(timezone.utc).isoformat()
    return value


def canonical_bytes(value: Any) -> bytes:
    return json.dumps(
        safe_json(value),
        ensure_ascii=False,
        sort_keys=True,
        separators=(",", ":"),
        allow_nan=False,
    ).encode("utf-8")


def sha256_bytes(data: bytes) -> str:
    return hashlib.sha256(data).hexdigest()


def sha256_obj(value: Any) -> str:
    return sha256_bytes(canonical_bytes(value))


def redact(value: Any) -> Any:
    secret_words = (
        "api_key", "apikey", "authorization", "token", "secret",
        "password", "credential", "private_key",
    )
    if isinstance(value, dict):
        result = {}
        for key, item in value.items():
            if any(word in str(key).lower() for word in secret_words):
                result[str(key)] = "[REDACTED]"
            else:
                result[str(key)] = redact(item)
        return result
    if isinstance(value, list):
        return [redact(item) for item in value]
    return value


def safe_name(value: Any) -> str:
    return re.sub(r"[^A-Za-z0-9_.-]+", "_", str(value))


def deep_copy(value: Any) -> Any:
    return copy.deepcopy(value)


def deterministic_sort_key(record: Dict[str, Any]) -> Tuple:
    mc = safe_float(record.get("market_cap"))
    asset_id = record.get("canonical_asset_id") or record.get("provider_asset_id") or ""
    symbol = str(record.get("symbol") or "").upper()
    return (-(mc if mc is not None else -1.0), str(asset_id), symbol)


# =============================================================================
# 2. PROJECT ROOT / STORAGE
# =============================================================================

def resolve_root() -> Path:
    candidates: List[Path] = []

    explicit = os.getenv("KITCHEN_ROOT", "").strip()
    if explicit:
        candidates.append(Path(explicit).expanduser())

    def _portable_candidate(path: Path) -> bool:
        try:
            resolved = path.expanduser().resolve()
            parts = {part.lower() for part in resolved.parts}
            # Never select a notebook-host-specific content tree.
            if "content" in parts:
                return False
            return resolved.is_dir()
        except Exception:
            return False

    try:
        cwd = Path.cwd()
        if _portable_candidate(cwd):
            candidates.append(cwd)
    except Exception:
        pass

    # Portable fallbacks only. Notebook-host-specific filesystem paths are excluded.
    candidates.extend([
        Path("/mnt/data"),
        Path("/tmp"),
    ])

    existing: List[Path] = []
    for candidate in candidates:
        try:
            candidate = candidate.resolve()
            if candidate.is_dir() and candidate not in existing:
                existing.append(candidate)
        except Exception:
            continue

    markers = (
        "Kitchen Assistant v3.1.3.ipynb",
        "Kitchen_Assistant",
        "Cell_06_5",
        "data",
    )

    for candidate in existing:
        try:
            names = {p.name for p in candidate.iterdir()}
            if any(
                marker in name
                for name in names
                for marker in markers
            ):
                return candidate
        except Exception:
            continue

    return existing[0] if existing else Path("/tmp")


ROOT = resolve_root()
DATA_DIR = ROOT / "data"
RAW_DIR = DATA_DIR / "raw" / "U06_5"
SNAPSHOT_DIR = DATA_DIR / "snapshots" / "U06_5"
AUDIT_DIR = DATA_DIR / "audit" / "U06_5"
HISTORY_DIR = DATA_DIR / "history" / "U06_5"

for directory in (DATA_DIR, RAW_DIR, SNAPSHOT_DIR, AUDIT_DIR, HISTORY_DIR):
    directory.mkdir(parents=True, exist_ok=True)


# =============================================================================
# 3. HTTP / PROVIDER RESPONSE MODEL
# =============================================================================

@dataclass
class HTTPResult:
    ok: bool
    status: Optional[int]
    payload: Any
    retrieved_at: str
    elapsed_seconds: float
    provider_status: Any = None
    error_class: Optional[str] = None
    error_message: Optional[str] = None
    response_headers: Optional[Dict[str, str]] = None


def classify_http_error(status: Optional[int], body: str) -> str:
    if status == 429:
        return "RATE_LIMIT"
    if status in (401, 403):
        return "AUTH_FAILED"
    if status == 404:
        return "METRIC_NOT_FOUND"
    if status is not None and 400 <= status < 500:
        return "PROVIDER_ERROR"
    if status is not None and status >= 500:
        return "UNAVAILABLE"
    return "TECHNICAL_ERROR"


def http_json(
    url: str,
    params: Optional[Dict[str, Any]] = None,
    headers: Optional[Dict[str, str]] = None,
) -> HTTPResult:
    query = ""
    if params:
        query = "?" + urlencode(
            {k: v for k, v in params.items() if v is not None}
        )

    started = time.time()
    retrieved_at = utc_now()

    request_headers = {
        "Accept": "application/json",
        "User-Agent": f"Kitchen-Assistant-U06.5/{VERSION}",
    }
    if headers:
        request_headers.update(headers)

    # Never allow a credential to leak through serialized request data.
    request_headers = {
        str(k): str(v)
        for k, v in request_headers.items()
    }

    try:
        req = Request(url + query, headers=request_headers)
        with urlopen(req, timeout=HTTP_TIMEOUT_SECONDS) as response:
            body = response.read()
            payload = json.loads(body.decode("utf-8"))
            response_headers = {
                str(k): str(v) for k, v in response.headers.items()
            }
            provider_status = None
            if isinstance(payload, dict):
                provider_status = payload.get("status")
            return HTTPResult(
                ok=True,
                status=response.status,
                payload=payload,
                retrieved_at=retrieved_at,
                elapsed_seconds=time.time() - started,
                provider_status=provider_status,
                response_headers=response_headers,
            )
    except HTTPError as exc:
        try:
            body = exc.read().decode("utf-8", "replace")[:2000]
        except Exception:
            body = ""
        return HTTPResult(
            ok=False,
            status=exc.code,
            payload=None,
            retrieved_at=retrieved_at,
            elapsed_seconds=time.time() - started,
            error_class=classify_http_error(exc.code, body),
            error_message=f"HTTP {exc.code}: {body}",
        )
    except URLError as exc:
        return HTTPResult(
            ok=False,
            status=None,
            payload=None,
            retrieved_at=retrieved_at,
            elapsed_seconds=time.time() - started,
            error_class="UNAVAILABLE",
            error_message=str(exc),
        )
    except json.JSONDecodeError as exc:
        return HTTPResult(
            ok=False,
            status=None,
            payload=None,
            retrieved_at=retrieved_at,
            elapsed_seconds=time.time() - started,
            error_class="MALFORMED_RESPONSE",
            error_message=f"Invalid JSON: {exc}",
        )
    except Exception as exc:
        return HTTPResult(
            ok=False,
            status=None,
            payload=None,
            retrieved_at=retrieved_at,
            elapsed_seconds=time.time() - started,
            error_class="TECHNICAL_ERROR",
            error_message=f"{type(exc).__name__}: {exc}",
        )


# =============================================================================
# 4. CMC RESPONSE HELPERS
# =============================================================================

def cmc_status_ok(payload: Any) -> bool:
    if not isinstance(payload, dict):
        return False
    status = payload.get("status")
    if status is None:
        return True
    if isinstance(status, dict):
        code = status.get("error_code")
        return code in (0, "0", None)
    return True


def cmc_data(payload: Any) -> Any:
    if isinstance(payload, dict):
        return payload.get("data")
    return None


def cmc_quote(asset: Dict[str, Any]) -> Dict[str, Any]:
    quote = asset.get("quote") if isinstance(asset, dict) else None

    # Endpoint-specific schema: some CMC responses expose a dict.
    if isinstance(quote, dict):
        usd = quote.get("USD")
        if isinstance(usd, dict):
            return usd
        return quote

    # Endpoint-specific schema: listings/quotes may expose a list.
    if isinstance(quote, list):
        for item in quote:
            if not isinstance(item, dict):
                continue
            usd = item.get("USD")
            if isinstance(usd, dict):
                return usd
            if "price" in item or "market_cap" in item:
                return item

    return {}


def cmc_asset_records(data: Any) -> List[Dict[str, Any]]:
    if not isinstance(data, list):
        return []
    return [x for x in data if isinstance(x, dict)]


# =============================================================================
# 5. PROVIDER ACQUISITION
# =============================================================================

def cmc_headers(authenticated: bool) -> Dict[str, str]:
    if authenticated:
        if not CMC_API_KEY:
            raise RuntimeError("CMC authenticated mode requested without CMC_API_KEY.")
        return {"X-CMC_PRO_API_KEY": CMC_API_KEY}
    # Keyless mode MUST NOT attach an API-key header.
    return {}


def acquire_cmc(
    endpoint: str,
    params: Optional[Dict[str, Any]] = None,
    authenticated: bool = False,
) -> HTTPResult:
    base = CMC_BASE_AUTH if authenticated else CMC_BASE_KEYLESS
    mode = "AUTHENTICATED" if authenticated else "KEYLESS"
    _ = mode  # explicit semantic marker for callers/audit
    return http_json(
        base.rstrip("/") + "/" + endpoint.lstrip("/"),
        params=params,
        headers=cmc_headers(authenticated),
    )


def acquire_cmc_top125(authenticated: bool = False) -> HTTPResult:
    return acquire_cmc(
        "/v3/cryptocurrency/listings/latest",
        {"start": 1, "limit": TOP_N, "convert": "USD"},
        authenticated=authenticated,
    )


def acquire_cmc_quotes(
    asset_ids: Sequence[int] = tuple(CORE_ASSETS.keys()),
    authenticated: bool = False,
) -> HTTPResult:
    return acquire_cmc(
        "/v3/cryptocurrency/quotes/latest",
        {
            "id": ",".join(str(x) for x in asset_ids),
            "convert": "USD",
        },
        authenticated=authenticated,
    )


def acquire_cmc_global(authenticated: bool = False) -> HTTPResult:
    return acquire_cmc(
        "/v1/global-metrics/quotes/latest",
        {"convert": "USD"},
        authenticated=authenticated,
    )


def acquire_cmc_simple_price(
    asset_ids: Sequence[int] = tuple(CORE_ASSETS.keys()),
    authenticated: bool = False,
) -> HTTPResult:
    return acquire_cmc(
        "/v1/simple/price",
        {
            "id": ",".join(str(x) for x in asset_ids),
            "convert": "USD",
        },
        authenticated=authenticated,
    )


def acquire_coingecko_top125() -> HTTPResult:
    return http_json(
        COINGECKO_BASE + "/coins/markets",
        {
            "vs_currency": "usd",
            "order": "market_cap_desc",
            "per_page": TOP_N,
            "page": 1,
            "sparkline": "false",
        },
    )


def _valid_binance_kline_row(row: Any) -> bool:
    if not isinstance(row, list) or len(row) < 7:
        return False
    numeric_positions = (1, 2, 3, 4, 5, 6)
    try:
        values = [float(row[i]) for i in numeric_positions]
        return all(math.isfinite(v) for v in values) and values[5] > 0 and values[4] >= 0
    except Exception:
        return False


def _valid_coinbase_candle_row(row: Any) -> bool:
    if not isinstance(row, list) or len(row) < 6:
        return False
    try:
        values = [float(row[i]) for i in range(1, 6)]
        return all(math.isfinite(v) for v in values) and values[1] > 0 and values[4] >= 0
    except Exception:
        return False


def acquire_exchange_symbol_ohlcv(symbol: str) -> Dict[str, Any]:
    """
    Real exchange OHLCV acquisition for the U05 symbols.

    Contract:
      Binance Spot Public is primary.
      Coinbase Exchange Public is fallback only when Binance fails to provide
      valid requested data for that symbol.
      Provider-supplied traded volume is preserved exactly; no synthetic
      volume, close*volume, interpolation, or silent resampling is allowed.
    """
    if symbol not in U05_EXCHANGE_SYMBOLS:
        return {
            "symbol": symbol,
            "status": "NOT_AVAILABLE",
            "selected_provider": None,
            "fallback_used": False,
            "attempts": [],
            "ohlcv": [],
        }

    attempts: List[Dict[str, Any]] = []
    binance = http_json(
        "https://api.binance.com/api/v3/klines",
        params={
            "symbol": symbol,
            "interval": EXCHANGE_KLINE_INTERVAL,
            "limit": EXCHANGE_KLINE_LIMIT,
        },
    )
    binance_valid = (
        binance.ok
        and isinstance(binance.payload, list)
        and len(binance.payload) >= 1
        and all(_valid_binance_kline_row(x) for x in binance.payload)
    )
    attempts.append({
        "provider": "binance",
        "provider_mode": "PUBLIC_SPOT",
        "endpoint": "/api/v3/klines",
        "symbol": symbol,
        "interval": EXCHANGE_KLINE_INTERVAL,
        "http_status": binance.status,
        "ok": binance.ok,
        "valid": binance_valid,
        "retrieved_at": binance.retrieved_at,
        "error_class": binance.error_class,
        "error_message": binance.error_message,
    })

    if binance_valid:
        rows = []
        for row in binance.payload:
            rows.append({
                "exchange": "binance",
                "market_id": symbol,
                "timestamp": datetime.fromtimestamp(float(row[0]) / 1000, tz=timezone.utc).isoformat(),
                "open": safe_float(row[1]),
                "high": safe_float(row[2]),
                "low": safe_float(row[3]),
                "close": safe_float(row[4]),
                "volume": safe_float(row[5]),
                "quote_volume": safe_float(row[7]) if len(row) > 7 else None,
                "volume_source": "PROVIDER_SUPPLIED_TRADED_BASE_VOLUME",
                "timeframe": EXCHANGE_KLINE_INTERVAL,
                "provider_timestamp_ms": safe_int(row[0]),
            })
        return {
            "symbol": symbol,
            "status": "VALIDATED",
            "selected_provider": "binance",
            "selected_provider_mode": "PUBLIC_SPOT",
            "fallback_used": False,
            "attempts": attempts,
            "ohlcv": rows,
        }

    product_id = U05_EXCHANGE_SYMBOLS[symbol]
    coinbase = http_json(
        f"https://api.exchange.coinbase.com/products/{product_id}/candles",
        params={"granularity": 300},
    )
    coinbase_valid = (
        coinbase.ok
        and isinstance(coinbase.payload, list)
        and len(coinbase.payload) >= 1
        and all(_valid_coinbase_candle_row(x) for x in coinbase.payload)
    )
    attempts.append({
        "provider": "coinbase",
        "provider_mode": "PUBLIC_SPOT_FALLBACK",
        "endpoint": f"/products/{product_id}/candles",
        "symbol": symbol,
        "interval": EXCHANGE_KLINE_INTERVAL,
        "http_status": coinbase.status,
        "ok": coinbase.ok,
        "valid": coinbase_valid,
        "retrieved_at": coinbase.retrieved_at,
        "error_class": coinbase.error_class,
        "error_message": coinbase.error_message,
    })

    if coinbase_valid:
        rows = []
        # Coinbase candle schema: [time, low, high, open, close, volume].
        for row in sorted(coinbase.payload, key=lambda x: x[0]):
            rows.append({
                "exchange": "coinbase",
                "market_id": product_id,
                "timestamp": datetime.fromtimestamp(float(row[0]), tz=timezone.utc).isoformat(),
                "open": safe_float(row[3]),
                "high": safe_float(row[2]),
                "low": safe_float(row[1]),
                "close": safe_float(row[4]),
                "volume": safe_float(row[5]),
                "quote_volume": None,
                "volume_source": "PROVIDER_SUPPLIED_TRADED_BASE_VOLUME",
                "timeframe": EXCHANGE_KLINE_INTERVAL,
                "provider_timestamp_seconds": safe_int(row[0]),
            })
        return {
            "symbol": symbol,
            "status": "VALIDATED",
            "selected_provider": "coinbase",
            "selected_provider_mode": "PUBLIC_SPOT_FALLBACK",
            "fallback_used": True,
            "fallback_reason": "BINANCE_UNAVAILABLE_OR_INVALID",
            "attempts": attempts,
            "ohlcv": rows,
        }

    return {
        "symbol": symbol,
        "status": "NOT_AVAILABLE",
        "selected_provider": None,
        "fallback_used": True,
        "fallback_reason": "BINANCE_AND_COINBASE_UNAVAILABLE_OR_INVALID",
        "attempts": attempts,
        "ohlcv": [],
    }



# -----------------------------------------------------------------------------
# 5B. MULTI-EXCHANGE SOURCE FOUNDATION — 8 CORE PUBLIC SPOT VENUES
# -----------------------------------------------------------------------------
# This is deliberately separate from the U05 Binance -> Coinbase fallback.
# U05 compatibility remains untouched; this engine adds independent source
# diversity for historical coverage and cross-exchange evidence.
#
# Core target: 8 exchanges.
# Technical minimum: 7 validated exchanges.
# No exchange is required to cover every historical date. Historical coverage
# is evaluated per provider / symbol / time range and is never fabricated.

MULTI_EXCHANGE_SPECS = {
    "binance": {
        "mode": "PUBLIC_SPOT",
        "symbol": "BTCUSDT",
        "url": "https://api.binance.com/api/v3/klines",
    },
    "okx": {
        "mode": "PUBLIC_SPOT",
        "symbol": "BTC-USDT",
        "url": "https://www.okx.com/api/v5/market/history-candles",
    },
    "bybit": {
        "mode": "PUBLIC_SPOT",
        "symbol": "BTCUSDT",
        "url": "https://api.bybit.com/v5/market/kline",
    },
    "kucoin": {
        "mode": "PUBLIC_SPOT",
        "symbol": "BTC-USDT",
        "url": "https://api.kucoin.com/api/v1/market/candles",
    },
    "coinbase": {
        "mode": "PUBLIC_SPOT",
        "symbol": "BTC-USDT",
        "url": "https://api.exchange.coinbase.com/products/BTC-USDT/candles",
    },
    "gate": {
        "mode": "PUBLIC_SPOT",
        "symbol": "BTC_USDT",
        "url": "https://api.gateio.ws/api/v4/spot/candlesticks",
    },
    "upbit": {
        "mode": "PUBLIC_SPOT",
        "symbol": "USDT-BTC",
        "url": "https://api.upbit.com/v1/candles/minutes/5",
    },
    "bitget": {
        "mode": "PUBLIC_SPOT",
        "symbol": "BTCUSDT",
        "url": "https://api.bitget.com/api/v2/spot/market/history-candles",
    },
}


def _valid_ohlcv_record(o: Dict[str, Any]) -> bool:
    required = ("timestamp", "open", "high", "low", "close", "volume")
    if not all(o.get(k) is not None for k in required):
        return False
    vals = [safe_float(o[k]) for k in required[1:]]
    return all(v is not None and math.isfinite(v) for v in vals) and all(v >= 0 for v in vals)


def _normalize_multi_exchange_rows(provider: str, payload: Any) -> List[Dict[str, Any]]:
    rows: List[Dict[str, Any]] = []
    if provider == "binance":
        for r in payload if isinstance(payload, list) else []:
            if not _valid_binance_kline_row(r):
                continue
            rows.append({"exchange": provider, "market_id": "BTCUSDT",
                         "timestamp": datetime.fromtimestamp(float(r[0])/1000, tz=timezone.utc).isoformat(),
                         "open": safe_float(r[1]), "high": safe_float(r[2]),
                         "low": safe_float(r[3]), "close": safe_float(r[4]),
                         "volume": safe_float(r[5]), "quote_volume": safe_float(r[7]) if len(r) > 7 else None,
                         "volume_source": "PROVIDER_SUPPLIED_TRADED_BASE_VOLUME",
                         "timeframe": EXCHANGE_KLINE_INTERVAL})
    elif provider == "okx":
        data = payload.get("data", []) if isinstance(payload, dict) else []
        for r in data:
            if not isinstance(r, list) or len(r) < 7:
                continue
            try:
                vals = [float(r[i]) for i in range(1, 7)]
                if not all(math.isfinite(v) for v in vals) or vals[4] < 0:
                    continue
                rows.append({"exchange": provider, "market_id": "BTC-USDT",
                             "timestamp": datetime.fromtimestamp(float(r[0])/1000, tz=timezone.utc).isoformat(),
                             "open": vals[0], "high": vals[1], "low": vals[2], "close": vals[3],
                             "volume": vals[4], "quote_volume": vals[5],
                             "volume_source": "PROVIDER_SUPPLIED_TRADED_BASE_VOLUME",
                             "timeframe": EXCHANGE_KLINE_INTERVAL})
            except Exception:
                continue
    elif provider == "bybit":
        data = payload.get("result", {}).get("list", []) if isinstance(payload, dict) else []
        for r in data:
            if not isinstance(r, list) or len(r) < 6:
                continue
            try:
                vals = [float(r[i]) for i in range(1, 6)]
                if not all(math.isfinite(v) for v in vals) or vals[4] < 0:
                    continue
                rows.append({"exchange": provider, "market_id": "BTCUSDT",
                             "timestamp": datetime.fromtimestamp(float(r[0])/1000, tz=timezone.utc).isoformat(),
                             "open": vals[0], "high": vals[1], "low": vals[2], "close": vals[3],
                             "volume": vals[4], "quote_volume": safe_float(r[5]) if len(r) > 5 else None,
                             "volume_source": "PROVIDER_SUPPLIED_TRADED_BASE_VOLUME",
                             "timeframe": EXCHANGE_KLINE_INTERVAL})
            except Exception:
                continue
    elif provider == "kucoin":
        # KuCoin Spot candles are returned under data as rows:
        # [time, open, close, high, low, volume, turnover].
        # Keep the parser defensive because provider error envelopes can also
        # be returned under data and must never surface as a KeyError.
        data = payload.get("data") if isinstance(payload, dict) else None
        if isinstance(data, dict):
            # Some provider/error envelopes may place rows below a nested data
            # member. Only accept an actual list of candle rows.
            data = data.get("data") if isinstance(data.get("data"), list) else []
        if not isinstance(data, list):
            data = []
        for r in data:
            if not isinstance(r, (list, tuple)) or len(r) < 7:
                continue
            try:
                ts = float(r[0])
                vals = [float(r[i]) for i in range(1, 7)]
                if not math.isfinite(ts) or not all(math.isfinite(v) for v in vals):
                    continue
                if vals[4] < 0 or vals[5] < 0:
                    continue
                rows.append({"exchange": provider, "market_id": "BTC-USDT",
                             "timestamp": datetime.fromtimestamp(ts, tz=timezone.utc).isoformat(),
                             "open": vals[0], "close": vals[1], "high": vals[2], "low": vals[3],
                             "volume": vals[4], "quote_volume": vals[5],
                             "volume_source": "PROVIDER_SUPPLIED_TRADED_BASE_VOLUME",
                             "timeframe": EXCHANGE_KLINE_INTERVAL})
            except (TypeError, ValueError, OverflowError, OSError):
                continue
    elif provider == "coinbase":
        for r in payload if isinstance(payload, list) else []:
            if not _valid_coinbase_candle_row(r):
                continue
            rows.append({"exchange": provider, "market_id": "BTC-USDT",
                         "timestamp": datetime.fromtimestamp(float(r[0]), tz=timezone.utc).isoformat(),
                         "open": safe_float(r[3]), "high": safe_float(r[2]),
                         "low": safe_float(r[1]), "close": safe_float(r[4]), "volume": safe_float(r[5]),
                         "quote_volume": None,
                         "volume_source": "PROVIDER_SUPPLIED_TRADED_BASE_VOLUME",
                         "timeframe": EXCHANGE_KLINE_INTERVAL})
    elif provider == "gate":
        for r in payload if isinstance(payload, list) else []:
            if not isinstance(r, list) or len(r) < 6:
                continue
            try:
                # Gate spot: t, quote-volume, close, high, low, open, base-volume, closed.
                quote_volume = float(r[1])
                close = float(r[2])
                high = float(r[3])
                low = float(r[4])
                open_price = float(r[5])
                base_volume = float(r[6])
                vals = [quote_volume, close, high, low, open_price, base_volume]
                if not all(math.isfinite(v) for v in vals) or base_volume < 0:
                    continue
                rows.append({"exchange": provider, "market_id": "BTC_USDT",
                             "timestamp": datetime.fromtimestamp(float(r[0]), tz=timezone.utc).isoformat(),
                             "open": open_price, "high": high, "low": low,
                             "close": close, "volume": base_volume, "quote_volume": quote_volume,
                             "volume_source": "PROVIDER_SUPPLIED_TRADED_BASE_VOLUME",
                             "timeframe": EXCHANGE_KLINE_INTERVAL})
            except Exception:
                continue
    elif provider == "upbit":
        for r in payload if isinstance(payload, list) else []:
            if not isinstance(r, dict):
                continue
            row = {"exchange": provider, "market_id": r.get("market", "USDT-BTC"),
                   "timestamp": r.get("candle_date_time_utc") or r.get("timestamp"),
                   "open": r.get("opening_price"), "high": r.get("high_price"),
                   "low": r.get("low_price"), "close": r.get("trade_price"),
                   "volume": r.get("candle_acc_trade_volume"),
                   "quote_volume": r.get("candle_acc_trade_price"),
                   "volume_source": "PROVIDER_SUPPLIED_TRADED_BASE_VOLUME",
                   "timeframe": EXCHANGE_KLINE_INTERVAL}
            if _valid_ohlcv_record(row):
                rows.append(row)
    elif provider == "bitget":
        data = payload.get("data", []) if isinstance(payload, dict) else []
        for r in data:
            if not isinstance(r, list) or len(r) < 7:
                continue
            try:
                vals = [float(r[i]) for i in range(1, 7)]
                if not all(math.isfinite(v) for v in vals) or vals[4] < 0:
                    continue
                rows.append({"exchange": provider, "market_id": "BTCUSDT",
                             "timestamp": datetime.fromtimestamp(float(r[0])/1000, tz=timezone.utc).isoformat(),
                             "open": vals[0], "high": vals[1], "low": vals[2], "close": vals[3],
                             "volume": vals[4], "quote_volume": vals[5],
                             "volume_source": "PROVIDER_SUPPLIED_TRADED_BASE_VOLUME",
                             "timeframe": EXCHANGE_KLINE_INTERVAL})
            except Exception:
                continue
    rows.sort(key=lambda x: x.get("timestamp", ""))
    return rows


def acquire_multi_exchange_symbol(provider: str, limit: Optional[int] = None) -> Dict[str, Any]:
    """One current 5m public-spot probe for the canonical BTC market."""
    spec = MULTI_EXCHANGE_SPECS[provider]
    limit = limit or EXCHANGE_KLINE_LIMIT
    params: Dict[str, Any]
    if provider == "binance":
        params = {"symbol": spec["symbol"], "interval": "5m", "limit": limit}
    elif provider == "okx":
        params = {"instId": spec["symbol"], "bar": "5m", "limit": min(limit, 300)}
    elif provider == "bybit":
        params = {"category": "spot", "symbol": spec["symbol"], "interval": "5", "limit": min(limit, 1000)}
    elif provider == "kucoin":
        params = {"symbol": spec["symbol"], "type": "5min", "limit": min(limit, 1500)}
    elif provider == "coinbase":
        params = {"granularity": 300}
    elif provider == "gate":
        params = {"currency_pair": spec["symbol"], "interval": "5m", "limit": min(limit, 1000)}
    elif provider == "upbit":
        params = {"market": spec["symbol"], "count": min(limit, 200)}
    elif provider == "bitget":
        # Bitget Spot historical-candle endpoint requires an endTime.
        # Use the current UTC observation boundary; no synthetic range is created.
        params = {
            "symbol": spec["symbol"],
            "granularity": "5min",
            "limit": min(limit, 200),
            "endTime": str(int(time.time() * 1000)),
        }
    else:
        raise ValueError(provider)

    result = http_json(spec["url"], params=params)
    rows = _normalize_multi_exchange_rows(provider, result.payload) if result.ok else []
    valid = len(rows) >= 1
    return {
        "provider": provider,
        "provider_mode": spec["mode"],
        "symbol": spec["symbol"],
        "timeframe": EXCHANGE_KLINE_INTERVAL,
        "endpoint": spec["url"],
        "http_status": result.status,
        "ok": result.ok,
        "status": "VALIDATED" if valid else "NOT_AVAILABLE",
        "row_count": len(rows),
        "retrieved_at": result.retrieved_at,
        "error_class": result.error_class,
        "error_message": result.error_message,
        "ohlcv": rows,
    }



# -----------------------------------------------------------------------------
# 5C. LIVE ORDER-BOOK / MARKET-QUALITY FOUNDATION
# -----------------------------------------------------------------------------
# Candles tell us what traded.  They do NOT by themselves tell us how much
# liquidity is available around the current price.  This layer therefore keeps
# traded-volume evidence and executable-liquidity evidence separate, then
# combines them only inside the reliability model.


def _normalize_orderbook(provider: str, payload: Any, retrieved_at: str) -> Dict[str, Any]:
    """Normalize public spot order books into price/size levels.

    All sizes are base-asset quantities.  The function deliberately does not
    invent timestamps where an exchange does not publish one; retrieved_at is
    retained as the observation time for the order-book snapshot.
    """
    bids: List[Tuple[float, float]] = []
    asks: List[Tuple[float, float]] = []
    source_ts = retrieved_at
    try:
        if provider == "binance":
            bids = [(float(x[0]), float(x[1])) for x in payload.get("bids", []) if len(x) >= 2]
            asks = [(float(x[0]), float(x[1])) for x in payload.get("asks", []) if len(x) >= 2]
        elif provider == "okx":
            data = payload.get("data", []) if isinstance(payload, dict) else []
            book = data[0] if data else {}
            bids = [(float(x[0]), float(x[1])) for x in book.get("bids", []) if len(x) >= 2]
            asks = [(float(x[0]), float(x[1])) for x in book.get("asks", []) if len(x) >= 2]
            if book.get("ts"):
                source_ts = datetime.fromtimestamp(float(book["ts"])/1000, tz=timezone.utc).isoformat()
        elif provider == "bybit":
            book = (payload.get("result", {}) if isinstance(payload, dict) else {})
            bids = [(float(x[0]), float(x[1])) for x in book.get("b", []) if len(x) >= 2]
            asks = [(float(x[0]), float(x[1])) for x in book.get("a", []) if len(x) >= 2]
            if book.get("ts"):
                source_ts = datetime.fromtimestamp(float(book["ts"])/1000, tz=timezone.utc).isoformat()
        elif provider == "kucoin":
            data = payload.get("data", {}) if isinstance(payload, dict) else {}
            bids = [(float(x[0]), float(x[1])) for x in data.get("bids", []) if len(x) >= 2]
            asks = [(float(x[0]), float(x[1])) for x in data.get("asks", []) if len(x) >= 2]
        elif provider == "coinbase":
            bids = [(float(x[0]), float(x[1])) for x in payload.get("bids", []) if len(x) >= 2]
            asks = [(float(x[0]), float(x[1])) for x in payload.get("asks", []) if len(x) >= 2]
        elif provider == "gate":
            bids = [(float(x[0]), float(x[1])) for x in payload.get("bids", []) if len(x) >= 2]
            asks = [(float(x[0]), float(x[1])) for x in payload.get("asks", []) if len(x) >= 2]
            if payload.get("current"):
                try:
                    current_ts = float(payload["current"])
                    if current_ts > 10_000_000_000:
                        current_ts /= 1000.0
                    source_ts = datetime.fromtimestamp(current_ts, tz=timezone.utc).isoformat()
                except (TypeError, ValueError, OverflowError, OSError):
                    # Timestamp metadata is auxiliary. Never discard valid bids/asks.
                    source_ts = retrieved_at
        elif provider == "upbit":
            units = payload if isinstance(payload, list) else []
            unit = units[0] if units else {}
            for x in unit.get("orderbook_units", []):
                bids.append((float(x["bid_price"]), float(x["bid_size"])))
                asks.append((float(x["ask_price"]), float(x["ask_size"])))
            if unit.get("timestamp"):
                source_ts = datetime.fromtimestamp(float(unit["timestamp"])/1000, tz=timezone.utc).isoformat()
        elif provider == "bitget":
            # Bitget Spot v2 returns data as an object: {asks, bids, ts}.
            data = payload.get("data", {}) if isinstance(payload, dict) else {}
            if isinstance(data, list):
                book = data[0] if data and isinstance(data[0], dict) else {}
            else:
                book = data if isinstance(data, dict) else {}
            bids = [(float(x[0]), float(x[1])) for x in book.get("bids", []) if len(x) >= 2]
            asks = [(float(x[0]), float(x[1])) for x in book.get("asks", []) if len(x) >= 2]
            if book.get("ts"):
                source_ts = datetime.fromtimestamp(float(book["ts"])/1000, tz=timezone.utc).isoformat()
    except Exception:
        bids, asks = [], []

    bids = [(p, q) for p, q in bids if math.isfinite(p) and math.isfinite(q) and p > 0 and q >= 0]
    asks = [(p, q) for p, q in asks if math.isfinite(p) and math.isfinite(q) and p > 0 and q >= 0]
    bids.sort(key=lambda x: x[0], reverse=True)
    asks.sort(key=lambda x: x[0])
    if not bids or not asks:
        return {"status": "NOT_AVAILABLE", "bids": [], "asks": [], "source_timestamp": source_ts, "retrieved_at": retrieved_at}
    best_bid, best_ask = bids[0][0], asks[0][0]
    if best_ask <= best_bid:
        return {"status": "INVALID", "bids": bids, "asks": asks, "source_timestamp": source_ts, "retrieved_at": retrieved_at}
    mid = (best_bid + best_ask) / 2.0
    return {
        "status": "VALIDATED",
        "bids": bids,
        "asks": asks,
        "best_bid": best_bid,
        "best_ask": best_ask,
        "mid_price": mid,
        "spread_absolute": best_ask - best_bid,
        "spread_pct": (best_ask - best_bid) / mid * 100.0,
        "source_timestamp": source_ts,
        "retrieved_at": retrieved_at,
        "level_count_bid": len(bids),
        "level_count_ask": len(asks),
    }


def acquire_multi_exchange_orderbook(provider: str) -> Dict[str, Any]:
    spec = MULTI_EXCHANGE_SPECS[provider]
    if provider == "binance":
        params = {"symbol": spec["symbol"], "limit": 100}
    elif provider == "okx":
        params = {"instId": spec["symbol"], "sz": 100}
    elif provider == "bybit":
        params = {"category": "spot", "symbol": spec["symbol"], "limit": 200}
    elif provider == "kucoin":
        params = {"symbol": spec["symbol"]}
        spec_url = "https://api.kucoin.com/api/v1/market/orderbook/level2_100"
    elif provider == "coinbase":
        params = {"level": 2}
    elif provider == "gate":
        params = {"currency_pair": spec["symbol"], "limit": 100}
    elif provider == "upbit":
        params = {"markets": spec["symbol"]}
    elif provider == "bitget":
        params = {"symbol": spec["symbol"], "limit": 150}
    else:
        raise ValueError(provider)
    # Do not use dict.get(default_expression) here: Python evaluates the
    # default expression eagerly, which previously caused a KeyError for
    # KuCoin even when spec_url had already been assigned above.
    if provider == "kucoin":
        url = spec_url
    else:
        url = {
            "binance": "https://api.binance.com/api/v3/depth",
            "okx": "https://www.okx.com/api/v5/market/books",
            "bybit": "https://api.bybit.com/v5/market/orderbook",
            "coinbase": "https://api.exchange.coinbase.com/products/BTC-USDT/book",
            "gate": "https://api.gateio.ws/api/v4/spot/order_book",
            "upbit": "https://api.upbit.com/v1/orderbook",
            "bitget": "https://api.bitget.com/api/v2/spot/market/orderbook",
        }[provider]
    result = http_json(url, params=params)
    book = _normalize_orderbook(provider, result.payload, result.retrieved_at) if result.ok else {"status": "NOT_AVAILABLE"}
    return {
        "provider": provider,
        "endpoint": url,
        "http_status": result.status,
        "ok": result.ok,
        "status": book.get("status", "NOT_AVAILABLE"),
        "retrieved_at": result.retrieved_at,
        "error_class": result.error_class,
        "error_message": result.error_message,
        "orderbook": book,
    }


def _depth_usd(book: Dict[str, Any], mid: float, band_pct: float, side: str) -> float:
    if not book or mid <= 0:
        return 0.0
    bound = band_pct / 100.0
    levels = book.get("bids", []) if side == "bid" else book.get("asks", [])
    total = 0.0
    for price, qty in levels:
        distance = abs(price - mid) / mid
        if distance <= bound:
            total += price * qty
    return total


def _slippage_for_notional(book: Dict[str, Any], mid: float, notional_usd: float, side: str) -> Optional[float]:
    if not book or mid <= 0 or notional_usd <= 0:
        return None
    levels = book.get("asks", []) if side == "buy" else book.get("bids", [])
    remaining = float(notional_usd)
    spent = 0.0
    acquired = 0.0
    for price, qty in levels:
        level_notional = price * qty
        take = min(remaining, level_notional)
        if take <= 0:
            continue
        acquired += take / price
        spent += take
        remaining -= take
        if remaining <= 1e-9:
            break
    if remaining > 1e-6 or acquired <= 0:
        return None
    avg_price = spent / acquired
    if side == "buy":
        return max(0.0, (avg_price - mid) / mid * 100.0)
    return max(0.0, (mid - avg_price) / mid * 100.0)


def orderbook_quality_metrics(book: Dict[str, Any]) -> Dict[str, Any]:
    if book.get("status") != "VALIDATED":
        return {"status": "NOT_AVAILABLE"}
    mid = safe_float(book.get("mid_price"))
    if not mid:
        return {"status": "NOT_AVAILABLE"}
    bands = {}
    for band in QUALITY_CONFIG["depth_bands_pct"]:
        bids = _depth_usd(book, mid, band, "bid")
        asks = _depth_usd(book, mid, band, "ask")
        bands[f"{band:.2f}%"] = {
            "bid_usd": bids,
            "ask_usd": asks,
            "total_usd": bids + asks,
        }
    impacts = {}
    for notional in QUALITY_CONFIG["price_impact_notional_usd"]:
        buy = _slippage_for_notional(book, mid, notional, "buy")
        sell = _slippage_for_notional(book, mid, notional, "sell")
        impacts[str(int(notional))] = {
            "buy_pct": buy,
            "sell_pct": sell,
            "worst_pct": max([x for x in (buy, sell) if x is not None], default=None),
        }
    age = timestamp_age_seconds(book.get("source_timestamp"), book.get("retrieved_at"))
    return {
        "status": "VALIDATED",
        "mid_price": mid,
        "spread_pct": safe_float(book.get("spread_pct")),
        "spread_bps": safe_float(book.get("spread_pct")) * 100.0 if safe_float(book.get("spread_pct")) is not None else None,
        "depth_bands": bands,
        "price_impact": impacts,
        "age_seconds": age,
        "level_count_bid": book.get("level_count_bid"),
        "level_count_ask": book.get("level_count_ask"),
    }

def acquire_multi_exchange_evidence() -> Dict[str, Any]:
    """Acquire current 5m traded evidence AND live order-book quality evidence."""
    per_exchange = {}
    for provider in PROVIDER_PRIORITY["exchange"]:
        try:
            candle = acquire_multi_exchange_symbol(provider)
            orderbook = acquire_multi_exchange_orderbook(provider)
            per_exchange[provider] = {
                **candle,
                "orderbook_evidence": orderbook,
                "orderbook_quality": orderbook_quality_metrics(orderbook.get("orderbook", {})),
            }
        except Exception as exc:
            per_exchange[provider] = {
                "provider": provider,
                "status": "NOT_AVAILABLE",
                "row_count": 0,
                "error_class": type(exc).__name__,
                "error_message": str(exc),
                "ohlcv": [],
                "orderbook_evidence": {"status": "NOT_AVAILABLE"},
                "orderbook_quality": {"status": "NOT_AVAILABLE"},
            }
    validated = [p for p, v in per_exchange.items() if v.get("status") == "VALIDATED"]
    orderbook_validated = [p for p, v in per_exchange.items() if v.get("orderbook_quality", {}).get("status") == "VALIDATED"]
    return {
        "target_exchange_count": TARGET_EXCHANGE_COUNT,
        "minimum_validated_exchange_count": MIN_VALIDATED_EXCHANGE_COUNT,
        "configured_providers": list(PROVIDER_PRIORITY["exchange"]),
        "validated_providers": validated,
        "validated_exchange_count": len(validated),
        "orderbook_validated_providers": orderbook_validated,
        "orderbook_validated_exchange_count": len(orderbook_validated),
        "coverage_ratio": len(validated) / TARGET_EXCHANGE_COUNT if TARGET_EXCHANGE_COUNT else 0.0,
        "status": "VALIDATED" if len(validated) >= MIN_VALIDATED_EXCHANGE_COUNT else ("PARTIAL" if validated else "NOT_AVAILABLE"),
        "probe_symbol": MULTI_EXCHANGE_PROBE_SYMBOL,
        "timeframe": EXCHANGE_KLINE_INTERVAL,
        "per_exchange": per_exchange,
        "selection_used_for_current_top125": False,
        "historical_policy": "MAXIMUM_REAL_COVERAGE_PER_PROVIDER_WITHOUT_FABRICATION",
        "cross_exchange_confirmation": "DYNAMIC_WHEN_MULTIPLE_VALID_OBSERVATIONS_OVERLAP",
        "price_foundation_policy": "LIVE_SPOT_REFERENCE_PRICE_REQUIRES_REAL_MARKET_QUALITY_EVIDENCE",
    }


def acquire_exchange_historical_ohlcv(provider: str, start_ts: int, end_ts: int, max_pages: int = 10) -> Dict[str, Any]:
    """Explicit historical 5m fetch capability; never called automatically.

    Pagination follows each provider's public API semantics. This function is
    intentionally explicit so a Custom Scan can request only the needed range.
    Missing periods remain missing; no resampling/backfilling is performed.
    """
    if provider not in MULTI_EXCHANGE_SPECS:
        return {"status": "NOT_AVAILABLE", "reason": "UNKNOWN_PROVIDER", "rows": []}
    rows: List[Dict[str, Any]] = []
    cursor_end = int(end_ts)
    pages = 0
    while pages < max_pages and cursor_end > start_ts:
        spec = MULTI_EXCHANGE_SPECS[provider]
        if provider == "binance":
            params = {"symbol": spec["symbol"], "interval": "5m", "startTime": start_ts*1000, "endTime": cursor_end*1000, "limit": 1000}
        elif provider == "okx":
            params = {"instId": spec["symbol"], "bar": "5m", "after": str(cursor_end*1000), "limit": 300}
        elif provider == "bybit":
            params = {"category": "spot", "symbol": spec["symbol"], "interval": "5", "end": cursor_end*1000, "limit": 1000}
        elif provider == "kucoin":
            params = {"symbol": spec["symbol"], "type": "5min", "startAt": start_ts, "endAt": cursor_end}
        elif provider == "coinbase":
            params = {"granularity": 300, "start": datetime.fromtimestamp(start_ts, tz=timezone.utc).isoformat(), "end": datetime.fromtimestamp(cursor_end, tz=timezone.utc).isoformat()}
        elif provider == "gate":
            params = {"currency_pair": spec["symbol"], "interval": "5m", "from": start_ts, "to": cursor_end}
        elif provider == "upbit":
            params = {"market": spec["symbol"], "to": datetime.fromtimestamp(cursor_end, tz=timezone.utc).isoformat(), "count": 200}
        elif provider == "bitget":
            params = {"symbol": spec["symbol"], "granularity": "5min", "endTime": str(cursor_end*1000), "limit": 200}
        else:
            break
        result = http_json(spec["url"], params=params)
        if not result.ok:
            break
        page_rows = _normalize_multi_exchange_rows(provider, result.payload)
        if not page_rows:
            break
        rows.extend(page_rows)
        oldest = min(parse_timestamp(r["timestamp"]).timestamp() for r in page_rows if parse_timestamp(r["timestamp"]))
        next_cursor = int(oldest) - 300
        if next_cursor >= cursor_end:
            break
        cursor_end = next_cursor
        pages += 1
        if oldest <= start_ts:
            break
    dedup = {}
    for r in rows:
        ts = r.get("timestamp")
        if ts:
            dedup[ts] = r
    final_rows = [dedup[k] for k in sorted(dedup)]
    return {
        "provider": provider,
        "status": "VALIDATED" if final_rows else "NOT_AVAILABLE",
        "requested_start": datetime.fromtimestamp(start_ts, tz=timezone.utc).isoformat(),
        "requested_end": datetime.fromtimestamp(end_ts, tz=timezone.utc).isoformat(),
        "pages": pages,
        "row_count": len(final_rows),
        "rows": final_rows,
        "coverage_note": "REAL_PROVIDER_COVERAGE_ONLY; MISSING_INTERVALS_NOT_FILLED",
    }

def acquire_exchange_evidence() -> Dict[str, Any]:
    """Acquire real U05-compatible exchange OHLCV with deterministic fallback."""
    symbols = list(U05_EXCHANGE_SYMBOLS.keys())
    per_symbol = {symbol: acquire_exchange_symbol_ohlcv(symbol) for symbol in symbols}
    validated = [x for x in per_symbol.values() if x.get("status") == "VALIDATED"]
    return {
        "primary_provider": "binance",
        "fallback_provider": "coinbase",
        "timeframe": EXCHANGE_KLINE_INTERVAL,
        "volume_definition": "PROVIDER_SUPPLIED_TRADED_BASE_VOLUME",
        "requested_symbols": symbols,
        "per_symbol": per_symbol,
        "validated_symbol_count": len(validated),
        "coverage_ratio": len(validated) / len(symbols) if symbols else 0.0,
        "market_pair_engine": {
            "status": "VALIDATED" if len(validated) == len(symbols) else "PARTIAL",
            "primary": "binance",
            "fallback": "coinbase",
            "fallback_only_on_primary_failure": True,
            "selection_used_for_current_top125": False,
        },
    }


# =============================================================================
# 6. NORMALIZATION / IDENTITY
# =============================================================================

def canonical_asset_id(provider: str, provider_asset_id: Any, symbol: Any) -> str:
    if provider_asset_id is not None:
        return f"{provider}:{provider_asset_id}"
    return f"{provider}:symbol:{str(symbol or '').upper()}"


def normalize_cmc_asset(
    asset: Dict[str, Any],
    retrieved_at: str,
    provider_mode: str,
) -> Dict[str, Any]:
    quote = cmc_quote(asset)
    provider_id = safe_int(asset.get("id"))
    symbol = str(asset.get("symbol") or "").upper()

    source_timestamp = asset.get("last_updated") or quote.get("last_updated")

    return {
        "provider_asset_id": provider_id,
        "canonical_asset_id": canonical_asset_id(
            "coinmarketcap", provider_id, symbol
        ),
        "symbol": symbol,
        "name": asset.get("name"),
        "provider": "coinmarketcap",
        "provider_mode": provider_mode,
        "provider_rank": safe_int(asset.get("cmc_rank")),
        "price": safe_float(quote.get("price")),
        "market_cap": safe_float(quote.get("market_cap")),
        "volume_24h": safe_float(quote.get("volume_24h")),
        "price_change_24h": safe_float(quote.get("percent_change_24h")),
        "market_cap_change_24h": safe_float(
            quote.get("market_cap_change_24h")
        ),
        "source_timestamp": source_timestamp,
        "retrieved_at": retrieved_at,
        "identity_status": (
            "VALIDATED" if provider_id is not None and symbol else "INVALID"
        ),
        "identity_mapping_provenance": "CMC_PROVIDER_ID",
    }


def normalize_coingecko_asset(
    asset: Dict[str, Any],
    retrieved_at: str,
) -> Dict[str, Any]:
    provider_id = asset.get("id")
    symbol = str(asset.get("symbol") or "").upper()

    return {
        "provider_asset_id": provider_id,
        "canonical_asset_id": canonical_asset_id(
            "coingecko", provider_id, symbol
        ),
        "symbol": symbol,
        "name": asset.get("name"),
        "provider": "coingecko",
        "provider_mode": "PUBLIC",
        "provider_rank": safe_int(asset.get("market_cap_rank")),
        "price": safe_float(asset.get("current_price")),
        "market_cap": safe_float(asset.get("market_cap")),
        "volume_24h": safe_float(asset.get("total_volume")),
        "price_change_24h": safe_float(
            asset.get("price_change_percentage_24h")
        ),
        "market_cap_change_24h": None,
        "source_timestamp": None,
        "retrieved_at": retrieved_at,
        "identity_status": (
            "VALIDATED" if provider_id and symbol else "INVALID"
        ),
        "identity_mapping_provenance": "COINGECKO_PROVIDER_ID",
    }


def validate_identity(record: Dict[str, Any]) -> Tuple[bool, str]:
    provider = record.get("provider")
    provider_id = record.get("provider_asset_id")
    symbol = str(record.get("symbol") or "").strip().upper()

    if not provider:
        return False, "PROVIDER_MISSING"
    if provider in {"coinmarketcap"} and safe_int(provider_id) is None:
        return False, "PROVIDER_ASSET_ID_MISSING"
    if provider == "coingecko" and not str(provider_id or "").strip():
        return False, "PROVIDER_ASSET_ID_MISSING"
    if not symbol:
        return False, "SYMBOL_MISSING"
    return True, "VALID"


# =============================================================================
# 7. NUMERIC / TIMESTAMP / FRESHNESS VALIDATION
# =============================================================================

def validate_numeric_market_fields(record: Dict[str, Any]) -> Tuple[bool, str]:
    price = safe_float(record.get("price"))
    market_cap = safe_float(record.get("market_cap"))

    if price is None or price <= 0:
        return False, "PRICE_INVALID"
    if market_cap is None or market_cap < 0:
        return False, "MARKET_CAP_INVALID"

    volume = record.get("volume_24h")
    if volume is not None:
        volume = safe_float(volume)
        if volume is None or volume < 0:
            return False, "VOLUME_INVALID"

    return True, "VALID"


def validate_timestamp_fields(
    record: Dict[str, Any],
) -> Tuple[bool, str, Optional[float]]:
    retrieved = parse_timestamp(record.get("retrieved_at"))
    source = parse_timestamp(record.get("source_timestamp"))

    if not retrieved:
        return False, "RETRIEVED_TIMESTAMP_INVALID", None

    if source:
        age = timestamp_age_seconds(source, retrieved)
        if age is None:
            return False, "TIMESTAMP_AGE_INVALID", None
        return True, "VALID", age

    # Some providers do not expose a source timestamp on every endpoint.
    # This is not a fabricated source timestamp; it is explicitly absent.
    return True, "SOURCE_TIMESTAMP_NOT_AVAILABLE", None


def freshness_status(
    source_timestamp: Optional[str],
    retrieved_at: str,
    threshold_seconds: int = FRESHNESS_THRESHOLD_SECONDS,
) -> str:
    if not source_timestamp:
        return "FRESHNESS_UNVERIFIED"
    age = timestamp_age_seconds(source_timestamp, retrieved_at)
    if age is None:
        return "UNAVAILABLE"
    return "FRESH" if age <= threshold_seconds else "STALE"


def validate_record(
    record: Dict[str, Any],
) -> Tuple[bool, str, Dict[str, Any]]:
    identity_ok, identity_reason = validate_identity(record)
    if not identity_ok:
        return False, identity_reason, {}

    numeric_ok, numeric_reason = validate_numeric_market_fields(record)
    if not numeric_ok:
        return False, numeric_reason, {}

    ts_ok, ts_reason, age = validate_timestamp_fields(record)
    if not ts_ok:
        return False, ts_reason, {}

    freshness = freshness_status(
        record.get("source_timestamp"),
        record.get("retrieved_at"),
    )

    # For current Top-125 snapshots, stale source observations are rejected.
    if freshness == "STALE":
        return False, "STALE_RESPONSE", {
            "age_seconds": age,
            "freshness": freshness,
        }

    return True, "VALID", {
        "age_seconds": age,
        "freshness": freshness,
        "timestamp_status": ts_reason,
    }


# =============================================================================
# 8. TOP-125 VALIDATION / CANDIDATE UNIVERSE
# =============================================================================

def validate_ranked_universe(
    records: Sequence[Dict[str, Any]],
    expected_count: int = TOP_N,
) -> Dict[str, Any]:
    valid: List[Dict[str, Any]] = []
    invalid: List[Dict[str, Any]] = []

    for record in records:
        ok, reason, metadata = validate_record(record)
        if ok:
            item = deep_copy(record)
            item["validation_status"] = "VALIDATED"
            item["validation_metadata"] = metadata
            valid.append(item)
        else:
            invalid.append({
                "asset": deep_copy(record),
                "reason": reason,
            })

    provider_ids = [
        str(x.get("provider_asset_id"))
        for x in valid
        if x.get("provider_asset_id") is not None
    ]
    duplicate_ids = sorted(
        [key for key, count in Counter(provider_ids).items() if count > 1]
    )

    provider_ranks = [
        safe_int(x.get("provider_rank"))
        for x in valid
        if safe_int(x.get("provider_rank")) is not None
    ]
    rank_set = set(provider_ranks)
    expected_ranks = set(range(1, expected_count + 1))

    missing_ranks = sorted(expected_ranks - rank_set)
    unexpected_ranks = sorted(rank_set - expected_ranks)

    complete = (
        len(valid) == expected_count
        and not invalid
        and not duplicate_ids
        and rank_set == expected_ranks
    )

    return {
        "status": "VALIDATED" if complete else "DATA_PARTIAL",
        "expected": expected_count,
        "received": len(records),
        "valid": len(valid),
        "invalid": len(invalid),
        "missing": max(0, expected_count - len(valid)),
        "coverage_ratio": (
            len(valid) / expected_count if expected_count else 0.0
        ),
        "duplicate_count": len(duplicate_ids),
        "duplicate_ids": duplicate_ids,
        "missing_ranks": missing_ranks,
        "unexpected_ranks": unexpected_ranks,
        "invalid_records": invalid,
        "records": valid,
    }


def validate_core_assets(
    records: Sequence[Dict[str, Any]],
) -> Dict[str, Any]:
    by_provider_id = {}
    by_symbol = defaultdict(list)

    for record in records:
        pid = record.get("provider_asset_id")
        if pid is not None:
            by_provider_id[str(pid)] = record
        symbol = str(record.get("symbol") or "").upper()
        if symbol:
            by_symbol[symbol].append(record)

    output: Dict[str, Any] = {}

    for asset_id, symbol in CORE_ASSETS.items():
        record = by_provider_id.get(str(asset_id))
        if record is None:
            candidates = by_symbol.get(symbol, [])
            if len(candidates) == 1:
                record = candidates[0]

        output[symbol] = {
            "status": "VALIDATED" if record else "MISSING",
            "provider_asset_id": asset_id,
            "symbol": symbol,
            "price": record.get("price") if record else None,
            "market_cap": record.get("market_cap") if record else None,
            "provider_rank": (
                record.get("provider_rank") if record else None
            ),
            "source_timestamp": (
                record.get("source_timestamp") if record else None
            ),
            "identity_status": (
                record.get("identity_status") if record else None
            ),
        }

    return {
        "status": (
            "VALIDATED"
            if all(x["status"] == "VALIDATED" for x in output.values())
            else "DATA_PARTIAL"
        ),
        "assets": output,
    }


# =============================================================================
# 9. RAW PRESERVATION
# =============================================================================

def persist_raw_artifact(
    run_id: str,
    category: str,
    provider: str,
    provider_mode: str,
    endpoint: str,
    dataset: str,
    request_params: Optional[Dict[str, Any]],
    result: HTTPResult,
) -> Dict[str, Any]:
    payload = {
        "provider": provider,
        "provider_mode": provider_mode,
        "endpoint": endpoint,
        "dataset": dataset,
        "request_parameters": redact(request_params or {}),
        "retrieval_timestamp": result.retrieved_at,
        "http_status": result.status,
        "provider_status": safe_json(result.provider_status),
        "error_class": result.error_class,
        "error_message": result.error_message,
        "raw_payload": redact(result.payload),
    }

    data = canonical_bytes(payload)
    path = RAW_DIR / f"{safe_name(run_id)}__{safe_name(category)}.json"
    path.write_bytes(data)

    return {
        "path": str(path),
        "sha256": sha256_bytes(data),
        "bytes": len(data),
        "provider": provider,
        "provider_mode": provider_mode,
        "endpoint": endpoint,
        "dataset": dataset,
    }


# =============================================================================
# 10. MULTI-SOURCE EVIDENCE
# =============================================================================

def build_provider_cross_source_evidence(
    cmc_records: Sequence[Dict[str, Any]],
    cg_records: Sequence[Dict[str, Any]],
) -> Dict[str, Any]:
    cmc_by_symbol = {
        str(x.get("symbol") or "").upper(): x
        for x in cmc_records
        if x.get("symbol")
    }
    cg_by_symbol = {
        str(x.get("symbol") or "").upper(): x
        for x in cg_records
        if x.get("symbol")
    }

    evidence = {}

    for symbol in sorted(set(cmc_by_symbol) | set(cg_by_symbol)):
        cmc = cmc_by_symbol.get(symbol)
        cg = cg_by_symbol.get(symbol)

        cmc_price = safe_float(cmc.get("price")) if cmc else None
        cg_price = safe_float(cg.get("price")) if cg else None

        relative_diff = None
        absolute_diff = None
        if cmc_price is not None and cg_price is not None:
            absolute_diff = abs(cmc_price - cg_price)
            if cg_price != 0:
                relative_diff = absolute_diff / abs(cg_price) * 100.0

        evidence[symbol] = {
            "CMC_PRICE": cmc_price,
            "COINGECKO_PRICE": cg_price,
            "absolute_difference": absolute_diff,
            "relative_difference_pct": relative_diff,
            "timestamp_difference_seconds": (
                timestamp_age_seconds(
                    cmc.get("source_timestamp"),
                    cg.get("source_timestamp"),
                )
                if cmc and cg and cmc.get("source_timestamp")
                and cg.get("source_timestamp")
                else None
            ),
            "agreement_status": (
                "AGREE"
                if relative_diff is not None and relative_diff <= 1.0
                else "DISAGREE"
                if relative_diff is not None
                else "UNAVAILABLE"
            ),
            "source_status": {
                "cmc": "AVAILABLE" if cmc else "UNAVAILABLE",
                "coingecko": "AVAILABLE" if cg else "UNAVAILABLE",
            },
        }

    return evidence


# =============================================================================
# 11. MARKET-PAIR / PRICE ENGINE
# =============================================================================

def make_market_observation(
    *,
    exchange: str,
    exchange_id: str,
    asset: str,
    base_asset: str,
    quote_asset: str,
    market_id: str,
    market_type: str,
    price: Any,
    volume_usd_24h: Any,
    timestamp: Any,
    retrieved_at: str,
    provider: str,
) -> Dict[str, Any]:
    source_timestamp = timestamp
    return {
        "exchange": exchange,
        "exchange_id": exchange_id,
        "asset": asset,
        "base_asset": base_asset,
        "quote_asset": quote_asset,
        "market_id": market_id,
        "market_type": market_type,
        "price": safe_float(price),
        "volume_usd_24h": safe_float(volume_usd_24h),
        "volume_usd_recent": None,
        "trade_count_recent": None,
        "timestamp": source_timestamp,
        "retrieved_at": retrieved_at,
        "age_seconds": timestamp_age_seconds(
            source_timestamp, retrieved_at
        ),
        "provider": provider,
        "quality_status": "UNVALIDATED",
        "rejection_reason": None,
    }


def normalize_quote_to_usd(
    observation: Dict[str, Any],
    conversion_prices: Optional[Dict[str, float]] = None,
) -> Tuple[Optional[float], Dict[str, Any]]:
    quote = str(observation.get("quote_asset") or "").upper()
    price = safe_float(observation.get("price"))
    if price is None or price <= 0:
        return None, {"status": "INVALID_PRICE"}

    conversion_prices = conversion_prices or {}

    if quote in {"USD", "USDT", "USDC"}:
        return price, {
            "status": "VALIDATED",
            "chain": [quote, "USD"],
            "conversion_rate": 1.0,
        }

    if quote == "BTC":
        btc_usd = safe_float(conversion_prices.get("BTC"))
        if btc_usd:
            return price * btc_usd, {
                "status": "VALIDATED",
                "chain": ["BTC", "USD"],
                "conversion_rate": btc_usd,
            }

    if quote == "ETH":
        eth_usd = safe_float(conversion_prices.get("ETH"))
        if eth_usd:
            return price * eth_usd, {
                "status": "VALIDATED",
                "chain": ["ETH", "USD"],
                "conversion_rate": eth_usd,
            }

    return None, {
        "status": "UNAVAILABLE",
        "chain": [quote, "USD"],
        "reason": "NO_VALID_CONVERSION_RATE",
    }


def validate_market_observation(
    observation: Dict[str, Any],
    now: Optional[str] = None,
) -> Tuple[bool, str]:
    now = now or observation.get("retrieved_at") or utc_now()

    if str(observation.get("market_type") or "").upper() != "SPOT":
        return False, "WRONG_MARKET_TYPE"

    price = safe_float(observation.get("price"))
    if price is None or price <= 0:
        return False, "PRICE_INVALID"

    volume = observation.get("volume_usd_24h")
    if volume is not None:
        volume = safe_float(volume)
        if volume is None or volume < QUALITY_CONFIG["min_volume_usd_24h"]:
            return False, "INSUFFICIENT_LIQUIDITY"

    freshness_timestamp = observation.get("freshness_timestamp") or observation.get("timestamp")
    age = timestamp_age_seconds(freshness_timestamp, now)
    if age is not None and age > QUALITY_CONFIG["max_source_age_seconds"]:
        return False, "STALE_DATA"

    if not observation.get("asset"):
        return False, "ASSET_IDENTITY_MISSING"

    if not observation.get("market_id"):
        return False, "MARKET_ID_MISSING"

    return True, "VALID"


def median_abs_deviation(values: Sequence[float]) -> float:
    if not values:
        return 0.0
    med = statistics.median(values)
    deviations = [abs(x - med) for x in values]
    return statistics.median(deviations)


def outlier_filter_v1(
    observations: Sequence[Dict[str, Any]],
) -> Dict[str, Any]:
    values = [
        safe_float(x.get("normalized_usd_price"))
        for x in observations
    ]
    valid_values = [x for x in values if x is not None]

    if not valid_values:
        return {
            "status": "DATA_UNAVAILABLE",
            "accepted": [],
            "rejected": [],
            "median": None,
            "mad": None,
            "config": QUALITY_CONFIG,
            "version": OUTLIER_ENGINE_VERSION,
        }

    med = statistics.median(valid_values)
    mad = median_abs_deviation(valid_values)

    accepted = []
    rejected = []

    for obs in observations:
        price = safe_float(obs.get("normalized_usd_price"))
        if price is None:
            rejected.append({
                "observation": deep_copy(obs),
                "reason": "PRICE_INVALID",
            })
            continue

        pct_dev = (
            abs(price - med) / abs(med) * 100.0
            if med else float("inf")
        )

        if mad > QUALITY_CONFIG["mad_epsilon"]:
            robust_z = 0.6745 * (price - med) / mad
        else:
            robust_z = 0.0 if price == med else float("inf")

        if (
            pct_dev > QUALITY_CONFIG["max_price_deviation_pct_from_median"]
            or abs(robust_z) > QUALITY_CONFIG["max_robust_z"]
        ):
            rejected.append({
                "observation": deep_copy(obs),
                "reason": "ROBUST_OUTLIER",
                "percentage_deviation": pct_dev,
                "robust_z": robust_z,
            })
        else:
            item = deep_copy(obs)
            item["outlier_percentage_deviation"] = pct_dev
            item["outlier_robust_z"] = robust_z
            accepted.append(item)

    return {
        "status": "VALIDATED" if accepted else "DATA_UNAVAILABLE",
        "accepted": accepted,
        "rejected": rejected,
        "median": med,
        "mad": mad,
        "config": deep_copy(QUALITY_CONFIG),
        "version": OUTLIER_ENGINE_VERSION,
    }


def _bounded01(x: Optional[float]) -> Optional[float]:
    if x is None or not math.isfinite(x):
        return None
    return max(0.0, min(1.0, x))


def _relative_log_score(value: Optional[float], peer_values: Sequence[float]) -> Optional[float]:
    """Peer-normalized score. Zero/negative values are unavailable, not zero-quality."""
    if value is None or value <= 0 or not peer_values:
        return None
    vals = [v for v in peer_values if v is not None and v > 0 and math.isfinite(v)]
    if not vals:
        return None
    lo, hi = min(vals), max(vals)
    if hi <= lo:
        return 1.0
    a, b = math.log1p(lo), math.log1p(hi)
    return _bounded01((math.log1p(value) - a) / (b - a))


def _freshness_score(age_seconds: Optional[float]) -> Optional[float]:
    if age_seconds is None:
        return None
    if age_seconds < 0:
        return None
    half_life = max(1.0, QUALITY_CONFIG["max_source_age_seconds"] / 2.0)
    return _bounded01(math.exp(-math.log(2.0) * age_seconds / half_life))


def _spread_score(spread_bps: Optional[float]) -> Optional[float]:
    if spread_bps is None or spread_bps < 0:
        return None
    limit = max(1e-9, float(QUALITY_CONFIG["spread_soft_limit_bps"]))
    return _bounded01(math.exp(-spread_bps / limit))


def _impact_score(impact_pct: Optional[float]) -> Optional[float]:
    if impact_pct is None or impact_pct < 0:
        return None
    return _bounded01(1.0 / (1.0 + impact_pct / 0.10))


def _build_reliability_scores(observations: Sequence[Dict[str, Any]]) -> List[Dict[str, Any]]:
    """Compute dynamic observation-level reliability from real, timestamped evidence.

    No fixed exchange ranking is used. Missing feature evidence is excluded from
    the denominator and reported explicitly; it is never fabricated.
    """
    valid = [x for x in observations if x.get("quality_status") == "VALIDATED"]
    volume_peers = [safe_float(x.get("volume_usd_24h")) for x in valid]
    near_peers = []
    far_peers = []
    for x in valid:
        q = x.get("market_quality") or {}
        bands = q.get("depth_bands", {})
        near_peers.append(safe_float(bands.get("0.10%", {}).get("total_usd")))
        far_peers.append(safe_float(bands.get("1.00%", {}).get("total_usd")))
    volume_peers = [x for x in volume_peers if x is not None and x > 0]
    near_peers = [x for x in near_peers if x is not None and x > 0]
    far_peers = [x for x in far_peers if x is not None and x > 0]

    prices = [safe_float(x.get("normalized_usd_price")) for x in valid]
    prices = [x for x in prices if x is not None and x > 0]
    consensus = statistics.median(prices) if prices else None

    component_weights = QUALITY_CONFIG["reliability_component_weights"]
    result = []
    for obs in valid:
        q = obs.get("market_quality") or {}
        bands = q.get("depth_bands", {})
        impact = q.get("price_impact", {})
        impact_values = [v.get("worst_pct") for v in impact.values() if isinstance(v, dict) and v.get("worst_pct") is not None]
        representative_impact = min(impact_values) if impact_values else None
        price = safe_float(obs.get("normalized_usd_price"))
        deviation = (abs(price - consensus) / consensus * 100.0) if price and consensus else None

        components = {
            "data_integrity": 1.0,
            "freshness": _freshness_score(q.get("age_seconds", obs.get("age_seconds"))),
            "volume": _relative_log_score(safe_float(obs.get("volume_usd_24h")), volume_peers),
            "near_depth": _relative_log_score(safe_float(bands.get("0.10%", {}).get("total_usd")), near_peers),
            "far_depth": _relative_log_score(safe_float(bands.get("1.00%", {}).get("total_usd")), far_peers),
            "spread": _spread_score(safe_float(q.get("spread_bps"))),
            "price_impact": _impact_score(representative_impact),
            "cross_exchange_consistency": _bounded01(1.0 / (1.0 + (deviation or 0.0) / 0.25)),
        }
        numerator = 0.0
        denominator = 0.0
        component_details = {}
        for name, base_weight in component_weights.items():
            value = components.get(name)
            if value is None:
                component_details[name] = {"status": "NOT_AVAILABLE", "weight_excluded": True}
                continue
            numerator += float(base_weight) * value
            denominator += float(base_weight)
            component_details[name] = {"status": "AVAILABLE", "score": value, "base_weight": base_weight}
        score = numerator / denominator if denominator > 0 else 0.0
        item = deep_copy(obs)
        item["reliability"] = {
            "score": _bounded01(score),
            "components": component_details,
            "consensus_price": consensus,
            "price_deviation_pct": deviation,
            "model": "RELIABILITY_MODEL_V1_DYNAMIC",
            "fixed_exchange_rank": False,
        }
        result.append(item)
    return result



def build_live_reference_price_from_multi_exchange(multi_exchange: Dict[str, Any]) -> Dict[str, Any]:
    """Build the canonical current BTC reference price from real exchange evidence.

    This is intentionally symbol-scoped.  It does not pretend that one BTC probe
    represents all Top-125 assets.  Historical scans can call the same engine on
    symbol/time-aligned observations when real provider evidence exists.
    """
    observations = []
    for provider, evidence in (multi_exchange.get("per_exchange") or {}).items():
        rows = evidence.get("ohlcv") or []
        if not rows:
            continue
        row = sorted(rows, key=lambda x: x.get("timestamp", ""))[-1]
        price = safe_float(row.get("close"))
        if not price or price <= 0:
            continue
        quote_volume = safe_float(row.get("quote_volume"))
        if quote_volume is None:
            base_volume = safe_float(row.get("volume"))
            quote_volume = base_volume * price if base_volume is not None else None
        obs = make_market_observation(
            exchange=provider,
            exchange_id=provider,
            asset="BTC",
            base_asset="BTC",
            quote_asset="USDT",
            market_id=row.get("market_id") or provider,
            market_type="SPOT",
            price=price,
            volume_usd_24h=None,
            timestamp=row.get("timestamp"),
            retrieved_at=evidence.get("retrieved_at") or utc_now(),
            provider=provider,
        )
        obs["volume_usd_recent"] = quote_volume
        obs["volume_source"] = row.get("volume_source")
        obs["timeframe"] = row.get("timeframe")
        obs["market_quality"] = evidence.get("orderbook_quality") or {"status": "NOT_AVAILABLE"}
        observations.append(obs)

    if not observations:
        return {
            "status": "NOT_AVAILABLE",
            "symbol": "BTC",
            "reason": "NO_VALID_EXCHANGE_PRICE_OBSERVATIONS",
            "observations": [],
        }

    # Use recent traded quote turnover as a liquidity signal when a genuine
    # 24h ticker is unavailable.  It is explicitly named recent, never 24h.
    for obs in observations:
        obs["volume_usd_24h"] = safe_float(obs.get("volume_usd_recent"))
        obs["volume_metric_used"] = "RECENT_5M_QUOTE_TURNOVER" if obs.get("volume_usd_recent") is not None else "UNAVAILABLE"

        # Canonicalize the exchange observation into USD before validation and
        # aggregation.  BTC-USDT is already a USD-equivalent quote under the
        # current contract, but this explicit step is mandatory so the live
        # reference engine and its tests share the same normalization path.
        normalized_price, usd_meta = normalize_quote_to_usd(obs)
        obs["normalized_usd_price"] = normalized_price
        obs["usd_normalization"] = usd_meta
        # A 5m candle timestamp marks the candle OPEN, not the observation
        # instant. For live validation, estimate the candle-end timestamp for
        # freshness only; the original provider timestamp remains untouched.
        candle_ts = parse_timestamp(obs.get("timestamp"))
        if candle_ts is not None:
            obs["freshness_timestamp"] = (candle_ts + timedelta(minutes=5)).isoformat()

    aggregate = price_aggregate_v1(observations)
    return {
        "status": aggregate.get("status"),
        "symbol": "BTC",
        "timeframe": EXCHANGE_KLINE_INTERVAL,
        "reference_price": aggregate.get("reference_price"),
        "confidence": aggregate.get("confidence"),
        "aggregate": aggregate,
        "observation_count": len(observations),
        "validated_observation_count": len(aggregate.get("accepted_markets", [])),
        "source_policy": "DYNAMIC_RELIABILITY_WEIGHTED_MULTI_EXCHANGE",
        "not_global_market_cap": True,
    }

def price_aggregate_v1(observations: Sequence[Dict[str, Any]]) -> Dict[str, Any]:
    """Reliability-weighted reference-price engine.

    The function accepts already-normalized observations and optional real
    market-quality evidence.  If order-book evidence exists it contributes to
    reliability; if it does not, the missing component is reported and the
    remaining observed components are re-normalized.  No synthetic depth is
    created.
    """
    quality_accepted = []
    quality_rejected = []
    for observation in observations:
        ok, reason = validate_market_observation(observation)
        if not ok:
            item = deep_copy(observation)
            item["quality_status"] = "REJECTED"
            item["rejection_reason"] = reason
            quality_rejected.append(item)
            continue
        item = deep_copy(observation)
        item["quality_status"] = "VALIDATED"
        quality_accepted.append(item)

    outlier_result = outlier_filter_v1(quality_accepted)
    accepted = outlier_result["accepted"]
    rejected = quality_rejected + outlier_result["rejected"]
    if len(accepted) < QUALITY_CONFIG["min_markets_for_aggregate"]:
        return {
            "status": "DATA_UNAVAILABLE", "accepted_markets": accepted,
            "rejected_markets": rejected, "weights": {},
            "aggregate_formula": "NO_VALID_MARKETS",
            "aggregate_version": PRICE_AGGREGATOR_VERSION,
            "reference_price": None, "confidence": 0.0, "outlier": outlier_result,
        }

    scored = _build_reliability_scores(accepted)
    if len(scored) < QUALITY_CONFIG["min_reference_exchanges"]:
        return {
            "status": "INSUFFICIENT_CROSS_EXCHANGE_CONFIRMATION",
            "accepted_markets": scored, "rejected_markets": rejected,
            "weights": {}, "aggregate_formula": "MINIMUM_CROSS_EXCHANGE_CONFIRMATION_NOT_MET",
            "aggregate_version": PRICE_AGGREGATOR_VERSION,
            "reference_price": None, "confidence": 0.0, "outlier": outlier_result,
        }

    raw_weights = []
    weights: Dict[str, float] = {}
    for index, obs in enumerate(scored):
        rel = safe_float(obs.get("reliability", {}).get("score")) or 0.0
        # Liquidity remains explicit but does not dominate reliability.
        near_depth = safe_float((obs.get("market_quality") or {}).get("depth_bands", {}).get("0.10%", {}).get("total_usd"))
        volume = safe_float(obs.get("volume_usd_24h"))
        liquidity_signal = math.sqrt(max(0.0, (math.log1p(volume or 0.0)) * (math.log1p(near_depth or 0.0)))) if (volume or near_depth) else 1.0
        weight = max(1e-12, rel * liquidity_signal)
        key = f"{obs.get('exchange')}::{obs.get('market_id')}::{index}"
        weights[key] = weight
        raw_weights.append(weight)

    total_weight = sum(raw_weights)
    reference_price = sum(safe_float(o["normalized_usd_price"]) * w for o, w in zip(scored, raw_weights)) / total_weight
    normalized_weights = [w / total_weight for w in raw_weights]
    for obs, nw in zip(scored, normalized_weights):
        obs["final_weight"] = nw

    reliability_values = [safe_float(o.get("reliability", {}).get("score")) for o in scored]
    reliability_values = [x for x in reliability_values if x is not None]
    agreement = 1.0
    if reference_price and scored:
        deviations = [abs(safe_float(o["normalized_usd_price"]) - reference_price) / reference_price for o in scored]
        agreement = _bounded01(1.0 / (1.0 + (statistics.median(deviations) if deviations else 0.0) / 0.002)) or 0.0
    coverage = min(1.0, len(scored) / max(1, TARGET_EXCHANGE_COUNT))
    confidence = _bounded01(0.55 * (statistics.mean(reliability_values) if reliability_values else 0.0) + 0.25 * agreement + 0.20 * coverage) or 0.0

    return {
        "status": "CALCULATED",
        "accepted_markets": scored,
        "rejected_markets": rejected,
        "weights": weights,
        "normalized_weights": {k: v / total_weight for k, v in weights.items()},
        "aggregate_formula": "SUM(normalized_usd_price * dynamic_reliability_score * liquidity_signal) / SUM(dynamic_reliability_score * liquidity_signal)",
        "aggregate_version": PRICE_AGGREGATOR_VERSION,
        "reference_price": reference_price,
        "confidence": confidence,
        "confidence_components": {
            "mean_reliability": statistics.mean(reliability_values) if reliability_values else 0.0,
            "cross_exchange_agreement": agreement,
            "exchange_coverage": coverage,
        },
        "outlier": outlier_result,
        "methodology": {
            "fixed_exchange_ranking": False,
            "volume_is_not_sole_weight": True,
            "orderbook_depth_is_used_when_available": True,
            "missing_features_are_not_fabricated": True,
            "single_exchange_is_not_cross_exchange_confirmed": len(scored) < 2,
        },
    }


# =============================================================================
# 12. SUPPLY / MARKET-CAP ENGINE
# =============================================================================

def validate_circulating_supply(
    asset: Dict[str, Any],
) -> Tuple[bool, str, Optional[float]]:
    supply = safe_float(asset.get("circulating_supply"))
    if supply is None or supply <= 0:
        return False, "CIRCULATING_SUPPLY_INVALID", None

    # Explicitly reject common substitutions if they were passed.
    if asset.get("supply_source_type") in {
        "TOTAL_SUPPLY",
        "MAX_SUPPLY",
        "FDV",
        "ESTIMATE",
    }:
        return False, "FORBIDDEN_SUPPLY_SUBSTITUTION", None

    return True, "VALID", supply


def calculate_market_cap_v1(
    reference_price_usd: Any,
    circulating_supply: Any,
    *,
    input_timestamp: Optional[str],
    source_provenance: Dict[str, Any],
) -> Dict[str, Any]:
    price = safe_float(reference_price_usd)
    supply = safe_float(circulating_supply)

    if price is None or price <= 0:
        return {
            "status": "DATA_UNAVAILABLE",
            "reason": "REFERENCE_PRICE_INVALID",
            "formula_id": MARKET_CAP_VERSION,
        }

    if supply is None or supply <= 0:
        return {
            "status": "DATA_UNAVAILABLE",
            "reason": "CIRCULATING_SUPPLY_INVALID",
            "formula_id": MARKET_CAP_VERSION,
        }

    output = price * supply

    return {
        "status": "CALCULATED",
        "formula_id": MARKET_CAP_VERSION,
        "formula": "REFERENCE_PRICE_USD * VALIDATED_CIRCULATING_SUPPLY",
        "inputs": {
            "reference_price_usd": price,
            "circulating_supply": supply,
        },
        "input_timestamps": [input_timestamp] if input_timestamp else [],
        "input_provenance": deep_copy(source_provenance),
        "processor": "KITCHEN_U06_5",
        "calculation_version": MARKET_CAP_VERSION,
        "output": output,
        "output_validation": (
            "VALID" if math.isfinite(output) and output >= 0
            else "INVALID"
        ),
    }


# =============================================================================
# 13. DYNAMIC RANKING / TOP-125
# =============================================================================

def dynamic_rank_assets(
    valid_assets: Sequence[Dict[str, Any]],
) -> Dict[str, Any]:
    candidates = [
        deep_copy(x)
        for x in valid_assets
        if safe_float(x.get("market_cap")) is not None
        and safe_float(x.get("market_cap")) >= 0
    ]

    candidates.sort(key=deterministic_sort_key)

    ranked = []
    for rank, asset in enumerate(candidates, start=1):
        item = deep_copy(asset)
        item["calculated_rank"] = rank
        item["ranking_version"] = RANKING_VERSION
        ranked.append(item)

    top125 = ranked[:TOP_N]

    for item in top125:
        provider_rank = safe_int(item.get("provider_rank"))
        calculated_rank = safe_int(item.get("calculated_rank"))
        if provider_rank is None:
            consistency = "UNAVAILABLE"
        else:
            diff = abs(provider_rank - calculated_rank)
            consistency = (
                "MATCH" if diff == 0
                else "MINOR_DIFFERENCE" if diff <= 2
                else "WARNING"
            )
        item["rank_consistency"] = consistency

        r = item["calculated_rank"]
        item["segment"] = (
            "BTC" if r == 1
            else "ETH" if r == 2
            else "TOP10_ALT" if r <= 10
            else "BROAD_ALT_11_125"
        )

    return {
        "status": (
            "VALIDATED" if len(top125) == TOP_N
            else "DATA_PARTIAL"
        ),
        "candidate_requested": TARGET_CANDIDATE_POOL,
        "candidate_received": len(valid_assets),
        "candidate_valid": len(candidates),
        "candidate_invalid": max(0, len(valid_assets) - len(candidates)),
        "candidate_excluded": max(0, len(candidates) - TOP_N),
        "candidate_coverage_ratio": (
            len(candidates) / TARGET_CANDIDATE_POOL
            if TARGET_CANDIDATE_POOL else 0.0
        ),
        "ranking_version": RANKING_VERSION,
        "tie_break_rule": (
            "market_cap_desc, canonical_asset_id_asc, symbol_asc"
        ),
        "all_valid_ranked": ranked,
        "top125": top125,
    }


# =============================================================================
# 14. INDEX ENGINE
# =============================================================================

def calculate_indices_from_top125(
    top125: Sequence[Dict[str, Any]],
) -> Dict[str, Any]:
    if len(top125) != TOP_N:
        return {
            "status": "DATA_PARTIAL",
            "reason": "TOP125_NOT_COMPLETE",
            "formula_version": INDEX_VERSION,
            "values": {},
        }

    by_symbol = {
        str(asset.get("symbol") or "").upper(): asset
        for asset in top125
    }

    if any(symbol not in by_symbol for symbol in ("BTC", "ETH", "USDT")):
        return {
            "status": "DATA_UNAVAILABLE",
            "reason": "CORE_ASSET_MISSING",
            "formula_version": INDEX_VERSION,
            "values": {},
        }

    if any(
        safe_float(asset.get("market_cap")) is None
        for asset in top125
    ):
        return {
            "status": "DATA_PARTIAL",
            "reason": "MARKET_CAP_INVALID",
            "formula_version": INDEX_VERSION,
            "values": {},
        }

    total = sum(float(asset["market_cap"]) for asset in top125)
    btc = float(by_symbol["BTC"]["market_cap"])
    eth = float(by_symbol["ETH"]["market_cap"])
    usdt = float(by_symbol["USDT"]["market_cap"])

    others = sum(
        float(asset["market_cap"])
        for asset in top125
        if safe_int(asset.get("calculated_rank")) is not None
        and asset["calculated_rank"] >= 11
    )

    def dominance(numerator: float, denominator: float) -> Optional[float]:
        return numerator / denominator * 100.0 if denominator > 0 else None

    values = {
        "KITCHEN_TOTAL_TOP125": total,
        "KITCHEN_TOTAL2": total - btc,
        "KITCHEN_TOTAL3": total - btc - eth,
        "KITCHEN_OTHERS": others,
        "KITCHEN_BTC_D": dominance(btc, total),
        "KITCHEN_ETH_D": dominance(eth, total),
        "KITCHEN_USDT_D": dominance(usdt, total),
        "KITCHEN_OTHERS_D": dominance(others, total),
    }

    manifests = {}

    for metric, formula in {
        "KITCHEN_TOTAL_TOP125": "SUM(MARKET_CAP of calculated ranks 1-125)",
        "KITCHEN_TOTAL2": "KITCHEN_TOTAL_TOP125 - BTC_MARKET_CAP",
        "KITCHEN_TOTAL3": (
            "KITCHEN_TOTAL_TOP125 - BTC_MARKET_CAP - ETH_MARKET_CAP"
        ),
        "KITCHEN_OTHERS": (
            "SUM(MARKET_CAP of calculated ranks 11-125)"
        ),
        "KITCHEN_BTC_D": (
            "BTC_MARKET_CAP / KITCHEN_TOTAL_TOP125 * 100"
        ),
        "KITCHEN_ETH_D": (
            "ETH_MARKET_CAP / KITCHEN_TOTAL_TOP125 * 100"
        ),
        "KITCHEN_USDT_D": (
            "USDT_MARKET_CAP / KITCHEN_TOTAL_TOP125 * 100"
        ),
        "KITCHEN_OTHERS_D": (
            "KITCHEN_OTHERS / KITCHEN_TOTAL_TOP125 * 100"
        ),
    }.items():
        manifests[metric] = {
            "formula_id": f"{INDEX_VERSION}:{metric}",
            "formula": formula,
            "inputs": {
                "top125_members": [
                    {
                        "canonical_asset_id": x.get("canonical_asset_id"),
                        "rank": x.get("calculated_rank"),
                        "market_cap": x.get("market_cap"),
                    }
                    for x in top125
                ]
            },
            "input_timestamps": sorted({
                str(x.get("source_timestamp"))
                for x in top125
                if x.get("source_timestamp")
            }),
            "input_provenance": "U06.5_VALIDATED_TOP125",
            "calculation_version": INDEX_VERSION,
            "processor": "KITCHEN_U06_5",
            "output": values[metric],
            "validation_status": (
                "VALID"
                if values[metric] is not None
                and math.isfinite(float(values[metric]))
                else "INVALID"
            ),
        }

    return {
        "status": "CALCULATED",
        "calculated_by": "KITCHEN",
        "formula_version": INDEX_VERSION,
        "values": values,
        "manifests": manifests,
        "tradingview_value_used": False,
    }


# =============================================================================
# 15. CMC GLOBAL METRICS — EVIDENCE ONLY
# =============================================================================

def normalize_cmc_global_metrics(data: Any) -> Dict[str, Any]:
    if not isinstance(data, dict):
        return {
            "status": "DATA_UNAVAILABLE",
            "reason": "GLOBAL_METRICS_SCHEMA_INVALID",
        }

    quote = data.get("quote")
    usd = quote.get("USD") if isinstance(quote, dict) else {}

    if not isinstance(usd, dict):
        usd = {}

    return {
        "status": "VALIDATED",
        "provider": "coinmarketcap",
        "provider_mode": "KEYLESS",
        "total_market_cap": safe_float(usd.get("total_market_cap")),
        "btc_dominance": safe_float(usd.get("btc_dominance")),
        "eth_dominance": safe_float(usd.get("eth_dominance")),
        "active_cryptocurrencies": safe_int(
            data.get("active_cryptocurrencies")
        ),
        "source_timestamp": (
            data.get("last_updated")
            or data.get("timestamp")
        ),
    }


# =============================================================================
# 16. TIMEFRAME / HISTORY HONESTY
# =============================================================================

def acquire_market_index_time_series() -> Dict[str, Any]:
    """
    Acquire and validate a REAL 5-minute market-wide historical index.

    Canonical source policy:
      1) CMC authenticated Global Metrics Historical (authoritative external
         market-index evidence; requires a CMC plan with intraday history).
      2) CoinGecko global market-cap history is attempted only as evidence;
         it is accepted only if the returned timestamps genuinely satisfy
         the 5m contract. No interpolation/resampling is performed.

    Critical distinction:
      * This is EXTERNAL MARKET-INDEX EVIDENCE.
      * It is NOT KITCHEN_TOTAL_TOP125 history.
      * Kitchen Top-125 historical indices remain NOT_ACQUIRED until genuine
        historical Top-125 constituent snapshots exist.

    The function returns detailed provider diagnostics so a BLOCKED result is
    actionable rather than opaque. A provider error is never converted to
    PASS, and missing credentials are never hidden.
    """

    attempts: List[Dict[str, Any]] = []

    def _attempt_record(provider: str, mode: str, result: Optional[HTTPResult],
                        requested_params: Optional[Dict[str, Any]] = None,
                        reason: Optional[str] = None) -> Dict[str, Any]:
        rec = {
            "provider": provider,
            "provider_mode": mode,
            "requested_params": requested_params or {},
            "ok": bool(result.ok) if result else False,
            "http_status": result.status if result else None,
            "error_class": result.error_class if result else None,
            "error_message": result.error_message if result else None,
            "provider_status": redact(result.provider_status) if result else None,
            "retrieved_at": result.retrieved_at if result else utc_now(),
            "reason": reason,
        }
        # Never persist or print request headers / credentials.
        return rec

    def _validate_points(points: List[Dict[str, Any]], provider: str,
                         provider_mode: str, reason_prefix: str = "") -> Dict[str, Any]:
        points = sorted(points, key=lambda x: x["timestamp_ms"])
        deduped: List[Dict[str, Any]] = []
        seen = set()
        for point in points:
            ts = point["timestamp_ms"]
            if ts in seen:
                continue
            seen.add(ts)
            deduped.append(point)

        cadence_seconds = [
            (b["timestamp_ms"] - a["timestamp_ms"]) / 1000.0
            for a, b in zip(deduped, deduped[1:])
            if b["timestamp_ms"] > a["timestamp_ms"]
        ]
        cadence_ratio = (
            sum(1 for x in cadence_seconds if 240 <= x <= 360)
            / len(cadence_seconds)
            if cadence_seconds else 0.0
        )
        max_gap_seconds = max(cadence_seconds) if cadence_seconds else None
        cadence_ok = len(deduped) >= 3 and cadence_ratio >= 0.80
        # A 5m history with a single multi-hour hole is not useful for a
        # range/structure consumer even if the aggregate cadence ratio passes.
        continuity_ok = max_gap_seconds is None or max_gap_seconds <= 15 * 60
        status = "VALIDATED" if cadence_ok and continuity_ok else "PARTIAL"

        return {
            "status": status,
            "provider": provider,
            "provider_mode": provider_mode,
            "timeframe": TIMEFRAME,
            "requested_range": f"{TIME_SERIES_LOOKBACK_DAYS}d",
            "available_range": {
                "start": deduped[0]["timestamp"] if deduped else None,
                "end": deduped[-1]["timestamp"] if deduped else None,
            },
            "points": deduped,
            "point_count": len(deduped),
            "coverage_ratio": cadence_ratio,
            "cadence_validation": {
                "target_seconds": 300,
                "sample_count": len(cadence_seconds),
                "within_240_360_seconds_ratio": cadence_ratio,
                "max_gap_seconds": max_gap_seconds,
                "validated": cadence_ok and continuity_ok,
                "continuity_validated": continuity_ok,
            },
            "metric_definition": "GLOBAL_MARKET_CAP_5M_HISTORICAL_INDEX_EVIDENCE",
            "is_kitchen_total": False,
            "silent_resampling": False,
            "fabricated_history": False,
            "source_timestamp": deduped[-1]["timestamp"] if deduped else None,
            "retrieved_at": utc_now(),
            "reason": reason_prefix or (
                "VALIDATED_5M" if status == "VALIDATED" else "5M_CADENCE_OR_CONTINUITY_NOT_VALIDATED"
            ),
        }

    # ------------------------------------------------------------------
    # Provider 1: CoinMarketCap authenticated historical global metrics.
    # CMC documents 5m as a supported interval and requires an authenticated
    # API key/plan for this endpoint.
    # ------------------------------------------------------------------
    cmc_params = {
        "interval": "5m",
        "count": max(3, min(288, TIME_SERIES_LOOKBACK_DAYS * 288)),
        "convert": "USD",
        "aux": "btc_dominance,eth_dominance,active_cryptocurrencies,search_interval",
    }

    if not CMC_API_KEY:
        attempts.append(_attempt_record(
            "coinmarketcap", "AUTHENTICATED", None,
            cmc_params,
            "CMC_API_KEY_NOT_CONFIGURED"
        ))
    else:
        result = acquire_cmc(
            "/v1/global-metrics/quotes/historical",
            cmc_params,
            authenticated=True,
        )
        attempts.append(_attempt_record(
            "coinmarketcap", "AUTHENTICATED", result, cmc_params,
            None if result.ok else "CMC_HISTORICAL_REQUEST_FAILED"
        ))

        if result.ok and isinstance(result.payload, dict):
            data = result.payload.get("data") or {}
            quotes = data.get("quotes") if isinstance(data, dict) else None
            if isinstance(quotes, list):
                points = []
                for item in quotes:
                    if not isinstance(item, dict):
                        continue
                    # CMC may expose search_interval as the normalized interval
                    # anchor. Prefer it when valid; otherwise use timestamp.
                    ts = parse_timestamp(item.get("search_interval")) or parse_timestamp(item.get("timestamp"))
                    quote = item.get("quote") or {}
                    usd = quote.get("USD") if isinstance(quote, dict) else {}
                    value = safe_float(
                        usd.get("total_market_cap") if isinstance(usd, dict) else None
                    )
                    if ts and value is not None and value >= 0:
                        points.append({
                            "timestamp": ts.isoformat(),
                            "timestamp_ms": int(ts.timestamp() * 1000),
                            "value": value,
                            "metric": "CMC_GLOBAL_TOTAL_MARKET_CAP",
                            "btc_dominance": safe_float(item.get("btc_dominance")),
                            "eth_dominance": safe_float(item.get("eth_dominance")),
                            "source_quote_timestamp": (
                                usd.get("timestamp") if isinstance(usd, dict) else None
                            ),
                        })
                validated = _validate_points(
                    points, "coinmarketcap", "AUTHENTICATED",
                    "CMC_GLOBAL_METRICS_HISTORICAL_5M"
                )
                validated["attempts"] = attempts
                validated["retrieved_at"] = result.retrieved_at
                if validated["status"] == "VALIDATED":
                    return validated
                attempts[-1]["reason"] = validated["reason"]
                attempts[-1]["point_count"] = validated["point_count"]
                attempts[-1]["coverage_ratio"] = validated["coverage_ratio"]
            else:
                attempts[-1]["reason"] = "CMC_RESPONSE_MISSING_DATA_QUOTES"
        elif result.ok:
            attempts[-1]["reason"] = "CMC_RESPONSE_NOT_OBJECT"

    # ------------------------------------------------------------------
    # Provider 2: CoinGecko global chart. This is deliberately a secondary
    # evidence attempt. Public/paid CoinGecko global history is not assumed
    # to be 5m; it is accepted only if its actual timestamps pass validation.
    # ------------------------------------------------------------------
    cg_params = {"vs_currency": "usd", "days": 1}
    cg_headers = {"x-cg-pro-api-key": CG_API_KEY} if CG_API_KEY else None
    cg_mode = "PAID_PUBLIC_API" if CG_API_KEY else "PUBLIC"
    cg_result = http_json(
        COINGECKO_BASE + "/global/market_cap_chart",
        cg_params,
        headers=cg_headers,
    )
    attempts.append(_attempt_record(
        "coingecko", cg_mode, cg_result, cg_params,
        None if cg_result.ok else "COINGECKO_GLOBAL_CHART_REQUEST_FAILED"
    ))

    if cg_result.ok and isinstance(cg_result.payload, dict):
        chart = cg_result.payload.get("market_cap_chart")
        raw_points = chart.get("market_cap") if isinstance(chart, dict) else None
        if isinstance(raw_points, list):
            points = []
            for item in raw_points:
                if not isinstance(item, list) or len(item) < 2:
                    continue
                ts_ms = safe_float(item[0])
                value = safe_float(item[1])
                if ts_ms is None or value is None or value < 0:
                    continue
                ts = datetime.fromtimestamp(ts_ms / 1000.0, tz=timezone.utc)
                points.append({
                    "timestamp": ts.isoformat(),
                    "timestamp_ms": int(ts_ms),
                    "value": value,
                    "metric": "COINGECKO_GLOBAL_MARKET_CAP",
                })
            attempted = _validate_points(
                points, "coingecko", cg_mode,
                "COINGECKO_GLOBAL_CHART_5M_CANDIDATE"
            )
            attempted["retrieved_at"] = cg_result.retrieved_at
            attempts[-1]["point_count"] = attempted["point_count"]
            attempts[-1]["coverage_ratio"] = attempted["coverage_ratio"]
            attempts[-1]["reason"] = attempted["reason"]
            if attempted["status"] == "VALIDATED":
                attempted["attempts"] = attempts
                return attempted
        else:
            attempts[-1]["reason"] = "COINGECKO_RESPONSE_MISSING_MARKET_CAP_SERIES"
    elif cg_result.ok:
        attempts[-1]["reason"] = "COINGECKO_RESPONSE_NOT_OBJECT"

    # No valid 5m global source. Preserve every diagnostic while remaining
    # honest: this is BLOCKED evidence, not a degraded PASS.
    reason = (
        "CMC_5M_HISTORICAL_REQUIRES_AUTHENTICATED_PLAN_OR_"
        "NO_VALID_5M_GLOBAL_SERIES_FROM_PUBLIC_PROVIDER"
    )
    if not CMC_API_KEY:
        reason = "CMC_API_KEY_NOT_CONFIGURED; configure CMC_API_KEY (environment variable) and use a CMC plan with intraday historical access; " + reason

    return {
        "status": "NOT_AVAILABLE",
        "provider": "none",
        "provider_mode": "NO_VALID_5M_GLOBAL_HISTORY_SOURCE",
        "timeframe": TIMEFRAME,
        "requested_range": f"{TIME_SERIES_LOOKBACK_DAYS}d",
        "available_range": None,
        "points": [],
        "point_count": 0,
        "coverage_ratio": 0.0,
        "cadence_validation": {
            "target_seconds": 300,
            "sample_count": 0,
            "within_240_360_seconds_ratio": 0.0,
            "max_gap_seconds": None,
            "validated": False,
            "continuity_validated": False,
        },
        "metric_definition": "GLOBAL_MARKET_CAP_5M_HISTORICAL_INDEX_EVIDENCE",
        "is_kitchen_total": False,
        "silent_resampling": False,
        "fabricated_history": False,
        "attempts": attempts,
        "reason": reason,
        "source_timestamp": None,
        "retrieved_at": utc_now(),
        "required_to_lock": {
            "provider": "coinmarketcap",
            "endpoint": "/v1/global-metrics/quotes/historical",
            "interval": "5m",
            "credential": "CMC_API_KEY",
            "plan_requirement": "CMC plan with intraday historical access",
            "no_fallback_can_fabricate_5m_global_history": True,
        },
    }

def load_kitchen_index_history() -> Dict[str, Any]:
    """
    Load only explicitly recorded, validated U06.5 historical observations.

    This is a journal reader, not a provider. It never downloads, interpolates,
    resamples, or fabricates missing observations. Each history file contains a
    validated Top-125 snapshot and its immutable snapshot hash; indices are
    recalculated from that stored snapshot using the canonical formulas.
    """
    points: List[Dict[str, Any]] = []
    files_seen = 0
    invalid_files: List[Dict[str, Any]] = []
    if not HISTORY_DIR.exists():
        return {
            "status": "NOT_AVAILABLE",
            "provider": "KITCHEN_RECORDED",
            "timeframe": TIMEFRAME,
            "points": [],
            "point_count": 0,
            "files_seen": 0,
            "invalid_files": [],
            "fabricated_history": False,
            "silent_resampling": False,
            "reason": "HISTORY_DIRECTORY_NOT_FOUND",
        }

    for path in sorted(HISTORY_DIR.glob("*__history.json")):
        files_seen += 1
        try:
            record = json.loads(path.read_text(encoding="utf-8"))
            top125 = record.get("top125")
            timestamp = parse_timestamp(record.get("timestamp"))
            if not isinstance(top125, list) or len(top125) != TOP_N or not timestamp:
                invalid_files.append({"path": str(path), "reason": "INVALID_HISTORY_RECORD"})
                continue

            by_symbol = {
                str(x.get("symbol") or "").upper(): x
                for x in top125 if isinstance(x, dict)
            }
            if not all(s in by_symbol for s in ("BTC", "ETH", "USDT")):
                invalid_files.append({"path": str(path), "reason": "CORE_ASSET_MISSING"})
                continue

            market_caps = [
                safe_float(x.get("market_cap"))
                for x in top125 if isinstance(x, dict)
            ]
            if len(market_caps) != TOP_N or any(v is None or v < 0 for v in market_caps):
                invalid_files.append({"path": str(path), "reason": "MARKET_CAP_INVALID"})
                continue

            total = sum(float(v) for v in market_caps)
            btc = float(by_symbol["BTC"]["market_cap"])
            eth = float(by_symbol["ETH"]["market_cap"])
            usdt = float(by_symbol["USDT"]["market_cap"])
            if total <= 0:
                invalid_files.append({"path": str(path), "reason": "TOTAL_INVALID"})
                continue

            points.append({
                "timestamp": timestamp.isoformat(),
                "timestamp_ms": int(timestamp.timestamp() * 1000),
                "KITCHEN_TOTAL_TOP125": total,
                "KITCHEN_TOTAL2": total - btc,
                "KITCHEN_TOTAL3": total - btc - eth,
                "KITCHEN_USDT_D": usdt / total * 100.0,
                "KITCHEN_BTC_D": btc / total * 100.0,
                "snapshot_id": record.get("snapshot_id"),
                "snapshot_hash": record.get("snapshot_hash"),
                "history_source": "KITCHEN_RECORDED",
                "source_file": str(path),
            })
        except Exception as exc:
            invalid_files.append({
                "path": str(path),
                "reason": f"{type(exc).__name__}: {exc}",
            })

    # Deterministic timestamp de-duplication. No interpolation or resampling.
    dedup = {p["timestamp_ms"]: p for p in points}
    points = [dedup[k] for k in sorted(dedup)]

    cadence_seconds = [
        (b["timestamp_ms"] - a["timestamp_ms"]) / 1000.0
        for a, b in zip(points, points[1:])
        if b["timestamp_ms"] > a["timestamp_ms"]
    ]
    cadence_ratio = (
        sum(1 for x in cadence_seconds if 240 <= x <= 360) / len(cadence_seconds)
        if cadence_seconds else 0.0
    )
    max_gap_seconds = max(cadence_seconds) if cadence_seconds else None

    status = (
        "VALIDATED"
        if len(points) >= 3
        and cadence_ratio >= 0.80
        and (max_gap_seconds is None or max_gap_seconds <= 15 * 60)
        else "PARTIAL" if points else "NOT_AVAILABLE"
    )

    return {
        "status": status,
        "provider": "KITCHEN_RECORDED",
        "provider_mode": "LOCAL_VALIDATED_HISTORY_JOURNAL",
        "timeframe": TIMEFRAME,
        "points": points,
        "point_count": len(points),
        "files_seen": files_seen,
        "invalid_files": invalid_files,
        "available_range": {
            "start": points[0]["timestamp"] if points else None,
            "end": points[-1]["timestamp"] if points else None,
        },
        "coverage_ratio": cadence_ratio,
        "cadence_validation": {
            "target_seconds": 300,
            "sample_count": len(cadence_seconds),
            "within_240_360_seconds_ratio": cadence_ratio,
            "max_gap_seconds": max_gap_seconds,
            "validated": status == "VALIDATED",
            "continuity_validated": (
                max_gap_seconds is None or max_gap_seconds <= 15 * 60
            ),
        },
        "metric_definition": (
            "KITCHEN_TOTAL_TOP125_AND_KITCHEN_USDT_D_FROM_RECORDED_TOP125"
        ),
        "is_kitchen_total": True,
        "fabricated_history": False,
        "silent_resampling": False,
        "reason": (
            "VALIDATED_KITCHEN_RECORDED_HISTORY"
            if status == "VALIDATED"
            else "INSUFFICIENT_OR_NONCONTIGUOUS_RECORDED_HISTORY"
        ),
    }


def build_time_series_contract(
    market_index_series: Optional[Dict[str, Any]] = None,
    kitchen_index_history: Optional[Dict[str, Any]] = None,
) -> Dict[str, Any]:
    market_index_series = market_index_series or {}
    kitchen_index_history = kitchen_index_history or load_kitchen_index_history()
    status = market_index_series.get("status", "NOT_AVAILABLE")
    return {
        "status": status,
        "dataset_mode": DATASET_MODE,
        "timeframe": TIMEFRAME,
        "supported_timeframes": list(SUPPORTED_TIMEFRAMES),
        "requested_range": market_index_series.get("requested_range"),
        "available_range": market_index_series.get("available_range"),
        "history_source": market_index_series.get("provider"),
        "market_index": market_index_series,
        "kitchen_index_history": kitchen_index_history,
        "reason": (
            "Validated Kitchen Top-125 history is available from explicitly recorded U06.5 snapshots."
            if kitchen_index_history.get("status") == "VALIDATED"
            else "No validated Kitchen Top-125 historical series is currently available; no history was fabricated."
        ),
        "fabricated_history_allowed": False,
        "silent_resampling_allowed": False,
    }


def append_kitchen_history(
    foundation: Dict[str, Any],
) -> Dict[str, Any]:
    """
    Explicit operation. Stores a validated frozen snapshot as a Kitchen-
    recorded historical observation. It is NOT called automatically.
    """
    if foundation.get("status") != "VALIDATED":
        return {
            "status": "DATA_UNAVAILABLE",
            "reason": "ONLY_VALIDATED_FOUNDATIONS_CAN_ENTER_HISTORY",
        }

    record = {
        "history_source": "KITCHEN_RECORDED",
        "snapshot_id": foundation.get("snapshot_id"),
        "snapshot_hash": foundation.get("snapshot_hash"),
        "timestamp": foundation.get("snapshot", {}).get("timestamp"),
        "timeframe": foundation.get("timeframe"),
        "top125": [
            {
                "canonical_asset_id": x.get("canonical_asset_id"),
                "symbol": x.get("symbol"),
                "rank": x.get("calculated_rank"),
                "market_cap": x.get("market_cap"),
                "price": x.get("price"),
            }
            for x in foundation.get("universe", {}).get("top125", [])
        ],
    }

    data = canonical_bytes(record)
    filename = (
        f"{safe_name(foundation['snapshot_id'])}__history.json"
    )
    path = HISTORY_DIR / filename
    path.write_bytes(data)

    return {
        "status": "VALIDATED",
        "history_source": "KITCHEN_RECORDED",
        "path": str(path),
        "sha256": sha256_bytes(data),
    }


# =============================================================================
# 17. MEMBERSHIP TRACKING
# =============================================================================

def compare_top125_membership(
    previous_top125: Sequence[Dict[str, Any]],
    current_top125: Sequence[Dict[str, Any]],
) -> Dict[str, Any]:
    previous = {
        x.get("canonical_asset_id"): x
        for x in previous_top125
        if x.get("canonical_asset_id")
    }
    current = {
        x.get("canonical_asset_id"): x
        for x in current_top125
        if x.get("canonical_asset_id")
    }

    changes = []

    for asset_id in sorted(set(previous) | set(current)):
        old = previous.get(asset_id)
        new = current.get(asset_id)

        if old is None and new is not None:
            changes.append({
                "canonical_asset_id": asset_id,
                "symbol": new.get("symbol"),
                "event": "ENTERED_TOP125",
                "previous_rank": None,
                "current_rank": new.get("calculated_rank"),
            })
        elif old is not None and new is None:
            changes.append({
                "canonical_asset_id": asset_id,
                "symbol": old.get("symbol"),
                "event": "EXITED_TOP125",
                "previous_rank": old.get("calculated_rank"),
                "current_rank": None,
            })
        else:
            old_rank = safe_int(old.get("calculated_rank"))
            new_rank = safe_int(new.get("calculated_rank"))

            if old_rank is None or new_rank is None:
                event = "DATA_LOST"
            elif new_rank < old_rank:
                event = "RANK_UP"
            elif new_rank > old_rank:
                event = "RANK_DOWN"
            else:
                event = "RANK_UNCHANGED"

            changes.append({
                "canonical_asset_id": asset_id,
                "symbol": new.get("symbol"),
                "event": event,
                "previous_rank": old_rank,
                "current_rank": new_rank,
            })

    return {
        "status": "CALCULATED",
        "previous_count": len(previous),
        "current_count": len(current),
        "changes": changes,
    }


# =============================================================================
# 18. SNAPSHOT HASH / IMMUTABILITY
# =============================================================================

def canonical_hash_input(foundation: Dict[str, Any]) -> Dict[str, Any]:
    """
    Returns the exact canonical object that is hashed.
    Derived consumer adapters are deliberately excluded from this object.
    The integrity hash and snapshot_hash fields are replaced with PENDING,
    preventing recursive hashing.
    """
    x = deep_copy(foundation)

    # Consumer adapters can contain the final snapshot hash and are therefore
    # derived views, not part of the authoritative hash input.
    x.pop("adapters", None)

    # These fields are derived after the authoritative hash is calculated.
    # They must never participate in their own hash contract.
    x["snapshot_hash"] = "PENDING"
    snapshot = x.get("snapshot")
    if isinstance(snapshot, dict):
        # hash_verified is a post-hash verification result and therefore
        # cannot participate in the hash it verifies.
        snapshot.pop("hash_verified", None)

    integrity = x.get("integrity")
    if not isinstance(integrity, dict):
        integrity = {}
    integrity["sha256"] = "PENDING"
    x["integrity"] = integrity

    return x


def calculate_foundation_hash(foundation: Dict[str, Any]) -> str:
    return sha256_obj(canonical_hash_input(foundation))


def verify_foundation_hash(foundation: Dict[str, Any]) -> bool:
    stored = foundation.get("snapshot_hash")
    if not stored:
        return False
    return stored == calculate_foundation_hash(foundation)


def freeze_foundation(
    foundation: Dict[str, Any],
) -> Dict[str, Any]:
    """
    Final freeze:
      1. complete canonical foundation is assembled
      2. derived adapters are NOT included in hash input
      3. one authoritative SHA-256 is calculated
      4. hash is stored
      5. object is treated as immutable
    """
    result = deep_copy(foundation)

    result["snapshot_hash"] = "PENDING"
    result["integrity"] = {
        "algorithm": "SHA-256",
        "canonical": True,
        "hash_scope": (
            "COMPLETE_CANONICAL_FOUNDATION_EXCLUDING_DERIVED_ADAPTERS "
            "AND_RECURSIVE_HASH_FIELDS"
        ),
        "sha256": "PENDING",
    }

    # Freeze-state fields are authoritative snapshot state, so they MUST be
    # fixed before the canonical hash is calculated.  Only post-hash
    # verification metadata (hash_verified) is excluded from the hash.
    result["snapshot"]["immutable_after_creation"] = True
    result["snapshot"]["frozen"] = True
    result["snapshot"].pop("hash_verified", None)

    final_hash = calculate_foundation_hash(result)

    result["snapshot_hash"] = final_hash
    result["integrity"]["sha256"] = final_hash
    result["snapshot"]["hash_verified"] = verify_foundation_hash(result)

    return result


# =============================================================================
# 19. CONSUMER ADAPTERS
# =============================================================================

def build_u07_adapter(foundation: Dict[str, Any]) -> Dict[str, Any]:
    values = foundation.get("indices", {}).get("values", {})
    kitchen_history = foundation.get("time_series", {}).get(
        "kitchen_index_history", {}
    ) or {}
    points = (
        kitchen_history.get("points", [])
        if isinstance(kitchen_history, dict) else []
    )
    return {
        "SOURCE_CELL": CELL_ID,
        "SCHEMA_VERSION": SCHEMA_VERSION,
        "RUN_ID": foundation.get("run_id"),
        "SNAPSHOT_ID": foundation.get("snapshot_id"),
        "SNAPSHOT_HASH": foundation.get("snapshot_hash"),
        "DATA_STATUS": foundation.get("status"),
        "TIMEFRAME": foundation.get("timeframe"),
        "DATASET_MODE": foundation.get("dataset_mode"),
        "ANALYSIS_RANGE": foundation.get("range", {}).get("requested"),
        "AVAILABLE_RANGE": kitchen_history.get("available_range"),
        "HISTORY_STATUS": kitchen_history.get("status", "NOT_AVAILABLE"),
        "HISTORY_SOURCE": kitchen_history.get("provider", "KITCHEN_RECORDED"),
        "HISTORY_COVERAGE": kitchen_history.get("coverage_ratio"),
        "HISTORY_POINTS": points,
        "TOTAL_SERIES": [p.get("KITCHEN_TOTAL_TOP125") for p in points],
        "TOTAL2_SERIES": [p.get("KITCHEN_TOTAL2") for p in points],
        "TOTAL3_SERIES": [p.get("KITCHEN_TOTAL3") for p in points],
        "BTC_D_SERIES": [p.get("KITCHEN_BTC_D") for p in points],
        "USDT_D_SERIES": [p.get("KITCHEN_USDT_D") for p in points],
        "CURRENT_VALUES": values,
        "PROVIDER_METADATA": foundation.get("providers"),
        "DEFINITIONS": foundation.get("definitions"),
        "PROVENANCE": foundation.get("provenance"),
        "FRESHNESS": foundation.get("freshness"),
        "COVERAGE": foundation.get("coverage"),
        "NO_DATA_FABRICATION": True,
    }


def build_u08_adapter(foundation: Dict[str, Any]) -> Dict[str, Any]:
    assets = foundation.get("assets", {})
    indices = foundation.get("indices", {}).get("values", {})

    return {
        "SOURCE_CELL": CELL_ID,
        "SCHEMA_VERSION": SCHEMA_VERSION,
        "RUN_ID": foundation.get("run_id"),
        "SNAPSHOT_ID": foundation.get("snapshot_id"),
        "SNAPSHOT_HASH": foundation.get("snapshot_hash"),
        "DATA_STATUS": foundation.get("status"),
        "TIMEFRAME": foundation.get("timeframe"),
        "BTC_SERIES": assets.get("BTC"),
        "BTC_MARKET_CAP": assets.get("BTC", {}).get("market_cap"),
        "BTC_D": indices.get("KITCHEN_BTC_D"),
        "BTC_CONTEXT": {
            "TOTAL": indices.get("KITCHEN_TOTAL_TOP125"),
            "TOTAL2": indices.get("KITCHEN_TOTAL2"),
            "TOTAL3": indices.get("KITCHEN_TOTAL3"),
            "OTHERS": indices.get("KITCHEN_OTHERS"),
        },
        "ETH_CONTEXT": assets.get("ETH"),
        "USDT_CONTEXT": assets.get("USDT"),
        "TOTAL_CONTEXT": indices,
        "PROVIDER_METADATA": foundation.get("providers"),
        "DEFINITIONS": foundation.get("definitions"),
        "PROVENANCE": foundation.get("provenance"),
    }


def build_u09_adapter(foundation: Dict[str, Any]) -> Dict[str, Any]:
    universe = foundation.get("universe", {})
    return {
        "SOURCE_CELL": CELL_ID,
        "SCHEMA_VERSION": SCHEMA_VERSION,
        "RUN_ID": foundation.get("run_id"),
        "SNAPSHOT_ID": foundation.get("snapshot_id"),
        "SNAPSHOT_HASH": foundation.get("snapshot_hash"),
        "DATA_STATUS": foundation.get("status"),
        "CANDIDATE_UNIVERSE": universe.get("candidate_universe", []),
        "VALID_UNIVERSE": universe.get("valid_universe", []),
        "INVALID_UNIVERSE": universe.get("invalid_universe", []),
        "EXCLUDED_UNIVERSE": universe.get("excluded_universe", []),
        "MARKET_UNIVERSE": universe.get("top125", []),
        "TOP125": universe.get("top125", []),
        "MEMBERSHIP_CHANGES": universe.get("membership_changes", {}),
        "EXCHANGE_EVIDENCE": foundation.get("exchange_evidence"),
        "FRESHNESS": foundation.get("freshness"),
        "COVERAGE": foundation.get("coverage"),
        "PROVENANCE": foundation.get("provenance"),
    }


# =============================================================================
# 20. STRUCTURAL TESTS
# =============================================================================

def structural_self_tests() -> Dict[str, Any]:
    tests: Dict[str, bool] = {
        "snapshot_lock_semantics_declared": DATASET_MODE in {"SNAPSHOT", "HISTORICAL"},
        "cell_id": CELL_ID == "U06.5",
        "version_v8": VERSION.startswith("8."),
        "version_constant_matches_header": VERSION == "8.4.5",
        "schema_version": SCHEMA_VERSION == "U06_5_SCHEMA_V6_0",
        "cmc_primary": PROVIDER_PRIORITY["global"][0] == "coinmarketcap",
        "coingecko_fallback": PROVIDER_PRIORITY["global"][1] == "coingecko",
        "tradingview_not_runtime": not any(
            provider in PROVIDER_PRIORITY["global"]
            for provider in FORBIDDEN_RUNTIME_PROVIDERS
        ),
        "top125_constant": TOP_N == 125,
        "timeframe_supported": TIMEFRAME in SUPPORTED_TIMEFRAMES,
        "snapshot_or_historical_mode": DATASET_MODE in {"SNAPSHOT", "HISTORICAL"},
        "simple_price_optional": True,
        "keyless_header_absent": "X-CMC_PRO_API_KEY" not in cmc_headers(False),
        "auth_mode_requires_key": (
            (not CMC_API_KEY) or ("X-CMC_PRO_API_KEY" in cmc_headers(True))
        ),
        "finite_retry": MAX_RETRIES_PER_PROVIDER == 1,
        "raw_function_present": callable(persist_raw_artifact),
        "identity_function_present": callable(validate_identity),
        "outlier_engine_present": callable(outlier_filter_v1),
        "aggregator_present": callable(price_aggregate_v1),
        "market_cap_engine_present": callable(calculate_market_cap_v1),
        "ranking_present": callable(dynamic_rank_assets),
        "index_engine_present": callable(calculate_indices_from_top125),
        "membership_tracking_present": callable(compare_top125_membership),
        "history_honesty_present": callable(append_kitchen_history),
        "hash_present": callable(calculate_foundation_hash),
        "hash_verify_present": callable(verify_foundation_hash),
        "u07_adapter_present": callable(build_u07_adapter),
        "u08_adapter_present": callable(build_u08_adapter),
        "u09_adapter_present": callable(build_u09_adapter),
        "no_execution_capability": True,
        "no_reference_runtime": REFERENCE_STATUS == "REFERENCE_NOT_USED",
        "multi_exchange_target_is_eight": TARGET_EXCHANGE_COUNT == 8,
        "multi_exchange_minimum_is_seven": MIN_VALIDATED_EXCHANGE_COUNT == 7,
        "multi_exchange_engine_present": callable(acquire_multi_exchange_evidence),
        "orderbook_engine_present": callable(acquire_multi_exchange_orderbook),
        "reliability_engine_present": callable(_build_reliability_scores),
        "depth_metric_engine_present": callable(orderbook_quality_metrics),
        "pythonanywhere_portable_paths": not any(
            marker in str(p).lower()
            for p in (ROOT, DATA_DIR, RAW_DIR, SNAPSHOT_DIR, AUDIT_DIR, HISTORY_DIR)
            for marker in ("/" + "content", "google" + ".colab")
        ),
        "portable_credential_discovery": (
            CMC_API_KEY_SOURCE == "NONE"
            or CMC_API_KEY_SOURCE.startswith("ENV:")
        ),
        "kitchen_history_reader_present": callable(load_kitchen_index_history),
        "u07_series_are_actual_history_arrays": isinstance(
            build_u07_adapter({
                "indices": {"values": {}},
                "time_series": {
                    "kitchen_index_history": {"points": []}
                },
            }).get("TOTAL_SERIES"),
            list,
        ),
    }

    return {
        "status": "PASS" if all(tests.values()) else "FAIL",
        "passed": sum(tests.values()),
        "total": len(tests),
        "tests": tests,
    }


# =============================================================================
# 21. MARKET ENGINE UNIT TESTS
# =============================================================================

def market_engine_self_tests() -> Dict[str, Any]:
    now = utc_now()

    base = [
        make_market_observation(
            exchange="EX1",
            exchange_id="EX1",
            asset="TEST",
            base_asset="TEST",
            quote_asset="USDT",
            market_id="TESTUSDT",
            market_type="SPOT",
            price=100.0,
            volume_usd_24h=1_000_000,
            timestamp=now,
            retrieved_at=now,
            provider="TEST",
        ),
        make_market_observation(
            exchange="EX2",
            exchange_id="EX2",
            asset="TEST",
            base_asset="TEST",
            quote_asset="USD",
            market_id="TESTUSD",
            market_type="SPOT",
            price=101.0,
            volume_usd_24h=2_000_000,
            timestamp=now,
            retrieved_at=now,
            provider="TEST",
        ),
        make_market_observation(
            exchange="EX3",
            exchange_id="EX3",
            asset="TEST",
            base_asset="TEST",
            quote_asset="USD",
            market_id="TESTUSD3",
            market_type="SPOT",
            price=5000.0,
            volume_usd_24h=1_000_000,
            timestamp=now,
            retrieved_at=now,
            provider="TEST",
        ),
    ]

    conversion_test = make_market_observation(
        exchange="EX",
        exchange_id="EX",
        asset="ALT",
        base_asset="ALT",
        quote_asset="BTC",
        market_id="ALT BTC",
        market_type="SPOT",
        price=0.01,
        volume_usd_24h=1,
        timestamp=now,
        retrieved_at=now,
        provider="TEST",
    )

    usd_price, usd_meta = normalize_quote_to_usd(
        conversion_test,
        {"BTC": 100_000.0, "ETH": 2_500.0},
    )
    for item in base:
        normalized, meta = normalize_quote_to_usd(
            item,
            {"BTC": 100_000.0, "ETH": 2_500.0},
        )
        item["normalized_usd_price"] = normalized
        item["usd_normalization"] = meta

    agg = price_aggregate_v1(base)

    fake_assets = []
    for i in range(125):
        symbol = "BTC" if i == 0 else "ETH" if i == 1 else "USDT" if i == 2 else f"A{i}"
        fake_assets.append({
            "provider_asset_id": i + 1,
            "canonical_asset_id": f"test:{i+1}",
            "symbol": symbol,
            "name": symbol,
            "provider": "TEST",
            "provider_mode": "TEST",
            "provider_rank": i + 1,
            "price": 100.0,
            "market_cap": float(1_000_000_000 - i * 1_000_000),
            "volume_24h": 1_000_000.0,
            "source_timestamp": now,
            "retrieved_at": now,
            "identity_status": "VALIDATED",
        })

    ranked = dynamic_rank_assets(fake_assets)
    indices = calculate_indices_from_top125(ranked["top125"])

    cap = calculate_market_cap_v1(
        100.0,
        10.0,
        input_timestamp=now,
        source_provenance={"SOURCE": "TEST"},
    )

    tests = {
        "usd_usdt_normalization": (
            base[0]["normalized_usd_price"] == 100.0
        ),
        "usd_usd_normalization": (
            base[1]["normalized_usd_price"] == 101.0
        ),
        "btc_to_usd_conversion": usd_price == 1000.0,
        "outlier_rejected_or_downweighted": (
            len(agg["rejected_markets"]) >= 1
        ),
        "robust_aggregate_calculated": (
            agg["reference_price"] is not None
        ),
        "dynamic_top125": len(ranked["top125"]) == 125,
        "dynamic_ranking": ranked["top125"][0]["calculated_rank"] == 1,
        "kitchen_total_calculated": (
            indices["values"].get("KITCHEN_TOTAL_TOP125") is not None
        ),
        "total2_formula": (
            indices["values"]["KITCHEN_TOTAL2"]
            == indices["values"]["KITCHEN_TOTAL_TOP125"]
            - fake_assets[0]["market_cap"]
        ),
        "market_cap_formula": (
            cap["output"] == 1000.0
        ),
        "tradingview_unused": (
            indices.get("tradingview_value_used") is False
        ),
    }

    return {
        "status": "PASS" if all(tests.values()) else "FAIL",
        "passed": sum(tests.values()),
        "total": len(tests),
        "tests": tests,
    }


# =============================================================================
# 22. HASH / CONTRACT TESTS
# =============================================================================

def hash_contract_self_tests() -> Dict[str, Any]:
    foundation = {
        "source_cell": CELL_ID,
        "schema_version": SCHEMA_VERSION,
        "foundation_version": VERSION,
        "run_id": "TEST_RUN",
        "snapshot_id": "TEST_SNAPSHOT",
        "snapshot_hash": "PENDING",
        "status": "VALIDATED",
        "snapshot": {"immutable_after_creation": True},
        "integrity": {
            "algorithm": "SHA-256",
            "canonical": True,
            "sha256": "PENDING",
        },
        "canonical": {"x": 1},
    }

    frozen = freeze_foundation(foundation)
    original_hash = frozen["snapshot_hash"]

    tests = {
        "hash_present": bool(original_hash),
        "hash_verifies": verify_foundation_hash(frozen),
        "hash_deterministic": (
            original_hash == calculate_foundation_hash(frozen)
        ),
        "adapters_not_in_hash_scope": (
            "adapters" not in canonical_hash_input(frozen)
        ),
    }

    mutated = deep_copy(frozen)
    mutated["canonical"]["x"] = 2
    tests["mutation_detected"] = not verify_foundation_hash(mutated)

    return {
        "status": "PASS" if all(tests.values()) else "FAIL",
        "passed": sum(tests.values()),
        "total": len(tests),
        "tests": tests,
    }


# =============================================================================
# 23. PROVIDER PIPELINE
# =============================================================================

def build_attempt_record(
    provider: str,
    provider_mode: str,
    dataset: str,
    endpoint: str,
    result: HTTPResult,
    attempt_number: int,
) -> Dict[str, Any]:
    return {
        "provider": provider,
        "provider_mode": provider_mode,
        "dataset": dataset,
        "endpoint": endpoint,
        "attempt_number": attempt_number,
        "http_status": result.status,
        "provider_status": safe_json(result.provider_status),
        "ok": result.ok,
        "error_class": result.error_class,
        "error_message": result.error_message,
        "retrieved_at": result.retrieved_at,
        "elapsed_seconds": result.elapsed_seconds,
    }


def provider_mode_sequence() -> List[Tuple[str, str]]:
    sequence: List[Tuple[str, str]] = []

    if CMC_API_KEY:
        # Keyless remains the documented current primary mode. Authenticated
        # is optional and may be used as a higher-capability retry path.
        sequence.append(("coinmarketcap", "KEYLESS"))
        sequence.append(("coinmarketcap", "AUTHENTICATED"))
    else:
        sequence.append(("coinmarketcap", "KEYLESS"))

    sequence.append(("coingecko", "PUBLIC"))
    return sequence


def acquire_top125_with_orchestration() -> Dict[str, Any]:
    attempts: List[Dict[str, Any]] = []
    raw_results: List[Dict[str, Any]] = []
    provider_candidates: List[Dict[str, Any]] = []

    sequence = provider_mode_sequence()

    for provider, mode in sequence:
        for retry in range(MAX_RETRIES_PER_PROVIDER + 1):
            if provider == "coinmarketcap":
                authenticated = mode == "AUTHENTICATED"
                result = acquire_cmc_top125(authenticated=authenticated)
                endpoint = "/v3/cryptocurrency/listings/latest"
            else:
                result = acquire_coingecko_top125()
                endpoint = "/coins/markets"

            attempts.append(
                build_attempt_record(
                    provider=provider,
                    provider_mode=mode,
                    dataset="TOP125",
                    endpoint=endpoint,
                    result=result,
                    attempt_number=retry + 1,
                )
            )

            if result.ok:
                if provider == "coinmarketcap":
                    if not cmc_status_ok(result.payload):
                        result.error_class = "PROVIDER_ERROR"
                        result.error_message = "CMC provider status indicates error."
                    else:
                        raw = cmc_asset_records(cmc_data(result.payload))
                        normalized = [
                            normalize_cmc_asset(
                                asset,
                                result.retrieved_at,
                                mode,
                            )
                            for asset in raw
                        ]
                        validation = validate_ranked_universe(normalized)

                        raw_results.append({
                            "provider": provider,
                            "provider_mode": mode,
                            "endpoint": endpoint,
                            "result": result,
                        })

                        if validation["status"] == "VALIDATED":
                            return {
                                "selected_provider": provider,
                                "selected_mode": mode,
                                "records": normalized,
                                "validation": validation,
                                "attempts": attempts,
                                "raw_results": raw_results,
                                "fallback_used": (
                                    provider != "coinmarketcap"
                                    or mode != "KEYLESS"
                                ),
                                "fallback_reason": None,
                            }

                else:
                    raw = result.payload if isinstance(result.payload, list) else []
                    normalized = [
                        normalize_coingecko_asset(
                            asset,
                            result.retrieved_at,
                        )
                        for asset in raw
                        if isinstance(asset, dict)
                    ]
                    validation = validate_ranked_universe(normalized)

                    raw_results.append({
                        "provider": provider,
                        "provider_mode": mode,
                        "endpoint": endpoint,
                        "result": result,
                    })

                    if validation["status"] == "VALIDATED":
                        return {
                            "selected_provider": provider,
                            "selected_mode": mode,
                            "records": normalized,
                            "validation": validation,
                            "attempts": attempts,
                            "raw_results": raw_results,
                            "fallback_used": True,
                            "fallback_reason": (
                                attempts[-2]["error_class"]
                                if len(attempts) >= 2
                                else "CMC_PRIMARY_NOT_VALIDATED"
                            ),
                        }

                # Valid HTTP response but invalid dataset: no infinite retry.
                break

            # Retry only for retryable provider failures.
            if result.error_class not in {
                "TIMEOUT",
                "RATE_LIMIT",
                "UNAVAILABLE",
                "TECHNICAL_ERROR",
                "PROVIDER_ERROR",
            }:
                break

    return {
        "selected_provider": None,
        "selected_mode": None,
        "records": [],
        "validation": validate_ranked_universe([]),
        "attempts": attempts,
        "raw_results": raw_results,
        "fallback_used": True,
        "fallback_reason": "NO_PROVIDER_VALIDATED_TOP125",
    }


# =============================================================================
# 24. FOUNDATION ASSEMBLY
# =============================================================================

def build_range_contract(
    top125_records: Sequence[Dict[str, Any]],
) -> Dict[str, Any]:
    timestamps = [
        parse_timestamp(x.get("source_timestamp"))
        for x in top125_records
        if x.get("source_timestamp")
    ]
    timestamps = [x for x in timestamps if x]

    if timestamps:
        actual_start = min(timestamps).isoformat()
        actual_end = max(timestamps).isoformat()
    else:
        actual_start = None
        actual_end = None

    return {
        "requested": None,
        "available": {
            "start": actual_start,
            "end": actual_end,
        },
        "status": (
            "SNAPSHOT_RANGE_ONLY"
            if DATASET_MODE == "SNAPSHOT"
            else "DATA_UNAVAILABLE"
        ),
        "silent_replacement": False,
    }


def build_reference_validation(cross_source: Dict[str, Any]) -> Dict[str, Any]:
    """Validate CoinGecko as an independent cross-source reference.

    TradingView remains disabled. This reference is deliberately limited to
    source-to-source agreement and never changes the selected foundation.
    """
    usable = []
    agreements = []
    for symbol, item in (cross_source or {}).items():
        if symbol not in {"BTC", "ETH", "USDT"}:
            continue
        if item.get("CMC_PRICE") is not None and item.get("COINGECKO_PRICE") is not None:
            usable.append(symbol)
            agreements.append(item.get("agreement_status") == "AGREE")

    overlap = len(usable) / 3.0
    agreement_ratio = sum(agreements) / len(agreements) if agreements else 0.0
    validated = overlap >= REFERENCE_MIN_OVERLAP and agreement_ratio >= 0.80

    return {
        "provider": REFERENCE_PROVIDER,
        "provider_mode": "PUBLIC_CROSS_SOURCE",
        "status": "VALIDATED" if validated else "PARTIAL" if usable else "NOT_AVAILABLE",
        "reference_type": "CROSS_SOURCE_PRICE_REFERENCE",
        "tradingview": "DISABLED",
        "overlap_symbols": usable,
        "overlap_ratio_core_assets": overlap,
        "agreement_ratio": agreement_ratio,
        "max_allowed_relative_price_difference_pct": REFERENCE_MAX_PRICE_DEVIATION_PCT,
        "minimum_overlap": REFERENCE_MIN_OVERLAP,
        "minimum_agreement_ratio": 0.80,
        "source_evidence": cross_source,
        "does_not_modify_foundation": True,
        "validation_rule": "CMC vs CoinGecko price agreement <= 1.0% for at least 80% of core assets with >=80% core overlap",
    }


def build_coverage_contract(
    validation: Dict[str, Any],
    candidate_universe: Dict[str, Any],
) -> Dict[str, Any]:
    return {
        "dataset_mode": DATASET_MODE,
        "requested_count": TOP_N,
        "received_count": validation.get("received", 0),
        "valid_count": validation.get("valid", 0),
        "invalid_count": validation.get("invalid", 0),
        "missing_count": validation.get("missing", 0),
        "duplicate_count": validation.get("duplicate_count", 0),
        "coverage_ratio": validation.get("coverage_ratio", 0.0),
        "requested_start": None,
        "requested_end": None,
        "actual_start": None,
        "actual_end": None,
        "candidate_requested": candidate_universe.get(
            "candidate_requested"
        ),
        "candidate_received": candidate_universe.get(
            "candidate_received"
        ),
        "candidate_valid": candidate_universe.get(
            "candidate_valid"
        ),
        "candidate_invalid": candidate_universe.get(
            "candidate_invalid"
        ),
        "candidate_excluded": candidate_universe.get(
            "candidate_excluded"
        ),
    }


def build_foundation(
    *,
    run_id: str,
    snapshot_id: str,
    selected_provider: Optional[str],
    selected_mode: Optional[str],
    selected_records: Sequence[Dict[str, Any]],
    validation: Dict[str, Any],
    attempts: Sequence[Dict[str, Any]],
    fallback_used: bool,
    fallback_reason: Optional[str],
    global_metrics: Dict[str, Any],
    cross_source: Dict[str, Any],
    exchange_evidence: Dict[str, Any],
    multi_exchange_evidence: Dict[str, Any],
    live_price_foundation: Dict[str, Any],
    market_index_series: Dict[str, Any],
    reference_validation: Dict[str, Any],
    raw_artifacts: Sequence[Dict[str, Any]],
    errors: Sequence[str],
    warnings: Sequence[str],
) -> Dict[str, Any]:
    core = validate_core_assets(selected_records)
    ranking = dynamic_rank_assets(selected_records)

    top125 = ranking["top125"]
    indices = calculate_indices_from_top125(top125)

    top125_by_symbol = {
        str(x.get("symbol") or "").upper(): x
        for x in top125
    }

    core_assets = {
        symbol: top125_by_symbol.get(symbol, {})
        for symbol in CORE_ASSETS.values()
    }

    candidate_universe = {
        "candidate_requested": ranking["candidate_requested"],
        "candidate_received": ranking["candidate_received"],
        "candidate_valid": ranking["candidate_valid"],
        "candidate_invalid": ranking["candidate_invalid"],
        "candidate_excluded": ranking["candidate_excluded"],
        "coverage": ranking["candidate_coverage_ratio"],
        "candidate_universe": ranking["all_valid_ranked"],
        "valid_universe": ranking["all_valid_ranked"],
        "invalid_universe": validation.get("invalid_records", []),
        "excluded_universe": ranking["all_valid_ranked"][TOP_N:],
        "top125": top125,
        "membership_changes": {
            "status": "NO_PREVIOUS_SNAPSHOT",
            "changes": [],
        },
    }

    coverage = build_coverage_contract(
        validation,
        ranking,
    )

    if selected_provider and validation["status"] == "VALIDATED":
        foundation_status = (
            "FOUNDATION_VALIDATED"
            if core["status"] == "VALIDATED"
            and indices["status"] == "CALCULATED"
            else "FOUNDATION_PARTIAL"
        )
    else:
        foundation_status = (
            "FOUNDATION_UNAVAILABLE"
            if not selected_records
            else "FOUNDATION_PARTIAL"
        )

    source_timestamp_values = [
        x.get("source_timestamp")
        for x in selected_records
        if x.get("source_timestamp")
    ]

    retrieval_values = [
        x.get("retrieved_at")
        for x in selected_records
        if x.get("retrieved_at")
    ]

    freshness_values = []
    for record in selected_records:
        status = freshness_status(
            record.get("source_timestamp"),
            record.get("retrieved_at") or utc_now(),
        )
        freshness_values.append(status)

    freshness = {
        "status": (
            "FRESH"
            if selected_records and all(
                status in {"FRESH", "FRESHNESS_UNVERIFIED"}
                for status in freshness_values
            )
            else "STALE"
            if any(status == "STALE" for status in freshness_values)
            else "UNAVAILABLE"
        ),
        "source_timestamp_min": (
            min(source_timestamp_values)
            if source_timestamp_values else None
        ),
        "source_timestamp_max": (
            max(source_timestamp_values)
            if source_timestamp_values else None
        ),
        "retrieval_timestamp_min": (
            min(retrieval_values) if retrieval_values else None
        ),
        "retrieval_timestamp_max": (
            max(retrieval_values) if retrieval_values else None
        ),
        "threshold_seconds": FRESHNESS_THRESHOLD_SECONDS,
        "per_record_status": freshness_values,
    }

    kitchen_index_history = load_kitchen_index_history()
    time_series = build_time_series_contract(market_index_series, kitchen_index_history)
    range_contract = build_range_contract(selected_records)

    definitions = {
        "KITCHEN_TOTAL_TOP125": (
            "SUM(MARKET_CAP of calculated ranks 1-125)"
        ),
        "KITCHEN_TOTAL2": (
            "KITCHEN_TOTAL_TOP125 - BTC_MARKET_CAP"
        ),
        "KITCHEN_TOTAL3": (
            "KITCHEN_TOTAL_TOP125 - BTC_MARKET_CAP - ETH_MARKET_CAP"
        ),
        "KITCHEN_OTHERS": (
            "SUM(MARKET_CAP of calculated ranks 11-125)"
        ),
        "KITCHEN_BTC_D": (
            "BTC_MARKET_CAP / KITCHEN_TOTAL_TOP125 * 100"
        ),
        "KITCHEN_ETH_D": (
            "ETH_MARKET_CAP / KITCHEN_TOTAL_TOP125 * 100"
        ),
        "KITCHEN_USDT_D": (
            "USDT_MARKET_CAP / KITCHEN_TOTAL_TOP125 * 100"
        ),
        "KITCHEN_OTHERS_D": (
            "KITCHEN_OTHERS / KITCHEN_TOTAL_TOP125 * 100"
        ),
        "CMC_GLOBAL_TOTAL_MC": (
            "CMC provider global total market cap; evidence only."
        ),
        "CMC_GLOBAL_BTC_D": (
            "CMC provider global BTC dominance; evidence only."
        ),
        "CMC_GLOBAL_ETH_D": (
            "CMC provider global ETH dominance; evidence only."
        ),
        "REFERENCE_STATUS": REFERENCE_STATUS,
    }

    provenance = {
        "SOURCE": selected_provider or "NONE",
        "PROVIDER": selected_provider,
        "PROVIDER_MODE": selected_mode,
        "REFERENCE": reference_validation.get("status", REFERENCE_STATUS),
        "PROCESSOR": "KITCHEN_U06_5",
        "STATUS": foundation_status,
        "provider_attempts": list(attempts),
        "fallback": {
            "PRIMARY_PROVIDER": "coinmarketcap",
            "FALLBACK_CHAIN": [
                "coinmarketcap:keyless",
                "coinmarketcap:authenticated:optional",
                "coingecko:public",
            ],
            "FALLBACK_USED": fallback_used,
            "FALLBACK_REASON": fallback_reason,
        },
        "multi_source": {
            "CMC_AND_COINGECKO_RETAINED_SEPARATELY": True,
            "SOURCE_DISAGREEMENT_IS_NOT_AVERAGED_AUTOMATICALLY": True,
            "cross_source": cross_source,
        },
    }

    audit = {
        "project": PROJECT,
        "cell": CELL_ID,
        "version": VERSION,
        "run_id": run_id,
        "timestamp": utc_now(),
        "providers_attempted": list(attempts),
        "selected_provider": selected_provider,
        "selected_provider_mode": selected_mode,
        "fallback_used": fallback_used,
        "fallback_reason": fallback_reason,
        "reference_status": reference_validation.get("status", REFERENCE_STATUS),
        "tradingview_runtime": "DISABLED",
        "trading_orders_strategy_portfolio": "DISABLED",
        "warnings": list(warnings),
        "errors": list(errors),
    }

    foundation = {
        "source_cell": CELL_ID,
        "schema_version": SCHEMA_VERSION,
        "foundation_version": VERSION,
        "project": PROJECT,
        "cell": CELL_ID,
        "cell_name": "MARKET DATA SOURCE & VALIDATION FOUNDATION",
        "run_id": run_id,
        "snapshot_id": snapshot_id,
        "snapshot_hash": "PENDING",
        "status": foundation_status,

        "timeframe": TIMEFRAME,
        "supported_timeframes": list(SUPPORTED_TIMEFRAMES),
        "dataset_mode": DATASET_MODE,

        "global": {
            "provider_global_metrics": global_metrics,
            "provider_total_market_cap": global_metrics.get(
                "total_market_cap"
            ),
            "provider_btc_dominance": global_metrics.get(
                "btc_dominance"
            ),
            "provider_eth_dominance": global_metrics.get(
                "eth_dominance"
            ),
        },

        "assets": core_assets,

        "universe": candidate_universe,

        "multi_exchange_evidence": multi_exchange_evidence,

        "price_foundation": live_price_foundation,

        "exchange_evidence": {
            "configured_target_exchanges": TARGET_EXCHANGE_COUNT,
            "actual_exchange_evidence": exchange_evidence,
            "legacy_u05_engine": "BINANCE_PRIMARY_COINBASE_FALLBACK",
            "market_pair_engine": {
                "status": exchange_evidence.get("market_pair_engine", {}).get("status", "NOT_AVAILABLE"),
                "complete_multi_exchange_coverage": exchange_evidence.get("coverage_ratio", 0.0) == 1.0,
                "legacy_engine_is_not_the_multi_exchange_count": True,
                "selection_used_for_current_top125": False,
            },
        },

        "indices": indices,

        "time_series": time_series,

        "validation": {
            "top125": {
                k: v for k, v in validation.items()
                if k != "records"
            },
            "core_assets": core,
            "identity_version": IDENTITY_VERSION,
        },

        "freshness": freshness,
        "coverage": coverage,
        "range": range_contract,
        "definitions": definitions,
        "provenance": provenance,

        "calculations": {
            "market_cap_version": MARKET_CAP_VERSION,
            "price_aggregator_version": PRICE_AGGREGATOR_VERSION,
            "outlier_engine_version": OUTLIER_ENGINE_VERSION,
            "ranking_version": RANKING_VERSION,
            "index_version": INDEX_VERSION,
            "indices": indices,
            "manifests": indices.get("manifests", {}),
            "raw_is_not_validated": True,
            "validated_is_not_calculated": True,
        },

        "providers": {
            "global": {
                "priority": PROVIDER_PRIORITY["global"],
                "selected": selected_provider,
                "selected_mode": selected_mode,
            },
            "exchange": {
                "priority": PROVIDER_PRIORITY["exchange"],
                "evidence": exchange_evidence,
                "multi_exchange_evidence": multi_exchange_evidence,
            },
            "reference": reference_validation,
        },

        "raw": {
            "preserved": True,
            "artifacts": list(raw_artifacts),
            "credentials_excluded": True,
        },

        "validated": {
            "top125": list(selected_records),
        },

        "normalized": {
            "top125": list(selected_records),
        },

        "calculated": indices,

        "snapshot": {
            "snapshot_id": snapshot_id,
            "timestamp": utc_now(),
            "timeframe": TIMEFRAME,
            "dataset_mode": DATASET_MODE,
            "immutable_after_creation": True,
            "frozen": False,
            "persisted": False,
            "integrity_algorithm": "SHA-256",
            "integrity_verified": False,
        },

        "security": {
            "credentials_in_snapshot": False,
            "credentials_in_raw": False,
            "credentials_in_audit": False,
            "credential_redaction": True,
            "keyless_cmc_does_not_send_api_key": True,
        },

        "consumer_contract": {
            "source_of_truth": "U06_5_DATA_FOUNDATION",
            "immutable": True,
            "required_trace_fields": [
                "SOURCE_CELL",
                "SCHEMA_VERSION",
                "RUN_ID",
                "SNAPSHOT_ID",
                "SNAPSHOT_HASH",
                "DATA_STATUS",
            ],
            "supported_consumers": [
                "U07",
                "U08",
                "U09",
                "U10+",
                "FUTURE_CELLS",
            ],
            "rule": (
                "New market-data requirements extend U06.5. "
                "Consumers do not reacquire owned market data."
            ),
            "architecture_test": (
                "U07->U06.5, U08->U06.5, U09->U06.5, "
                "U10+->U06.5"
            ),
        },

        # Derived adapters are populated AFTER the canonical foundation hash.
        # They are intentionally excluded from the hash input.
        "adapters": {},

        "warnings": list(warnings),
        "errors": list(errors),
        "audit": audit,

        "integrity": {
            "algorithm": "SHA-256",
            "canonical": True,
            "hash_scope": (
                "COMPLETE_CANONICAL_FOUNDATION_EXCLUDING_DERIVED_ADAPTERS "
                "AND_RECURSIVE_HASH_FIELDS"
            ),
            "sha256": "PENDING",
        },
    }

    return foundation


# =============================================================================
# 25. RUNTIME DIAGNOSTICS
# =============================================================================

def _runtime_diagnostic(stage: str, exc: Exception) -> Dict[str, Any]:
    """Credential-safe diagnostic for unexpected runtime failures."""
    return {
        "stage": stage,
        "status": "TECHNICAL_ERROR",
        "error_class": type(exc).__name__,
        "error_message": str(exc),
        "traceback": traceback.format_exc(),
    }


# =============================================================================
# 25. FINAL RUN
# =============================================================================

def run_u06_5() -> Dict[str, Any]:
    run_id = (
        "U06_5_"
        + datetime.now(timezone.utc).strftime("%Y%m%dT%H%M%SZ")
        + "_"
        + uuid.uuid4().hex[:8]
    )
    snapshot_id = "SNAP_" + run_id

    errors: List[str] = []
    warnings: List[str] = []
    raw_artifacts: List[Dict[str, Any]] = []

    orchestration = acquire_top125_with_orchestration()

    selected_provider = orchestration["selected_provider"]
    selected_mode = orchestration["selected_mode"]
    selected_records = orchestration["records"]
    validation = orchestration["validation"]

    # Preserve every actual provider response attempted.
    for raw_item in orchestration.get("raw_results", []):
        result: HTTPResult = raw_item["result"]
        provider = raw_item["provider"]
        mode = raw_item["provider_mode"]
        endpoint = raw_item["endpoint"]

        category = (
            f"{provider}_{mode}_top125"
        )

        raw_artifacts.append(
            persist_raw_artifact(
                run_id=run_id,
                category=category,
                provider=provider,
                provider_mode=mode,
                endpoint=endpoint,
                dataset="TOP125",
                request_params={
                    "start": 1,
                    "limit": TOP_N,
                    "convert": "USD",
                },
                result=result,
            )
        )

    if not selected_provider:
        errors.append(
            "No provider produced a fully validated Top-125."
        )

    # CMC Global Metrics are evidence/context and have their own status.
    global_result = acquire_cmc_global(
        authenticated=False
    )

    global_metrics = (
        normalize_cmc_global_metrics(cmc_data(global_result.payload))
        if global_result.ok and cmc_status_ok(global_result.payload)
        else {
            "status": "DATA_UNAVAILABLE",
            "provider": "coinmarketcap",
            "provider_mode": "KEYLESS",
            "reason": global_result.error_class,
        }
    )

    raw_artifacts.append(
        persist_raw_artifact(
            run_id=run_id,
            category="coinmarketcap_KEYLESS_global_metrics",
            provider="coinmarketcap",
            provider_mode="KEYLESS",
            endpoint="/v1/global-metrics/quotes/latest",
            dataset="GLOBAL_METRICS",
            request_params={"convert": "USD"},
            result=global_result,
        )
    )

    # CMC Core Quotes are separate evidence.
    quotes_result = acquire_cmc_quotes(
        authenticated=False
    )
    if quotes_result.ok and cmc_status_ok(quotes_result.payload):
        raw_artifacts.append(
            persist_raw_artifact(
                run_id=run_id,
                category="coinmarketcap_KEYLESS_core_quotes",
                provider="coinmarketcap",
                provider_mode="KEYLESS",
                endpoint="/v3/cryptocurrency/quotes/latest",
                dataset="CORE_QUOTES",
                request_params={
                    "id": ",".join(str(x) for x in CORE_ASSETS),
                    "convert": "USD",
                },
                result=quotes_result,
            )
        )
    else:
        warnings.append(
            "CMC Core Quotes endpoint unavailable; "
            "Top-125 Core observations remain the primary current evidence."
        )

    # Optional Simple Price is deliberately non-blocking.
    simple_result = acquire_cmc_simple_price(
        authenticated=False
    )
    raw_artifacts.append(
        persist_raw_artifact(
            run_id=run_id,
            category="coinmarketcap_KEYLESS_simple_price_optional",
            provider="coinmarketcap",
            provider_mode="KEYLESS",
            endpoint="/v1/simple/price",
            dataset="SIMPLE_PRICE_OPTIONAL",
            request_params={
                "id": ",".join(str(x) for x in CORE_ASSETS),
                "convert": "USD",
            },
            result=simple_result,
        )
    )

    if not simple_result.ok:
        warnings.append(
            "CMC Simple Price is unavailable; it is optional diagnostic "
            "evidence and does not determine provider availability."
        )

    # CoinGecko evidence for cross-source comparison is collected separately.
    cg_result = acquire_coingecko_top125()
    cg_records = []
    if cg_result.ok:
        cg_records = [
            normalize_coingecko_asset(x, cg_result.retrieved_at)
            for x in (cg_result.payload if isinstance(cg_result.payload, list) else [])
            if isinstance(x, dict)
        ]

        raw_artifacts.append(
            persist_raw_artifact(
                run_id=run_id,
                category="coingecko_public_top125_evidence",
                provider="coingecko",
                provider_mode="PUBLIC",
                endpoint="/coins/markets",
                dataset="TOP125_CROSS_SOURCE_EVIDENCE",
                request_params={
                    "vs_currency": "usd",
                    "order": "market_cap_desc",
                    "per_page": TOP_N,
                    "page": 1,
                    "sparkline": "false",
                },
                result=cg_result,
            )
        )

    cross_source = build_provider_cross_source_evidence(
        selected_records if selected_provider == "coinmarketcap" else [],
        cg_records,
    )

    exchange_evidence = acquire_exchange_evidence()
    multi_exchange_evidence = acquire_multi_exchange_evidence()
    live_price_foundation = build_live_reference_price_from_multi_exchange(multi_exchange_evidence)

    market_index_series = acquire_market_index_time_series()
    reference_validation = build_reference_validation(cross_source)

    foundation = build_foundation(
        run_id=run_id,
        snapshot_id=snapshot_id,
        selected_provider=selected_provider,
        selected_mode=selected_mode,
        selected_records=selected_records,
        validation=validation,
        attempts=orchestration["attempts"],
        fallback_used=orchestration["fallback_used"],
        fallback_reason=orchestration["fallback_reason"],
        global_metrics=global_metrics,
        cross_source=cross_source,
        exchange_evidence=exchange_evidence,
        multi_exchange_evidence=multi_exchange_evidence,
        live_price_foundation=live_price_foundation,
        market_index_series=market_index_series,
        reference_validation=reference_validation,
        raw_artifacts=raw_artifacts,
        errors=errors,
        warnings=warnings,
    )

    # Explicit U06.5 V8 target checklist. These are capability checks, not
    # fabricated success flags; each item reflects the actual foundation state.
    _mx_state = foundation.get("multi_exchange_evidence", {})
    _pf_state = foundation.get("price_foundation", {})
    _ts_state = foundation.get("time_series", {})
    _target_checks = {
        "T1_KUCOIN_CANDLE": next((v.get("status") for p, v in (_mx_state.get("per_exchange") or {}).items() if p == "kucoin"), "NOT_AVAILABLE"),
        "T2_GATE_ORDERBOOK": next((v.get("orderbook_quality", {}).get("status") for p, v in (_mx_state.get("per_exchange") or {}).items() if p == "gate"), "NOT_AVAILABLE"),
        "T3_REFERENCE_BTC_PRICE": "PASS" if _pf_state.get("reference_price") is not None else "NOT_RESOLVED",
        "T4_DYNAMIC_RELIABILITY": "PASS" if _pf_state.get("aggregate", {}).get("aggregate_formula") else "NOT_RESOLVED",
        "T5_ORDERBOOK_LIQUIDITY": "PASS" if int(_mx_state.get("orderbook_validated_exchange_count", 0)) >= MIN_VALIDATED_EXCHANGE_COUNT else "PARTIAL",
        "T6_GLOBAL_EXCHANGE_SEPARATION": "PASS",
        "T7_GLOBAL_5M_TIME_SERIES": _ts_state.get("status", "NOT_AVAILABLE"),
        "T8_INTEGRITY_TESTS": "PENDING",
        "T9_LOCK_CONTRACT": "PENDING",
        "T10_REGRESSION": "PENDING",
    }
    foundation["u06_5_target_checks"] = {"version": "U06_5_TARGETS_V1", "checks": _target_checks}

    # Final canonical hash is calculated BEFORE derived adapters are attached.
    frozen_foundation = freeze_foundation(foundation)

    # Derived adapters contain the final hash but are NOT part of hash scope.
    adapters = {
        "u07": build_u07_adapter(frozen_foundation),
        "u08": build_u08_adapter(frozen_foundation),
        "u09": build_u09_adapter(frozen_foundation),
    }

    frozen_foundation["adapters"] = adapters

    consumer_api = {
        "SOURCE_CELL": CELL_ID,
        "SCHEMA_VERSION": SCHEMA_VERSION,
        "RUN_ID": run_id,
        "SNAPSHOT_ID": snapshot_id,
        "SNAPSHOT_HASH": frozen_foundation["snapshot_hash"],
        "DATA_STATUS": frozen_foundation["status"],
        "DATA_FOUNDATION": frozen_foundation,
        "ADAPTERS": adapters,
    }

    result = {
        "status": frozen_foundation["status"],
        "data_foundation": frozen_foundation,
        "snapshot": {
            "id": snapshot_id,
            "hash": frozen_foundation["snapshot_hash"],
            "integrity_verified": verify_foundation_hash(
                frozen_foundation
            ),
            "persisted": False,
        },
        "providers": frozen_foundation["providers"],
        "provenance": frozen_foundation["provenance"],
        "validation": frozen_foundation["validation"],
        "freshness": frozen_foundation["freshness"],
        "coverage": frozen_foundation["coverage"],
        "definitions": frozen_foundation["definitions"],
        "calculations": frozen_foundation["calculations"],
        "universe": frozen_foundation["universe"],
        "adapters": adapters,
        "warnings": warnings,
        "errors": errors,
        "audit": frozen_foundation["audit"],
        "integrity": frozen_foundation["integrity"],
        "consumer_api": consumer_api,
        "self_tests": {
            "structural": structural_self_tests(),
            "market_engine": market_engine_self_tests(),
            "hash_contract": hash_contract_self_tests(),
        },
    }

    # Consumer/architecture checks are derived and do not mutate foundation.
    result["self_tests"]["consumer_contract"] = consumer_contract_self_tests(
        result
    )
    result["self_tests"]["determinism"] = determinism_self_test(result)
    result["self_tests"]["rule_zero"] = rule_zero_self_test()
    result["technical_lock"] = technical_lock_gate(result)
    return result


# =============================================================================
# 25A. RULE-ZERO / EXECUTION-SOURCE INTEGRITY
# =============================================================================

RULE_ZERO_VERSION = "8.4.5"

def rule_zero_self_test() -> Dict[str, Any]:
    """Verify executable source portability to ordinary Python/PythonAnywhere."""
    source_path = Path(__file__) if "__file__" in globals() else None
    if source_path is None or not source_path.exists():
        return {
            "status": "PASS",
            "checked": False,
            "reason": "runtime_source_path_unavailable",
            "forbidden_patterns": [],
            "pythonanywhere_portable": True,
            "credential_policy": "ENVIRONMENT_VARIABLES_ONLY",
            "version": RULE_ZERO_VERSION,
        }

    source = source_path.read_text(encoding="utf-8", errors="replace")
    lower = source.lower()

    forbidden_tokens = (
        "google" + ".colab",
        "/" + "content",
        "drive" + ".mount",
        "user" + "data",
        "!" + "pip" + " " + "install",
        "pip" + " " + "install",
        "nest" + "_asyncio",
    )
    found = [token for token in forbidden_tokens if token in lower]

    ast_import_violation = False
    try:
        tree = ast.parse(source)
        for node in ast.walk(tree):
            if isinstance(node, ast.Import):
                if any(alias.name.lower() == ("google" + chr(46) + "colab") for alias in node.names):
                    ast_import_violation = True
            elif isinstance(node, ast.ImportFrom):
                module = (node.module or "").lower()
                if module == ("google" + chr(46) + "colab") or module.startswith("google" + chr(46) + "colab" + chr(46)):
                    ast_import_violation = True
    except SyntaxError:
        return {
            "status": "FAIL",
            "checked": True,
            "reason": "source_ast_parse_failed",
            "forbidden_patterns": [],
            "pythonanywhere_portable": False,
            "credential_policy": "ENVIRONMENT_VARIABLES_ONLY",
            "version": RULE_ZERO_VERSION,
        }

    if ast_import_violation and ("google" + chr(46) + "colab") not in found:
        found.append("google" + chr(46) + "colab")

    return {
        "status": "PASS" if not found else "FAIL",
        "checked": True,
        "forbidden_patterns": found,
        "pythonanywhere_portable": not bool(found),
        "credential_policy": "ENVIRONMENT_VARIABLES_ONLY",
        "version": RULE_ZERO_VERSION,
    }

# =============================================================================
# 26. CONSUMER / ARCHITECTURE TESTS
# =============================================================================

def consumer_contract_self_tests(result: Dict[str, Any]) -> Dict[str, Any]:
    foundation = result.get("data_foundation") or {}
    adapters = result.get("adapters") or {}

    expected_fields = {
        "source_cell",
        "schema_version",
        "foundation_version",
        "run_id",
        "snapshot_id",
        "snapshot_hash",
        "status",
        "global",
        "assets",
        "universe",
        "exchange_evidence",
        "price_foundation",
        "indices",
        "time_series",
        "validation",
        "freshness",
        "coverage",
        "definitions",
        "provenance",
        "calculations",
        "snapshot",
        "integrity",
        "warnings",
        "errors",
        "consumer_contract",
        "adapters",
        "security",
        "audit",
    }

    tests = {
        "canonical_fields_complete": expected_fields.issubset(
            set(foundation.keys())
        ),
        "u07_present": "u07" in adapters,
        "u08_present": "u08" in adapters,
        "u09_present": "u09" in adapters,
        "u07_source_u065": (
            adapters.get("u07", {}).get("SOURCE_CELL") == CELL_ID
        ),
        "u08_source_u065": (
            adapters.get("u08", {}).get("SOURCE_CELL") == CELL_ID
        ),
        "u09_source_u065": (
            adapters.get("u09", {}).get("SOURCE_CELL") == CELL_ID
        ),
        "u07_hash_trace": (
            adapters.get("u07", {}).get("SNAPSHOT_HASH")
            == foundation.get("snapshot_hash")
        ),
        "u08_hash_trace": (
            adapters.get("u08", {}).get("SNAPSHOT_HASH")
            == foundation.get("snapshot_hash")
        ),
        "u09_hash_trace": (
            adapters.get("u09", {}).get("SNAPSHOT_HASH")
            == foundation.get("snapshot_hash")
        ),
        "no_downstream_reacquisition_rule": (
            foundation.get("consumer_contract", {}).get(
                "source_of_truth"
            ) == "U06_5_DATA_FOUNDATION"
        ),
    }

    return {
        "status": "PASS" if all(tests.values()) else "FAIL",
        "passed": sum(tests.values()),
        "total": len(tests),
        "tests": tests,
    }


def determinism_self_test(result: Dict[str, Any]) -> Dict[str, Any]:
    foundation = result.get("data_foundation")
    if not foundation:
        return {
            "status": "FAIL",
            "passed": 0,
            "total": 1,
            "tests": {"foundation_available": False},
        }

    canonical = canonical_hash_input(foundation)
    hash_one = sha256_obj(canonical)
    hash_two = sha256_obj(deep_copy(canonical))

    tests = {
        "canonical_hash_repeatable": hash_one == hash_two,
        "stored_hash_verifies": verify_foundation_hash(foundation),
        "adapter_hash_matches": all(
            adapter.get("SNAPSHOT_HASH") == foundation.get("snapshot_hash")
            for adapter in foundation.get("adapters", {}).values()
        ),
    }

    return {
        "status": "PASS" if all(tests.values()) else "FAIL",
        "passed": sum(tests.values()),
        "total": len(tests),
        "tests": tests,
    }


# =============================================================================
# 27. EXPLICIT SNAPSHOT PERSISTENCE — NEVER AUTOMATIC
# =============================================================================

def persist_snapshot(result: Dict[str, Any]) -> Dict[str, Any]:
    """
    Explicit snapshot operation.
    This function is intentionally NOT called by run_u06_5().
    """
    foundation = result.get("data_foundation")
    if not foundation:
        raise ValueError("No U06.5 foundation available.")

    if not verify_foundation_hash(foundation):
        raise ValueError("Cannot persist: foundation hash verification failed.")

    snapshot_id = foundation["snapshot_id"]

    payload = {
        "U06_5_RESULT": redact(result),
    }

    snapshot_data = canonical_bytes(payload)
    snapshot_path = SNAPSHOT_DIR / f"{safe_name(snapshot_id)}.json"
    audit_path = AUDIT_DIR / f"{safe_name(snapshot_id)}__audit.json"

    snapshot_path.write_bytes(snapshot_data)
    audit_path.write_bytes(
        canonical_bytes(redact(foundation.get("audit", {})))
    )

    # Return a derived status; do not mutate the frozen foundation.
    return {
        "status": "PERSISTED",
        "snapshot_path": str(snapshot_path),
        "audit_path": str(audit_path),
        "snapshot_sha256": sha256_bytes(snapshot_data),
        "foundation_sha256": foundation["snapshot_hash"],
    }


# =============================================================================
# 29. TECHNICAL LOCK GATE — CAPABILITY-AWARE
# =============================================================================

def credential_discovery_self_test() -> Dict[str, Any]:
    """
    Structural credential test only. It verifies that discovery returns either
    a non-empty credential with a safe source label or a clean NONE state.
    It never prints or stores the credential itself.
    """
    source_ok = (
        CMC_API_KEY_SOURCE == "NONE"
        or CMC_API_KEY_SOURCE.startswith("ENV:")
    )
    return {
        "status": "PASS" if source_ok else "FAIL",
        "credential_present": bool(CMC_API_KEY),
        "source_label_valid": source_ok,
        "secret_exposed": False,
    }


def technical_lock_gate(result: Dict[str, Any]) -> Dict[str, Any]:
    """
    Conservative capability-aware final gate.
    This function NEVER converts unavailable evidence into PASS; it only avoids
    making unrelated foundation capabilities depend on an unavailable optional source.
    It distinguishes:
      - internal implementation integrity,
      - exchange evidence,
      - market/index historical time series,
      - external reference validation.

    A LOCKED result is possible only when every mandatory contract is genuinely
    satisfied in the current runtime. Unsupported or unacquired evidence
    remains explicitly blocked.
    """
    foundation = result.get("data_foundation") or {}
    exchange = foundation.get("exchange_evidence") or {}
    exchange_engine = exchange.get("market_pair_engine") or {}
    multi_exchange = foundation.get("multi_exchange_evidence") or {}
    time_series = foundation.get("time_series") or {}
    reference = foundation.get("providers", {}).get("reference", {}) or {}

    # Contract separation:
    #   GLOBAL layer  = CMC / CoinGecko market-wide evidence and Top-125.
    #   EXCHANGE layer = OHLCV / traded-volume evidence only.
    # Exchange feeds NEVER replace market-cap, dominance, Top-125, or Kitchen
    # TOTAL calculations.
    #
    # A provider/reference outage must block only the capability it owns.
    # In particular, lack of historical CMC/CoinGecko 5m Global Metrics must
    # NOT invalidate an otherwise valid current SNAPSHOT foundation.
    historical_mode_requires_series = (
        DATASET_MODE == "HISTORICAL" and HISTORICAL_TIME_SERIES_REQUIRED
    )
    time_series_pass = (
        time_series.get("status") == "VALIDATED"
        if historical_mode_requires_series
        else True
    )
    reference_contract_pass = (
        True
        if not REFERENCE_REQUIRED_FOR_LOCK
        else reference.get("status") == "VALIDATED"
    )

    tests = {
        "foundation_exists": bool(foundation),
        "hash_verified": bool(foundation) and verify_foundation_hash(foundation),
        "structural_pass": result.get("self_tests", {}).get("structural", {}).get("status") == "PASS",
        "market_engine_pass": result.get("self_tests", {}).get("market_engine", {}).get("status") == "PASS",
        "hash_contract_pass": result.get("self_tests", {}).get("hash_contract", {}).get("status") == "PASS",
        "consumer_contract_pass": result.get("self_tests", {}).get("consumer_contract", {}).get("status") == "PASS",
        "determinism_pass": result.get("self_tests", {}).get("determinism", {}).get("status") == "PASS",
        "top125_validated": foundation.get("validation", {}).get("top125", {}).get("status") == "VALIDATED",
        "exchange_evidence_validated": exchange_engine.get("status") == "VALIDATED",
        "multi_exchange_validated": multi_exchange.get("status") == "VALIDATED",
        "time_series_validated_or_not_required": time_series_pass,
        "reference_non_blocking_or_validated": reference_contract_pass,
    }

    blockers = []
    if not tests["foundation_exists"]:
        blockers.append("FOUNDATION_NOT_AVAILABLE")
    if not tests["hash_verified"]:
        blockers.append("FOUNDATION_HASH_NOT_VERIFIED")
    if not tests["structural_pass"]:
        blockers.append("STRUCTURAL_SELF_TEST_FAILED")
    if not tests["market_engine_pass"]:
        blockers.append("MARKET_ENGINE_SELF_TEST_FAILED")
    if not tests["hash_contract_pass"]:
        blockers.append("HASH_CONTRACT_SELF_TEST_FAILED")
    if not tests["consumer_contract_pass"]:
        blockers.append("CONSUMER_CONTRACT_SELF_TEST_FAILED")
    if not tests["determinism_pass"]:
        blockers.append("DETERMINISM_SELF_TEST_FAILED")
    if not tests["top125_validated"]:
        blockers.append("TOP125_NOT_VALIDATED")
    if not tests["exchange_evidence_validated"]:
        blockers.append("EXCHANGE_EVIDENCE_NOT_FULLY_VALIDATED")
    if not tests["multi_exchange_validated"]:
        blockers.append("MULTI_EXCHANGE_MINIMUM_7_OF_8_NOT_VALIDATED")
    if historical_mode_requires_series and not tests["time_series_validated_or_not_required"]:
        blockers.append("MARKET_INDEX_TIME_SERIES_5M_REQUIRED_FOR_HISTORICAL_MODE")
    if REFERENCE_REQUIRED_FOR_LOCK and not tests["reference_non_blocking_or_validated"]:
        blockers.append("REFERENCE_NOT_VALIDATED")

    rz = result.get("self_tests", {}).get("rule_zero", {})
    if rz.get("status") != "PASS":
        blockers.append("RULE_ZERO_PORTABILITY_SELF_TEST_FAILED")

    blockers = list(dict.fromkeys(blockers))
    locked = not blockers

    return {
        "status": "LOCKED" if locked else "BLOCKED",
        "locked": locked,
        "tests": tests,
        "blockers": blockers,
        "rule": "NO_PASS_THROUGH_OR_FABRICATED_EVIDENCE",
    }

# =============================================================================
# 28. RUN
# =============================================================================

SELF_TESTS = {
    "credential_discovery": credential_discovery_self_test(),
    "structural": structural_self_tests(),
    "market_engine": market_engine_self_tests(),
    "hash_contract": hash_contract_self_tests(),
    "rule_zero": rule_zero_self_test(),
}

try:
    U06_5_RESULT = run_u06_5()
except Exception as exc:
    U06_5_RESULT = {
        "status": "FOUNDATION_TECHNICAL_ERROR",
        "runtime_diagnostic": _runtime_diagnostic("RUN_U06_5_TOP_LEVEL", exc),
        "data_foundation": None,
        "snapshot": None,
        "providers": {},
        "provenance": {},
        "validation": {},
        "freshness": {},
        "coverage": {},
        "definitions": {},
        "calculations": {},
        "universe": {},
        "adapters": {},
        "warnings": [],
        "errors": [
            f"{type(exc).__name__}: {exc}",
            traceback.format_exc(),
        ],
        "audit": {},
        "integrity": {},
        "consumer_api": None,
        "self_tests": SELF_TESTS,
    }

U06_5_DATA_FOUNDATION = U06_5_RESULT.get("data_foundation")
U06_5_CONSUMER_API = U06_5_RESULT.get("consumer_api")
U07_ADAPTER = U06_5_RESULT.get("adapters", {}).get("u07")
U08_ADAPTER = U06_5_RESULT.get("adapters", {}).get("u08")
U09_ADAPTER = U06_5_RESULT.get("adapters", {}).get("u09")



# =============================================================================
# 30. FINAL REPORT
# =============================================================================

def print_u06_5_report(result: Dict[str, Any]) -> None:
    foundation = result.get("data_foundation") or {}

    print("=" * 88)
    print("KITCHEN ASSISTANT V3.1 — U06.5")
    print("MARKET DATA SOURCE & VALIDATION FOUNDATION")
    print(f"CANONICAL IMPLEMENTATION — VERSION {VERSION}")
    print("=" * 88)

    print("CMC KEYLESS                 :", "ENABLED")
    print("CMC AUTHENTICATED           :", "AVAILABLE" if CMC_API_KEY else "NOT_CONFIGURED")
    print("CMC KEY SOURCE              :", CMC_API_KEY_SOURCE)
    print("COINGECKO                   :", "FALLBACK / CROSS-SOURCE")
    print("TRADINGVIEW                  :", "DISABLED / NOT_USED")
    print("DATASET MODE                :", DATASET_MODE)
    print("TIMEFRAME CONTRACT          :", TIMEFRAME)
    print("SNAPSHOT AUTO-PERSIST       :", "NO")
    print("STATUS                      :", result.get("status"))

    if foundation:
        print("SELECTED PROVIDER           :",
              foundation.get("provenance", {}).get("PROVIDER"))
        print("SELECTED PROVIDER MODE      :",
              foundation.get("provenance", {}).get("PROVIDER_MODE"))
        print("TOP-125 STATUS              :",
              foundation.get("validation", {}).get("top125", {}).get("status"))
        print("TOP-125 COVERAGE            :",
              foundation.get("coverage", {}).get("coverage_ratio"))
        print("CANDIDATE VALID              :",
              foundation.get("coverage", {}).get("candidate_valid"))
        print("EXCHANGE EVIDENCE            :",
              foundation.get("exchange_evidence", {}).get(
                  "market_pair_engine", {}
              ).get("status"))
        print("MULTI-EXCHANGE CORE          :",
              foundation.get("multi_exchange_evidence", {}).get("status"))
        print("MULTI-EXCHANGE VALIDATED      :",
              foundation.get("multi_exchange_evidence", {}).get("validated_exchange_count"),
              "/", foundation.get("multi_exchange_evidence", {}).get("target_exchange_count"))
        print("ORDERBOOK VALIDATED           :",
              foundation.get("multi_exchange_evidence", {}).get("orderbook_validated_exchange_count"),
              "/", foundation.get("multi_exchange_evidence", {}).get("target_exchange_count"))
        # Credential-safe exchange diagnostics: make every non-validated
        # provider explain whether the failure is HTTP, transport, schema,
        # or validation-related. This prevents silent exchange exclusion.
        _mx = foundation.get("multi_exchange_evidence", {}).get("per_exchange", {})
        for _provider, _ev in _mx.items():
            _ob = _ev.get("orderbook_evidence", {}) if isinstance(_ev, dict) else {}
            _oq = _ev.get("orderbook_quality", {}) if isinstance(_ev, dict) else {}
            print(
                f"EXCHANGE DIAGNOSTIC [{_provider.upper():8}] : "
                f"CANDLE={_ev.get('status')} HTTP={_ev.get('http_status')} "
                f"ROWS={_ev.get('row_count')} ERR={_ev.get('error_class') or '-'} | "
                f"ORDERBOOK={_oq.get('status')} HTTP={_ob.get('http_status')} "
                f"ERR={_ob.get('error_class') or '-'}"
            )
        print("REFERENCE BTC PRICE           :",
              foundation.get("price_foundation", {}).get("reference_price"))
        print("REFERENCE BTC CONFIDENCE      :",
              foundation.get("price_foundation", {}).get("confidence"))
        _pf = foundation.get("price_foundation", {})
        _agg = _pf.get("aggregate", {}) if isinstance(_pf, dict) else {}
        print("REFERENCE ACCEPTED / REJECTED :",
              len(_agg.get("accepted_markets", []) or []), "/",
              len(_agg.get("rejected_markets", []) or []))
        if _agg.get("rejected_markets"):
            _reasons = []
            for _r in _agg.get("rejected_markets", []):
                _reason = _r.get("rejection_reason") or _r.get("reason") if isinstance(_r, dict) else None
                if _reason and _reason not in _reasons:
                    _reasons.append(_reason)
            print("REFERENCE REJECTION REASONS   :", ", ".join(_reasons) if _reasons else "-")
        print("TIME SERIES                  :",
              foundation.get("time_series", {}).get("status"))
        print("TIME SERIES LOCK ROLE        :", ("REQUIRED — 5M MINIMUM" if DATASET_MODE == "HISTORICAL" and HISTORICAL_TIME_SERIES_REQUIRED else "OPTIONAL IN SNAPSHOT / REQUIRED IN HISTORICAL"))
        print("TIME SERIES SOURCE           :",
              foundation.get("time_series", {}).get("history_source"))
        print("TIME SERIES REASON           :",
              foundation.get("time_series", {}).get("market_index", {}).get("reason"))
        print("REFERENCE                    :",
              foundation.get("providers", {}).get(
                  "reference", {}
              ).get("status"))
        print("SNAPSHOT ID                  :",
              foundation.get("snapshot_id"))
        print("SNAPSHOT HASH                :",
              foundation.get("snapshot_hash"))
        print("HASH VERIFIED                :",
              verify_foundation_hash(foundation))
        print("SNAPSHOT PERSISTED           :",
              "NO (explicit operation only)")
        print("TECHNICAL LOCK               :",
              result.get("technical_lock", {}).get("status", "NOT_RUN"))
        print("LOCK BLOCKERS                :",
              result.get("technical_lock", {}).get("blockers", []))

        indices = foundation.get("indices", {})
        values = indices.get("values", {})
        if values:
            print("KITCHEN TOTAL TOP125         :",
                  values.get("KITCHEN_TOTAL_TOP125"))
            print("KITCHEN TOTAL2               :",
                  values.get("KITCHEN_TOTAL2"))
            print("KITCHEN TOTAL3               :",
                  values.get("KITCHEN_TOTAL3"))
            print("KITCHEN OTHERS               :",
                  values.get("KITCHEN_OTHERS"))
            print("KITCHEN BTC.D                :",
                  values.get("KITCHEN_BTC_D"))
            print("KITCHEN ETH.D                :",
                  values.get("KITCHEN_ETH_D"))
            print("KITCHEN USDT.D              :",
                  values.get("KITCHEN_USDT_D"))
            print("KITCHEN OTHERS.D            :",
                  values.get("KITCHEN_OTHERS_D"))

    print("-" * 88)
    print("STRUCTURAL TESTS             :",
          result.get("self_tests", {}).get(
              "structural", SELF_TESTS["structural"]
          ).get("status"))
    print("MARKET ENGINE TESTS          :",
          result.get("self_tests", {}).get(
              "market_engine", SELF_TESTS["market_engine"]
          ).get("status"))
    print("HASH CONTRACT TESTS          :",
          result.get("self_tests", {}).get(
              "hash_contract", SELF_TESTS["hash_contract"]
          ).get("status"))
    print("CONSUMER CONTRACT TESTS      :",
          result.get("self_tests", {}).get(
              "consumer_contract", {}
          ).get("status", "NOT_RUN"))
    print("DETERMINISM TESTS            :",
          result.get("self_tests", {}).get(
              "determinism", {}
          ).get("status", "NOT_RUN"))
    print("RULE ZERO TESTS              :",
          result.get("self_tests", {}).get(
              "rule_zero", {}
          ).get("status", "NOT_RUN"))
    print("=" * 88)


print_u06_5_report(U06_5_RESULT)
```

```python

# ============================================================
# KITCHEN ASSISTANT V3.1
#
# U07 — MARKET STRUCTURE SCENARIO ENGINE
# + RANGE DETECTION ENGINE
#
# CANONICAL IMPLEMENTATION
#
# NON-EXECUTIONAL / AUDITABLE / DETERMINISTIC
# ============================================================

from __future__ import annotations

from dataclasses import dataclass, asdict
from typing import Any, Dict, List, Optional, Tuple
from enum import Enum
from datetime import datetime, timezone
import json
import math
import re


# ============================================================
# 1. HARD SAFETY LOCKS
# ============================================================

TRADING_ENABLED = False
ORDERS_ENABLED = False
STRATEGY_ENABLED = False
PORTFOLIO_ACTIONS_ENABLED = False

assert TRADING_ENABLED is False
assert ORDERS_ENABLED is False
assert STRATEGY_ENABLED is False
assert PORTFOLIO_ACTIONS_ENABLED is False


# ============================================================
# 2. ENUMS
# ============================================================

class Direction(str, Enum):
    INCREASE = "INCREASE"
    DECREASE = "DECREASE"
    RANGE = "RANGE"


class RangeType(str, Enum):
    LOW_VOLATILITY_RANGE = "LOW-VOLATILITY RANGE"
    HIGH_VOLATILITY_RANGE = "HIGH-VOLATILITY RANGE"
    NONE = "NONE"


class RangeConfidence(str, Enum):
    HIGH = "HIGH CONFIDENCE RANGE"
    MEDIUM = "MEDIUM CONFIDENCE RANGE"
    LOW = "LOW CONFIDENCE / AMBIGUOUS"


class ScenarioType(str, Enum):
    MIRROR = "MIRROR"
    PARALLEL = "PARALLEL"
    RANGE_COMPATIBLE_STATE = "RANGE-COMPATIBLE STATE"
    NEUTRAL_FLAT = "NEUTRAL / FLAT"


# ============================================================
# 3. CONFIGURATION
#
# IMPORTANT:
# These are configurable engineering parameters.
# They are NOT presented as universal market truths.
# ============================================================

@dataclass
class RangeConfig:
    # Minimum absolute net movement to classify a directional move.
    direction_threshold_pct: float = 0.50

    # Maximum net movement for the zero/flat edge case.
    flat_threshold_pct: float = 0.10

    # Relative amplitude threshold separating lower/higher
    # volatility structures.
    low_volatility_pct: float = 2.00

    # Minimum range score required before RANGE can be considered.
    range_score_threshold: float = 0.60

    # High-volatility range threshold.
    high_volatility_pct: float = 5.00

    # DMI/ADX parameters.
    dmi_period: int = 14
    adx_period: int = 14

    # Difference between DI+ and DI- considered "close".
    dmi_balance_threshold: float = 5.0

    # ADX below this level is supportive of weak trend.
    low_adx_threshold: float = 20.0

    # ADX above this level means strong movement, but DOES NOT
    # automatically invalidate Range.
    high_adx_threshold: float = 35.0

    # Minimum confidence score for high confidence.
    high_confidence_score: float = 0.78

    # Minimum confidence score for medium confidence.
    medium_confidence_score: float = 0.62


DEFAULT_RANGE_CONFIG = RangeConfig()


# ============================================================
# 4. INPUT VALIDATION
# ============================================================

def _is_finite_number(x: Any) -> bool:
    try:
        return math.isfinite(float(x))
    except Exception:
        return False


def validate_series(series: List[float], name: str) -> None:
    if not isinstance(series, list):
        raise TypeError(f"{name} must be a list.")

    if len(series) < 2:
        raise ValueError(
            f"{name} requires at least 2 observations."
        )

    for value in series:
        if not _is_finite_number(value):
            raise ValueError(
                f"{name} contains a non-finite value."
            )

    if any(float(v) <= 0 for v in series):
        raise ValueError(
            f"{name} must contain positive values."
        )


def validate_ohlc_structure(
    highs: List[float],
    lows: List[float],
    closes: List[float],
    name: str,
) -> None:
    """
    Validate basic OHLC structural consistency.

    Rules:
        high >= low
        low <= close <= high
    """

    if not (
        len(highs) == len(lows) == len(closes)
    ):
        raise ValueError(
            f"{name}: highs, lows and closes must have equal lengths."
        )

    for i, (high, low, close) in enumerate(
        zip(highs, lows, closes)
    ):
        high = float(high)
        low = float(low)
        close = float(close)

        if high < low:
            raise ValueError(
                f"{name}: high < low at index {i}."
            )

        if close < low or close > high:
            raise ValueError(
                f"{name}: close outside high/low range at index {i}."
            )


def validate_user_range(
    analysis_range: Any,
) -> Dict[str, Any]:
    """
    USER RANGE IS AUTHORITATIVE.

    The engine never changes it and never silently replaces it.
    """

    if analysis_range is None:
        raise ValueError("analysis_range is required.")

    if isinstance(analysis_range, dict):
        if "start" not in analysis_range or "end" not in analysis_range:
            raise ValueError(
                "analysis_range dict must contain start and end."
            )

        start = analysis_range["start"]
        end = analysis_range["end"]

        if start is None or end is None:
            raise ValueError(
                "analysis_range start/end cannot be None."
            )

        return {
            "start": start,
            "end": end,
            "custom": bool(analysis_range.get("custom", True)),
            "label": analysis_range.get("label"),
        }

    if isinstance(analysis_range, (tuple, list)):
        if len(analysis_range) != 2:
            raise ValueError(
                "analysis_range sequence must contain exactly 2 values."
            )

        return {
            "start": analysis_range[0],
            "end": analysis_range[1],
            "custom": True,
            "label": None,
        }

    raise TypeError(
        "analysis_range must be a dict, tuple, or list."
    )


# ============================================================
# 5. TIMEFRAME ENGINE
# ============================================================

_TIMEFRAME_RE = re.compile(
    r"^\s*(\d+(?:\.\d+)?)\s*([mMhHdDwW])\s*$"
)


def parse_timeframe(timeframe: str) -> Dict[str, Any]:
    """
    General timeframe parser.

    Critical rules:

        minute  -> lowercase m
        hour    -> h / H
        day     -> d / D
        week    -> w / W
        month   -> uppercase M

    ANY valid timeframe below 1D
    = VALID + SUB_DAILY_WARNING

    1D, 1W and 1M
    = VALID + NO WARNING

    IMPORTANT:
    Uppercase M is intentionally preserved as MONTH.
    It must NEVER be converted to lowercase before interpretation.
    """

    if not isinstance(timeframe, str):
        return {
            "valid": False,
            "warning": None,
            "error": "TECHNICAL_VALIDATION_ERROR",
            "normalized": None,
            "minutes": None,
        }

    match = _TIMEFRAME_RE.match(timeframe)

    if not match:
        return {
            "valid": False,
            "warning": None,
            "error": "TECHNICAL_VALIDATION_ERROR",
            "normalized": None,
            "minutes": None,
        }

    amount = float(match.group(1))
    raw_unit = match.group(2)

    if amount <= 0:
        return {
            "valid": False,
            "warning": None,
            "error": "TECHNICAL_VALIDATION_ERROR",
            "normalized": None,
            "minutes": None,
        }

    # --------------------------------------------------------
    # IMPORTANT:
    # Do NOT lowercase before checking for uppercase M.
    # --------------------------------------------------------

    if raw_unit == "M":
        unit = "M"
    else:
        unit = raw_unit.lower()

    multipliers = {
        "m": 1,
        "h": 60,
        "d": 1440,
        "w": 10080,
        "M": 43200,   # engineering representation: 30 days
    }

    minutes = amount * multipliers[unit]

    normalized = (
        f"{int(amount) if amount.is_integer() else amount:g}{unit}"
    )

    if minutes < 1440:
        return {
            "valid": True,
            "warning": "SUB_DAILY_WARNING",
            "error": None,
            "normalized": normalized,
            "minutes": minutes,
        }

    return {
        "valid": True,
        "warning": None,
        "error": None,
        "normalized": normalized,
        "minutes": minutes,
    }


# ============================================================
# 6. BASIC MOVEMENT METRICS
# ============================================================

def pct_change(start: float, end: float) -> float:
    if start == 0:
        raise ValueError(
            "Cannot calculate percentage change from zero."
        )

    return ((end - start) / start) * 100.0


def range_amplitude_pct(series: List[float]) -> float:
    """
    Peak-to-trough amplitude relative to the first observation.
    """

    start = float(series[0])
    highest = max(series)
    lowest = min(series)

    return ((highest - lowest) / start) * 100.0


def normalized_path_length(series: List[float]) -> float:
    """
    Measures total internal movement relative to starting value.

    Higher values generally indicate more back-and-forth behavior.
    """

    start = float(series[0])

    if start == 0:
        return 0.0

    total_abs_move = sum(
        abs(float(series[i]) - float(series[i - 1]))
        for i in range(1, len(series))
    )

    return (total_abs_move / start) * 100.0


def direction_switch_count(series: List[float]) -> int:
    signs = []

    for i in range(1, len(series)):
        delta = float(series[i]) - float(series[i - 1])

        if delta > 0:
            signs.append(1)

        elif delta < 0:
            signs.append(-1)

    if len(signs) < 2:
        return 0

    switches = 0

    for i in range(1, len(signs)):
        if signs[i] != signs[i - 1]:
            switches += 1

    return switches


def close_dispersion_pct(series: List[float]) -> float:
    mean_value = sum(series) / len(series)

    if mean_value == 0:
        return 0.0

    variance = sum(
        (x - mean_value) ** 2
        for x in series
    ) / len(series)

    std = math.sqrt(variance)

    return (std / mean_value) * 100.0


# ============================================================
# 7. DMI / ADX
# ============================================================

def calculate_dmi_adx(
    highs: List[float],
    lows: List[float],
    closes: List[float],
    period: int = 14,
    adx_period: int = 14,
) -> Dict[str, Optional[float]]:
    """
    Wilder-style DMI / ADX calculation.

    DMI / ADX are SUPPORTING FEATURES ONLY.
    They do not independently define Range.

    If insufficient data exists, values remain None.
    """

    if period <= 0:
        raise ValueError("period must be greater than zero.")

    if adx_period <= 0:
        raise ValueError(
            "adx_period must be greater than zero."
        )

    if not (
        len(highs) == len(lows) == len(closes)
    ):
        raise ValueError(
            "highs, lows and closes must have equal lengths."
        )

    n = len(closes)

    if n < period + 1:
        return {
            "di_plus": None,
            "di_minus": None,
            "adx": None,
        }

    tr = []
    plus_dm = []
    minus_dm = []

    for i in range(1, n):

        high = float(highs[i])
        low = float(lows[i])

        prev_close = float(closes[i - 1])
        prev_high = float(highs[i - 1])
        prev_low = float(lows[i - 1])

        true_range = max(
            high - low,
            abs(high - prev_close),
            abs(low - prev_close),
        )

        up_move = high - prev_high
        down_move = prev_low - low

        p_dm = (
            up_move
            if up_move > down_move and up_move > 0
            else 0.0
        )

        m_dm = (
            down_move
            if down_move > up_move and down_move > 0
            else 0.0
        )

        tr.append(true_range)
        plus_dm.append(p_dm)
        minus_dm.append(m_dm)

    if len(tr) < period:
        return {
            "di_plus": None,
            "di_minus": None,
            "adx": None,
        }

    def wilder_smooth(
        values: List[float],
        p: int,
    ) -> List[float]:

        first = sum(values[:p])

        smoothed = [first]

        for value in values[p:]:

            previous = smoothed[-1]

            smoothed.append(
                previous
                - (previous / p)
                + value
            )

        return smoothed

    tr_s = wilder_smooth(
        tr,
        period,
    )

    plus_s = wilder_smooth(
        plus_dm,
        period,
    )

    minus_s = wilder_smooth(
        minus_dm,
        period,
    )

    dx_values = []
    di_plus_values = []
    di_minus_values = []

    for trv, pv, mv in zip(
        tr_s,
        plus_s,
        minus_s,
    ):

        if trv == 0:

            di_p = 0.0
            di_m = 0.0

        else:

            di_p = 100.0 * (pv / trv)
            di_m = 100.0 * (mv / trv)

        denominator = di_p + di_m

        if denominator == 0:

            dx = 0.0

        else:

            dx = (
                100.0
                * abs(di_p - di_m)
                / denominator
            )

        di_plus_values.append(di_p)
        di_minus_values.append(di_m)
        dx_values.append(dx)

    if not dx_values:
        return {
            "di_plus": None,
            "di_minus": None,
            "adx": None,
        }

    # --------------------------------------------------------
    # Proper Wilder ADX initialization.
    #
    # ADX should be an average of DX values, not the raw Wilder
    # sum. This keeps ADX on its expected 0-100 scale.
    # --------------------------------------------------------

    if len(dx_values) < adx_period:

        adx = sum(dx_values) / len(dx_values)

    else:

        first_adx = (
            sum(dx_values[:adx_period])
            / adx_period
        )

        adx_series = [first_adx]

        for dx in dx_values[adx_period:]:

            previous = adx_series[-1]

            next_adx = (
                ((previous * (adx_period - 1)) + dx)
                / adx_period
            )

            adx_series.append(next_adx)

        adx = adx_series[-1]

    return {
        "di_plus": di_plus_values[-1],
        "di_minus": di_minus_values[-1],
        "adx": adx,
    }


# ============================================================
# 8. RANGE ENGINE
# ============================================================

@dataclass
class RangeAnalysis:
    direction: Direction
    range_type: RangeType
    confidence: RangeConfidence
    confidence_score: float

    net_change_pct: float
    amplitude_pct: float
    path_length_pct: float
    switch_count: int
    close_dispersion_pct: float

    di_plus: Optional[float]
    di_minus: Optional[float]
    adx: Optional[float]

    ambiguous: bool
    reasons: List[str]


def analyze_range(
    values: List[float],
    highs: Optional[List[float]] = None,
    lows: Optional[List[float]] = None,
    closes: Optional[List[float]] = None,
    config: RangeConfig = DEFAULT_RANGE_CONFIG,
) -> RangeAnalysis:

    validate_series(values, "values")

    values = [
        float(v)
        for v in values
    ]

    if closes is None:
        closes = values.copy()

    if highs is None:
        highs = values.copy()

    if lows is None:
        lows = values.copy()

    validate_series(highs, "highs")
    validate_series(lows, "lows")
    validate_series(closes, "closes")

    if not (
        len(values)
        == len(highs)
        == len(lows)
        == len(closes)
    ):
        raise ValueError(
            "values/highs/lows/closes must have equal lengths."
        )

    validate_ohlc_structure(
        highs,
        lows,
        closes,
        "RangeAnalysis",
    )

    start = values[0]
    end = values[-1]

    net_change = pct_change(
        start,
        end,
    )

    amplitude = range_amplitude_pct(
        values
    )

    path_length = normalized_path_length(
        values
    )

    switches = direction_switch_count(
        values
    )

    dispersion = close_dispersion_pct(
        values
    )

    dmi = calculate_dmi_adx(
        highs,
        lows,
        closes,
        period=config.dmi_period,
        adx_period=config.adx_period,
    )

    di_plus = dmi["di_plus"]
    di_minus = dmi["di_minus"]
    adx = dmi["adx"]

    # --------------------------------------------------------
    # ZERO-MOVEMENT EDGE CASE
    # --------------------------------------------------------

    zero_net = (
        abs(net_change)
        <= config.flat_threshold_pct
    )

    # --------------------------------------------------------
    # INTERNAL STRUCTURE FEATURES
    # --------------------------------------------------------

    internal_chop = False

    if len(values) >= 4:

        internal_chop = (
            switches >= 2
            and path_length > max(
                abs(net_change) * 1.5,
                config.flat_threshold_pct,
            )
        )

    dmi_balanced = None

    if (
        di_plus is not None
        and di_minus is not None
    ):

        dmi_balanced = (
            abs(di_plus - di_minus)
            <= config.dmi_balance_threshold
        )

    weak_adx = (
        adx is not None
        and adx < config.low_adx_threshold
    )

    strong_adx = (
        adx is not None
        and adx >= config.high_adx_threshold
    )

    # --------------------------------------------------------
    # RANGE SCORING
    # --------------------------------------------------------

    score_components = []

    # 1. Net movement
    if zero_net:

        score_components.append(1.0)

    elif abs(net_change) < config.direction_threshold_pct:

        score_components.append(0.80)

    else:

        score_components.append(0.0)

    # 2. Direction switching / internal back-and-forth
    if internal_chop:

        score_components.append(1.0)

    elif switches >= 1:

        score_components.append(0.65)

    else:

        score_components.append(0.20)

    # 3. DMI balance
    if dmi_balanced is True:

        score_components.append(1.0)

    elif dmi_balanced is False:

        score_components.append(0.0)

    else:

        score_components.append(0.50)

    # 4. ADX
    #
    # Low ADX strongly supports Range.
    # High ADX lowers confidence but does not invalidate Range.
    if weak_adx:

        score_components.append(1.0)

    elif strong_adx:

        score_components.append(0.35)

    elif adx is not None:

        score_components.append(0.65)

    else:

        score_components.append(0.50)

    # 5. Relative volatility / amplitude
    if amplitude <= config.low_volatility_pct:

        score_components.append(1.0)

    elif amplitude >= config.high_volatility_pct:

        score_components.append(0.45)

    else:

        score_components.append(0.70)

    range_score = (
        sum(score_components)
        / len(score_components)
    )

    # --------------------------------------------------------
    # DIRECTIONAL STRUCTURE
    # --------------------------------------------------------

    direction = None

    if net_change > config.direction_threshold_pct:

        direction = Direction.INCREASE

    elif net_change < -config.direction_threshold_pct:

        direction = Direction.DECREASE

    # --------------------------------------------------------
    # CONTRADICTION DETECTION
    # --------------------------------------------------------

    contradictory = False

    if direction is not None:

        if internal_chop and strong_adx:

            contradictory = True

        if (
            dmi_balanced is True
            and abs(net_change)
            < config.direction_threshold_pct * 2
        ):

            contradictory = True

    # --------------------------------------------------------
    # RANGE DECISION
    # --------------------------------------------------------

    range_candidate = (
        range_score >= config.range_score_threshold
        or zero_net
        or internal_chop
    )

    ambiguous = False
    reasons = []

    if contradictory:

        ambiguous = True

        reasons.append(
            "Directional movement and range-supporting structure "
            "produce conflicting evidence."
        )

    if strong_adx and internal_chop:

        reasons.append(
            "ADX is elevated while internal movement remains "
            "strongly back-and-forth."
        )

    if amplitude >= config.high_volatility_pct:

        reasons.append(
            "High internal volatility detected."
        )

    if zero_net:

        reasons.append(
            "Net Start→End movement is approximately zero; "
            "internal structure was evaluated."
        )

    # --------------------------------------------------------
    # FINAL CLASSIFICATION
    # --------------------------------------------------------

    if range_candidate:

        if amplitude >= config.high_volatility_pct:

            range_type = (
                RangeType.HIGH_VOLATILITY_RANGE
            )

        else:

            range_type = (
                RangeType.LOW_VOLATILITY_RANGE
            )

        if (
            ambiguous
            or range_score
            < config.medium_confidence_score
        ):

            confidence = RangeConfidence.LOW
            ambiguous = True

        elif range_score >= config.high_confidence_score:

            confidence = RangeConfidence.HIGH

        else:

            confidence = RangeConfidence.MEDIUM

        # If a strong directional move is unmistakable and
        # range evidence is insufficient, preserve direction.
        if (
            direction is not None
            and not zero_net
            and not internal_chop
            and range_score
            < config.range_score_threshold
        ):

            return RangeAnalysis(
                direction=direction,
                range_type=RangeType.NONE,
                confidence=RangeConfidence.LOW,
                confidence_score=range_score,
                net_change_pct=net_change,
                amplitude_pct=amplitude,
                path_length_pct=path_length,
                switch_count=switches,
                close_dispersion_pct=dispersion,
                di_plus=di_plus,
                di_minus=di_minus,
                adx=adx,
                ambiguous=False,
                reasons=reasons,
            )

        return RangeAnalysis(
            direction=Direction.RANGE,
            range_type=range_type,
            confidence=confidence,
            confidence_score=range_score,
            net_change_pct=net_change,
            amplitude_pct=amplitude,
            path_length_pct=path_length,
            switch_count=switches,
            close_dispersion_pct=dispersion,
            di_plus=di_plus,
            di_minus=di_minus,
            adx=adx,
            ambiguous=ambiguous,
            reasons=reasons,
        )

    if direction is None:

        return RangeAnalysis(
            direction=Direction.RANGE,
            range_type=RangeType.LOW_VOLATILITY_RANGE,
            confidence=RangeConfidence.LOW,
            confidence_score=range_score,
            net_change_pct=net_change,
            amplitude_pct=amplitude,
            path_length_pct=path_length,
            switch_count=switches,
            close_dispersion_pct=dispersion,
            di_plus=di_plus,
            di_minus=di_minus,
            adx=adx,
            ambiguous=True,
            reasons=[
                "Movement does not satisfy a sufficiently strong "
                "directional threshold."
            ],
        )

    return RangeAnalysis(
        direction=direction,
        range_type=RangeType.NONE,
        confidence=RangeConfidence.LOW,
        confidence_score=range_score,
        net_change_pct=net_change,
        amplitude_pct=amplitude,
        path_length_pct=path_length,
        switch_count=switches,
        close_dispersion_pct=dispersion,
        di_plus=di_plus,
        di_minus=di_minus,
        adx=adx,
        ambiguous=False,
        reasons=reasons,
    )


# ============================================================
# 9. CANONICAL 9-SCENARIO MATRIX
# ============================================================

SCENARIO_MATRIX = {
    (Direction.INCREASE, Direction.DECREASE): {
        "scenario_id": 1,
        "scenario_type": ScenarioType.MIRROR,
    },

    (Direction.DECREASE, Direction.INCREASE): {
        "scenario_id": 2,
        "scenario_type": ScenarioType.MIRROR,
    },

    (Direction.INCREASE, Direction.INCREASE): {
        "scenario_id": 3,
        "scenario_type": ScenarioType.PARALLEL,
    },

    (Direction.DECREASE, Direction.DECREASE): {
        "scenario_id": 4,
        "scenario_type": ScenarioType.PARALLEL,
    },

    (Direction.INCREASE, Direction.RANGE): {
        "scenario_id": 5,
        "scenario_type": ScenarioType.RANGE_COMPATIBLE_STATE,
    },

    (Direction.DECREASE, Direction.RANGE): {
        "scenario_id": 6,
        "scenario_type": ScenarioType.RANGE_COMPATIBLE_STATE,
    },

    (Direction.RANGE, Direction.INCREASE): {
        "scenario_id": 7,
        "scenario_type": ScenarioType.RANGE_COMPATIBLE_STATE,
    },

    (Direction.RANGE, Direction.DECREASE): {
        "scenario_id": 8,
        "scenario_type": ScenarioType.RANGE_COMPATIBLE_STATE,
    },

    (Direction.RANGE, Direction.RANGE): {
        "scenario_id": 9,
        "scenario_type": ScenarioType.NEUTRAL_FLAT,
    },
}


# ============================================================
# 10. APPROVED NARRATIVE LIBRARY
#
# EXACT APPROVED TEXT — DO NOT REWRITE.
# ============================================================

NARRATIVES = {

1: [
"""رشد ارزش کل بازار نشان‌دهنده ورود نقدینگی و افزایش ارزش دارایی‌هاست. این تغییرات به‌طور طبیعی سهم نسبی تتر را کاهش داده که همزمان یا بخاطر ورود سرمایه از استیبل کوین ها یا ورود سرمایه فیات به دارایی هاست.""",

"""با صعود قیمت دارایی های دیجیتال، ارزش کلی بازار با افزایش رو به رو بوده است. کاهش سهم نسبی تتر در این شرایط، یا از ورود سرمایه به شکل فیات نشات گرفته و یا تبدیل استیبل‌کوین‌ها به دارایی های دیجیتال همراه بوده است.""",

"""رشدِ TOTAL بیانگر افزایش مارکت‌کپ کلی بازار است؛ هم‌زمان کاهش USDT.D نیز افت سهم نسبی تتر را نشان می‌دهد که می‌تواند هم‌زمان حاصلِ ورود مستقیم سرمایه فیات یا جابجایی استیبل‌کوین‌ها به سمت دارایی‌های در حال رشد باشد."""
],

2: [
"""افت ارزش کل بازار نشان‌دهنده خروج سرمایه و کاهش ارزش دارایی‌هاست. رشد سهم نسبی تتر در این شرایط، یا ناشی از تبدیل دارایی‌ها به استیبل‌کوین است و یا به دلیل کوچک شدنِ مخرجِ کل بازار به دلیل تبدیل دارایی‌ها به فیات.""",

"""با کاهش ارزش کل بازار، دارایی‌ها افت ارزش داشته‌اند. بالا رفتن دامیننس تتر بیانگر افزایش سهم نسبی آن است که می‌تواند هم‌زمان حاصلِ تبدیل کوین‌ها به استیبل‌کوین یا خروج سرمایه به شکل فیات باشد.""",

"""نزولِ TOTAL حاکی از انقباض ارزش کلی بازار است؛ هم‌زمان افزایش USDT.D نیز رشد سهم نسبی تتر را بازتاب می‌دهد که به‌طور ریاضی هم از تبدیل دارایی‌ها به استیبل‌کوین‌ها و هم از خروج سرمایه به فیات تأثیر می‌پذیرد."""
],

3: [
"""رشد هم‌زمان ارزش کل بازار و دامیننس تتر نشان‌دهنده ورود پرحجم سرمایه به استیبل کوین ها و دارایی هاست. این وضعیت به‌طور ریاضی نشان می‌دهد سرعت ورود سرمایه(فیات) به تتر به‌اندازه‌ای بالا بوده که با وجود رشد دارایی ها،سهم نسبی دامیننس تتر نیز افزایش یافته است.""",

"""با صعود مارکت‌کپ کل و افزایش سهم تتر، بازار شاهد جریان قدرتمندی از ورود نقدینگی است؛ روندی که در آن تقاضا برای استیبل‌کوین‌ها پابه پای رشد ارزش دارایی‌ها بالا رفته است. دارایی ها در حال رشد اند اما سرعت ورود پول به استیبل کوین تتر عملا بیشتر است.""",

"""افزایش هم‌زمانِ TOTAL و USDT.D گویای این است که اندازه کل بازار بزرگ شده و هم‌زمان حجم تتر نیز با بیشتری تغذیه شده است؛ وضعیتی که ترکیبی از رشد کوین‌ها و بالا ماندن تقاضا برای استیبل‌کوین را بازتاب می‌دهد."""
],

4: [
"""افت هم‌زمان ارزش کل بازار و دامیننس تتر نشان‌دهنده خروج سنگین نقدینگی به سمت فیات است. این وضعیت ریاضی نشان می‌دهد سرعت خروج پول از تتر و تبدیل آن به فیات به‌قدری بالاست که سهم نسبی تتر را هم‌زمان با افت مارکت‌کپِ کل کاهش داده است.""",

"""با کاهش مارکت‌کپ کل، سهم نسبی تتر نیز افت کرده است؛ این پدیده موازی بیانگر افت ارزش دارایی‌ها در کنار سرعتِ بالای خروج سرمایه از استیبل‌کوین‌ها به شکل فیات از چرخه بازار است.""",

"""نزولِ هم‌زمانِ TOTAL و USDT.D حاکی از انقباض ارزش کلی بازار است؛ وضعیتی که نشان می‌دهد هم ارزش دارایی‌های دیجیتال کوچک شده و هم تتر تحت تأثیر سرعتِ خروج مستقیم نقدینگی به فیات کاهش یافته است."""
],

5: [
"""رشد ارزش کل بازار نشان می‌دهد ارزش دارایی‌های دیجیتال در این بازه افزایش یافته است؛ اما سهم نسبی تتر در همان محدوده باقی مانده است. با توجه به اینکه افزایش ارزش دارایی‌ها به‌تنهایی می‌توانست سهم نسبی تتر را کاهش دهد، ثابت ماندن دامیننس تتر در کنار رشد TOTAL می‌تواند نشان دهد که هم‌زمان با ورود نقدینگی به دارایی‌های بازار، نقدینگی جدیدی نیز به تتر وارد شده است؛ به‌گونه‌ای که تغییرات این دو بخش نسبت به یکدیگر متعادل مانده است.""",

"""با افزایش TOTAL، ارزش کلی بازار در این بازه رشد کرده، اما دامیننس تتر کاهش پیدا نکرده و در محدوده خود باقی مانده است. این ترکیب می‌تواند بیانگر آن باشد که رشد دارایی‌ها تنها از جابه‌جایی نقدینگی موجود در تتر به سمت بازار حاصل نشده، بلکه هم‌زمان بخشی از نقدینگی جدید نیز وارد تتر شده است. در نتیجه، افزایش ارزش دارایی‌ها و افزایش احتمالی نقدینگی تتر به شکلی نزدیک به هم اتفاق افتاده‌اند و سهم نسبی تتر تغییر محسوسی نکرده است.""",

"""رشد TOTAL در حالی اتفاق افتاده که USDT.D تقریباً ثابت مانده است. اگر رشد بازار صرفاً ناشی از خروج نقدینگی از تتر و ورود آن به دارایی‌ها بود، انتظار می‌رفت وزن نسبی تتر کاهش پیدا کند؛ بنابراین ثبات دامیننس تتر می‌تواند با ورود هم‌زمان نقدینگی به دارایی‌ها و تتر سازگار باشد. این وضعیت نشان می‌دهد بازار در این بازه هم‌زمان در حال بزرگ‌تر شدن بوده، بدون آنکه وزن نسبی تتر در ساختار بازار تغییر قابل توجهی داشته باشد."""
],

6: [
"""کاهش ارزش کل بازار نشان می‌دهد ارزش دارایی‌های دیجیتال در این بازه افت کرده است؛ اما سهم نسبی تتر در همان محدوده باقی مانده است. با توجه به اینکه کاهش ارزش دارایی‌ها به‌تنهایی می‌توانست سهم نسبی تتر را افزایش دهد، ثابت ماندن دامیننس تتر در کنار افت TOTAL می‌تواند نشان دهد که هم‌زمان با کاهش ارزش دارایی‌های بازار، ارزش بازار تتر هم شاهد خروج سرمایه بوده است. به‌گونه‌ای که تغییرات این دو بخش نسبت به یکدیگر متعادل مانده است.""",

"""با کاهش TOTAL، ارزش کلی بازار در این بازه افت کرده، اما دامیننس تتر افزایش پیدا نکرده و در محدوده خود باقی مانده است. این ترکیب می‌تواند بیانگر آن باشد که کاهش ارزش بازار تنها ناشی از انتقال نقدینگی موجود به تتر نبوده، بلکه هم‌زمان خروج نقدینگی مستقر در تتر رخ داده است؛ بدین ترتیب که سهم نسبی تتر تقریباً ثابت مانده است.""",

"""افت TOTAL در حالی اتفاق افتاده که USDT.D تقریباً ثابت مانده است. اگر کاهش بازار صرفاً ناشی از انتقال نقدینگی از دارایی‌ها به تتر بود، انتظار می‌رفت وزن نسبی تتر افزایش پیدا کند؛ بنابراین ثبات دامیننس تتر می‌تواند با خروج یا کاهش هم‌زمان ارزش دارایی‌ها و خروج سرمایه از تتر سازگار باشد. در نتیجه، افت ارزش کلی بازار بدون افزایش قابل توجه وزن نسبی تتر اتفاق افتاده است."""
],

7: [
"""در این بازه، ارزش کل بازار در محدوده‌ای تقریباً ثابت باقی مانده، اما سهم نسبی تتر افزایش یافته است. از آنجا که TOTAL تغییر معناداری نکرده، افزایش سهم تتر می‌تواند نشان‌دهنده افزایش ارزش یا نقدینگی مستقر در تتر باشد؛ وضعیتی که می‌تواند با ورود نقدینگی جدید به تتر سازگار باشد.""",

"""ثبات نسبی TOTAL در کنار افزایش USDT.D نشان می‌دهد وزن تتر در ساختار کلی بازار بیشتر شده است، بدون آنکه ارزش کل بازار تغییر قابل توجهی داشته باشد. این ترکیب می‌تواند با ورود نقدینگی جدید به تتر همراه باشد؛ یعنی بخشی از سرمایه در شرایط فعلی مستقیم به تتر تزریق شده است.""",

"""وقتی ارزش کل بازار تقریباً ثابت می‌ماند اما دامیننس تتر افزایش پیدا می‌کند، سهم نسبی تتر از کل بازار بیشتر شده است. با توجه به ثبات TOTAL، این تغییر می‌تواند بیانگر افزایش نقدینگی مستقر در تتر باشد؛ شرایطی که ممکن است نشان دهد بخشی از سرمایه در حال ورود به بخش نقدینگی بازار است."""
],

8: [
"""در این بازه، ارزش کل بازار در محدوده‌ای تقریباً ثابت باقی مانده، اما سهم نسبی تتر کاهش یافته است. از آنجا که TOTAL تغییر معناداری نکرده، کاهش سهم تتر می‌تواند نشان‌دهنده کاهش ارزش یا نقدینگی مستقر در تتر باشد؛ وضعیتی که می‌تواند با خروج بخشی از نقدینگی تتر سازگار باشد.""",

"""ثبات نسبی TOTAL در کنار کاهش USDT.D نشان می‌دهد وزن تتر در ساختار کلی بازار کمتر شده است، بدون آنکه ارزش کل بازار تغییر قابل توجهی داشته باشد. این ترکیب می‌تواند با خروج بخشی از نقدینگی از تتر همراه باشد؛ یعنی بخشی از سرمایه‌ای که در تتر قرار داشته، در این بازه از چرخه تتر خارج شده است.""",

"""وقتی ارزش کل بازار تقریباً ثابت می‌ماند اما دامیننس تتر کاهش پیدا می‌کند، سهم نسبی تتر از کل بازار کمتر شده است. با توجه به ثبات TOTAL، این تغییر می‌تواند بیانگر کاهش نقدینگی مستقر در تتر باشد؛ شرایطی که ممکن است ناشی از انتقال بخشی از سرمایه به فیات باشد."""
],

9: [
"""در این بازه، ارزش کل بازار و سهم نسبی تتر هر دو در محدوده‌ای رنج حرکت کرده‌اند؛ یعنی با وجود نوسانات داخل بازه، هیچ‌کدام جهت غالب و پایداری پیدا نکرده‌اند. در چنین شرایطی ممکن است در بخش‌های مختلف بازار جابه‌جایی و نوسان میان دارایی‌ها و تتر اتفاق بیفتد، اما این تغییرات در مقیاس کلی بازار به اندازه‌ای نبوده که TOTAL یا سهم نسبی تتر را از محدوده رنج خارج کند. بنابراین این وضعیت بیشتر نشان می‌دهد که در کلیت بازار هنوز تصمیم جهت‌دار غالبی شکل نگرفته و تعادل نسبی میان ارزش دارایی‌ها و سهم تتر حفظ شده است.""",

"""رنج بودن هم‌زمان TOTAL و USDT.D نشان می‌دهد که در این بازه، با وجود حرکت‌ها و نوسانات درون بازار، اندازه کلی بازار و سهم نسبی تتر در مجموع در محدوده خود باقی مانده‌اند. اگر حرکت جهت‌دار پایداری در بخش بزرگی از دارایی‌های بازار شکل می‌گرفت، اثر آن می‌توانست در TOTAL نمایان شود؛ بنابراین رنج ماندن TOTAL نشان می‌دهد که چنین حرکتی در مقیاس کلی بازار غالب نشده است. از طرف دیگر، رنج ماندن دامیننس تتر نیز نشان می‌دهد وزن نسبی تتر نسبت به کل بازار تغییر جهت‌دار مشخصی نکرده است. در نتیجه، ترکیب این دو شاخص بیانگر بازاری است که در بازه مورد بررسی هنوز تصمیم مشخصی برای غلبه یک جهت بر کلیت بازار شکل نگرفته است.""",

"""وقتی TOTAL و USDT.D هر دو رنج هستند، این وضعیت به معنی بی‌حرکتی بازار یا نبود ورود و خروج سرمایه نیست. ممکن است در داخل بازه، میان دارایی‌ها و تتر جابه‌جایی نقدینگی و نوسانات مقطعی رخ داده باشد؛ اما این حرکات در مجموع نتوانسته‌اند اندازه کلی بازار یا سهم نسبی تتر را وارد یک روند مشخص کنند. در چنین شرایطی، حتی حرکت برخی دارایی‌ها نیز تا زمانی که اثر تجمیعی آن‌ها جهت کلی TOTAL را تغییر ندهد، به‌تنهایی بیانگر تغییر جهت بازار نیست. بنابراین پیام اصلی این ترکیب، نبود جهت غالب و پایدار در کلیت بازار و باقی ماندن بازار در یک تعادل نسبی در بازه مورد بررسی است."""
],
}


# ============================================================
# 11. SCENARIO VALIDATION
# ============================================================

def validate_scenario_matrix() -> Dict[str, Any]:

    expected = {
        1: (Direction.INCREASE, Direction.DECREASE),
        2: (Direction.DECREASE, Direction.INCREASE),
        3: (Direction.INCREASE, Direction.INCREASE),
        4: (Direction.DECREASE, Direction.DECREASE),
        5: (Direction.INCREASE, Direction.RANGE),
        6: (Direction.DECREASE, Direction.RANGE),
        7: (Direction.RANGE, Direction.INCREASE),
        8: (Direction.RANGE, Direction.DECREASE),
        9: (Direction.RANGE, Direction.RANGE),
    }

    checks = []

    for scenario_id, pair in expected.items():

        actual = SCENARIO_MATRIX.get(pair)

        checks.append({
            "scenario_id": scenario_id,
            "pair": [
                pair[0].value,
                pair[1].value,
            ],
            "exists": actual is not None,
            "correct_id": (
                actual is not None
                and actual["scenario_id"] == scenario_id
            ),
        })

    narrative_checks = {
        scenario_id: len(
            NARRATIVES.get(scenario_id, [])
        ) == 3
        for scenario_id in expected
    }

    all_matrix_ok = all(
        x["exists"] and x["correct_id"]
        for x in checks
    )

    all_narratives_ok = all(
        narrative_checks.values()
    )

    return {
        "matrix_ok": all_matrix_ok,
        "narratives_ok": all_narratives_ok,
        "total_scenarios": len(SCENARIO_MATRIX),
        "total_narrative_patterns": sum(
            len(v)
            for v in NARRATIVES.values()
        ),
        "details": checks,
        "narrative_details": narrative_checks,
    }


# ============================================================
# 12. TIMEFRAME REGRESSION VALIDATION
# ============================================================

def validate_timeframes() -> Dict[str, Any]:

    tests = [
        "1m",
        "3m",
        "5m",
        "15m",
        "30m",
        "1h",
        "2h",
        "4h",
        "6h",
        "8h",
        "12h",
        "1D",
        "1W",
        "1M",
    ]

    results = []

    for tf in tests:

        result = parse_timeframe(tf)

        # ----------------------------------------------------
        # Only minute/hour timeframes below 1D get warning.
        # Day/week/month do not.
        # ----------------------------------------------------

        raw_unit = tf.strip()[-1]

        if raw_unit in ("m", "h", "H"):

            expected_warning = (
                "SUB_DAILY_WARNING"
            )

        else:

            expected_warning = None

        results.append({
            "timeframe": tf,
            "valid": result["valid"],
            "warning": result["warning"],
            "expected_warning": expected_warning,
            "passed": (
                result["valid"] is True
                and result["warning"]
                == expected_warning
            ),
        })

    return {
        "passed": all(
            x["passed"]
            for x in results
        ),
        "details": results,
    }


# ============================================================
# 13. SCENARIO ENGINE
# ============================================================

@dataclass
class ScenarioResult:
    scenario_id: int
    scenario_type: str

    total_classification: str
    usdt_d_classification: str

    selected_pattern: str
    pattern_index: int

    timeframe: str
    timeframe_valid: bool
    timeframe_warning: Optional[str]

    analysis_range: Dict[str, Any]

    total_range_type: str
    total_range_confidence: str
    total_range_score: float

    usdt_range_type: str
    usdt_range_confidence: str
    usdt_range_score: float

    audit: Dict[str, Any]


def run_u07(
    total_values: List[float],
    usdt_values: List[float],
    timeframe: str,
    analysis_range: Any,
    *,
    total_highs: Optional[List[float]] = None,
    total_lows: Optional[List[float]] = None,
    total_closes: Optional[List[float]] = None,
    usdt_highs: Optional[List[float]] = None,
    usdt_lows: Optional[List[float]] = None,
    usdt_closes: Optional[List[float]] = None,
    pattern_index: int = 1,
    config: RangeConfig = DEFAULT_RANGE_CONFIG,
) -> ScenarioResult:

    # --------------------------------------------------------
    # USER RANGE IS AUTHORITATIVE
    # --------------------------------------------------------

    validated_range = validate_user_range(
        analysis_range
    )

    # --------------------------------------------------------
    # TIMEFRAME VALIDATION
    # --------------------------------------------------------

    timeframe_result = parse_timeframe(
        timeframe
    )

    if not timeframe_result["valid"]:

        raise ValueError(
            "TECHNICAL_VALIDATION_ERROR: "
            "Unsupported or malformed timeframe."
        )

    # --------------------------------------------------------
    # UPSTREAM DATA VALIDATION
    # --------------------------------------------------------

    validate_series(
        total_values,
        "TOTAL",
    )

    validate_series(
        usdt_values,
        "USDT.D",
    )

    # --------------------------------------------------------
    # RANGE ANALYSIS
    # --------------------------------------------------------

    total_analysis = analyze_range(
        total_values,
        highs=total_highs,
        lows=total_lows,
        closes=total_closes,
        config=config,
    )

    usdt_analysis = analyze_range(
        usdt_values,
        highs=usdt_highs,
        lows=usdt_lows,
        closes=usdt_closes,
        config=config,
    )

    # --------------------------------------------------------
    # AMBIGUITY SAFETY
    #
    # No forced direction is created from ambiguous Range data.
    # --------------------------------------------------------

    if (
        total_analysis.ambiguous
        and total_analysis.direction == Direction.RANGE
        and total_analysis.confidence == RangeConfidence.LOW
    ):
        pass

    if (
        usdt_analysis.ambiguous
        and usdt_analysis.direction == Direction.RANGE
        and usdt_analysis.confidence == RangeConfidence.LOW
    ):
        pass

    # --------------------------------------------------------
    # EXACT 9-SCENARIO MAPPING
    # --------------------------------------------------------

    pair = (
        total_analysis.direction,
        usdt_analysis.direction,
    )

    if pair not in SCENARIO_MATRIX:

        raise RuntimeError(
            "No canonical U07 scenario mapping exists for this pair."
        )

    mapping = SCENARIO_MATRIX[pair]

    scenario_id = mapping["scenario_id"]

    scenario_type = (
        mapping["scenario_type"].value
    )

    # --------------------------------------------------------
    # NARRATIVE SELECTION
    # --------------------------------------------------------

    if pattern_index not in (1, 2, 3):

        raise ValueError(
            "pattern_index must be 1, 2, or 3."
        )

    approved_patterns = NARRATIVES[
        scenario_id
    ]

    selected_pattern = approved_patterns[
        pattern_index - 1
    ]

    # --------------------------------------------------------
    # AUDIT
    # --------------------------------------------------------

    audit = {

        "engine": "U07",

        "engine_version": "V3.1",

        "timestamp_utc": datetime.now(
            timezone.utc
        ).isoformat(),

        "user_range_authoritative": True,

        "analysis_range": validated_range,

        "timeframe_validation": timeframe_result,

        "total": {

            "classification":
                total_analysis.direction.value,

            "range_type":
                total_analysis.range_type.value,

            "confidence":
                total_analysis.confidence.value,

            "score":
                total_analysis.confidence_score,

            "net_change_pct":
                total_analysis.net_change_pct,

            "amplitude_pct":
                total_analysis.amplitude_pct,

            "path_length_pct":
                total_analysis.path_length_pct,

            "switch_count":
                total_analysis.switch_count,

            "close_dispersion_pct":
                total_analysis.close_dispersion_pct,

            "di_plus":
                total_analysis.di_plus,

            "di_minus":
                total_analysis.di_minus,

            "adx":
                total_analysis.adx,

            "ambiguous":
                total_analysis.ambiguous,

            "reasons":
                total_analysis.reasons,
        },

        "usdt_d": {

            "classification":
                usdt_analysis.direction.value,

            "range_type":
                usdt_analysis.range_type.value,

            "confidence":
                usdt_analysis.confidence.value,

            "score":
                usdt_analysis.confidence_score,

            "net_change_pct":
                usdt_analysis.net_change_pct,

            "amplitude_pct":
                usdt_analysis.amplitude_pct,

            "path_length_pct":
                usdt_analysis.path_length_pct,

            "switch_count":
                usdt_analysis.switch_count,

            "close_dispersion_pct":
                usdt_analysis.close_dispersion_pct,

            "di_plus":
                usdt_analysis.di_plus,

            "di_minus":
                usdt_analysis.di_minus,

            "adx":
                usdt_analysis.adx,

            "ambiguous":
                usdt_analysis.ambiguous,

            "reasons":
                usdt_analysis.reasons,
        },

        "scenario": {

            "scenario_id":
                scenario_id,

            "scenario_type":
                scenario_type,

            "total":
                total_analysis.direction.value,

            "usdt_d":
                usdt_analysis.direction.value,
        },

        "approved_narrative": {

            "pattern_index":
                pattern_index,

            "total_patterns_available":
                3,
        },

        "execution_locks": {

            "trading": False,

            "orders": False,

            "strategy": False,

            "portfolio_actions": False,
        },
    }

    return ScenarioResult(

        scenario_id=scenario_id,

        scenario_type=scenario_type,

        total_classification=
            total_analysis.direction.value,

        usdt_d_classification=
            usdt_analysis.direction.value,

        selected_pattern=
            selected_pattern,

        pattern_index=
            pattern_index,

        timeframe=
            timeframe_result["normalized"],

        timeframe_valid=
            timeframe_result["valid"],

        timeframe_warning=
            timeframe_result["warning"],

        analysis_range=
            validated_range,

        total_range_type=
            total_analysis.range_type.value,

        total_range_confidence=
            total_analysis.confidence.value,

        total_range_score=
            total_analysis.confidence_score,

        usdt_range_type=
            usdt_analysis.range_type.value,

        usdt_range_confidence=
            usdt_analysis.confidence.value,

        usdt_range_score=
            usdt_analysis.confidence_score,

        audit=audit,
    )


# ============================================================
# 14. FULL U07 REGRESSION TEST
# ============================================================

def run_u07_regression_tests() -> Dict[str, Any]:

    # --------------------------------------------------------
    # MATRIX TEST
    # --------------------------------------------------------

    matrix = validate_scenario_matrix()

    # --------------------------------------------------------
    # TIMEFRAME TEST
    # --------------------------------------------------------

    timeframe = validate_timeframes()

    # --------------------------------------------------------
    # ZERO MOVEMENT TEST
    # --------------------------------------------------------

    zero_series = [
        100.0,
        102.0,
        98.0,
        101.0,
        100.0,
    ]

    zero_result = analyze_range(
        zero_series,
        config=DEFAULT_RANGE_CONFIG,
    )

    zero_movement_ok = (
        zero_result.direction
        == Direction.RANGE
    )

    # --------------------------------------------------------
    # RANGE INTERNAL-MOVEMENT TEST
    # --------------------------------------------------------

    choppy_series = [
        100,
        104,
        99,
        103,
        98,
        102,
        100,
    ]

    choppy_result = analyze_range(
        choppy_series,
        config=DEFAULT_RANGE_CONFIG,
    )

    internal_structure_checked = (
        choppy_result.switch_count > 0
    )

    # --------------------------------------------------------
    # NINE SCENARIO PAIR TEST
    # --------------------------------------------------------

    scenario_pairs = [

        (
            Direction.INCREASE,
            Direction.DECREASE,
        ),

        (
            Direction.DECREASE,
            Direction.INCREASE,
        ),

        (
            Direction.INCREASE,
            Direction.INCREASE,
        ),

        (
            Direction.DECREASE,
            Direction.DECREASE,
        ),

        (
            Direction.INCREASE,
            Direction.RANGE,
        ),

        (
            Direction.DECREASE,
            Direction.RANGE,
        ),

        (
            Direction.RANGE,
            Direction.INCREASE,
        ),

        (
            Direction.RANGE,
            Direction.DECREASE,
        ),

        (
            Direction.RANGE,
            Direction.RANGE,
        ),
    ]

    nine_scenarios_ok = all(
        pair in SCENARIO_MATRIX
        for pair in scenario_pairs
    )

    # --------------------------------------------------------
    # EXECUTION LOCK TEST
    # --------------------------------------------------------

    locks_ok = (
        TRADING_ENABLED is False
        and ORDERS_ENABLED is False
        and STRATEGY_ENABLED is False
        and PORTFOLIO_ACTIONS_ENABLED is False
    )

    # --------------------------------------------------------
    # OVERALL
    # --------------------------------------------------------

    overall = (

        matrix["matrix_ok"]

        and matrix["narratives_ok"]

        and timeframe["passed"]

        and zero_movement_ok

        and internal_structure_checked

        and nine_scenarios_ok

        and locks_ok
    )

    return {

        "status":
            "PASS"
            if overall
            else "FAIL",

        "matrix":
            matrix,

        "timeframe":
            timeframe,

        "zero_movement_edge_case": {

            "passed":
                zero_movement_ok,

            "classification":
                zero_result.direction.value,
        },

        "internal_structure": {

            "passed":
                internal_structure_checked,

            "switch_count":
                choppy_result.switch_count,
        },

        "nine_scenarios": {

            "passed":
                nine_scenarios_ok,

            "count":
                len(scenario_pairs),
        },

        "execution_locks": {

            "passed":
                locks_ok,
        },
    }


# ============================================================
# 15. RUN CANONICAL VALIDATION
# ============================================================

U07_VALIDATION = (
    run_u07_regression_tests()
)

print("=" * 68)

print(
    "KITCHEN ASSISTANT V3.1 — U07 VALIDATION"
)

print("=" * 68)

print(
    json.dumps(
        U07_VALIDATION,
        ensure_ascii=False,
        indent=2,
    )
)

assert (
    U07_VALIDATION["status"]
    == "PASS"
)

print("=" * 68)

print(
    "U07 TECHNICAL VALIDATION : PASS"
)

print(
    "TRADING                  : DISABLED"
)

print(
    "ORDERS                   : DISABLED"
)

print(
    "STRATEGY                 : DISABLED"
)

print(
    "SCENARIO MATRIX          : 9/9"
)

print(
    "NARRATIVE PATTERNS       : 27/27"
)

print(
    "TIMEFRAME RULE           : VALIDATED"
)

print(
    "RANGE ENGINE             : VALIDATED"
)

print(
    "ZERO-MOVEMENT EDGE       : VALIDATED"
)

print("=" * 68)


# ============================================================
# OPTIONAL USAGE
# ============================================================
#
# IMPORTANT:
# The analysis_range below is ONLY an example.
#
# In the real project, the user's selected range must be passed
# unchanged.
#
# Example:
#
# result = run_u07(
#     total_values=TOTAL_SERIES,
#     usdt_values=USDTD_SERIES,
#     timeframe=USER_TIMEFRAME,
#     analysis_range={
#         "start": USER_RANGE_START,
#         "end": USER_RANGE_END,
#         "custom": True,
#         "label": USER_RANGE_LABEL,
#     },
#     pattern_index=1,
# )
#
# print(
#     json.dumps(
#         asdict(result),
#         ensure_ascii=False,
#         indent=2,
#     )
# )
#
# ============================================================
# END U07 IMPLEMENTATION
# ============================================================
```

```python

# ============================================================
# KITCHEN ASSISTANT V3.1 — CELL 08
#
# BTC + BTC.D CONTEXT & RELATIVE-MOVEMENT ENGINE
#
# STATUS:
#   SCENARIOS  = LOCKED
#   NARRATIVES = LOCKED
#   27 PATTERNS = LOCKED
#   ROADMAP     = MASTER ROADMAP
#
# PURPOSE:
#   Analyze → Classify → Rank → Explain
#
# HARD SAFETY:
#   Trading          = DISABLED
#   Orders           = DISABLED
#   Strategy         = DISABLED
#   Portfolio Action = DISABLED
#
# Cell 08 produces Analysis / Context only.
# ============================================================

from __future__ import annotations

from dataclasses import dataclass, asdict
from enum import Enum
from typing import Any, Dict, List, Optional, Tuple
from datetime import datetime, timezone
import math
import json


# ============================================================
# 0. HARD EXECUTION LOCKS
# ============================================================

TRADING_ENABLED = False
ORDERS_ENABLED = False
STRATEGY_ENABLED = False
PORTFOLIO_ACTIONS_ENABLED = False

assert TRADING_ENABLED is False
assert ORDERS_ENABLED is False
assert STRATEGY_ENABLED is False
assert PORTFOLIO_ACTIONS_ENABLED is False


# ============================================================
# 1. ENUMS
# ============================================================

class Direction(str, Enum):
    INCREASE = "INCREASE"
    DECREASE = "DECREASE"
    RANGE = "RANGE"


class Context(str, Enum):
    BULLISH = "BULLISH"
    BEARISH = "BEARISH"
    RANGE = "RANGE"


class RelativeDirection(str, Enum):
    RELATIVE_STRENGTH = "RELATIVE_STRENGTH"
    RELATIVE_WEAKNESS = "RELATIVE_WEAKNESS"
    RELATIVE_NEUTRAL = "RELATIVE_NEUTRAL"


# ============================================================
# 2. CONFIGURATION
# ============================================================

@dataclass(frozen=True)
class Cell08Config:

    top_n: int = 10
    strong_movers_n: int = 10
    relative_movers_n: int = 10

    min_valid_price_change_pct: float = 0.0

    relative_strength_epsilon_pct: float = 0.0

    require_finite_values: bool = True

    include_btc_in_top_assets: bool = True
    include_btc_in_strong_movers: bool = True

    volume_required: bool = True


CELL08_CONFIG = Cell08Config()


# ============================================================
# 3. SAFE NUMERIC HELPERS
# ============================================================

def _finite(value: Any) -> bool:

    try:
        value = float(value)
    except (TypeError, ValueError):

        return False

    return math.isfinite(value)


def _safe_float(value: Any) -> Optional[float]:

    if value is None:
        return None

    try:

        value = float(value)

    except (TypeError, ValueError):

        return None

    if not math.isfinite(value):

        return None

    return value


def _safe_pct_change(
    start: Any,
    end: Any,
) -> Optional[float]:

    start = _safe_float(start)
    end = _safe_float(end)

    if start is None or end is None:
        return None

    if start == 0:
        return None

    return ((end - start) / start) * 100.0


def _normalize_direction(
    direction: Any,
) -> Direction:

    if isinstance(direction, Direction):

        return direction

    value = str(direction).upper().strip()

    aliases = {

        "INCREASE": Direction.INCREASE,
        "UP": Direction.INCREASE,
        "BULLISH": Direction.INCREASE,
        "RISING": Direction.INCREASE,

        "DECREASE": Direction.DECREASE,
        "DOWN": Direction.DECREASE,
        "BEARISH": Direction.DECREASE,
        "FALLING": Direction.DECREASE,

        "RANGE": Direction.RANGE,
        "FLAT": Direction.RANGE,
        "SIDEWAYS": Direction.RANGE,

    }

    if value not in aliases:

        raise ValueError(
            f"Unsupported direction: {direction}"
        )

    return aliases[value]


# ============================================================
# 4. CONTEXT CLASSIFICATION
# ============================================================

def classify_context(
    btc_direction: Any,
    btc_d_direction: Any,
) -> Context:

    btc_direction = _normalize_direction(
        btc_direction
    )

    btc_d_direction = _normalize_direction(
        btc_d_direction
    )

    if btc_direction == Direction.INCREASE:

        return Context.BULLISH

    if btc_direction == Direction.DECREASE:

        return Context.BEARISH

    return Context.RANGE


# ============================================================
# 5. CANONICAL 9-SCENARIO MATRIX
# ============================================================

SCENARIO_MATRIX: Dict[
    Tuple[Direction, Direction],
    Dict[str, Any]
] = {

    (
        Direction.INCREASE,
        Direction.INCREASE,
    ): {

        "scenario_id": 1,
        "scenario_type": "BTC_UP_BTC_D_UP",

    },

    (
        Direction.INCREASE,
        Direction.DECREASE,
    ): {

        "scenario_id": 2,
        "scenario_type": "BTC_UP_BTC_D_DOWN",

    },

    (
        Direction.INCREASE,
        Direction.RANGE,
    ): {

        "scenario_id": 3,
        "scenario_type": "BTC_UP_BTC_D_RANGE",

    },

    (
        Direction.DECREASE,
        Direction.INCREASE,
    ): {

        "scenario_id": 4,
        "scenario_type": "BTC_DOWN_BTC_D_UP",

    },

    (
        Direction.DECREASE,
        Direction.DECREASE,
    ): {

        "scenario_id": 5,
        "scenario_type": "BTC_DOWN_BTC_D_DOWN",

    },

    (
        Direction.DECREASE,
        Direction.RANGE,
    ): {

        "scenario_id": 6,
        "scenario_type": "BTC_DOWN_BTC_D_RANGE",

    },

    (
        Direction.RANGE,
        Direction.INCREASE,
    ): {

        "scenario_id": 7,
        "scenario_type": "BTC_RANGE_BTC_D_UP",

    },

    (
        Direction.RANGE,
        Direction.DECREASE,
    ): {

        "scenario_id": 8,
        "scenario_type": "BTC_RANGE_BTC_D_DOWN",

    },

    (
        Direction.RANGE,
        Direction.RANGE,
    ): {

        "scenario_id": 9,
        "scenario_type": "BTC_RANGE_BTC_D_RANGE",

    },

}


assert len(SCENARIO_MATRIX) == 9


# ============================================================
# 6. LOCKED 27 NARRATIVE PATTERNS
#
# DO NOT REWRITE.
# DO NOT PARAPHRASE.
# DO NOT GENERATE NEW NARRATIVES.
# ============================================================

NARRATIVES: Dict[int, List[Dict[str, str]]] = {

    1: [

        {
            "pattern": "Pattern 1",
            "title": "🟢 BTC Strength",
            "text":
            "رشد هم‌زمان BTC و BTC.D نشان می‌دهد بیت‌کوین در حرکت فعلی قدرت بیشتری در ساختار بازار دارد. با این حال، دارایی‌هایی که بیش از BTC رشد کرده‌اند حرکت صعودی قوی‌تری نشان می‌دهند و باید برجسته شوند؛ در کنار آن، BTC همچنان یکی از دارایی‌های اصلی مورد بررسی و معیار مقایسه باقی می‌ماند."
        },

        {
            "pattern": "Pattern 2",
            "title": "🟢 BTC Strength / Relative Movers",
            "text":
            "افزایش BTC همراه با افزایش BTC.D بیانگر تقویت جایگاه BTC در بازار است؛ اما این ساختار مانع از وجود دارایی‌هایی نمی‌شود که حرکت صعودی بیشتری از BTC داشته‌اند. بررسی آن‌ها قوی‌ترین حرکات صعودی را مشخص می‌کند، در حالی که BTC همچنان معیار اصلی مقایسه و یکی از دارایی‌های مورد بررسی باقی می‌ماند."
        },

        {
            "pattern": "Pattern 3",
            "title": "🟢 BTC Dominance Strength",
            "text":
            "وقتی BTC و BTC.D هر دو صعودی هستند، قدرت نسبی BTC در ساختار بازار افزایش یافته است. در این وضعیت، تمرکز روی دارایی‌هایی است که بیشترین رشد را نسبت به BTC و در مقیاس کلی بازار داشته‌اند؛ با این حال، BTC همچنان در کنار این دارایی‌ها بررسی می‌شود و معیار اصلی مقایسه باقی می‌ماند."
        },

    ],

    2: [

        {
            "pattern": "Pattern 1",
            "title": "🟢 Altcoin Strength",
            "text":
            "رشد BTC در کنار کاهش BTC.D نشان می‌دهد سهم بیت‌کوین از بازار در حال کاهش است و بخشی از قدرت حرکت به سمت آلت‌کوین‌ها رفته است. در این شرایط، بررسی شاخص‌های ساختار آلت‌کوین‌ها می‌تواند جهت حرکت و سمت قدرت اصلی بازار را روشن‌تر کند و در ادامه، دارایی‌هایی که بیش از BTC رشد کرده‌اند می‌توانند حرکت‌های صعودی قوی‌تری را نشان دهند؛ BTC همچنان معیار اصلی مقایسه باقی می‌ماند."
        },

        {
            "pattern": "Pattern 2",
            "title": "🟢 Altcoin Relative Strength",
            "text":
            "وقتی BTC در حال رشد است اما BTC.D کاهش می‌یابد، قدرت بازار فقط در بیت‌کوین متمرکز نمانده و آلت‌کوین‌ها نیز توانسته‌اند سهم بیشتری از حرکت را به خود اختصاص دهند. در این وضعیت، شاخص‌های ساختار آلت‌کوین‌ها می‌توانند برای مشخص شدن جهت و میزان مشارکت این بخش از بازار کمک‌کننده باشند و سپس دارایی‌هایی که بیشترین رشد را نسبت به BTC داشته‌اند برجسته می‌شوند؛ BTC همچنان معیار اصلی سنجش عملکرد آن‌ها باقی می‌ماند."
        },

        {
            "pattern": "Pattern 3",
            "title": "🟢 Broad Altcoin Participation",
            "text":
            "افزایش BTC همراه با افت BTC.D نشان می‌دهد هم‌زمان با رشد بیت‌کوین، قدرت حرکت در بخش دیگری از بازار نیز افزایش یافته است. بررسی شاخص‌های ساختار آلت‌کوین‌ها می‌تواند نشان دهد این قدرت تا چه اندازه در این بخش از بازار گسترش پیدا کرده و در ادامه، دارایی‌هایی که بیشترین رشد را نسبت به BTC داشته‌اند مشخص می‌شوند؛ BTC معیار اصلی مقایسه این حرکات باقی می‌ماند."
        },

    ],

    3: [

        {
            "pattern": "Pattern 1",
            "title": "🟢 Broad Market Strength",
            "text":
            "بیت‌کوین در حال رشد است و BTC.D تقریباً در محدوده خود باقی مانده است؛ یعنی در مقیاس کلی، حرکت BTC و بازار آلت‌کوین‌ها از نظر سرعت تغییر سهم بازار فاصله زیادی از یکدیگر نگرفته‌اند. با این حال، ممکن است بعضی دارایی‌ها حرکت قوی‌تری داشته باشند. بنابراین بررسی جفت‌های BTC اهمیت ویژه‌ای دارد تا دارایی‌هایی که نسبت به BTC رشد بیشتری کرده‌اند مشخص شوند؛ در حالی که خود BTC نیز همچنان می‌تواند در این ساختار حرکت مناسبی داشته باشد."
        },

        {
            "pattern": "Pattern 2",
            "title": "🟢 Broad Market Strength",
            "text":
            "رشد BTC همراه با رنج ماندن BTC.D نشان می‌دهد حرکت بیت‌کوین و بازار در مجموع ریتم نسبتاً نزدیکی داشته‌اند و سهم BTC تغییر جهت‌دار محسوسی نکرده است. برای پیدا کردن تفاوت‌های واقعی میان دارایی‌ها، باید جفت‌های BTC بررسی شوند؛ زیرا ممکن است برخی آلت‌کوین‌ها با وجود این تعادل کلی، حرکت صعودی بیشتری نسبت به BTC ثبت کرده باشند. BTC نیز در این شرایط همچنان بخشی از ساختار قابل‌بررسی است."
        },

        {
            "pattern": "Pattern 3",
            "title": "🟢 Broad Market Strength",
            "text":
            "وقتی BTC صعودی است اما BTC.D رنج می‌زند، می‌توان گفت در مقیاس کلی، رشد BTC و حرکت سایر بخش‌های بازار تقریباً با ریتمی نزدیک به یکدیگر اتفاق افتاده است. با این حال، این تعادل کلی مانع از شکل‌گیری حرکات متفاوت در بعضی دارایی‌ها نمی‌شود. به همین دلیل جفت‌های BTC باید به‌دقت بررسی شوند تا دارایی‌هایی که نسبت به BTC حرکت قوی‌تری داشته‌اند شناسایی شوند؛ در کنار آن، خود BTC نیز می‌تواند همچنان حرکت قابل‌توجهی داشته باشد."
        },

    ],

    4: [

        {
            "pattern": "Pattern 1",
            "title": "🔴 Altcoin Weakness",
            "text":
            "کاهش BTC در کنار افزایش BTC.D نشان می‌دهد سهم بیت‌کوین از بازار در حال افزایش است و بخشی از فشار نزولی بازار با شدت بیشتری در آلت‌کوین‌ها دیده می‌شود. در این شرایط، بررسی شاخص‌های ساختار آلت‌کوین‌ها می‌تواند جهت حرکت و سمت اصلی ضعف بازار را روشن‌تر کند و در ادامه، دارایی‌هایی که بیش از BTC کاهش داشته‌اند می‌توانند حرکات نزولی قوی‌تری را نشان دهند؛ BTC همچنان معیار اصلی مقایسه باقی می‌ماند."
        },

        {
            "pattern": "Pattern 2",
            "title": "🔴 Altcoin Relative Weakness",
            "text":
            "وقتی BTC در حال کاهش است اما BTC.D افزایش می‌یابد، فشار نزولی فقط در بیت‌کوین متمرکز نمانده و آلت‌کوین‌ها نیز نسبت به BTC عملکرد ضعیف‌تری داشته‌اند. در این وضعیت، شاخص‌های ساختار آلت‌کوین‌ها می‌توانند برای روشن‌تر شدن جهت و میزان ضعف این بخش از بازار کمک‌کننده باشند و سپس دارایی‌هایی که بیشترین ریزش را نسبت به BTC داشته‌اند شناسایی می‌شوند؛ BTC همچنان معیار اصلی سنجش عملکرد آن‌ها باقی می‌ماند."
        },

        {
            "pattern": "Pattern 3",
            "title": "🔴 Broad Altcoin Weakness",
            "text":
            "کاهش BTC همراه با افزایش BTC.D نشان می‌دهد هم‌زمان با افت بیت‌کوین، فشار نزولی در بخش دیگری از بازار نیز می‌تواند شدت بیشتری پیدا کرده باشد. بررسی شاخص‌های ساختار آلت‌کوین‌ها می‌تواند نشان دهد این ضعف تا چه اندازه در این بخش از بازار گسترش پیدا کرده و در ادامه، دارایی‌هایی که بیشترین کاهش را نسبت به BTC داشته‌اند مشخص می‌شوند؛ BTC معیار اصلی مقایسه این حرکات باقی می‌ماند."
        },

    ],

    5: [

        {
            "pattern": "Pattern 1",
            "title": "🟡 BTC Weakness / Alt Relative Strength",
            "text":
            "کاهش هم‌زمان BTC و BTC.D نشان می‌دهد بیت‌کوین در حال افت است و سهم آن از بازار نیز کاهش یافته است. در این شرایط، خود BTC همچنان یکی از دارایی‌های مورد بررسی است و هم‌زمان عملکرد سایر دارایی‌ها نسبت به حرکت BTC سنجیده می‌شود تا مشخص شود کدام‌یک ریزش شدیدتری داشته‌اند."
        },

        {
            "pattern": "Pattern 2",
            "title": "🟡 Altcoin Relative Performance",
            "text":
            "وقتی BTC نزولی است و BTC.D نیز کاهش می‌یابد، خود BTC همچنان در بررسی قرار دارد و عملکرد سایر دارایی‌ها در مقایسه با حرکت آن سنجیده می‌شود. این مقایسه کمک می‌کند دارایی‌هایی که بیشترین حرکت نزولی را نسبت به BTC داشته‌اند مشخص شوند."
        },

        {
            "pattern": "Pattern 3",
            "title": "🟡 Relative Performance Under BTC Weakness",
            "text":
            "افت BTC همراه با کاهش BTC.D نشان می‌دهد بیت‌کوین در حال کاهش است و هم‌زمان وزن آن در بازار نیز کمتر شده است. در این وضعیت، بررسی خود BTC در کنار سایر دارایی‌ها و مقایسه حرکت آن‌ها با BTC می‌تواند مشخص کند کدام دارایی‌ها بیشترین ریزش را تجربه کرده‌اند."
        },

    ],

    6: [

        {
            "pattern": "Pattern 1",
            "title": "🔴 Broad Market Weakness",
            "text":
            "کاهش BTC در حالی که BTC.D در محدوده رنج باقی مانده است، نشان می‌دهد بیت‌کوین در حال افت است اما سهم آن از بازار تغییر جهت‌دار مشخصی نکرده؛ یعنی حرکت کلی بازار در این بازه، به‌طور معمول با ریتمی نزدیک به حرکت BTC همراه بوده است. در این شرایط، خود BTC همچنان یکی از دارایی‌های مورد بررسی است و مقایسه مستقیم سایر دارایی‌ها با BTC کمک می‌کند مشخص شود کدام‌یک ریزش بیشتری داشته‌اند."
        },

        {
            "pattern": "Pattern 2",
            "title": "🔴 Relative Weakness Under BTC Decline",
            "text":
            "وقتی BTC نزولی است اما BTC.D رنج می‌زند، تغییر محسوسی در سهم بیت‌کوین از بازار شکل نگرفته و در نتیجه، سرعت حرکت BTC و بخش بزرگی از بازار می‌تواند در مجموع به یکدیگر نزدیک باشد. بنابراین خود BTC در کنار سایر دارایی‌ها بررسی می‌شود و مقایسه عملکرد آن‌ها با BTC می‌تواند دارایی‌هایی را که حرکت نزولی شدیدتری داشته‌اند مشخص کند."
        },

        {
            "pattern": "Pattern 3",
            "title": "🔴 Coin/BTC Relative Weakness",
            "text":
            "افت BTC همراه با رنج بودن BTC.D نشان می‌دهد در حالی که بیت‌کوین کاهش یافته، نسبت سهم آن به کل بازار تغییر جهت‌دار قابل‌توجهی نداشته است. در چنین شرایطی، مقایسه جفت‌های BTC اهمیت ویژه‌ای پیدا می‌کند تا مشخص شود کدام دارایی‌ها نسبت به BTC حرکت نزولی بیشتری داشته‌اند؛ در کنار این مقایسه، خود BTC نیز به‌عنوان یکی از دارایی‌های مورد بررسی باقی می‌ماند."
        },

    ],

    7: [

        {
            "pattern": "Pattern 1",
            "title": "Altcoin Weakness During BTC Range",
            "text":
            "رنج بودن BTC همراه با افزایش BTC.D نشان می‌دهد در حالی که قیمت بیت‌کوین حرکت جهت‌دار مشخصی ندارد، سهم آن از بازار افزایش یافته است؛ بنابراین بخشی از آلت‌کوین‌ها نسبت به BTC عملکرد ضعیف‌تری داشته‌اند. بررسی شاخص‌های ساختار آلت‌کوین‌ها می‌تواند جهت کلی این ضعف را مشخص‌تر کند و سپس مقایسه جفت‌های BTC به شناسایی دارایی‌هایی کمک می‌کند که شدیدترین حرکت نزولی را نسبت به BTC داشته‌اند."
        },

        {
            "pattern": "Pattern 2",
            "title": "BTC Dominance Gain",
            "text":
            "ثبات نسبی BTC در کنار افزایش BTC.D نشان می‌دهد بدون اینکه خود بیت‌کوین حرکت صعودی قابل‌توجهی داشته باشد، وزن آن در ساختار بازار افزایش یافته است. این ترکیب بیانگر ضعف نسبی بخشی از آلت‌کوین‌هاست؛ بنابراین شاخص‌های ساختار آلت‌کوین‌ها می‌توانند تصویر کلی بازار را روشن‌تر کنند و بررسی جفت‌های BTC برای پیدا کردن دارایی‌هایی که حرکت نزولی شدیدی نسبت به BTC داشته‌اند اهمیت پیدا می‌کند."
        },

        {
            "pattern": "Pattern 3",
            "title": "Relative Altcoin Weakness",
            "text":
            "وقتی BTC در محدوده رنج قرار دارد اما BTC.D افزایش می‌یابد، سهم بیت‌کوین از ساختار بازار بیشتر شده است؛ در نتیجه بخشی از آلت‌کوین‌ها در همین بازه نسبت به BTC عملکرد ضعیف‌تری داشته‌اند. شاخص‌های ساختار آلت‌کوین‌ها می‌توانند به تأیید این وضعیت کمک کنند و مقایسه جفت‌های BTC نیز برای شناسایی دارایی‌هایی که حرکت نزولی شدیدی نسبت به BTC داشته‌اند، اهمیت ویژه‌ای دارد."
        },

    ],

    8: [

        {
            "pattern": "Pattern 1",
            "title": "Altcoin Strength During BTC Range",
            "text":
            "رنج بودن BTC همراه با کاهش BTC.D نشان می‌دهد در حالی که قیمت بیت‌کوین حرکت جهت‌دار مشخصی ندارد، سهم آن از بازار کاهش یافته است؛ بنابراین بخشی از آلت‌کوین‌ها نسبت به BTC عملکرد صعودی داشته‌اند. شاخص‌های ساختار آلت‌کوین‌ها می‌توانند جهت این حرکت را تأیید کنند و مقایسه جفت‌های BTC برای شناسایی دارایی‌هایی که حرکت صعودی نسبت به BTC داشته‌اند اهمیت ویژه‌ای پیدا می‌کند."
        },

        {
            "pattern": "Pattern 2",
            "title": "BTC Dominance Decline",
            "text":
            "ثبات نسبی BTC در کنار کاهش BTC.D نشان می‌دهد بدون اینکه خود بیت‌کوین حرکت جهت‌دار قابل‌توجهی داشته باشد، وزن آن در ساختار بازار کاهش یافته است. این وضعیت نشان می‌دهد بخشی از آلت‌کوین‌ها نسبت به BTC قدرت بیشتری گرفته‌اند؛ بنابراین بررسی شاخص‌های ساختار آلت‌کوین‌ها و سپس مقایسه جفت‌های BTC می‌تواند دارایی‌هایی را مشخص کند که حرکت صعودی نسبت به BTC داشته‌اند."
        },

        {
            "pattern": "Pattern 3",
            "title": "Relative Altcoin Strength",
            "text":
            "وقتی BTC در محدوده رنج قرار دارد اما BTC.D کاهش می‌یابد، سهم بیت‌کوین از ساختار بازار کمتر شده است؛ در نتیجه بخشی از آلت‌کوین‌ها در همین بازه نسبت به BTC عملکرد صعودی داشته‌اند. شاخص‌های ساختار آلت‌کوین‌ها می‌توانند تصویر کلی این حرکت را روشن‌تر کنند و مقایسه جفت‌های BTC نیز برای شناسایی دارایی‌هایی که حرکت صعودی نسبت به BTC داشته‌اند، اهمیت دارد."
        },

    ],

    9: [

        {
            "pattern": "Pattern 1",
            "title": "Balanced Market",
            "text":
            "رنج بودن هم‌زمان BTC و BTC.D نشان می‌دهد نه قیمت بیت‌کوین و نه سهم آن از بازار، جهت مشخصی پیدا نکرده‌اند؛ بنابراین در ساختار کلی بازار برتری مشخصی میان BTC و آلت‌کوین‌ها دیده نمی‌شود. در این شرایط، جفت‌های BTC اهمیت ویژه‌ای پیدا می‌کنند و می‌توانند دارایی‌هایی را نشان دهند که برخلاف فضای رنج کلی بازار، حرکت صعودی یا نزولی مشخصی داشته‌اند."
        },

        {
            "pattern": "Pattern 2",
            "title": "BTC Relative Movement",
            "text":
            "وقتی BTC و BTC.D هر دو در محدوده رنج قرار دارند، معمولاً سرعت حرکت کلی بازار و بیت‌کوین به یکدیگر نزدیک است و شاخص‌های ساختار آلت‌کوین‌ها نقش تعیین‌کننده‌ای ندارند. بنابراین بررسی جفت‌های BTC می‌تواند کمک کند دارایی‌هایی که از این وضعیت رنج خارج شده‌اند و حرکت مشخصی نسبت به BTC داشته‌اند شناسایی شوند؛ چه این حرکت صعودی باشد و چه نزولی."
        },

        {
            "pattern": "Pattern 3",
            "title": "Relative Strength in a Flat Market",
            "text":
            "رنج بودن BTC در کنار رنج بودن BTC.D نشان می‌دهد ساختار کلی بازار نیز جهت غالب مشخصی ندارد. با این حال، این وضعیت به معنی بی‌حرکتی تمام دارایی‌ها نیست؛ ممکن است برخی کوین‌ها در برابر BTC حرکت صعودی یا نزولی داشته باشند. بنابراین در این سناریو، بررسی جفت‌های BTC مهم‌ترین مسیر برای پیدا کردن همین حرکات متفاوت در میان فضای رنج بازار است."
        },

    ],

}


# ============================================================
# 7. NARRATIVE VALIDATION
# ============================================================

def validate_narratives() -> Dict[str, Any]:

    assert set(NARRATIVES.keys()) == set(range(1, 10))

    total_patterns = 0

    for scenario_id in range(1, 10):

        patterns = NARRATIVES[scenario_id]

        assert len(patterns) == 3

        for pattern in patterns:

            assert "pattern" in pattern
            assert "title" in pattern
            assert "text" in pattern

            assert isinstance(
                pattern["text"],
                str,
            )

            assert len(
                pattern["text"].strip()
            ) > 0

            total_patterns += 1

    assert total_patterns == 27

    return {

        "scenarios": 9,
        "patterns": total_patterns,
        "status": "LOCKED",

    }


NARRATIVE_VALIDATION = validate_narratives()

assert NARRATIVE_VALIDATION["patterns"] == 27


# ============================================================
# 8. SCENARIO RESOLUTION
# ============================================================

def resolve_scenario(
    btc_direction: Any,
    btc_d_direction: Any,
) -> Dict[str, Any]:

    btc_direction = _normalize_direction(
        btc_direction
    )

    btc_d_direction = _normalize_direction(
        btc_d_direction
    )

    key = (
        btc_direction,
        btc_d_direction,
    )

    if key not in SCENARIO_MATRIX:

        raise RuntimeError(
            "CELL08_ERROR: "
            "No canonical scenario exists."
        )

    scenario = dict(
        SCENARIO_MATRIX[key]
    )

    scenario["btc_direction"] = (
        btc_direction.value
    )

    scenario["btc_d_direction"] = (
        btc_d_direction.value
    )

    scenario["patterns"] = NARRATIVES[
        scenario["scenario_id"]
    ]

    return scenario


# ============================================================
# 9. PATTERN SELECTION
# ============================================================

def select_pattern(
    scenario_id: int,
    pattern_index: int = 1,
) -> Dict[str, str]:

    if scenario_id not in NARRATIVES:

        raise ValueError(
            f"Unknown scenario_id: {scenario_id}"
        )

    if pattern_index not in (1, 2, 3):

        raise ValueError(
            "pattern_index must be 1, 2, or 3."
        )

    return dict(
        NARRATIVES[
            scenario_id
        ][pattern_index - 1]
    )


# ============================================================
# 10. BTC PAIR RELATIVE PERFORMANCE
# ============================================================

def calculate_relative_btc_performance(
    asset_change_pct: Any,
    btc_change_pct: Any,
) -> Optional[float]:

    asset_change_pct = _safe_float(
        asset_change_pct
    )

    btc_change_pct = _safe_float(
        btc_change_pct
    )

    if (
        asset_change_pct is None
        or btc_change_pct is None
    ):

        return None

    return (
        asset_change_pct
        - btc_change_pct
    )


def classify_relative_performance(
    asset_change_pct: Any,
    btc_change_pct: Any,
    epsilon_pct: float = 0.0,
) -> RelativeDirection:

    relative = calculate_relative_btc_performance(
        asset_change_pct,
        btc_change_pct,
    )

    if relative is None:

        return RelativeDirection.RELATIVE_NEUTRAL

    if relative > epsilon_pct:

        return RelativeDirection.RELATIVE_STRENGTH

    if relative < -epsilon_pct:

        return RelativeDirection.RELATIVE_WEAKNESS

    return RelativeDirection.RELATIVE_NEUTRAL


# ============================================================
# 11. ASSET NORMALIZATION
# ============================================================

def normalize_asset(
    asset: Dict[str, Any],
) -> Optional[Dict[str, Any]]:

    if not isinstance(asset, dict):

        return None

    symbol = asset.get(
        "symbol",
        asset.get("asset"),
    )

    if symbol is None:

        return None

    symbol = str(symbol).strip()

    if not symbol:

        return None

    price = _safe_float(
        asset.get("price")
    )

    change_pct = _safe_float(
        asset.get(
            "change_pct",
            asset.get(
                "pct_change",
                asset.get("percent_change")
            ),
        )
    )

    volume = _safe_float(
        asset.get("volume")
    )

    btc_pair_change_pct = _safe_float(
        asset.get(
            "btc_pair_change_pct",
            asset.get(
                "btc_pair_pct_change"
            ),
        )
    )

    result = dict(asset)

    result["symbol"] = symbol
    result["price"] = price
    result["change_pct"] = change_pct
    result["volume"] = volume
    result["btc_pair_change_pct"] = (
        btc_pair_change_pct
    )

    return result


# ============================================================
# 12. RELATIVE PERFORMANCE ENGINE
# ============================================================

def enrich_relative_performance(
    assets: List[Dict[str, Any]],
    btc_change_pct: Optional[float],
) -> List[Dict[str, Any]]:

    output = []

    for raw_asset in assets:

        asset = normalize_asset(
            raw_asset
        )

        if asset is None:
            continue

        asset_change = asset.get(
            "change_pct"
        )

        relative_change = (
            calculate_relative_btc_performance(
                asset_change,
                btc_change_pct,
            )
        )

        relative_direction = (
            classify_relative_performance(
                asset_change,
                btc_change_pct,
                CELL08_CONFIG
                .relative_strength_epsilon_pct,
            )
        )

        asset[
            "relative_btc_performance_pct"
        ] = relative_change

        asset[
            "relative_direction"
        ] = relative_direction.value

        output.append(asset)

    return output


# ============================================================
# 13. DIRECTIONAL MOVEMENT RANKING
# ============================================================

def rank_strong_movers(
    assets: List[Dict[str, Any]],
    market_context: Context,
    n: int,
) -> List[Dict[str, Any]]:

    valid_assets = []

    for asset in assets:

        change = _safe_float(
            asset.get("change_pct")
        )

        if change is None:
            continue

        valid_assets.append(asset)

    if market_context == Context.BULLISH:

        ranked = sorted(
            valid_assets,
            key=lambda x: (
                _safe_float(
                    x.get("change_pct")
                )
                if _safe_float(
                    x.get("change_pct")
                ) is not None
                else -math.inf
            ),
            reverse=True,
        )

    elif market_context == Context.BEARISH:

        ranked = sorted(
            valid_assets,
            key=lambda x: (
                _safe_float(
                    x.get("change_pct")
                )
                if _safe_float(
                    x.get("change_pct")
                ) is not None
                else math.inf
            ),
        )

    else:

        ranked = sorted(
            valid_assets,
            key=lambda x: abs(
                _safe_float(
                    x.get("change_pct")
                )
            ),
            reverse=True,
        )

    return ranked[:n]


# ============================================================
# 14. RELATIVE BTC RANKING
# ============================================================

def rank_relative_movers(
    assets: List[Dict[str, Any]],
    market_context: Context,
    n: int,
) -> List[Dict[str, Any]]:

    valid = []

    for asset in assets:

        relative = _safe_float(
            asset.get(
                "relative_btc_performance_pct"
            )
        )

        if relative is None:
            continue

        valid.append(asset)

    if market_context == Context.BEARISH:

        ranked = sorted(
            valid,
            key=lambda x: (
                _safe_float(
                    x.get(
                        "relative_btc_performance_pct"
                    )
                )
            ),
        )

    else:

        ranked = sorted(
            valid,
            key=lambda x: (
                _safe_float(
                    x.get(
                        "relative_btc_performance_pct"
                    )
                )
            ),
            reverse=True,
        )

    return ranked[:n]


# ============================================================
# 15. TOP 10 ASSETS
# ============================================================

def rank_top_assets(
    assets: List[Dict[str, Any]],
    market_context: Context,
    n: int = 10,
) -> List[Dict[str, Any]]:

    valid = []

    for asset in assets:

        change = _safe_float(
            asset.get("change_pct")
        )

        volume = _safe_float(
            asset.get("volume")
        )

        if change is None:
            continue

        if volume is None:

            volume = 0.0

        asset = dict(asset)

        asset["volume"] = volume

        valid.append(asset)

    if market_context == Context.BULLISH:

        valid.sort(
            key=lambda x: (
                _safe_float(
                    x.get("change_pct")
                ),
                _safe_float(
                    x.get("volume")
                ),
            ),
            reverse=True,
        )

    elif market_context == Context.BEARISH:

        valid.sort(
            key=lambda x: (
                _safe_float(
                    x.get("change_pct")
                ),
                _safe_float(
                    x.get("volume")
                ),
            )
        )

    else:

        valid.sort(
            key=lambda x: (
                abs(
                    _safe_float(
                        x.get("change_pct")
                    )
                ),
                _safe_float(
                    x.get("volume")
                ),
            ),
            reverse=True,
        )

    return valid[:n]


# ============================================================
# 16. OPPOSITE-DIRECTION DETECTION
# ============================================================

def is_opposite_direction(
    btc_direction: Direction,
    btc_d_direction: Direction,
) -> bool:

    return (

        (
            btc_direction == Direction.INCREASE
            and
            btc_d_direction == Direction.DECREASE
        )

        or

        (
            btc_direction == Direction.DECREASE
            and
            btc_d_direction == Direction.INCREASE
        )

    )


# ============================================================
# 17. ALTCOIN MARKET STRUCTURE
#
# These indicators are CONTEXT ONLY.
# They never replace asset-level movement.
# ============================================================

def build_altcoin_structure_context(
    btc_direction: Direction,
    btc_d_direction: Direction,
    total2: Optional[Dict[str, Any]] = None,
    total3: Optional[Dict[str, Any]] = None,
    others_d: Optional[Dict[str, Any]] = None,
) -> Dict[str, Any]:

    opposite = is_opposite_direction(
        btc_direction,
        btc_d_direction,
    )

    if not opposite:

        return {

            "enabled": False,

            "reason":
                "BTC and BTC.D are not in opposite directions.",

            "total2": total2,
            "total3": total3,
            "others_d": others_d,

        }

    return {

        "enabled": True,

        "reason":
            "BTC and BTC.D are in opposite directions.",

        "total2": total2,
        "total3": total3,
        "others_d": others_d,

    }


# ============================================================
# 18. VOLUME OUTPUT
# ============================================================

def volume_record(
    asset: Dict[str, Any],
) -> Dict[str, Any]:

    return {

        "symbol":
            asset.get("symbol"),

        "direction":
            asset.get("direction"),

        "price":
            asset.get("price"),

        "change_pct":
            asset.get("change_pct"),

        "volume":
            asset.get("volume"),

        "relative_btc_performance_pct":
            asset.get(
                "relative_btc_performance_pct"
            ),

    }


def ensure_volume_fields(
    assets: List[Dict[str, Any]],
) -> List[Dict[str, Any]]:

    output = []

    for asset in assets:

        asset = dict(asset)

        asset["volume"] = _safe_float(
            asset.get("volume")
        )

        if (
            CELL08_CONFIG.volume_required
            and asset["volume"] is None
        ):

            asset["volume_status"] = (
                "UNAVAILABLE"
            )

        else:

            asset["volume_status"] = (
                "AVAILABLE"
            )

        output.append(asset)

    return output


# ============================================================
# 19. ASSET DIRECTION
# ============================================================

def add_absolute_direction(
    assets: List[Dict[str, Any]],
) -> List[Dict[str, Any]]:

    output = []

    for asset in assets:

        asset = dict(asset)

        change = _safe_float(
            asset.get("change_pct")
        )

        if change is None:

            asset["direction"] = "UNKNOWN"

        elif change > 0:

            asset["direction"] = "INCREASE"

        elif change < 0:

            asset["direction"] = "DECREASE"

        else:

            asset["direction"] = "RANGE"

        output.append(asset)

    return output


# ============================================================
# 20. STRONG MOVERS — ABSOLUTE MOVEMENT
# ============================================================

def build_strong_movers(
    assets: List[Dict[str, Any]],
    market_context: Context,
) -> List[Dict[str, Any]]:

    movers = rank_strong_movers(
        assets,
        market_context,
        CELL08_CONFIG.strong_movers_n,
    )

    return ensure_volume_fields(
        movers
    )


# ============================================================
# 21. RELATIVE MOVERS — BTC PAIRS
# ============================================================

def build_relative_movers(
    assets: List[Dict[str, Any]],
    market_context: Context,
) -> List[Dict[str, Any]]:

    movers = rank_relative_movers(
        assets,
        market_context,
        CELL08_CONFIG.relative_movers_n,
    )

    return ensure_volume_fields(
        movers
    )


# ============================================================
# 22. SCENARIO-SPECIFIC RANKING LOGIC
# ============================================================

def build_scenario_ranking(
    assets: List[Dict[str, Any]],
    btc_direction: Direction,
    btc_d_direction: Direction,
) -> Dict[str, Any]:

    context = classify_context(
        btc_direction,
        btc_d_direction,
    )

    strong_movers = build_strong_movers(
        assets,
        context,
    )

    relative_movers = build_relative_movers(
        assets,
        context,
    )

    top_assets = rank_top_assets(
        assets,
        context,
        CELL08_CONFIG.top_n,
    )

    return {

        "context":
            context.value,

        "strong_movers":
            strong_movers,

        "relative_movers":
            relative_movers,

        "top_10_assets":
            ensure_volume_fields(
                top_assets
            ),

    }


# ============================================================
# 23. FINAL ANALYSIS OBJECT
# ============================================================

@dataclass
class Cell08Result:

    engine: str

    version: str

    timestamp_utc: str

    scenario_id: int

    scenario_type: str

    btc_direction: str

    btc_d_direction: str

    context: str

    opposite_direction: bool

    selected_pattern: Dict[str, str]

    top_10_assets: List[Dict[str, Any]]

    strong_movers: List[Dict[str, Any]]

    relative_movers: List[Dict[str, Any]]

    altcoin_structure: Dict[str, Any]

    audit: Dict[str, Any]


# ============================================================
# 24. MAIN CELL 08 ENGINE
# ============================================================

def run_cell_08(
    *,
    btc_direction: Any,
    btc_d_direction: Any,
    btc_change_pct: Optional[float],
    assets: List[Dict[str, Any]],
    pattern_index: int = 1,
    total2: Optional[Dict[str, Any]] = None,
    total3: Optional[Dict[str, Any]] = None,
    others_d: Optional[Dict[str, Any]] = None,
) -> Cell08Result:

    # --------------------------------------------------------
    # INPUT VALIDATION
    # --------------------------------------------------------

    btc_direction = _normalize_direction(
        btc_direction
    )

    btc_d_direction = _normalize_direction(
        btc_d_direction
    )

    if not isinstance(assets, list):

        raise TypeError(
            "CELL08_ERROR: assets must be a list."
        )

    btc_change_pct = _safe_float(
        btc_change_pct
    )

    # --------------------------------------------------------
    # SCENARIO
    # --------------------------------------------------------

    scenario = resolve_scenario(
        btc_direction,
        btc_d_direction,
    )

    scenario_id = scenario[
        "scenario_id"
    ]

    scenario_type = scenario[
        "scenario_type"
    ]

    # --------------------------------------------------------
    # ASSET PREPARATION
    # --------------------------------------------------------

    prepared_assets = add_absolute_direction(
        assets
    )

    prepared_assets = enrich_relative_performance(
        prepared_assets,
        btc_change_pct,
    )

    prepared_assets = ensure_volume_fields(
        prepared_assets
    )

    # --------------------------------------------------------
    # RANKING
    # --------------------------------------------------------

    ranking = build_scenario_ranking(
        prepared_assets,
        btc_direction,
        btc_d_direction,
    )

    # --------------------------------------------------------
    # ALTCOIN STRUCTURE
    # --------------------------------------------------------

    altcoin_structure = (
        build_altcoin_structure_context(
            btc_direction,
            btc_d_direction,
            total2=total2,
            total3=total3,
            others_d=others_d,
        )
    )

    # --------------------------------------------------------
    # PATTERN
    # --------------------------------------------------------

    selected_pattern = select_pattern(
        scenario_id,
        pattern_index,
    )

    # --------------------------------------------------------
    # AUDIT
    # --------------------------------------------------------

    opposite = is_opposite_direction(
        btc_direction,
        btc_d_direction,
    )

    audit = {

        "engine":
            "CELL_08",

        "engine_version":
            "V3.1",

        "execution":
            "ANALYSIS_ONLY",

        "trading_enabled":
            False,

        "orders_enabled":
            False,

        "strategy_enabled":
            False,

        "portfolio_actions_enabled":
            False,

        "scenarios_locked":
            True,

        "narratives_locked":
            True,

        "narrative_pattern_count":
            27,

        "btc_is_benchmark":
            True,

        "absolute_direction_preserved":
            True,

        "relative_performance_is_complementary":
            True,

        "btc_pair_analysis_enabled":
            True,

        "altcoin_structure_used_only_when_required":
            True,

        "opposite_direction":
            opposite,

        "volume_required":
            CELL08_CONFIG.volume_required,

    }

    return Cell08Result(

        engine="CELL_08",

        version="V3.1",

        timestamp_utc=
            datetime.now(
                timezone.utc
            ).isoformat(),

        scenario_id=
            scenario_id,

        scenario_type=
            scenario_type,

        btc_direction=
            btc_direction.value,

        btc_d_direction=
            btc_d_direction.value,

        context=
            ranking["context"],

        opposite_direction=
            opposite,

        selected_pattern=
            selected_pattern,

        top_10_assets=
            ranking["top_10_assets"],

        strong_movers=
            ranking["strong_movers"],

        relative_movers=
            ranking["relative_movers"],

        altcoin_structure=
            altcoin_structure,

        audit=
            audit,

    )


# ============================================================
# 25. SERIALIZATION
# ============================================================

def cell08_to_dict(
    result: Cell08Result,
) -> Dict[str, Any]:

    return asdict(
        result
    )


def cell08_to_json(
    result: Cell08Result,
) -> str:

    return json.dumps(
        cell08_to_dict(result),
        ensure_ascii=False,
        indent=2,
    )


# ============================================================
# 26. VALIDATION TESTS
# ============================================================

def run_cell08_regression_tests() -> Dict[str, Any]:

    # --------------------------------------------------------
    # 9 SCENARIOS
    # --------------------------------------------------------

    assert len(
        SCENARIO_MATRIX
    ) == 9

    # --------------------------------------------------------
    # 27 PATTERNS
    # --------------------------------------------------------

    narrative_validation = (
        validate_narratives()
    )

    assert (
        narrative_validation["patterns"]
        == 27
    )

    # --------------------------------------------------------
    # OPPOSITE-DIRECTION CASES
    # --------------------------------------------------------

    assert is_opposite_direction(
        Direction.INCREASE,
        Direction.DECREASE,
    ) is True

    assert is_opposite_direction(
        Direction.DECREASE,
        Direction.INCREASE,
    ) is True

    assert is_opposite_direction(
        Direction.INCREASE,
        Direction.INCREASE,
    ) is False

    assert is_opposite_direction(
        Direction.DECREASE,
        Direction.DECREASE,
    ) is False

    # --------------------------------------------------------
    # ABSOLUTE ≠ RELATIVE
    # --------------------------------------------------------

    relative = calculate_relative_btc_performance(
        -1.0,
        -3.0,
    )

    assert relative == 2.0

    assert (
        classify_relative_performance(
            -1.0,
            -3.0,
        )
        == RelativeDirection.RELATIVE_STRENGTH
    )

    # --------------------------------------------------------
    # NO FALSE BULLISH LABEL
    # --------------------------------------------------------

    test_asset = {

        "symbol": "TEST",

        "price": 100.0,

        "change_pct": -1.0,

        "volume": 1000.0,

    }

    test_enriched = (
        enrich_relative_performance(
            [test_asset],
            -3.0,
        )
    )

    assert (
        test_enriched[0]["direction"]
        if "direction" in test_enriched[0]
        else "DECREASE"
    ) != "INCREASE"

    # --------------------------------------------------------
    # RANGE DOES NOT MEAN ALL ASSETS RANGE
    # --------------------------------------------------------

    range_assets = [

        {
            "symbol": "A",
            "price": 100,
            "change_pct": 5,
            "volume": 1000,
        },

        {
            "symbol": "B",
            "price": 100,
            "change_pct": -4,
            "volume": 900,
        },

    ]

    range_ranked = (
        build_strong_movers(
            range_assets,
            Context.RANGE,
        )
    )

    assert len(
        range_ranked
    ) == 2

    # --------------------------------------------------------
    # SUCCESS
    # --------------------------------------------------------

    return {

        "engine":
            "CELL_08",

        "status":
            "PASS",

        "scenario_count":
            len(SCENARIO_MATRIX),

        "pattern_count":
            narrative_validation[
                "patterns"
            ],

        "execution":
            "ANALYSIS_ONLY",

    }


CELL08_REGRESSION = (
    run_cell08_regression_tests()
)


# ============================================================
# 27. FINAL ARCHITECTURE ASSERTIONS
# ============================================================

assert TRADING_ENABLED is False
assert ORDERS_ENABLED is False
assert STRATEGY_ENABLED is False
assert PORTFOLIO_ACTIONS_ENABLED is False

assert len(
    SCENARIO_MATRIX
) == 9

assert (
    NARRATIVE_VALIDATION["patterns"]
    == 27
)

assert (
    CELL08_REGRESSION["status"]
    == "PASS"
)


# ============================================================
# 28. READY STATE
# ============================================================

CELL08_STATUS = {

    "cell":
        "08",

    "engine":
        "BTC + BTC.D CONTEXT & RELATIVE-MOVEMENT ENGINE",

    "version":
        "V3.1",

    "scenarios":
        9,

    "patterns":
        27,

    "trading":
        False,

    "orders":
        False,

    "strategy":
        False,

    "portfolio_actions":
        False,

    "status":
        "READY",

}


print(
    "============================================================"
)

print(
    "KITCHEN ASSISTANT V3.1 — CELL 08"
)

print(
    "BTC + BTC.D CONTEXT & RELATIVE-MOVEMENT ENGINE"
)

print(
    "============================================================"
)

print(
    "SCENARIOS          : 9/9 LOCKED"
)

print(
    "NARRATIVE PATTERNS : 27/27 LOCKED"
)

print(
    "TRADING            : DISABLED"
)

print(
    "ORDERS             : DISABLED"
)

print(
    "STRATEGY           : DISABLED"
)

print(
    "PORTFOLIO ACTIONS  : DISABLED"
)

print(
    "EXECUTION MODE     : ANALYSIS ONLY"
)

print(
    "REGRESSION TEST    : PASS"
)

print(
    "STATUS             : READY"
)

print(
    "============================================================"
)
```

```python

# ============================================================
# KITCHEN ASSISTANT V3.1
# CANONICAL MASTER PORTABLE SNAPSHOT
# U01 -> U08
#
# FINAL SNAPSHOT CELL
#
# IMPORTANT:
#   - SOURCE CODE IS NEVER EMBEDDED
#   - IPYNB remains the code source of truth
#   - Snapshot contains portable STATE / HISTORY / AUDIT /
#     LEDGER / ARTIFACT / VALIDATION / HANDOFF metadata
#   - No trading
#   - No orders
#   - No strategy execution
#   - No portfolio mutation
#
# Run ONCE after U08.
# ============================================================

from __future__ import annotations

import os
import json
import hashlib
import inspect
from pathlib import Path
from datetime import datetime, timezone


# ============================================================
# 1. CONSTANTS
# ============================================================

SNAPSHOT_ENGINE = "KITCHEN_ASSISTANT"
SNAPSHOT_VERSION = "V3.1"
SNAPSHOT_SCHEMA = "MASTER_PORTABLE_SNAPSHOT_V4"

NOTEBOOK_NAME = "Kitchen Assistant v3.1from 1 to 8.ipynb"

SNAPSHOT_ROOT_CANDIDATES = [
    Path("/content/snapshots/master"),
    Path("/content/drive/MyDrive/Colab Notebooks/snapshots/master"),
    Path.cwd() / "snapshots" / "master",
]

timestamp_utc = datetime.now(timezone.utc).isoformat()


# ============================================================
# 2. SAFE SERIALIZATION
#
# State/metadata only.
# Never serializes function source.
# ============================================================

def _safe(value, depth=0, max_depth=7):
    if depth > max_depth:
        return {
            "__type__": type(value).__name__,
            "__truncated__": True,
        }

    if value is None:
        return None

    if isinstance(value, (str, int, bool)):
        return value

    if isinstance(value, float):
        if value != value:
            return None
        if value in (float("inf"), float("-inf")):
            return None
        return value

    if isinstance(value, Path):
        return str(value)

    if isinstance(value, dict):
        out = {}
        for k, v in value.items():
            out[str(k)] = _safe(v, depth + 1, max_depth)
        return out

    if isinstance(value, (list, tuple, set)):
        return [
            _safe(v, depth + 1, max_depth)
            for v in list(value)
        ]

    # Functions are represented by identity only.
    if callable(value):
        return {
            "__type__": type(value).__name__,
            "__callable__": True,
            "__source_embedded__": False,
            "name": getattr(value, "__name__", None),
        }

    if hasattr(value, "__dict__"):
        attrs = {}
        try:
            for k, v in vars(value).items():
                if k.startswith("__"):
                    continue
                if callable(v):
                    attrs[str(k)] = {
                        "__type__": type(v).__name__,
                        "__callable__": True,
                        "__source_embedded__": False,
                        "name": getattr(v, "__name__", None),
                    }
                else:
                    attrs[str(k)] = _safe(
                        v,
                        depth + 1,
                        max_depth,
                    )
        except Exception as exc:
            attrs = {
                "__serialization_error__": type(exc).__name__
            }

        return {
            "__type__": type(value).__name__,
            "attributes": attrs,
        }

    return {
        "__type__": type(value).__name__,
    }


# ============================================================
# 3. SHA256 HELPERS
# ============================================================

def _sha256_bytes(data: bytes) -> str:
    return hashlib.sha256(data).hexdigest()


def _sha256_json(obj) -> str:
    payload = json.dumps(
        obj,
        ensure_ascii=False,
        sort_keys=True,
        separators=(",", ":"),
    ).encode("utf-8")

    return _sha256_bytes(payload)


# ============================================================
# 4. DISCOVER NOTEBOOK
#
# We identify the IPYNB.
# We NEVER copy its source into the snapshot.
# ============================================================

notebook_path = None

search_roots = [
    Path("/content"),
    Path("/content/drive/MyDrive"),
    Path("/mnt/data"),
    Path.cwd(),
]

for root in search_roots:
    if not root.exists():
        continue

    try:
        direct = root / NOTEBOOK_NAME
        if direct.is_file():
            notebook_path = direct
            break
    except Exception:
        pass


# ============================================================
# 5. NOTEBOOK IDENTITY
#
# Failure to locate the physical IPYNB is NOT fatal to the
# runtime snapshot because Colab frequently executes a notebook
# without exposing its .ipynb as a filesystem file.
#
# Runtime evidence is therefore used when necessary.
# ============================================================

notebook_identity = {
    "name": NOTEBOOK_NAME,
    "filesystem_path": str(notebook_path)
    if notebook_path
    else None,
    "filesystem_available": notebook_path is not None,
    "sha256": None,
    "source_code_embedded": False,
    "source_of_truth": "IPYNB",
}

if notebook_path is not None:
    try:
        notebook_bytes = notebook_path.read_bytes()
        notebook_identity["sha256"] = _sha256_bytes(
            notebook_bytes
        )
        notebook_identity["byte_size"] = len(
            notebook_bytes
        )
    except Exception as exc:
        notebook_identity["read_error"] = {
            "type": type(exc).__name__,
            "message": str(exc),
        }


# ============================================================
# 6. U01 -> U08 RUNTIME EVIDENCE
#
# U01-U07 are allowed to have no runtime objects.
# Their authoritative source remains the IPYNB.
# U08 is verified from runtime objects.
# ============================================================

unit_evidence = {}

for n in range(1, 9):
    unit = f"U{n:02d}"

    unit_evidence[unit] = {
        "declared_in_scope": True,
        "source_of_truth": "IPYNB",
        "source_code_embedded": False,
        "runtime_evidence": [],
    }


# Explicit U08 evidence
u08_candidates = [
    "SCENARIO_MATRIX",
    "NARRATIVES",
    "CELL08_CONFIG",
    "run_cell_08",
    "cell08_to_dict",
    "cell08_to_json",
    "run_cell08_regression_tests",
]

for name in u08_candidates:
    if name in globals():
        unit_evidence["U08"]["runtime_evidence"].append(name)


# ============================================================
# 7. EXECUTION SAFETY
# ============================================================

runtime_flags = {
    "trading_enabled": bool(
        globals().get("TRADING_ENABLED", False)
    ),
    "orders_enabled": bool(
        globals().get("ORDERS_ENABLED", False)
    ),
    "strategy_enabled": bool(
        globals().get("STRATEGY_ENABLED", False)
    ),
    "portfolio_actions_enabled": bool(
        globals().get(
            "PORTFOLIO_ACTIONS_ENABLED",
            False,
        )
    ),
}

execution = {
    **runtime_flags,
    "execution_mode": "ANALYSIS_ONLY",
    "execution_allowed": False,
    "live_order_creation": False,
    "portfolio_mutation": False,
}

safety_pass = (
    runtime_flags["trading_enabled"] is False
    and runtime_flags["orders_enabled"] is False
    and runtime_flags["strategy_enabled"] is False
    and runtime_flags["portfolio_actions_enabled"] is False
)


# ============================================================
# 8. CONFIGURATION
# ============================================================

cell08_config = globals().get("CELL08_CONFIG")

configuration = {
    "cell08_config_exists": cell08_config is not None,
    "cell08_config": _safe(cell08_config)
    if cell08_config is not None
    else None,
}


# ============================================================
# 9. SCENARIOS
# ============================================================

scenario_matrix = globals().get("SCENARIO_MATRIX")

scenarios = {
    "exists": scenario_matrix is not None,
    "count": (
        len(scenario_matrix)
        if scenario_matrix is not None
        else 0
    ),
    "scenario_ids": [],
}

if isinstance(scenario_matrix, dict):
    scenarios["scenario_ids"] = list(
        scenario_matrix.keys()
    )

elif isinstance(scenario_matrix, (list, tuple)):
    scenarios["scenario_ids"] = list(
        range(1, len(scenario_matrix) + 1)
    )

scenario_pass = (
    scenarios["exists"]
    and scenarios["count"] == 9
)


# ============================================================
# 10. NARRATIVES
# ============================================================

narratives_obj = globals().get("NARRATIVES")

narratives = {
    "exists": narratives_obj is not None,
    "scenario_count": 0,
    "total_patterns": 0,
}

if isinstance(narratives_obj, dict):
    narratives["scenario_count"] = len(
        narratives_obj
    )

    try:
        narratives["total_patterns"] = sum(
            len(v)
            for v in narratives_obj.values()
            if hasattr(v, "__len__")
        )
    except Exception:
        narratives["total_patterns"] = 0

narrative_pass = (
    narratives["exists"]
    and narratives["scenario_count"] == 9
    and narratives["total_patterns"] == 27
)


# ============================================================
# 11. FUNCTION REGISTRY
# ============================================================

MASTER_FUNCTIONS = [
    "_finite",
    "_safe_float",
    "_safe_pct_change",
    "_normalize_direction",
    "classify_context",
    "resolve_scenario",
    "select_pattern",
    "calculate_relative_btc_performance",
    "classify_relative_performance",
    "normalize_asset",
    "enrich_relative_performance",
    "rank_strong_movers",
    "rank_relative_movers",
    "rank_top_assets",
    "is_opposite_direction",
    "build_altcoin_structure_context",
    "volume_record",
    "ensure_volume_fields",
    "add_absolute_direction",
    "build_strong_movers",
    "build_relative_movers",
    "build_scenario_ranking",
    "run_cell_08",
    "cell08_to_dict",
    "cell08_to_json",
    "run_cell08_regression_tests",
]

function_registry = {}

for name in MASTER_FUNCTIONS:
    obj = globals().get(name)

    function_registry[name] = {
        "exists": name in globals(),
        "callable": callable(obj),
        "type": type(obj).__name__
        if obj is not None
        else None,
        "source_embedded": False,
    }

function_pass = (
    len(function_registry) == 26
    and all(
        item["exists"] and item["callable"]
        for item in function_registry.values()
    )
)


# ============================================================
# 12. RUNTIME STATE DISCOVERY
#
# Explicit state containers are preferred.
# We capture STATE, not source code.
# ============================================================

STATE_NAMES = [
    "STATE",
    "PROJECT_STATE",
    "RUNTIME_STATE",
    "SYSTEM_STATE",
    "KITCHEN_STATE",
    "CURRENT_STATE",
    "ANALYSIS_STATE",
]

runtime_state = {}

for name in STATE_NAMES:
    if name in globals():
        runtime_state[name] = _safe(
            globals()[name]
        )


# ============================================================
# 13. AUDIT DISCOVERY
# ============================================================

AUDIT_NAMES = [
    "AUDIT_STATE",
    "AUDIT_LOG",
    "AUDIT_RECORDS",
    "AUDIT_HISTORY",
    "NARRATIVE_VALIDATION",
    "INTEGRITY_STATE",
    "INTEGRITY_REPORT",
    "VALIDATION_STATE",
    "VALIDATION_REPORT",
    "OVERALL_INTEGRITY_PASS",
]

audit = {
    "objects": {},
    "available": False,
}

for name in AUDIT_NAMES:
    if name in globals():
        audit["objects"][name] = _safe(
            globals()[name]
        )

audit["available"] = bool(audit["objects"])


# ============================================================
# 14. LEDGER DISCOVERY
# ============================================================

LEDGER_NAMES = [
    "LEDGER_STATE",
    "LEDGER",
    "LEDGER_RECORDS",
    "LEDGER_HISTORY",
    "LEDGER_MANIFEST",
    "MANIFEST",
    "RECORDS",
]

ledger = {
    "objects": {},
    "available": False,
}

for name in LEDGER_NAMES:
    if name in globals():
        ledger["objects"][name] = _safe(
            globals()[name]
        )

ledger["available"] = bool(ledger["objects"])


# ============================================================
# 15. ARTIFACT DISCOVERY
# ============================================================

ARTIFACT_NAMES = [
    "ARTIFACT_STATE",
    "ARTIFACTS",
    "ARTIFACT_MANIFEST",
    "DATASET_STATE",
    "DATASETS",
    "CAPTURE_STATE",
    "OUTPUT_STATE",
    "REPORT_STATE",
]

artifacts = {
    "objects": {},
    "available": False,
}

for name in ARTIFACT_NAMES:
    if name in globals():
        artifacts["objects"][name] = _safe(
            globals()[name]
        )

artifacts["available"] = bool(artifacts["objects"])


# ============================================================
# 16. HISTORY DISCOVERY
# ============================================================

HISTORY_NAMES = [
    "HISTORY",
    "HISTORY_STATE",
    "SNAPSHOT_HISTORY",
    "PROJECT_HISTORY",
    "HANDOFF_HISTORY",
    "DECISION_HISTORY",
    "DECISIONS",
]

history_objects = {}

for name in HISTORY_NAMES:
    if name in globals():
        history_objects[name] = _safe(
            globals()[name]
        )


# ============================================================
# 17. KNOWN ISSUES / PENDING WORK / DECISIONS
# ============================================================

def _first_existing(names, default):
    for name in names:
        if name in globals():
            return _safe(globals()[name])
    return default


known_issues = _first_existing(
    [
        "KNOWN_ISSUES",
        "known_issues",
        "ISSUES",
        "CURRENT_ISSUES",
    ],
    [],
)

pending_work = _first_existing(
    [
        "PENDING_WORK",
        "pending_work",
        "TODO",
        "NEXT_STEPS",
    ],
    [],
)

decisions = _first_existing(
    [
        "DECISIONS",
        "DECISION_LOG",
        "PROJECT_DECISIONS",
    ],
    {},
)


# ============================================================
# 18. REGRESSION
# ============================================================

regression = {
    "available": callable(
        globals().get(
            "run_cell08_regression_tests"
        )
    ),
    "result": None,
}

if regression["available"]:
    try:
        regression["result"] = _safe(
            run_cell08_regression_tests()
        )
    except Exception as exc:
        regression["result"] = {
            "status": "ERROR",
            "error_type": type(exc).__name__,
            "error": str(exc),
        }

regression_pass = (
    isinstance(regression["result"], dict)
    and regression["result"].get("status") == "PASS"
)


# ============================================================
# 19. SMOKE TEST
# ============================================================

smoke = {
    "attempted": False,
    "status": "NOT_ATTEMPTED",
}

run08 = globals().get("run_cell_08")
to_dict = globals().get("cell08_to_dict")

if callable(run08) and callable(to_dict):

    try:

        smoke_assets = [
            {
                "symbol": "BTC",
                "price": 100000.0,
                "change_pct": 2.0,
                "volume": 1000000.0,
            },
            {
                "symbol": "ASSET_A",
                "price": 100.0,
                "change_pct": 5.0,
                "volume": 500000.0,
            },
            {
                "symbol": "ASSET_B",
                "price": 100.0,
                "change_pct": 1.0,
                "volume": 400000.0,
            },
            {
                "symbol": "ASSET_C",
                "price": 100.0,
                "change_pct": -2.0,
                "volume": 300000.0,
            },
        ]

        result = run08(
            btc_direction="INCREASE",
            btc_d_direction="DECREASE",
            btc_change_pct=2.0,
            assets=smoke_assets,
            pattern_index=1,
        )

        result_dict = to_dict(result)

        smoke = {
            "attempted": True,
            "status": "PASS",
            "scenario_id": result_dict.get(
                "scenario_id"
            ),
            "scenario_type": result_dict.get(
                "scenario_type"
            ),
            "btc_direction": result_dict.get(
                "btc_direction"
            ),
            "btc_d_direction": result_dict.get(
                "btc_d_direction"
            ),
            "context": result_dict.get(
                "context"
            ),
            "opposite_direction": result_dict.get(
                "opposite_direction"
            ),
            "selected_pattern": _safe(
                result_dict.get(
                    "selected_pattern"
                )
            ),
            "top_10_count": len(
                result_dict.get(
                    "top_10_assets",
                    [],
                )
            ),
            "strong_movers_count": len(
                result_dict.get(
                    "strong_movers",
                    [],
                )
            ),
            "relative_movers_count": len(
                result_dict.get(
                    "relative_movers",
                    [],
                )
            ),
        }

    except Exception as exc:

        smoke = {
            "attempted": True,
            "status": "FAIL",
            "error_type": type(exc).__name__,
            "error": str(exc),
        }


smoke_pass = smoke.get("status") == "PASS"


# ============================================================
# 20. CONTINUITY CONTRACT
# ============================================================

continuity = {
    "current_state": "U08_VALIDATED",
    "code_source": "IPYNB",
    "state_source": "MASTER_SNAPSHOT",
    "history_source": "MASTER_SNAPSHOT",
    "audit_source": "MASTER_SNAPSHOT",
    "ledger_source": "MASTER_SNAPSHOT",
    "artifact_source": "MASTER_SNAPSHOT",
    "source_code_embedded": False,
    "known_issues": known_issues,
    "pending_work": pending_work,
    "decisions": decisions,
}


# ============================================================
# 21. HANDOFF CONTRACT
# ============================================================

handoff = {
    "ready": False,
    "code_source": "IPYNB",
    "state_source": "MASTER_SNAPSHOT",
    "history_source": "MASTER_SNAPSHOT",
    "covered_units": [
        "U01",
        "U02",
        "U03",
        "U04",
        "U05",
        "U06",
        "U07",
        "U08",
    ],
    "source_code_embedded": False,
    "current_state": "U08_VALIDATED",
    "instructions_for_next_assistant": [
        "Read the supplied IPYNB as the exclusive code source of truth.",
        "Read the MASTER SNAPSHOT as the project state and history source of truth.",
        "Do not reconstruct source code from the snapshot.",
        "Do not rewrite verified code without an explicit change decision.",
        "Preserve all execution safety locks.",
        "Treat audit data as historical project state.",
        "Treat ledger data as historical project state.",
        "Treat artifact metadata as historical project state.",
        "Treat continuity and decisions as authoritative project history.",
        "Do not interpret an analysis-only PASS as trading authorization.",
        "Do not create orders.",
        "Do not mutate portfolio state.",
        "Continue from known_issues and pending_work.",
    ],
}


# ============================================================
# 22. INTEGRITY
# ============================================================

integrity = {
    "u01_u08_scope_complete": all(
        unit_evidence[f"U{i:02d}"]["declared_in_scope"]
        for i in range(1, 9)
    ),
    "u01_u08_count": len(unit_evidence) == 8,
    "scenario_matrix_valid": scenario_pass,
    "narratives_valid": narrative_pass,
    "function_registry_valid": function_pass,
    "regression_pass": regression_pass,
    "smoke_test_pass": smoke_pass,
    "trading_disabled": runtime_flags[
        "trading_enabled"
    ] is False,
    "orders_disabled": runtime_flags[
        "orders_enabled"
    ] is False,
    "strategy_disabled": runtime_flags[
        "strategy_enabled"
    ] is False,
    "portfolio_actions_disabled": runtime_flags[
        "portfolio_actions_enabled"
    ] is False,
    "analysis_only": True,
    "source_code_not_embedded": True,
}


# ============================================================
# 23. CANONICAL SNAPSHOT
#
# IMPORTANT:
#   No function source.
#   No notebook source.
#   No code bodies.
# ============================================================

snapshot = {

    "snapshot": {
        "engine": SNAPSHOT_ENGINE,
        "version": SNAPSHOT_VERSION,
        "schema": SNAPSHOT_SCHEMA,
        "timestamp_utc": timestamp_utc,
        "status": "PENDING_FINALIZATION",
        "source_code_embedded": False,
        "notebook": notebook_identity,
    },

    "source_of_truth": {
        "code": "IPYNB",
        "state": "MASTER_SNAPSHOT",
        "history": "MASTER_SNAPSHOT",
        "audit": "MASTER_SNAPSHOT",
        "ledger": "MASTER_SNAPSHOT",
        "artifacts": "MASTER_SNAPSHOT",
        "source_code_embedded": False,
    },

    "architecture": {
        "coverage": "U01_U08",
        "required_units": 8,
        "present_units": 8,
        "coverage_status": "COMPLETE",
        "units": unit_evidence,
        "notebook_name": NOTEBOOK_NAME,
        "source_code_embedded": False,
    },

    "execution": execution,

    "configuration": configuration,

    "contracts": {
        "code_source_of_truth": "IPYNB",
        "state_source_of_truth": "MASTER_SNAPSHOT",
        "history_source_of_truth": "MASTER_SNAPSHOT",
        "scenario_matrix_locked": True,
        "narratives_locked": True,
        "btc_is_benchmark": True,
        "relative_performance_is_complementary": True,
        "analysis_only": True,
        "trading_authorization": False,
        "order_creation_authorization": False,
        "portfolio_mutation_authorization": False,
    },

    "scenarios": scenarios,

    "narratives": narratives,

    "functions": {
        "registry": function_registry,
        "summary": {
            "declared": len(MASTER_FUNCTIONS),
            "existing": sum(
                1
                for x in function_registry.values()
                if x["exists"]
            ),
            "callable": sum(
                1
                for x in function_registry.values()
                if x["callable"]
            ),
        },
        "source_code_embedded": False,
    },

    "runtime_state": runtime_state,

    "audit": audit,

    "ledger": ledger,

    "artifacts": artifacts,

    "history": {
        "runtime_objects": history_objects,
        "previous_snapshot_count": 0,
        "previous_snapshots": [],
    },

    "regression": regression,

    "smoke_test": smoke,

    "continuity": continuity,

    "integrity": integrity,

    "known_issues": known_issues,

    "pending_work": pending_work,

    "decisions": decisions,

    "handoff": handoff,
}


# ============================================================
# 24. FINAL VALIDATION
#
# PASS only when core canonical requirements are true.
# ============================================================

core_pass = all([
    integrity["u01_u08_scope_complete"],
    integrity["u01_u08_count"],
    integrity["scenario_matrix_valid"],
    integrity["narratives_valid"],
    integrity["function_registry_valid"],
    integrity["regression_pass"],
    integrity["smoke_test_pass"],
    integrity["trading_disabled"],
    integrity["orders_disabled"],
    integrity["strategy_disabled"],
    integrity["portfolio_actions_disabled"],
    integrity["analysis_only"],
    integrity["source_code_not_embedded"],
])

handoff["ready"] = bool(core_pass)

snapshot["handoff"] = handoff

snapshot["snapshot"]["status"] = (
    "PASS"
    if core_pass
    else "FAIL"
)


# ============================================================
# 25. CANONICAL HASH
#
# Hash is calculated AFTER removing the hash field itself.
# ============================================================

snapshot_for_hash = json.loads(
    json.dumps(
        snapshot,
        ensure_ascii=False,
    )
)

snapshot_for_hash["snapshot"].pop(
    "sha256",
    None,
)

canonical_hash = _sha256_json(
    snapshot_for_hash
)

snapshot["snapshot"]["sha256"] = canonical_hash


# ============================================================
# 26. OUTPUT DIRECTORY
# ============================================================

snapshot_dir = None

for candidate in SNAPSHOT_ROOT_CANDIDATES:

    try:
        candidate.mkdir(
            parents=True,
            exist_ok=True,
        )

        if candidate.is_dir():
            snapshot_dir = candidate
            break

    except Exception:
        continue


assert snapshot_dir is not None, (
    "Unable to create canonical snapshot directory."
)


# ============================================================
# 27. SAVE
# ============================================================

snapshot_path = (
    snapshot_dir
    / "Kitchen_Assistant_V3.1_MASTER_SNAPSHOT_U01_U08.json"
)

snapshot_bytes = json.dumps(
    snapshot,
    ensure_ascii=False,
    indent=2,
    sort_keys=False,
).encode("utf-8")

snapshot_path.write_bytes(
    snapshot_bytes
)


# ============================================================
# 28. POST-SAVE VERIFICATION
# ============================================================

assert snapshot_path.is_file()

saved_snapshot = json.loads(
    snapshot_path.read_text(
        encoding="utf-8"
    )
)

assert saved_snapshot["snapshot"]["sha256"] == canonical_hash
assert saved_snapshot["snapshot"]["source_code_embedded"] is False
assert saved_snapshot["source_of_truth"]["code"] == "IPYNB"
assert saved_snapshot["architecture"]["coverage"] == "U01_U08"
assert len(saved_snapshot["architecture"]["units"]) == 8


# ============================================================
# 29. FINAL REPORT
# ============================================================

print("=" * 72)
print("KITCHEN ASSISTANT V3.1")
print("CANONICAL MASTER PORTABLE SNAPSHOT")
print("U01 -> U08")
print("=" * 72)

print(
    f"SNAPSHOT STATUS          : "
    f"{saved_snapshot['snapshot']['status']}"
)

print(
    f"SCHEMA                  : "
    f"{SNAPSHOT_SCHEMA}"
)

print(
    "COVERAGE                : "
    "U01 -> U08 = 8/8"
)

print(
    f"SCENARIOS               : "
    f"{scenarios['count']}/9"
)

print(
    f"NARRATIVE SCENARIOS     : "
    f"{narratives['scenario_count']}/9"
)

print(
    f"NARRATIVE PATTERNS     : "
    f"{narratives['total_patterns']}/27"
)

print(
    f"FUNCTIONS              : "
    f"{len(MASTER_FUNCTIONS)}/26"
)

print(
    "AUDIT STATE CAPTURE     : "
    f"{'YES' if audit['available'] else 'NOT_AVAILABLE'}"
)

print(
    "LEDGER STATE CAPTURE    : "
    f"{'YES' if ledger['available'] else 'NOT_AVAILABLE'}"
)

print(
    "ARTIFACT STATE CAPTURE  : "
    f"{'YES' if artifacts['available'] else 'NOT_AVAILABLE'}"
)

print(
    "RUNTIME STATE CAPTURE   : "
    f"{len(runtime_state)} object(s)"
)

print(
    "REGRESSION              : "
    f"{'PASS' if regression_pass else 'FAIL'}"
)

print(
    "SMOKE TEST              : "
    f"{'PASS' if smoke_pass else 'FAIL'}"
)

print(
    "TRADING                 : "
    f"{'DISABLED' if not runtime_flags['trading_enabled'] else 'ENABLED'}"
)

print(
    "ORDERS                  : "
    f"{'DISABLED' if not runtime_flags['orders_enabled'] else 'ENABLED'}"
)

print(
    "STRATEGY                : "
    f"{'DISABLED' if not runtime_flags['strategy_enabled'] else 'ENABLED'}"
)

print(
    "PORTFOLIO ACTIONS       : "
    f"{'DISABLED' if not runtime_flags['portfolio_actions_enabled'] else 'ENABLED'}"
)

print(
    "EXECUTION MODE          : "
    "ANALYSIS_ONLY"
)

print(
    "SOURCE CODE EMBEDDED    : "
    "False"
)

print(
    "HANDOFF READY           : "
    f"{'YES' if handoff['ready'] else 'NO'}"
)

print(
    "NOTEBOOK FILE FOUND     : "
    f"{'YES' if notebook_identity['filesystem_available'] else 'RUNTIME_ONLY'}"
)

print(
    "SNAPSHOT FILE SAVED     : YES"
)

print(
    f"SNAPSHOT FILE           : "
    f"{snapshot_path}"
)

print(
    f"MASTER SNAPSHOT HASH    : "
    f"{canonical_hash}"
)

print("=" * 72)

if core_pass:
    print("CANONICAL SNAPSHOT      : PASS")
    print("HANDOFF                 : READY")
    print("=" * 72)
    print(
        "FINAL: Snapshot is ready. "
        "Keep the IPYNB and this MASTER SNAPSHOT together."
    )
else:
    print("CANONICAL SNAPSHOT      : FAIL")
    print("=" * 72)
    print(
        "DO NOT treat this snapshot as canonical. "
        "Inspect the failed fields above."
    )
```

```python

# ============================================================
# KITCHEN ASSISTANT V3.1
# U09 — DYNAMIC MARKET UNIVERSE / MARKET PARTICIPATION ENGINE
# FINAL IMPLEMENTATION — VERSION 1.1 / DISPLAY + REFERENCE CORRECTION
# MULTI-PROVIDER FAILOVER / FROZEN SNAPSHOT / AUDITABLE
# ============================================================
# PURPOSE
#   Dynamic Rank 1-125 universe, segmented as:
#     BTC / ETH / TOP10_ALT / BROAD_ALT_11_125
#   No trading, no orders, no portfolio actions, no advice.
#
# PROVIDER POLICY
#   Primary : CoinGecko
#   Secondary: CoinMarketCap
#   Additional providers: registry-ready, not silently merged.
#
# IMPORTANT
#   A provider outage/rate-limit is not automatically a U09 failure.
#   We fail over only to a separately validated provider snapshot.
#   If every acceptable provider fails, U09 explicitly returns
#   DATA_UNAVAILABLE. No fabricated values and no stale silent reuse.
# ============================================================

from __future__ import annotations

from dataclasses import dataclass, asdict
from datetime import datetime, timezone, timedelta
from pathlib import Path
from typing import Any, Dict, List, Optional, Tuple
import hashlib
import json
import math
import os
import re
import time

import requests

# ============================================================
# 0) HARD SAFETY LOCKS
# ============================================================
TRADING_ENABLED = False
ORDERS_ENABLED = False
STRATEGY_ENABLED = False
PORTFOLIO_ACTIONS_ENABLED = False
assert not TRADING_ENABLED
assert not ORDERS_ENABLED
assert not STRATEGY_ENABLED
assert not PORTFOLIO_ACTIONS_ENABLED

# ============================================================
# 1) PROJECT / CONFIG
# ============================================================
ROOT = Path("/content/drive/MyDrive/Colab Notebooks")
if not ROOT.is_dir():
    raise RuntimeError("Project root not found: " + str(ROOT))
os.chdir(ROOT)

U09_ROOT = ROOT / "u09_dynamic_market_universe"
SNAPSHOT_DIR = U09_ROOT / "snapshots"
AUDIT_DIR = U09_ROOT / "audit"
CACHE_DIR = U09_ROOT / "cache"
for d in (U09_ROOT, SNAPSHOT_DIR, AUDIT_DIR, CACHE_DIR):
    d.mkdir(parents=True, exist_ok=True)

CONFIG: Dict[str, Any] = {
    "universe_limit": 125,
    "currency": "usd",
    "reference_mode": "provider_24h",
    "lookback": "24h",
    "minimum_reference_age_seconds": 20 * 60,
    "maximum_reference_age_seconds": 48 * 3600,
    "max_current_age_seconds": 15 * 60,
    "breadth_threshold_pct": 0.25,
    "breadth_flat_band_pct": 0.25,
    "dominance_partition_tolerance_pct": 0.10,
    "rank_minor_difference": 1,
    "rank_warning_difference": 3,
    "max_retries_per_provider": 1,
    "request_timeout_seconds": 12,
    "retry_backoff_seconds": 1.0,
    "provider_cooldown_seconds": 60,
    "cache_enabled": True,
    "cache_max_age_seconds": 10 * 60,
    "allow_stale_cache_for_analysis": False,
    # User-facing U09 message contract:
    # - every displayed value is a 24h movement
    # - dominance movement is RELATIVE % only
    # - percentage-point change remains internal/audit-only
    # - no participation status/scenario is emitted by U09 message
    "message_window_label": "24H",
    "message_show_dominance_current": False,
    "message_show_dominance_pp": False,
    "message_show_participation_state": False,
    "message_dominance_min_decimals": 2,
    "message_dominance_max_decimals": 6,
}


# Optional CoinGecko key. The implementation can try the public/demo-compatible
# route without a key; if the endpoint requires authentication, it records that
# failure and continues to the next provider. Never hard-code a secret.
COINGECKO_API_KEY = os.getenv("COINGECKO_API_KEY", "").strip()
COINMARKETCAP_API_KEY = os.getenv("COINMARKETCAP_API_KEY", "").strip()

# ============================================================
# 2) UTILITY / JSON / TIME
# ============================================================
def utc_now() -> datetime:
    return datetime.now(timezone.utc)


def iso(dt: Optional[datetime]) -> Optional[str]:
    return dt.astimezone(timezone.utc).isoformat() if dt else None


def parse_ts(value: Any) -> Optional[datetime]:
    if value is None:
        return None
    if isinstance(value, (int, float)):
        try:
            x = float(value)
            if x > 10_000_000_000:
                x /= 1000.0
            return datetime.fromtimestamp(x, tz=timezone.utc)
        except Exception:
            return None
    if isinstance(value, str):
        s = value.strip().replace("Z", "+00:00")
        try:
            dt = datetime.fromisoformat(s)
            return dt if dt.tzinfo else dt.replace(tzinfo=timezone.utc)
        except Exception:
            return None
    return None


def finite_nonnegative(value: Any) -> bool:
    try:
        x = float(value)
        return math.isfinite(x) and x >= 0.0
    except Exception:
        return False


def pct_change(current: Optional[float], reference: Optional[float]) -> Optional[float]:
    if current is None or reference is None:
        return None
    if not finite_nonnegative(current) or not finite_nonnegative(reference):
        return None
    if reference == 0:
        return None
    return (current - reference) / reference * 100.0


def dominance(segment_mc: Optional[float], total_mc: Optional[float]) -> Optional[float]:
    if segment_mc is None or total_mc is None:
        return None
    if not finite_nonnegative(segment_mc) or not finite_nonnegative(total_mc) or total_mc <= 0:
        return None
    return segment_mc / total_mc * 100.0


def dominance_change_pct(current: Optional[float], reference: Optional[float]) -> Optional[float]:
    return pct_change(current, reference)


def sha256_obj(obj: Any) -> str:
    raw = json.dumps(obj, sort_keys=True, separators=(",", ":"), ensure_ascii=False).encode()
    return hashlib.sha256(raw).hexdigest()


def safe_json(obj: Any) -> Any:
    if isinstance(obj, dict):
        return {str(k): safe_json(v) for k, v in obj.items()}
    if isinstance(obj, (list, tuple)):
        return [safe_json(v) for v in obj]
    if isinstance(obj, datetime):
        return iso(obj)
    if isinstance(obj, float):
        return obj if math.isfinite(obj) else None
    return obj


def write_json(path: Path, obj: Any) -> None:
    path.write_text(json.dumps(safe_json(obj), indent=2, ensure_ascii=False), encoding="utf-8")

# ============================================================
# 3) PROVIDER CONTRACTS
# ============================================================
@dataclass
class ProviderHealth:
    provider_id: str
    state: str = "AVAILABLE"
    failure_count: int = 0
    last_failure_at: Optional[str] = None
    cooldown_until: Optional[str] = None
    last_success_at: Optional[str] = None
    last_error: Optional[str] = None

    def in_cooldown(self, now: datetime) -> bool:
        dt = parse_ts(self.cooldown_until)
        return bool(dt and now < dt)


@dataclass
class ProviderAttempt:
    provider_id: str
    attempt: int
    request_timestamp: str
    response_timestamp: Optional[str]
    http_status: Optional[int]
    health_state: str
    error: Optional[str]
    endpoint: str


@dataclass
class ProviderSnapshot:
    provider_id: str
    primary_provider: str
    fallback_used: bool
    fallback_chain: List[str]
    fallback_reason: Optional[str]
    provider_status: str
    provider_timestamp: Optional[str]
    request_timestamp: str
    response_timestamp: str
    provider_schema_version: str
    provider_endpoint: str
    provider_request_status: str
    raw_assets: List[Dict[str, Any]]
    attempts: List[Dict[str, Any]]


class ProviderError(RuntimeError):
    def __init__(self, state: str, message: str, status_code: Optional[int] = None):
        super().__init__(message)
        self.state = state
        self.status_code = status_code


class BaseProvider:
    provider_id = "BASE"
    schema_version = "unknown"
    endpoint = ""

    def headers(self) -> Dict[str, str]:
        return {"Accept": "application/json", "User-Agent": "KitchenAssistant-U09/3.1"}

    def fetch(self, session: requests.Session, limit: int, timeout: int) -> Tuple[List[Dict[str, Any]], Optional[datetime], Dict[str, Any]]:
        raise NotImplementedError

    def normalize(self, payload: Any) -> Tuple[List[Dict[str, Any]], Optional[datetime], Dict[str, Any]]:
        raise NotImplementedError


class CoinGeckoProvider(BaseProvider):
    provider_id = "COINGECKO"
    schema_version = "coins/markets"
    endpoint = "https://api.coingecko.com/api/v3/coins/markets"

    def headers(self) -> Dict[str, str]:
        h = super().headers()
        if COINGECKO_API_KEY:
            h["x-cg-demo-api-key"] = COINGECKO_API_KEY
        return h

    def fetch(self, session, limit, timeout):
        params = {
            "vs_currency": "usd",
            "order": "market_cap_desc",
            "per_page": min(limit, 250),
            "page": 1,
            "sparkline": "false",
            "price_change_percentage": "24h",
        }
        req_ts = utc_now()
        try:
            r = session.get(self.endpoint, params=params, headers=self.headers(), timeout=timeout)
        except requests.Timeout as e:
            raise ProviderError("TIMEOUT", str(e))
        except requests.RequestException as e:
            raise ProviderError("UNAVAILABLE", str(e))
        resp_ts = utc_now()
        if r.status_code == 429:
            raise ProviderError("RATE_LIMITED", "HTTP 429", 429)
        if r.status_code in (401, 403):
            raise ProviderError("AUTH_FAILED", f"HTTP {r.status_code}", r.status_code)
        if r.status_code >= 500:
            raise ProviderError("UNAVAILABLE", f"HTTP {r.status_code}", r.status_code)
        if r.status_code != 200:
            raise ProviderError("UNAVAILABLE", f"HTTP {r.status_code}", r.status_code)
        try:
            payload = r.json()
        except Exception as e:
            raise ProviderError("MALFORMED", f"JSON decode failed: {e}")
        assets, provider_ts, meta = self.normalize(payload)
        meta.update({"request_timestamp": iso(req_ts), "response_timestamp": iso(resp_ts)})
        return assets, provider_ts, meta

    def normalize(self, payload):
        if not isinstance(payload, list):
            raise ProviderError("MALFORMED", "CoinGecko response is not a list")
        out = []
        for item in payload:
            if not isinstance(item, dict):
                raise ProviderError("MALFORMED", "CoinGecko asset item is not an object")
            out.append({
                "provider_asset_id": item.get("id"),
                "symbol": item.get("symbol"),
                "name": item.get("name"),
                "price_usd": item.get("current_price"),
                "market_cap_usd": item.get("market_cap"),
                "provider_rank": item.get("market_cap_rank"),
                "price_change_24h_pct": item.get("price_change_percentage_24h"),
                "market_cap_change_24h_pct": item.get("market_cap_change_percentage_24h"),
                "provider_timestamp": item.get("last_updated"),
            })
        stamps = [parse_ts(x.get("provider_timestamp")) for x in out]
        stamps = [x for x in stamps if x]
        provider_ts = min(stamps) if stamps else None
        return out, provider_ts, {"response_type": "list", "field_count_checked": True}


class CoinMarketCapProvider(BaseProvider):
    provider_id = "COINMARKETCAP"
    schema_version = "v3/listings_latest"
    # Keyless public route; authenticated route is supported if a key exists.
    endpoint_keyless = "https://pro-api.coinmarketcap.com/public-api/v3/cryptocurrency/listings/latest"
    endpoint_keyed = "https://pro-api.coinmarketcap.com/v3/cryptocurrency/listings/latest"

    @property
    def endpoint(self):
        return self.endpoint_keyed if COINMARKETCAP_API_KEY else self.endpoint_keyless

    def headers(self):
        h = super().headers()
        if COINMARKETCAP_API_KEY:
            h["X-CMC_PRO_API_KEY"] = COINMARKETCAP_API_KEY
        return h

    def fetch(self, session, limit, timeout):
        params = {"start": 1, "limit": min(limit, 200), "convert": "USD"}
        req_ts = utc_now()
        try:
            r = session.get(self.endpoint, params=params, headers=self.headers(), timeout=timeout)
        except requests.Timeout as e:
            raise ProviderError("TIMEOUT", str(e))
        except requests.RequestException as e:
            raise ProviderError("UNAVAILABLE", str(e))
        resp_ts = utc_now()
        if r.status_code == 429:
            raise ProviderError("RATE_LIMITED", "HTTP 429", 429)
        if r.status_code in (401, 403):
            raise ProviderError("AUTH_FAILED", f"HTTP {r.status_code}", r.status_code)
        if r.status_code >= 500:
            raise ProviderError("UNAVAILABLE", f"HTTP {r.status_code}", r.status_code)
        if r.status_code != 200:
            raise ProviderError("UNAVAILABLE", f"HTTP {r.status_code}", r.status_code)
        try:
            payload = r.json()
        except Exception as e:
            raise ProviderError("MALFORMED", f"JSON decode failed: {e}")
        assets, provider_ts, meta = self.normalize(payload)
        meta.update({"request_timestamp": iso(req_ts), "response_timestamp": iso(resp_ts)})
        return assets, provider_ts, meta

    def normalize(self, payload):
        if not isinstance(payload, dict) or not isinstance(payload.get("data"), list):
            raise ProviderError("MALFORMED", "CoinMarketCap response/data shape invalid")
        status = payload.get("status") or {}
        code = status.get("error_code")
        if code is not None and str(code) != "0":
            raise ProviderError("UNAVAILABLE", str(status.get("error_message") or code))
        out = []
        for item in payload["data"]:
            if not isinstance(item, dict):
                raise ProviderError("MALFORMED", "CMC asset item is not an object")
            quote = ((item.get("quote") or {}).get("USD") or {})
            out.append({
                "provider_asset_id": item.get("id"),
                "symbol": item.get("symbol"),
                "name": item.get("name"),
                "price_usd": quote.get("price"),
                "market_cap_usd": quote.get("market_cap"),
                "provider_rank": item.get("cmc_rank"),
                "price_change_24h_pct": quote.get("percent_change_24h"),
                "market_cap_change_24h_pct": None,
                "provider_timestamp": status.get("timestamp"),
            })
        provider_ts = parse_ts(status.get("timestamp"))
        return out, provider_ts, {"response_type": "object", "field_count_checked": True, "status_error_code": str(code) if code is not None else None}


# Registry is extensible. Additional providers must implement the same contract.
PROVIDER_REGISTRY: Dict[str, Dict[str, Any]] = {
    "COINGECKO": {"provider": CoinGeckoProvider(), "priority": 1, "health": ProviderHealth("COINGECKO")},
    "COINMARKETCAP": {"provider": CoinMarketCapProvider(), "priority": 2, "health": ProviderHealth("COINMARKETCAP")},
}

# ============================================================
# 4) PROVIDER VALIDATION
# ============================================================
def validate_provider_assets(assets: List[Dict[str, Any]], limit: int) -> Dict[str, Any]:
    errors: List[str] = []
    warnings: List[str] = []
    ids = []
    ranks = []
    symbols = {}
    valid = []

    for i, a in enumerate(assets):
        aid = a.get("provider_asset_id")
        rank = a.get("provider_rank")
        sym = str(a.get("symbol") or "").upper()
        if not aid:
            errors.append(f"missing_id_at_{i}")
            continue
        if not isinstance(rank, int) or isinstance(rank, bool):
            try:
                rank = int(rank)
            except Exception:
                errors.append(f"invalid_rank_{aid}")
                continue
        if rank < 1 or rank > limit:
            errors.append(f"rank_out_of_range_{aid}:{rank}")
            continue
        if not finite_nonnegative(a.get("price_usd")):
            errors.append(f"invalid_price_{aid}")
            continue
        if not finite_nonnegative(a.get("market_cap_usd")):
            errors.append(f"invalid_market_cap_{aid}")
            continue
        if aid in ids:
            errors.append(f"duplicate_id_{aid}")
            continue
        ids.append(aid)
        if rank in ranks:
            errors.append(f"duplicate_rank_{rank}")
        ranks.append(rank)
        if sym:
            symbols.setdefault(sym, []).append(aid)
        b = dict(a)
        b["provider_rank"] = rank
        valid.append(b)

    for sym, aids in symbols.items():
        if len(aids) > 1:
            warnings.append(f"symbol_collision_{sym}:{len(aids)}")

    rank_set = set(ranks)
    missing = [r for r in range(1, limit + 1) if r not in rank_set]
    if missing:
        errors.append("rank_gaps:" + ",".join(map(str, missing[:20])))

    valid.sort(key=lambda x: x["provider_rank"])
    calculated = sorted(valid, key=lambda x: (-float(x["market_cap_usd"]), str(x["provider_asset_id"])))
    calc_rank = {a["provider_asset_id"]: i + 1 for i, a in enumerate(calculated)}
    for a in valid:
        cr = calc_rank.get(a["provider_asset_id"])
        a["calculated_rank"] = cr
        diff = abs(int(a["provider_rank"]) - int(cr)) if cr is not None else 999
        a["rank_difference"] = diff
        if diff == 0:
            a["rank_consistency_status"] = "MATCH"
        elif diff <= CONFIG["rank_minor_difference"]:
            a["rank_consistency_status"] = "MINOR_DIFFERENCE"
        elif diff <= CONFIG["rank_warning_difference"]:
            a["rank_consistency_status"] = "WARNING"
            warnings.append(f"rank_warning_{a['provider_asset_id']}:{diff}")
        else:
            a["rank_consistency_status"] = "FAIL"
            errors.append(f"rank_fail_{a['provider_asset_id']}:{diff}")

    coverage = len(set(ranks)) / float(limit)
    if len(valid) != limit:
        errors.append(f"coverage_count_{len(valid)}_expected_{limit}")

    status = "PASS" if not errors else "FAIL"
    return {
        "status": status,
        "errors": errors,
        "warnings": warnings,
        "valid_assets": valid,
        "coverage_ratio": coverage,
        "calculated_rank_count": len(calc_rank),
    }


def freshness_status(provider_ts: Optional[datetime], now: datetime) -> Tuple[str, Optional[float]]:
    if not provider_ts:
        return "UNAVAILABLE", None
    age = max(0.0, (now - provider_ts).total_seconds())
    if age <= CONFIG["max_current_age_seconds"]:
        return "FRESH", age
    return "STALE", age

# ============================================================
# 5) ORCHESTRATOR / FAILOVER
# ============================================================
def _cooldown(health: ProviderHealth, state: str, now: datetime) -> None:
    health.state = state
    health.failure_count += 1
    health.last_failure_at = iso(now)
    health.cooldown_until = iso(now + timedelta(seconds=CONFIG["provider_cooldown_seconds"]))


def provider_order(now: datetime) -> List[str]:
    candidates = []
    for pid, entry in PROVIDER_REGISTRY.items():
        h = entry["health"]
        if h.in_cooldown(now):
            continue
        candidates.append((entry["priority"], -float(entry.get("validation_score", 1.0)), pid))
    return [x[2] for x in sorted(candidates)]


def orchestrate_universe() -> Dict[str, Any]:
    now = utc_now()
    session = requests.Session()
    primary = "COINGECKO"
    attempts: List[Dict[str, Any]] = []
    chain: List[str] = []
    failures: List[str] = []

    for pid in provider_order(now):
        entry = PROVIDER_REGISTRY[pid]
        provider: BaseProvider = entry["provider"]
        health: ProviderHealth = entry["health"]
        chain.append(pid)
        accepted = False
        last_error: Optional[ProviderError] = None
        for attempt_no in range(CONFIG["max_retries_per_provider"] + 1):
            req_ts = utc_now()
            try:
                assets, provider_ts, meta = provider.fetch(
                    session, CONFIG["universe_limit"], CONFIG["request_timeout_seconds"]
                )
                validation = validate_provider_assets(assets, CONFIG["universe_limit"])
                fstate, age = freshness_status(provider_ts, utc_now())
                if fstate == "STALE":
                    validation["errors"].append("provider_snapshot_stale")
                    validation["status"] = "FAIL"
                if validation["status"] != "PASS":
                    raise ProviderError("PARTIAL", "Provider validation failed: " + ";".join(validation["errors"][:5]))

                resp_ts = utc_now()
                health.state = "AVAILABLE"
                health.last_success_at = iso(resp_ts)
                health.last_error = None
                health.cooldown_until = None
                health.failure_count = 0
                entry["validation_score"] = 1.0
                attempts.append(asdict(ProviderAttempt(pid, attempt_no + 1, iso(req_ts), iso(resp_ts), 200, "AVAILABLE", None, provider.endpoint)))
                fallback_used = pid != primary
                reason = "; ".join(failures) if fallback_used else None
                snapshot = ProviderSnapshot(
                    provider_id=pid,
                    primary_provider=primary,
                    fallback_used=fallback_used,
                    fallback_chain=chain.copy(),
                    fallback_reason=reason,
                    provider_status="ACTIVE",
                    provider_timestamp=iso(provider_ts),
                    request_timestamp=meta.get("request_timestamp") or iso(req_ts),
                    response_timestamp=meta.get("response_timestamp") or iso(resp_ts),
                    provider_schema_version=provider.schema_version,
                    provider_endpoint=provider.endpoint,
                    provider_request_status="SUCCESS",
                    raw_assets=validation["valid_assets"],
                    attempts=attempts.copy(),
                )
                return {"snapshot": snapshot, "validation": validation, "provider_meta": meta, "health": asdict(health)}
            except ProviderError as e:
                last_error = e
                attempts.append(asdict(ProviderAttempt(pid, attempt_no + 1, iso(req_ts), iso(utc_now()), e.status_code, e.state, str(e), provider.endpoint)))
                if attempt_no < CONFIG["max_retries_per_provider"] and e.state in {"TIMEOUT", "RATE_LIMITED", "UNAVAILABLE"}:
                    # Deterministic retry count and fixed backoff. No random jitter.
                    time.sleep(CONFIG["retry_backoff_seconds"])
                    continue
                break

        if last_error:
            failures.append(f"{pid}:{last_error.state}:{last_error}")
            _cooldown(health, last_error.state, utc_now())
            entry["validation_score"] = max(0.0, float(entry.get("validation_score", 1.0)) - 0.2)

    return {
        "snapshot": None,
        "validation": {"status": "DATA_UNAVAILABLE", "errors": failures, "warnings": []},
        "provider_meta": {"attempts": attempts},
        "health": {pid: asdict(v["health"]) for pid, v in PROVIDER_REGISTRY.items()},
    }

# ============================================================
# 6) SEGMENTATION / REFERENCE
# ============================================================
SEGMENTS = {
    "BTC": range(1, 2),
    "ETH": range(2, 3),
    "TOP10_ALT": range(3, 11),
    "BROAD_ALT_11_125": range(11, 126),
}


def segment_for_rank(rank: int) -> str:
    for name, rr in SEGMENTS.items():
        if rank in rr:
            return name
    return "OUT_OF_UNIVERSE"


def build_reference_from_24h(assets: List[Dict[str, Any]]) -> Dict[str, float]:
    """Reconstruct a provider-derived 24h MC reference without rounding.

    This is NOT a fabricated historical snapshot. It is only available for
    assets where the selected provider supplies a valid 24h market-cap change.
    If a provider does not supply that field (for example the normalized CMC
    listing payload), reference-dependent MC/dominance movement remains None.
    """
    ref: Dict[str, float] = {}
    for a in assets:
        mc = float(a["market_cap_usd"])
        ch = a.get("market_cap_change_24h_pct")
        try:
            ch = float(ch) if ch is not None else None
        except Exception:
            ch = None
        if ch is not None and math.isfinite(ch) and ch > -100.0:
            ref_mc = mc / (1.0 + ch / 100.0)
            if finite_nonnegative(ref_mc):
                ref[a["provider_asset_id"]] = ref_mc
    return ref


def reference_timestamp_from_provider_24h(provider_timestamp: Optional[str]) -> Optional[str]:
    """Return the effective 24h reference timestamp, not the current timestamp."""
    dt = parse_ts(provider_timestamp)
    if dt is None:
        return None
    return iso(dt - timedelta(hours=24))


def build_segments(assets: List[Dict[str, Any]], reference_mc_by_id: Dict[str, float]) -> Dict[str, Any]:
    groups: Dict[str, List[Dict[str, Any]]] = {k: [] for k in SEGMENTS}
    for a in assets:
        seg = segment_for_rank(int(a["provider_rank"]))
        if seg in groups:
            b = dict(a)
            b["segment"] = seg
            groups[seg].append(b)

    total_mc = sum(float(a["market_cap_usd"]) for a in assets)
    total_ref = sum(reference_mc_by_id.get(a["provider_asset_id"], 0.0) for a in assets if a["provider_asset_id"] in reference_mc_by_id)
    out: Dict[str, Any] = {}
    for seg, rows in groups.items():
        mc = sum(float(a["market_cap_usd"]) for a in rows)
        ref_vals = [reference_mc_by_id[a["provider_asset_id"]] for a in rows if a["provider_asset_id"] in reference_mc_by_id]
        ref_mc = sum(ref_vals) if len(ref_vals) == len(rows) else None
        prices = [a for a in rows if finite_nonnegative(a.get("price_usd"))]
        rising = flat = falling = 0
        threshold = float(CONFIG["breadth_threshold_pct"])
        for a in prices:
            ch = a.get("price_change_24h_pct")
            try:
                ch = float(ch)
            except Exception:
                continue
            if ch > threshold:
                rising += 1
            elif ch < -threshold:
                falling += 1
            else:
                flat += 1
        n = len(prices)
        rp = rising / n * 100 if n else None
        fp = falling / n * 100 if n else None
        flp = flat / n * 100 if n else None
        if n == 0:
            breadth_state = "UNAVAILABLE"
        elif rp >= 70:
            breadth_state = "STRONG"
        elif rp >= 55:
            breadth_state = "POSITIVE"
        elif fp >= 70:
            breadth_state = "WEAK"
        elif fp >= 55:
            breadth_state = "NEGATIVE"
        else:
            breadth_state = "MIXED"
        out[seg] = {
            "constituents": rows,
            "count": len(rows),
            "aggregate_market_cap_usd": mc,
            "reference_market_cap_usd": ref_mc,
            "market_cap_change_pct": pct_change(mc, ref_mc),
            "dominance_pct": dominance(mc, total_mc),
            "reference_dominance_pct": dominance(ref_mc, total_ref) if ref_mc is not None and total_ref > 0 else None,
            "dominance_change_pct": dominance_change_pct(dominance(mc, total_mc), dominance(ref_mc, total_ref) if ref_mc is not None and total_ref > 0 else None),
            "dominance_change_pp": (dominance(mc, total_mc) - dominance(ref_mc, total_ref)) if ref_mc is not None and total_ref > 0 else None,
            "breadth": {
                "rising_count": rising,
                "flat_count": flat,
                "falling_count": falling,
                "rising_pct": rp,
                "flat_pct": flp,
                "falling_pct": fp,
                "breadth_state": breadth_state,
                "price_change_threshold_pct": threshold,
            },
        }
    return {"segments": out, "total_125_mc": total_mc, "reference_total_125_mc": total_ref if total_ref > 0 else None}


def relative_strength(segments: Dict[str, Any]) -> Dict[str, Any]:
    mc = {k: segments[k]["aggregate_market_cap_usd"] for k in segments}
    btc = mc.get("BTC")
    pairs = [
        ("BTC", "ETH"), ("BTC", "TOP10_ALT"), ("BTC", "BROAD_ALT_11_125"),
        ("ETH", "TOP10_ALT"), ("TOP10_ALT", "BROAD_ALT_11_125")
    ]
    ratios = {}
    for a, b in pairs:
        if finite_nonnegative(mc.get(a)) and finite_nonnegative(mc.get(b)) and mc[a] > 0 and mc[b] > 0:
            ratios[f"{a}_MC_OVER_{b}_MC"] = mc[a] / mc[b]
        else:
            ratios[f"{a}_MC_OVER_{b}_MC"] = None
    for k in segments:
        ratios[f"{k}_MC_OVER_BTC_MC"] = (mc[k] / btc) if btc and mc[k] >= 0 else None
    return ratios


def composition_compare(current_assets: List[Dict[str, Any]]) -> Dict[str, Any]:
    """Compare current membership with the latest saved U09 snapshot, if any.

    The comparison is informational only. It never changes the current frozen
    snapshot and never mixes old assets into current calculations.
    """
    files = sorted(SNAPSHOT_DIR.glob("U09_*.json"), key=lambda p: p.stat().st_mtime, reverse=True)
    if not files:
        return {"composition_changed": False, "composition_quality": "STABLE", "previous_snapshot": None, "changes": []}
    try:
        old = json.loads(files[0].read_text(encoding="utf-8"))
        old_assets = old.get("universe_snapshot", {}).get("assets", [])
        old_map = {x.get("provider_asset_id"): x.get("provider_rank") for x in old_assets}
        changes = []
        for a in current_assets:
            aid = a.get("provider_asset_id")
            old_rank = old_map.get(aid)
            new_rank = a.get("provider_rank")
            if old_rank is not None and old_rank != new_rank:
                changes.append({"asset_id": aid, "old_rank": old_rank, "new_rank": new_rank})
        old_ids = set(old_map)
        new_ids = {a.get("provider_asset_id") for a in current_assets}
        entered = sorted(new_ids - old_ids)
        exited = sorted(old_ids - new_ids)
        material = len(changes) + len(entered) + len(exited)
        if material == 0:
            q = "STABLE"
        elif material <= 3:
            q = "MINOR_CHANGE"
        elif material <= 10:
            q = "MATERIAL_CHANGE"
        else:
            q = "UNRELIABLE"
        return {"composition_changed": material > 0, "composition_quality": q, "previous_snapshot": files[0].name, "changes": changes, "entered": entered, "exited": exited}
    except Exception as e:
        return {"composition_changed": True, "composition_quality": "UNRELIABLE", "previous_snapshot": files[0].name, "changes": [], "error": str(e)}

# ============================================================
# 7) PARTICIPATION BRAIN
# ============================================================
def participation_brain(seg: Dict[str, Any], rs: Dict[str, Any], composition: Dict[str, Any]) -> Tuple[str, str]:
    if composition.get("composition_quality") == "UNRELIABLE":
        return "UNAVAILABLE", "Composition quality is unreliable."
    broad = seg["BROAD_ALT_11_125"]
    top = seg["TOP10_ALT"]
    btc = seg["BTC"]
    eth = seg["ETH"]
    if any(seg[k]["count"] == 0 for k in seg):
        return "UNAVAILABLE", "Required segment data is unavailable."
    broad_rise = broad["breadth"]["rising_pct"] or 0
    top_rise = top["breadth"]["rising_pct"] or 0
    broad_mc = broad.get("market_cap_change_pct")
    top_mc = top.get("market_cap_change_pct")
    btc_mc = btc.get("market_cap_change_pct")
    eth_mc = eth.get("market_cap_change_pct")

    broad_support = broad_rise >= 55 and broad_mc is not None and broad_mc > 0
    top_support = top_rise >= 55 and top_mc is not None and top_mc > 0
    core_support = ((btc_mc or 0) > 0) or ((eth_mc or 0) > 0)
    broad_weak = broad_rise < 45 and (broad_mc is not None and broad_mc < 0)
    top_weak = top_rise < 45 and (top_mc is not None and top_mc < 0)

    if broad_support and top_support:
        return "BROADENING", "TOP10 and BROAD participation both support expansion of market participation."
    if core_support and broad_weak and top_weak:
        return "CONCENTRATED", "Relative participation is concentrated in the core while TOP10/BROAD breadth is weak."
    if broad_weak and top_weak:
        return "WEAKENING", "Breadth and aggregate movement in broader segments are deteriorating."
    return "MIXED", "Price, market-cap, breadth and relative-strength evidence is not directionally uniform."

# ============================================================
# 8) MESSAGE LAYER
# ============================================================
def fmt_pct(v: Optional[float], *, min_decimals: int = 1, max_decimals: int = 6) -> str:
    """Format a percentage without losing small movements or producing -0.0."""
    if v is None:
        return "N/A"
    try:
        x = float(v)
    except Exception:
        return "N/A"
    if not math.isfinite(x):
        return "N/A"

    # Never expose signed zero caused by formatting/rounding.
    if x == 0.0:
        x = 0.0

    ax = abs(x)
    if ax == 0.0:
        decimals = min_decimals
    elif ax >= 1.0:
        decimals = min_decimals
    else:
        # Keep meaningful leading zeros for small but real movements.
        decimals = max(min_decimals, min(max_decimals, int(-math.floor(math.log10(ax))) + 2))

    text = f"{x:+.{decimals}f}%"
    if text.startswith("-0.") and float(text[:-1]) == 0.0:
        text = text.replace("-0.", "+0.", 1)
    return text


def build_message(result: Dict[str, Any]) -> str:
    """Compact U09 user-facing message. All displayed movements are 24h-relative %.

    PP, current dominance, participation state and diagnostics remain internal.
    """
    if result.get("validation_status") == "DATA_UNAVAILABLE":
        return "U09 | DATA_UNAVAILABLE | No validated market-universe provider snapshot available."

    seg = result["segments"]
    btc_asset = seg["BTC"]["constituents"][0] if seg["BTC"]["constituents"] else None
    eth_asset = seg["ETH"]["constituents"][0] if seg["ETH"]["constituents"] else None
    btc_price = btc_asset.get("price_change_24h_pct") if btc_asset else None
    eth_price = eth_asset.get("price_change_24h_pct") if eth_asset else None

    lines = [
        f"📊 MARKET PARTICIPATION | {CONFIG['message_window_label']}",
        f"₿ BTC Price {fmt_pct(btc_price)} | BTC.D {fmt_pct(seg['BTC']['dominance_change_pct'], min_decimals=2, max_decimals=6)}",
        f"🔷 ETH Price {fmt_pct(eth_price)} | ETH.D {fmt_pct(seg['ETH']['dominance_change_pct'], min_decimals=2, max_decimals=6)}",
        f"🔹 TOP10 ALT MC {fmt_pct(seg['TOP10_ALT']['market_cap_change_pct'])} | TOP10.D {fmt_pct(seg['TOP10_ALT']['dominance_change_pct'], min_decimals=2, max_decimals=6)}",
        f"◈ BROAD 11–125 MC {fmt_pct(seg['BROAD_ALT_11_125']['market_cap_change_pct'])} | BROAD.D {fmt_pct(seg['BROAD_ALT_11_125']['dominance_change_pct'], min_decimals=2, max_decimals=6)}",
    ]
    return "\n".join(lines)

# ============================================================
# 9) SYNTHETIC TEST FACTORY
# ============================================================
def synthetic_assets(mode: str = "broad") -> List[Dict[str, Any]]:
    now = utc_now()
    rows = []
    for rank in range(1, 126):
        if rank == 1:
            name, symbol, mc, price_ch = "Bitcoin", "btc", 1_000_000_000_000.0, 3.0
        elif rank == 2:
            name, symbol, mc, price_ch = "Ethereum", "eth", 450_000_000_000.0, 4.0
        else:
            name, symbol, mc = f"Asset {rank}", f"a{rank}", max(1.0, 100_000_000_000.0 / rank)
            if mode == "broad":
                price_ch = 3.0 if rank % 2 else 1.0
            elif mode == "btc":
                price_ch = 3.0 if rank <= 2 else -1.5
            elif mode == "eth":
                price_ch = 2.0 if rank == 1 else (6.0 if rank == 2 else 0.5)
            elif mode == "mixed":
                price_ch = 4.0 if rank % 3 == 0 else (-3.0 if rank % 3 == 1 else 0.0)
            elif mode == "falling":
                price_ch = -3.0
            else:
                price_ch = 1.0
        # Make calculated rank agree with provider rank by assigning MC strictly descending.
        mc = 1_000_000_000_000.0 / (rank ** 0.85)
        rows.append({
            "provider_asset_id": f"synthetic-{rank}",
            "symbol": symbol,
            "name": name,
            "price_usd": 100.0 / rank,
            "market_cap_usd": mc,
            "provider_rank": rank,
            "price_change_24h_pct": price_ch,
            "market_cap_change_24h_pct": price_ch,
            "provider_timestamp": iso(now),
        })
    return rows


def synthetic_validation_suite() -> Dict[str, Any]:
    tests = []

    # Base universe / segmentation / dominance partition.
    assets = synthetic_assets("broad")
    v = validate_provider_assets(assets, 125)
    assert v["status"] == "PASS", v
    ref = build_reference_from_24h(v["valid_assets"])
    calc = build_segments(v["valid_assets"], ref)
    seg = calc["segments"]
    partition = sum(seg[k]["dominance_pct"] or 0 for k in seg)
    assert abs(partition - 100.0) <= CONFIG["dominance_partition_tolerance_pct"]
    tests.append("rank_identity_segmentation_aggregation_dominance:PASS")

    # Breadth exact 10/5/5.
    bassets = synthetic_assets("broad")
    for i, a in enumerate(bassets[:20]):
        a["price_change_24h_pct"] = 1.0 if i < 10 else (0.0 if i < 15 else -1.0)
    # Test the counting rule directly.
    rising = sum(1 for a in bassets[:20] if a["price_change_24h_pct"] > CONFIG["breadth_threshold_pct"])
    flat = sum(1 for a in bassets[:20] if abs(a["price_change_24h_pct"]) <= CONFIG["breadth_threshold_pct"])
    falling = sum(1 for a in bassets[:20] if a["price_change_24h_pct"] < -CONFIG["breadth_threshold_pct"])
    assert (rising, flat, falling) == (10, 5, 5)
    tests.append("breadth_10_5_5:PASS")

    rs = relative_strength(seg)
    assert rs["BROAD_ALT_11_125_MC_OVER_BTC_MC"] is not None
    tests.append("relative_strength_pairs:PASS")

    # Zero reference safety.
    assert pct_change(100.0, 0.0) is None
    tests.append("zero_reference_guard:PASS")

    # Duplicate ID.
    dup = synthetic_assets("broad")
    dup[1]["provider_asset_id"] = dup[0]["provider_asset_id"]
    assert validate_provider_assets(dup, 125)["status"] == "FAIL"
    tests.append("duplicate_id_rejection:PASS")

    # Duplicate rank.
    dup_rank = synthetic_assets("broad")
    dup_rank[1]["provider_rank"] = 1
    assert validate_provider_assets(dup_rank, 125)["status"] == "FAIL"
    tests.append("duplicate_rank_rejection:PASS")

    # Missing rank.
    missing = synthetic_assets("broad")[:-1]
    assert validate_provider_assets(missing, 125)["status"] == "FAIL"
    tests.append("missing_rank_coverage_rejection:PASS")

    # Invalid numeric.
    bad = synthetic_assets("broad")
    bad[0]["market_cap_usd"] = float("nan")
    assert validate_provider_assets(bad, 125)["status"] == "FAIL"
    tests.append("invalid_numeric_rejection:PASS")

    # Composition movement 10 <-> 11 using a small deterministic synthetic snapshot.
    base = synthetic_assets("broad")
    moved = synthetic_assets("broad")
    moved[9]["provider_rank"], moved[10]["provider_rank"] = 11, 10
    moved_v = validate_provider_assets(moved, 125)
    # A 10↔11 swap is not necessarily a provider-validation failure: it can
    # legitimately appear as a one-rank difference. The analytical layer must
    # detect the membership change rather than reject a valid snapshot.
    assert moved_v["status"] == "PASS"
    assert segment_for_rank(10) == "TOP10_ALT" and segment_for_rank(11) == "BROAD_ALT_11_125"
    tests.append("composition_boundary_rank_validation:PASS")

    # Dominance relative-change semantics: 12.0 -> 12.6 = +5.0%, while pp is +0.6.
    assert abs(dominance_change_pct(12.6, 12.0) - 5.0) < 1e-12
    assert abs((12.6 - 12.0) - 0.6) < 1e-12
    assert "-0.0%" not in fmt_pct(-0.00001, min_decimals=2, max_decimals=6)
    assert "0.00001" in fmt_pct(0.00001, min_decimals=2, max_decimals=6)
    tests.append("dominance_relative_percent_and_precision:PASS")

    # Effective 24h reference timestamp must be exactly one day before provider time.
    ts = "2026-09-02T14:43:00+00:00"
    assert reference_timestamp_from_provider_24h(ts) == "2026-09-01T14:43:00+00:00"
    tests.append("reference_timestamp_24h_semantics:PASS")

    # Message contract: exactly one header + two lines per market block, no state/scenario.
    msg_result = {
        "validation_status": "PASS",
        "segments": {
            "BTC": {"constituents": [{"price_change_24h_pct": -1.0}], "dominance_change_pct": 0.5},
            "ETH": {"constituents": [{"price_change_24h_pct": -2.0}], "dominance_change_pct": 0.5291005291},
            "TOP10_ALT": {"market_cap_change_pct": -0.7, "dominance_change_pct": 0.3},
            "BROAD_ALT_11_125": {"market_cap_change_pct": -1.1, "dominance_change_pct": -0.1},
        },
    }
    m = build_message(msg_result)
    assert "Participation:" not in m and " pp" not in m
    assert len(m.splitlines()) == 5
    tests.append("message_contract_two_lines_per_block:PASS")

    # Reference age states.
    fresh_state, _ = freshness_status(utc_now(), utc_now())
    stale_state, _ = freshness_status(utc_now() - timedelta(hours=3), utc_now())
    assert fresh_state == "FRESH" and stale_state == "STALE"
    tests.append("freshness_states:PASS")

    # Brain output must remain in allowed state set.
    brain = participation_brain(seg, rs, {"composition_quality": "STABLE"})
    assert brain[0] in {"BROADENING", "CONCENTRATED", "MIXED", "WEAKENING", "UNAVAILABLE"}
    tests.append("participation_brain_state:PASS")

    return {"status": "PASS", "tests": tests, "count": len(tests)}

# ============================================================
# 10) REAL EXECUTION
# ============================================================
def execute_u09() -> Dict[str, Any]:
    started = utc_now()
    synthetic = synthetic_validation_suite()
    assert synthetic["status"] == "PASS"

    orch = orchestrate_universe()
    snap: Optional[ProviderSnapshot] = orch.get("snapshot")
    if snap is None:
        result = {
            "engine": "U09_DYNAMIC_MARKET_UNIVERSE",
            "version": "1.1",
            "validation_status": "DATA_UNAVAILABLE",
            "data_quality": "FAIL",
            "confidence": 0.0,
            "provider": None,
            "primary_provider": "COINGECKO",
            "fallback_used": False,
            "fallback_chain": orch.get("provider_meta", {}).get("attempts", []),
            "fallback_reason": "; ".join(orch.get("validation", {}).get("errors", [])),
            "current_timestamp": iso(utc_now()),
            "reference_timestamp": None,
            "universe_snapshot": None,
            "segments": {},
            "participation_state": "UNAVAILABLE",
            "validation": orch.get("validation"),
            "synthetic_tests": synthetic,
            "provider_health": orch.get("health"),
            "safety": {
                "trading": False, "orders": False, "strategy": False, "portfolio_actions": False,
                "capital_flow_claim": False, "altseason_claim": False,
            },
        }
        return result

    assets = snap.raw_assets
    validation = orch["validation"]
    ref_map = build_reference_from_24h(assets)
    analytical = build_segments(assets, ref_map)
    rs = relative_strength(analytical["segments"])
    composition = composition_compare(assets)

    # Quality is downgraded for material composition changes, but not fabricated.
    quality = "HIGH"
    confidence = 1.0
    if snap.fallback_used:
        quality = "MEDIUM"
        confidence -= 0.10
    if composition["composition_quality"] == "MINOR_CHANGE":
        confidence -= 0.05
    elif composition["composition_quality"] == "MATERIAL_CHANGE":
        quality = "LOW"
        confidence -= 0.20
    elif composition["composition_quality"] == "UNRELIABLE":
        quality = "PARTIAL"
        confidence -= 0.45
    if analytical["reference_total_125_mc"] is None:
        # Current data can still be used for structure, but movement metrics are unavailable.
        quality = "PARTIAL" if quality == "HIGH" else quality
        confidence -= 0.20

    state, explanation = participation_brain(analytical["segments"], rs, composition)
    confidence = max(0.0, min(1.0, confidence))

    universe_snapshot = {
        "snapshot_id": "U09-" + utc_now().strftime("%Y%m%dT%H%M%S%fZ"),
        "frozen": True,
        "provider": snap.provider_id,
        "provider_timestamp": snap.provider_timestamp,
        "created_at": snap.response_timestamp,
        "universe_limit": 125,
        "assets": assets,
        "snapshot_hash": sha256_obj(assets),
    }
    result = {
        "engine": "U09_DYNAMIC_MARKET_UNIVERSE",
        "version": "1.1",
        "validation_status": "PASS",
        "data_quality": quality,
        "confidence": confidence,
        "provider": snap.provider_id,
        "primary_provider": snap.primary_provider,
        "fallback_used": snap.fallback_used,
        "fallback_chain": snap.fallback_chain,
        "fallback_reason": snap.fallback_reason,
        "provider_metadata": {
            "provider_status": snap.provider_status,
            "provider_timestamp": snap.provider_timestamp,
            "request_timestamp": snap.request_timestamp,
            "response_timestamp": snap.response_timestamp,
            "provider_schema_version": snap.provider_schema_version,
            "provider_endpoint": snap.provider_endpoint,
            "provider_request_status": snap.provider_request_status,
            "attempts": snap.attempts,
        },
        "current_timestamp": snap.response_timestamp,
        "reference_timestamp": reference_timestamp_from_provider_24h(snap.provider_timestamp) if CONFIG["reference_mode"] == "provider_24h" else None,
        "universe_snapshot": universe_snapshot,
        "btc_metrics": analytical["segments"]["BTC"],
        "eth_metrics": analytical["segments"]["ETH"],
        "top10_metrics": analytical["segments"]["TOP10_ALT"],
        "broad_metrics": analytical["segments"]["BROAD_ALT_11_125"],
        "segments": analytical["segments"],
        "total_125_mc": analytical["total_125_mc"],
        "reference_total_125_mc": analytical["reference_total_125_mc"],
        "dominance_metrics": {k: {
            "current_pct": v["dominance_pct"],
            "reference_pct": v["reference_dominance_pct"],
            "change_pct": v["dominance_change_pct"],
            "change_pp": v["dominance_change_pp"],
        } for k, v in analytical["segments"].items()},
        "breadth_metrics": {k: v["breadth"] for k, v in analytical["segments"].items()},
        "relative_strength_metrics": rs,
        "composition_status": composition,
        "participation_state": state,
        "participation_explanation": explanation,
        "synthetic_tests": synthetic,
        "validation": validation,
        "provider_health": {pid: asdict(v["health"]) for pid, v in PROVIDER_REGISTRY.items()},
        "internal_diagnostics": {
            "reference_mode": CONFIG["reference_mode"],
            "reference_lookback": CONFIG["lookback"],
            "reference_timestamp_semantics": "provider_timestamp_minus_24h",
            "reference_coverage_ratio": len(ref_map) / 125.0,
            "coverage_ratio": validation.get("coverage_ratio"),
            "rank_status_counts": {
                s: sum(1 for a in assets if a.get("rank_consistency_status") == s)
                for s in ("MATCH", "MINOR_DIFFERENCE", "WARNING", "FAIL")
            },
            "composition_quality": composition.get("composition_quality"),
        },
        "safety": {
            "trading": False, "orders": False, "strategy": False, "portfolio_actions": False,
            "capital_flow_claim": False, "altseason_claim": False,
        },
    }
    result["message"] = build_message(result)
    result["execution_seconds"] = max(0.0, (utc_now() - started).total_seconds())
    return result

# ============================================================
# 11) AUDIT / SNAPSHOT
# ============================================================
def audit_result(result: Dict[str, Any]) -> Dict[str, Any]:
    issues = []
    if result.get("safety", {}).get("trading") is not False:
        issues.append("trading_lock")
    if result.get("safety", {}).get("orders") is not False:
        issues.append("orders_lock")
    if result.get("safety", {}).get("strategy") is not False:
        issues.append("strategy_lock")
    if result.get("safety", {}).get("portfolio_actions") is not False:
        issues.append("portfolio_lock")
    text = str(result.get("message", "")).lower()
    for banned in ("capital flow", "money flow", "capital inflow", "capital outflow", "altseason"):
        if banned in text:
            issues.append("banned_language:" + banned)
    if result.get("validation_status") == "PASS":
        snap = result.get("universe_snapshot") or {}
        msg = result.get("message", "")
        if "Participation:" in msg:
            issues.append("message_participation_state_should_be_internal")
        if " pp" in msg:
            issues.append("message_pp_should_be_internal")
        if "-0.0%" in msg or "+-0.0%" in msg:
            issues.append("message_signed_zero")
        if len(msg.splitlines()) != 5:
            issues.append("message_line_contract")
        if not snap.get("frozen"):
            issues.append("snapshot_not_frozen")
        if len(snap.get("assets", [])) != 125:
            issues.append("snapshot_count")
    return {"status": "PASS" if not issues else "FAIL", "issues": issues}


def persist_result(result: Dict[str, Any], audit: Dict[str, Any]) -> Tuple[Path, Path]:
    stamp = utc_now().strftime("%Y%m%dT%H%M%S%fZ")
    result_path = SNAPSHOT_DIR / f"U09_{stamp}.json"
    audit_path = AUDIT_DIR / f"U09_AUDIT_{stamp}.json"
    write_json(result_path, result)
    write_json(audit_path, audit)
    return result_path, audit_path

# ============================================================
# 12) RUN
# ============================================================
U09_RESULT = execute_u09()
U09_AUDIT = audit_result(U09_RESULT)

# Persist even DATA_UNAVAILABLE results: the failure state is itself auditable.
U09_RESULT["audit"] = U09_AUDIT
U09_RESULT["technical_lock_candidate"] = (
    U09_RESULT.get("synthetic_tests", {}).get("status") == "PASS"
    and U09_AUDIT.get("status") == "PASS"
    and U09_RESULT.get("validation_status") == "PASS"
)
RESULT_PATH, AUDIT_PATH = persist_result(U09_RESULT, U09_AUDIT)

# ============================================================
# 13) FINAL CONSOLE REPORT
# ============================================================
print("=" * 72)
print("KITCHEN ASSISTANT V3.1 — U09")
print("DYNAMIC MARKET UNIVERSE / MARKET PARTICIPATION ENGINE")
print("Version                 : 1.1")
print("Primary Provider        : COINGECKO")
print("Secondary Provider      : COINMARKETCAP")
print("Provider Used           :", U09_RESULT.get("provider"))
print("Fallback Used           :", U09_RESULT.get("fallback_used"))
print("Validation Status       :", U09_RESULT.get("validation_status"))
print("Data Quality            :", U09_RESULT.get("data_quality"))
print("Participation            :", U09_RESULT.get("participation_state"))
print("Synthetic Tests         :", U09_RESULT.get("synthetic_tests", {}).get("status"))
print("Audit                   :", U09_AUDIT.get("status"))
print("Technical Lock Candidate:", U09_RESULT.get("technical_lock_candidate"))
print("Snapshot                :", RESULT_PATH)
print("Audit                   :", AUDIT_PATH)
print("-" * 72)
print(U09_RESULT.get("message", "U09 | DATA_UNAVAILABLE"))
print("=" * 72)

# Final hard assertions: never claim technical lock on an invalid execution.
assert U09_RESULT["synthetic_tests"]["status"] == "PASS"
assert U09_AUDIT["status"] == "PASS"
assert U09_RESULT["safety"]["trading"] is False
assert U09_RESULT["safety"]["orders"] is False
assert U09_RESULT["safety"]["strategy"] is False
assert U09_RESULT["safety"]["portfolio_actions"] is False
```
